## 🔍 Patrón 1: Exploración de Datos (EDA)

### Template Base
```
Explora la tabla/DataFrame [NOMBRE].
Muestra:
- Dimensiones (filas x columnas)
- Tipos de datos
- Valores nulos por columna
- Estadísticas descriptivas
- Primeras 5 filas de muestra
```

### Ejemplo Real: Tabla de Clientes
```
Explora la tabla 'clientes_corporativos'.
Muestra:
- Total de clientes
- Distribución por industria
- Rango de ingresos anuales
- Clientes con datos faltantes en 'email'
- Top 5 clientes por revenue
```

### Variantes Comunes

**Para Data Quality:**
```
Analiza la calidad de datos en 'ventas_2024':
- % de valores nulos por columna
- Duplicados (basado en 'id_transaccion')
- Outliers en 'monto_venta' (método IQR)
- Valores fuera de rango en 'descuento' (debe ser 0-100)
```

**Para Entendimiento de Negocio:**
```
Crea un reporte ejecutivo de la tabla 'inventario':
- Total de productos
- Categorías con más SKUs
- Productos con stock < 10 unidades
- Valor total del inventario
- Producto más caro y más barato
```

## 📊 Patrón 2: Cálculo de Métricas y KPIs

### Template Base
```
Usando [TABLA/DF], calcula:
- [METRICA_1]: [definición]
- [METRICA_2]: [definición]
Agrupa por [DIMENSION].
Ordena por [CRITERIO].
```

### Ejemplo Real: Métricas de Marketing
```
Usando la tabla 'campanas_marketing', calcula:
- CAC (Customer Acquisition Cost): costo_campana / nuevos_clientes
- ROAS (Return on Ad Spend): ingresos_generados / costo_campana
- CTR (Click-Through Rate): clicks / impresiones * 100
Agrupa por canal (email, social, paid_search).
Ordena por ROAS descendente.
Muestra solo canales con ROAS > 2.0
```

### Métricas Financieras
```
Calcula el EBITDA mensual de 'estados_financieros':
- Ingresos: suma de 'revenue'
- COGS: suma de 'cost_of_goods'
- Gastos operativos: suma de 'opex'
- EBITDA = Ingresos - COGS - Gastos operativos
Agrupa por mes del año 2024.
Crea un gráfico waterfall mostrando la evolución mes a mes.
```

### Métricas SaaS
```
Calcula métricas SaaS de 'subscripciones':
- MRR (Monthly Recurring Revenue): suma de 'plan_price' donde status='activo'
- Churn Rate: (cancelaciones del mes / usuarios iniciales del mes) * 100
- ARPU: MRR / total de usuarios activos
- LTV: ARPU / Churn Rate
Compara mes actual vs mes anterior.
Highlight cambios > 10%.
```

## 🔗 Patrón 3: Joins y Combinación de Tablas

### Template Base
```
Combina [TABLA_1] con [TABLA_2] usando [TIPO_JOIN].
Clave: [COLUMNA_COMUN].
Selecciona: [COLUMNAS_FINALES].
Filtra: [CONDICIONES].
```

### Ejemplo Real: Análisis de Pedidos
```
Combina 'pedidos' con 'clientes' usando LEFT JOIN.
Clave: pedidos.customer_id = clientes.id
Además, combina con 'productos' usando LEFT JOIN.
Clave: pedidos.product_id = productos.sku

Selecciona:
- pedidos.order_id
- clientes.nombre
- clientes.segmento
- productos.categoria
- pedidos.cantidad
- pedidos.precio_total

Filtra solo pedidos de Q4 2024.
Agrupa por segmento de cliente y categoría de producto.
Muestra el total de ventas por combinación.
```

### Join con Verificación de Calidad
```
Combina 'transacciones' con 'usuarios' (INNER JOIN en user_id).
Antes de combinar:
- Verifica duplicados en ambas tablas por las claves
- Identifica registros en 'transacciones' sin match en 'usuarios'
- Muestra ¿cuántos registros se perderían con INNER vs LEFT JOIN?
Después, procede con el join que preserve todas las transacciones.
```

## 🎯 Patrón 4: Filtrado Complejo y Segmentación

### Template Base
```
Filtra [TABLA/DF] donde:
- [CONDICION_1]
- [CONDICION_2]
- [CONDICION_3]
Retorna: [CAMPOS]
```

### Segmentación RFM (Recency, Frequency, Monetary)
```
Segmenta la tabla 'clientes' usando análisis RFM:

1. Recency: días desde última compra
   - Alta: < 30 días
   - Media: 30-90 días  
   - Baja: > 90 días

2. Frequency: número de compras en últimos 12 meses
   - Alta: >= 10 compras
   - Media: 5-9 compras
   - Baja: < 5 compras

3. Monetary: total gastado en últimos 12 meses
   - Alto: >= $5,000
   - Medio: $1,000-$4,999
   - Bajo: < $1,000

Crea una columna 'segmento' combinando los tres:
Ejemplo: 'Alta_Recency_Alta_Frequency_Alto_Monetary' = 'Champions'

Muestra la distribución de clientes por segmento con gráfico de barras.
```

### Filtrado de Series de Tiempo
```
De la tabla 'ventas_diarias', filtra:
- Solo días laborables (lunes a viernes)
- Excluye festivos de [lista_festivos]
- Solo transacciones con monto > percentil 25 (eliminar ruido)
- Periodo: Q4 2023 y Q4 2024 (para comparación YoY)

Calcula las ventas promedio diarias para cada Q4.
Muestra el % de cambio.
```

## 📈 Patrón 5: Análisis de Series de Tiempo

### Template Base
```
Analiza la serie de tiempo en [TABLA] con columna fecha [COLUMNA_FECHA].
Métrica: [METRICA].
Calcula:
- Tendencia: [método]
- Estacionalidad: [periodo]
- Comparaciones: [YoY/MoM/WoW]
```

### Ejemplo: Análisis de Revenue
```
Analiza el revenue mensual de 'ventas' (2022-2024).

Calcula:
1. Revenue total por mes
2. Promedio móvil de 3 meses (suavizar volatilidad)
3. Crecimiento MoM (Month-over-Month): (mes_actual - mes_anterior) / mes_anterior
4. Crecimiento YoY (Year-over-Year): comparar mismo mes año anterior
5. Detecta meses atípicos (outliers) usando desviación estándar

Visualiza todo en un gráfico de líneas múltiple:
- Línea azul: revenue real
- Línea roja: promedio móvil
- Banda gris: rango de confianza ± 1 std
```

### Detección de Patrones
```
Identifica patrones estacionales en 'trafico_web':
- Agrupa por día de la semana (lun-dom)
- Agrupa por hora del día (0-23)
- Crea un heatmap 2D: día x hora
- Valores: promedio de visitas
- Destaca las celdas con tráfico > promedio global + 1 std

Respuesta esperada: ¿Cuándo deberíamos programar mantenimiento? (horarios de bajo tráfico)
```

## 🎨 Patrón 6: Generación de Visualizaciones

### Template Base
```
Crea un [TIPO_GRAFICO] de [DATOS].
Eje X: [VARIABLE_X]
Eje Y: [VARIABLE_Y]
Color/Agrupación: [DIMENSION]
Estilo: [PREFERENCIAS]
```

### Dashboard Ejecutivo Completo
```
Crea un dashboard de ventas con 4 visualizaciones:

1. **KPI Cards** (arriba):
   - Revenue Total (número grande + % cambio vs mes anterior)
   - Número de Pedidos (número + % cambio)
   - Ticket Promedio (número + % cambio)
   - Tasa de Conversión (número + % cambio)

2. **Gráfico de Líneas** (izquierda):
   - Título: "Revenue Mensual vs Objetivo"
   - Línea azul sólida: revenue real
   - Línea verde punteada: objetivo
   - Periodo: últimos 12 meses

3. **Gráfico de Barras Horizontal** (derecha arriba):
   - Título: "Top 10 Productos por Ventas"
   - Color degradado según valor
   - Etiquetas con valores formateados ($XX,XXX)

4. **Heatmap** (derecha abajo):
   - Título: "Ventas por Región y Categoría"
   - Filas: regiones
   - Columnas: categorías de producto
   - Escala: verde (bajo) a rojo (alto)

Usa Plotly. Estilo corporativo: colores #2C3E50, #E74C3C, #3498DB.
```

### Waterfall Chart (P&L)
```
Crea un gráfico waterfall del P&L trimestral:

Datos de entrada:
- Ingresos: $500,000 (barra verde, hacia arriba)
- COGS: -$200,000 (barra roja, hacia abajo)
- Margen Bruto: $300,000 (barra azul, total parcial)
- Gastos de Ventas: -$80,000 (roja)
- Gastos Administrativos: -$60,000 (roja)
- EBITDA: $160,000 (barra azul, total parcial)
- Depreciación: -$20,000 (roja)
- Utilidad Neta: $140,000 (barra verde oscura, total final)

Conectores entre barras. Etiquetas con valores. Título: "P&L Q4 2024".
```

## 🏭 Biblioteca de Prompts por Industria

### 💰 Finanzas / Banca
```
# Análisis de Cartera de Créditos
"Analiza la cartera de préstamos:
- Segmenta por score crediticio (300-579: malo, 580-669: regular, 670-739: bueno, 740+: excelente)
- Calcula tasa de morosidad (% con atraso > 30 días) por segmento
- Identifica correlación entre monto del préstamo y morosidad
- Proyecta pérdida esperada (morosidad * exposición)
- Visual: scatter plot con tamaño de punto = monto, color = morosidad"

# Detección de Fraude
"Identifica transacciones sospechosas:
- Monto > 3 desviaciones estándar del promedio del usuario
- Transacciones en horario inusual (2am-6am)
- Cambios bruscos de geografía (transacción en 2 países en < 1 hora)
- Patrones de compras inconsistentes con historial
- Marca como 'alto_riesgo', 'medio_riesgo', 'bajo_riesgo'"
```

### 🛍️ Retail / E-commerce
```
# Market Basket Analysis
"Analiza combinaciones frecuentes de productos:
- Encuentra pares de productos comprados juntos (mínimo 50 transacciones)
- Calcula soporte: % de transacciones con ambos productos
- Calcula confianza: P(producto_B | producto_A)
- Calcula lift: confianza / P(producto_B)
- Recomienda 5 bundles con mayor lift > 1.5
- Estima incremento de revenue si se implementan"

# Optimización de Inventario
"Analiza rotación de inventario:
- Calcula días de inventario: (stock_actual / ventas_promedio_dia)
- Identifica productos con > 90 días (overstocked)
- Identifica productos con < 7 días (riesgo de stockout)
- Sugiere cantidad óptima de reorden por producto
- Visualiza en matriz: rotación vs margen (priorizar acciones)"
```

### 📦 Logística / Supply Chain
```
# Análisis de Eficiencia de Rutas
"Analiza las rutas de entrega:
- Calcula distancia promedio por entrega
- Calcula tiempo promedio por entrega
- Identifica rutas con tiempo/distancia > percentil 75 (ineficientes)
- Agrupa entregas por zona geográfica (usar clustering k-means)
- Sugiere consolidación de rutas para reducir 15% los kms recorridos
- Estima ahorro en combustible ($3.50/galón, rendimiento 8 km/L)"
```

### 🏭 SaaS / Tech
```
# Cohort Retention Analysis
"Analiza retención de usuarios por cohorte de signup:
- Agrupa usuarios por mes de registro (cohortes)
- Para cada cohorte, calcula % de usuarios activos en mes 1, 2, 3, ..., 12
- Crea heatmap de retención: filas = cohortes, columnas = meses desde signup
- Identifica el mes con mayor churn (mayor caída %)
- Calcula LTV promedio por cohorte (asumiendo ARPU = $50/mes)
- Compara cohortes pre y post lanzamiento de feature X"
```
