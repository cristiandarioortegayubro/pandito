````python
import pandas as pd
import numpy as np

print("📊 DETECCIÓN DE OUTLIERS CON IQR (Rango Intercuartílico)")
print("="*70)

# Dataset de salarios con outliers
df = pd.DataFrame({
    'Empleado': ['E1', 'E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10'],
    'Salario': [45000, 48000, 50000, 52000, 55000, 58000, 60000, 62000, 65000, 500000]
})

print("\n📄 Salarios de empleados:")
print(df)
print(f"\nPromedio: ${df['Salario'].mean():,.0f}")
print(f"Mediana: ${df['Salario'].median():,.0f}")
print("⚠️  E10 tiene salario atípico (CEO?)")

print("\n" + "="*70)
print("\n1️⃣  MÉTODO IQR (Interquartile Range)")
print("\nFórmula:")
print("  Q1 = percentil 25")
print("  Q3 = percentil 75")
print("  IQR = Q3 - Q1")
print("  Límite inferior = Q1 - 1.5 * IQR")
print("  Límite superior = Q3 + 1.5 * IQR")

Q1 = df['Salario'].quantile(0.25)
Q3 = df['Salario'].quantile(0.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

print(f"\nCálculos:")
print(f"  Q1 (percentil 25): ${Q1:,.0f}")
print(f"  Q3 (percentil 75): ${Q3:,.0f}")
print(f"  IQR: ${IQR:,.0f}")
print(f"  Límite inferior: ${limite_inferior:,.0f}")
print(f"  Límite superior: ${limite_superior:,.0f}")

print("\n" + "-"*70)
print("\n2️⃣  DETECTAR OUTLIERS")
outliers = df[(df['Salario'] < limite_inferior) | (df['Salario'] > limite_superior)]
print(f"\nOutliers detectados: {len(outliers)}")
if len(outliers) > 0:
    print(outliers)

print("\n" + "-"*70)
print("\n3️⃣  VISUALIZACIÓN CON BOXPLOT (conceptual)")
print("\nBoxplot muestra:")
print("  - Caja: Q1 a Q3 (50% central de datos)")
print("  - Línea en caja: Mediana")
print("  - Bigotes: Hasta 1.5*IQR")
print("  - Puntos fuera: Outliers")

print("\n" + "="*70)
print("✅ Detección IQR completada")
````

````python
import pandas as pd
import numpy as np
from scipy import stats

print("📉 DETECCIÓN DE OUTLIERS CON Z-SCORE")
print("="*70)

df = pd.DataFrame({
    'Empleado': ['E1', 'E2', 'E3', 'E4', 'E5', 'E6', 'E7', 'E8', 'E9', 'E10'],
    'Salario': [45000, 48000, 50000, 52000, 55000, 58000, 60000, 62000, 65000, 500000]
})

print("\n📄 Salarios:")
print(df[['Empleado', 'Salario']])

print("\n" + "="*70)
print("\n1️⃣  MÉTODO Z-SCORE (estandarización)")
print("\nFórmula:")
print("  Z = (X - media) / desviación_estándar")
print("\nInterpretación:")
print("  |Z| > 3   : Outlier extremo")
print("  |Z| > 2   : Outlier moderado")
print("  |Z| < 2   : Normal")

media = df['Salario'].mean()
std = df['Salario'].std()

print(f"\nEstadísticas:")
print(f"  Media: ${media:,.0f}")
print(f"  Desviación estándar: ${std:,.0f}")

print("\n" + "-"*70)
print("\n2️⃣  CALCULAR Z-SCORE")
df['Z_Score'] = (df['Salario'] - media) / std
print(df[['Empleado', 'Salario', 'Z_Score']])

print("\n" + "-"*70)
print("\n3️⃣  DETECTAR OUTLIERS (umbral |Z| > 2)")
UMBRAL = 2
outliers = df[np.abs(df['Z_Score']) > UMBRAL]
print(f"\nOutliers detectados: {len(outliers)}")
if len(outliers) > 0:
    print(outliers[['Empleado', 'Salario', 'Z_Score']])
    print(f"\n⚠️  {outliers['Empleado'].values[0]} está a {outliers['Z_Score'].values[0]:.1f} desviaciones del promedio")

print("\n" + "="*70)
print("✅ Detección Z-Score completada")
````

````python
import pandas as pd
import numpy as np

print("🛠️ MÉTODOS DE TRATAMIENTO DE OUTLIERS")
print("="*70)

df = pd.DataFrame({
    'Salario': [45000, 48000, 50000, 52000, 55000, 58000, 60000, 62000, 65000, 500000]
})

print("\n📄 Salarios originales:")
print(df['Salario'].values)
print(f"Media: ${df['Salario'].mean():,.0f}")
print(f"Mediana: ${df['Salario'].median():,.0f}")

print("\n" + "="*70)
print("\n1️⃣  MÉTODO: ELIMINACIÓN")

Q1 = df['Salario'].quantile(0.25)
Q3 = df['Salario'].quantile(0.75)
IQR = Q3 - Q1
limite_superior = Q3 + 1.5 * IQR

df_eliminado = df[df['Salario'] <= limite_superior].copy()
print(f"\nAntes: {len(df)} filas")
print(f"Después: {len(df_eliminado)} filas")
print(f"Media sin outlier: ${df_eliminado['Salario'].mean():,.0f}")

print("\n" + "-"*70)
print("\n2️⃣  MÉTODO: CAPPING (límite duro)")

df_capped = df.copy()
df_capped['Salario'] = df_capped['Salario'].clip(upper=limite_superior)
print(f"\nLímite superior: ${limite_superior:,.0f}")
print(f"Valores después de capping:")
print(df_capped['Salario'].values)
print(f"Media con capping: ${df_capped['Salario'].mean():,.0f}")

print("\n" + "-"*70)
print("\n3️⃣  MÉTODO: WINSORIZE (reemplazo con percentil)")

p95 = df['Salario'].quantile(0.95)
df_winsorized = df.copy()
df_winsorized['Salario'] = df_winsorized['Salario'].apply(lambda x: p95 if x > p95 else x)
print(f"\nPercentil 95: ${p95:,.0f}")
print(f"Valores después de winsorize:")
print(df_winsorized['Salario'].values)
print(f"Media con winsorize: ${df_winsorized['Salario'].mean():,.0f}")

print("\n" + "="*70)
print("\n📊 COMPARACIÓN DE MÉTODOS")
print("-"*70)
print(f"{'Método':<20} {'Media':>15} {'Filas':>10}")
print("-"*70)
print(f"{'Original':<20} ${df['Salario'].mean():>14,.0f} {len(df):>10}")
print(f"{'Eliminación':<20} ${df_eliminado['Salario'].mean():>14,.0f} {len(df_eliminado):>10}")
print(f"{'Capping':<20} ${df_capped['Salario'].mean():>14,.0f} {len(df_capped):>10}")
print(f"{'Winsorize':<20} ${df_winsorized['Salario'].mean():>14,.0f} {len(df_winsorized):>10}")
print("-"*70)

print("\n" + "="*70)
print("✅ Tratamientos completados")
````

````python
import pandas as pd
import numpy as np

print("🔄 CONVERSIÓN DE TIPOS DE DATOS")
print("="*70)

df = pd.DataFrame({
    'ID': ['1', '2', '3', '4', '5'],
    'Fecha': ['2024-01-15', '2024-02-20', '2024-03-10', '2024-04-05', '2024-05-12'],
    'Monto': ['$50,000', '$12,500.50', '$8,900', '$105,000.00', '$75,200'],
    'Activo': ['True', 'False', 'True', 'True', 'False']
})

print("\n📄 Dataset original:")
print(df)
print(f"\nTipos de datos:")
print(df.dtypes)
print("\n⚠️  TODO es 'object' (string) - necesitamos convertir")

print("\n" + "="*70)
print("\n1️⃣  CONVERTIR ID A ENTERO (.astype())")
df['ID'] = df['ID'].astype(int)
print(f"Tipo de ID: {df['ID'].dtype}")
print(f"Ahora podemos: ID + 10 = {(df['ID'] + 10).values}")

print("\n" + "-"*70)
print("\n2️⃣  CONVERTIR FECHA A DATETIME (pd.to_datetime())")
df['Fecha'] = pd.to_datetime(df['Fecha'])
print(f"Tipo de Fecha: {df['Fecha'].dtype}")
print(f"Ahora podemos extraer: año = {df['Fecha'].dt.year.values}")
print(f"                       mes = {df['Fecha'].dt.month.values}")

print("\n" + "-"*70)
print("\n3️⃣  CONVERTIR MONTO A FLOAT (limpiar primero)")
df['Monto'] = df['Monto'].str.replace('$', '', regex=False)
df['Monto'] = df['Monto'].str.replace(',', '', regex=False)
df['Monto'] = pd.to_numeric(df['Monto'])
print(f"Tipo de Monto: {df['Monto'].dtype}")
print(f"Ahora podemos: suma = ${df['Monto'].sum():,.2f}")

print("\n" + "-"*70)
print("\n4️⃣  CONVERTIR ACTIVO A BOOLEANO")
df['Activo'] = df['Activo'].map({'True': True, 'False': False})
print(f"Tipo de Activo: {df['Activo'].dtype}")
print(f"Activos: {df['Activo'].sum()} de {len(df)}")

print("\n" + "="*70)
print("\n✅ TIPOS CORRECTOS")
print(df.dtypes)
print("\n👉 Ahora podemos operar matemáticamente, ordenar fechas, filtrar booleanos")

print("\n" + "="*70)
print("✅ Conversión completada")
````

````python
import pandas as pd
import numpy as np

print("💼 CASO INTEGRADOR: PIPELINE COMPLETO DE LIMPIEZA")
print("="*70)

# Dataset con TODOS los problemas del módulo
df_raw = pd.DataFrame({
    'ID': ['1', '2', '3', '3', '4', '5', '6'],
    'Fecha': ['2024-01-15', '2024-02-20', '2024-03-10', '2024-03-10', None, '2024-05-12', '2024-06-01'],
    'Cliente': ['  ACME CORP  ', 'techstart srl', 'FinPlus SA', 'FinPlus SA', 'DATACO SRL', None, 'LogiExp SA'],
    'Email': ['contacto@acme.com', 'info@techstart', 'ventas@finplus.com', 'ventas@finplus.com', 'admin@dataco.com', 'soporte@logiexp.com', 'ventas@logiexp.com'],
    'Monto': ['$50,000', '$ 12,500', np.nan, '$8,900', '$105,000', '$75,200', '$5,000,000']
})

print("\n📄 DATOS CRUDOS (problemas múltiples):")
print(df_raw)
print(f"\nShape: {df_raw.shape}")

print("\n" + "="*70)
print("\n🔍 DIAGNÓSTICO")
print("-"*70)
print(f"1. Tipos de datos: {df_raw.dtypes.to_dict()}")
print(f"2. Nulos: {df_raw.isnull().sum().sum()}")
print(f"3. Duplicados: {df_raw.duplicated().sum()}")
print(f"4. Espacios en Cliente: Sí")
print(f"5. Email sin validar: Sí")
print(f"6. Montos como string: Sí")
print(f"7. Outliers en Monto: Posiblemente")

print("\n" + "="*70)
print("\n🧹 PIPELINE DE LIMPIEZA (15 PASOS)")
print("-"*70)

df = df_raw.copy()

print("\n1. Eliminar duplicados completos")
df = df.drop_duplicates()
print(f"   Filas: {df_raw.shape[0]} → {df.shape[0]}")

print("\n2. Convertir ID a entero")
df['ID'] = df['ID'].astype(int)

print("\n3. Normalizar Cliente (strip + upper)")
df['Cliente'] = df['Cliente'].str.strip().str.upper()

print("\n4. Eliminar filas donde Cliente es nulo")
df = df.dropna(subset=['Cliente'])
print(f"   Filas: {df.shape[0]}")

print("\n5. Limpiar Monto (quitar $ y ,)")
df['Monto'] = df['Monto'].str.replace('$', '', regex=False)
df['Monto'] = df['Monto'].str.replace(',', '', regex=False)
df['Monto'] = df['Monto'].str.strip()
df['Monto'] = pd.to_numeric(df['Monto'], errors='coerce')

print("\n6. Imputar Monto con mediana")
monto_mediana = df['Monto'].median()
df['Monto'] = df['Monto'].fillna(monto_mediana)
print(f"   Mediana: ${monto_mediana:,.0f}")

print("\n7. Detectar outliers en Monto (IQR)")
Q1 = df['Monto'].quantile(0.25)
Q3 = df['Monto'].quantile(0.75)
IQR = Q3 - Q1
limite_superior = Q3 + 1.5 * IQR
outliers_mask = df['Monto'] > limite_superior
print(f"   Outliers detectados: {outliers_mask.sum()}")

print("\n8. Aplicar capping a outliers")
df.loc[outliers_mask, 'Monto'] = limite_superior
print(f"   Monto máximo: ${df['Monto'].max():,.0f}")

print("\n9. Convertir Fecha a datetime")
df['Fecha'] = pd.to_datetime(df['Fecha'], errors='coerce')

print("\n10. Eliminar filas con Fecha nula")
df = df.dropna(subset=['Fecha'])
print(f"   Filas finales: {df.shape[0]}")

print("\n11. Extraer año y mes")
df['Anio'] = df['Fecha'].dt.year
df['Mes'] = df['Fecha'].dt.month

print("\n12. Validar Email (simplificado)")
df['Email_valido'] = df['Email'].str.contains('@', na=False) & df['Email'].str.contains('\.', na=False)

print("\n13. Ordenar por Fecha")
df = df.sort_values('Fecha').reset_index(drop=True)

print("\n" + "="*70)
print("\n✅ DATOS LIMPIOS (RESULTADO FINAL)")
print("-"*70)
print(df)

print("\n" + "="*70)
print("\n📊 REPORTE FINAL")
print("-"*70)
print(f"Registros originales: {df_raw.shape[0]}")
print(f"Registros finales: {df.shape[0]}")
print(f"Eliminados: {df_raw.shape[0] - df.shape[0]}")
print(f"\nNulos restantes: {df.isnull().sum().sum()}")
print(f"Total facturado: ${df['Monto'].sum():,.0f}")
print(f"Monto promedio: ${df['Monto'].mean():,.0f}")
print(f"Emails válidos: {df['Email_valido'].sum()} de {len(df)}")
print(f"\nRango de fechas: {df['Fecha'].min().date()} a {df['Fecha'].max().date()}")
print(f"Clientes únicos: {df['Cliente'].nunique()}")

print("\n" + "="*70)
print("✅ Pipeline completo de limpieza ejecutado exitosamente")
print("🎓 MÓDULO 04 - LIMPIEZA Y PREPARACIÓN DE DATOS COMPLETADO")
````

### 📊 Pipeline de Limpieza Completo

```python
# 1. Duplicados
df = df.drop_duplicates()

# 2. Nulos críticos
df = df.dropna(subset=['columnas_criticas'])

# 3. Normalizar strings
df['col'] = df['col'].str.strip().str.upper()

# 4. Limpiar y convertir tipos
df['monto'] = pd.to_numeric(df['monto'].str.replace('$', ''), errors='coerce')
df['fecha'] = pd.to_datetime(df['fecha'])

# 5. Imputar nulos por grupo
df['col'] = df.groupby('grupo')['col'].transform(lambda x: x.fillna(x.mean()))

# 6. Tratar outliers
Q1 = df['col'].quantile(0.25)
Q3 = df['col'].quantile(0.75)
IQR = Q3 - Q1
df['col'] = df['col'].clip(upper=Q3 + 1.5*IQR)

# 7. Validar
print(df.info())
print(df.isnull().sum())
```
