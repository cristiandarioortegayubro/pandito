## 📝 Anatomía de un Prompt Efectivo

### 🎯 Estructura Recomendada

```
[CONTEXTO] + [ACCIÓN] + [DETALLES] + [FORMATO DE SALIDA]
```

### Ejemplos Comparativos

#### ❌ Prompt Vago
```
"Necesito analizar ventas"
```
**Problema:** Genie no sabe qué análisis, qué periodo, qué métricas.

#### ✅ Prompt Específico
```
"Analiza las ventas del Q4 2023 de la tabla 'ventas_retail'.
Calcula: total de ingresos, ticket promedio, producto más vendido.
Agrupa por mes y muestra un gráfico de líneas."
```
**Resultado:** Código preciso, ejecutable inmediatamente.

---

### 🔑 Elementos Clave de un Buen Prompt

1. **Datos de Entrada**
   ```
   "Usando la tabla unity_catalog.schema.clientes..."
   "Con este DataFrame llamado 'df_ventas'..."
   "Crea un DataFrame de ejemplo con..."
   ```

2. **Transformaciones**
   ```
   "Filtra las filas donde estado = 'ACTIVO'"
   "Agrupa por región y calcula el promedio"
   "Combina con la tabla de productos usando LEFT JOIN"
   ```

3. **Métricas/Cálculos**
   ```
   "Calcula el ARPU (ingreso promedio por usuario)"
   "Determina el Churn Rate mensual"
   "Computa el EBITDA por sucursal"
   ```

4. **Formato de Salida**
   ```
   "Muestra los top 10 resultados"
   "Genera un gráfico de barras horizontal"
   "Exporta a formato Parquet"
   ```

---

### 💡 Tip Pro: Refinamiento Iterativo

No necesitas un prompt perfecto al primer intento. Genie entiende conversaciones:

```
Prompt 1: "Crea un DataFrame de empleados"
→ Genie genera DataFrame básico

Prompt 2: "Agrega una columna de 'departamento'"
→ Genie modifica el DataFrame anterior

Prompt 3: "Calcula el salario promedio por departamento"
→ Genie añade la agregación

Prompt 4: "Visualízalo en un gráfico de barras"
→ Genie agrega el código de Plotly
```
