## 📚 Teoría: Margen Bruto

### 🎯 Definición

**Margen Bruto** = Ventas Netas - Costo de Ventas (COGS)

**Margen Bruto %** = (Margen Bruto / Ventas Netas) × 100

---

### 📉 Ejemplo Conceptual

**Empresa de Software:**

```
Ventas Netas:        $1,000,000
Costo de Ventas:     $  250,000 (licencias, hosting, soporte)
-----------------------------------
Margen Bruto:        $  750,000
Margen Bruto %:      75%
```

**Interpretación:**
* Por cada $100 de venta, quedan $75 después de costos directos
* Ese $75 debe cubrir: sueldos, marketing, alquiler, etc.

---

### 📈 Componentes

| Concepto | Incluye | Excluye |
|----------|---------|--------|
| **Ventas Netas** | Ventas brutas - Descuentos - Devoluciones | Impuestos (IVA) |
| **Costo de Ventas** | Materias primas, Manufactura, Envío | Sueldos admin, Marketing |
| **Margen Bruto** | Rentabilidad antes de gastos operativos | Intereses, Impuestos |

---

### 🎯 Benchmarks por Industria

| Industria | Margen Bruto % Típico |
|-----------|------------------------|
| Software (SaaS) | 70-90% |
| Retail | 20-40% |
| Manufactura | 25-45% |
| Servicios | 50-70% |
| E-commerce | 30-50% |

**Un margen bruto alto indica:**
* ✅ Producto con alto valor agregado
* ✅ Ventaja competitiva
* ✅ Poder de pricing

**Un margen bruto bajo indica:**
* ⚠️ Costos altos
* ⚠️ Competencia intensa
* ⚠️ Producto comoditizado

````python
import pandas as pd

print("💰 CÁLCULO DE MARGEN BRUTO")
print("="*70)

# Estado de resultados simplificado
ventas = pd.DataFrame({
    'Linea_Producto': ['Software', 'Hardware', 'Servicios'],
    'Ventas_Netas': [1000000, 500000, 300000],
    'Costo_Ventas': [250000, 375000, 180000]
})

print("\n📊 Estado de Resultados (simplificado):")
print(ventas)

print("\n" + "="*70)
print("\n1️⃣  CALCULAR MARGEN BRUTO")
print("-"*70)

ventas['Margen_Bruto'] = ventas['Ventas_Netas'] - ventas['Costo_Ventas']

print("\nCon Margen Bruto:")
print(ventas)

print("\n" + "-"*70)
print("\n2️⃣  CALCULAR MARGEN BRUTO %")
print("-"*70)

ventas['Margen_Bruto_%'] = (ventas['Margen_Bruto'] / ventas['Ventas_Netas'] * 100).round(2)

print("\nCon Margen Bruto %:")
print(ventas)

print("\n" + "-"*70)
print("\n3️⃣  ANÁLISIS DE RENTABILIDAD")
print("-"*70)

ventas_ordenadas = ventas.sort_values('Margen_Bruto_%', ascending=False)
print("\nLíneas ordenadas por rentabilidad:")
print(ventas_ordenadas[['Linea_Producto', 'Margen_Bruto_%', 'Margen_Bruto']])

print("\n👉 INTERPRETACIÓN:")
for idx, row in ventas_ordenadas.iterrows():
    producto = row['Linea_Producto']
    margen_pct = row['Margen_Bruto_%']
    margen_abs = row['Margen_Bruto']
    
    if margen_pct >= 60:
        emoji = "✅"
        status = "EXCELENTE"
    elif margen_pct >= 40:
        emoji = "🟡"
        status = "BUENO"
    else:
        emoji = "⚠️"
        status = "BAJO"
    
    print(f"{emoji} {producto}: {margen_pct}% ({status}) - ${margen_abs:,.0f}")

print("\n" + "-"*70)
print("\n4️⃣  TOTALES CONSOLIDADOS")
print("-"*70)

total_ventas = ventas['Ventas_Netas'].sum()
total_costo = ventas['Costo_Ventas'].sum()
total_margen = ventas['Margen_Bruto'].sum()
total_margen_pct = (total_margen / total_ventas * 100)

print(f"\nVentas Totales:    ${total_ventas:,.0f}")
print(f"Costo Total:       ${total_costo:,.0f}")
print(f"Margen Bruto:      ${total_margen:,.0f}")
print(f"Margen Bruto %:    {total_margen_pct:.2f}%")

print("\n" + "="*70)
print("✅ Margen Bruto calculado")
````

## 📈 Teoría: EBITDA

### 🎯 Definición

**EBITDA** = Earnings Before Interest, Taxes, Depreciation & Amortization

**En español:** Utilidad Antes de Intereses, Impuestos, Depreciación y Amortización

---

### 🧠 ¿Por Qué EBITDA?

**Pregunta clave:** ¿Cuánto efectivo genera mi operación **ANTES** de:
* 💰 Intereses (financiamiento)
* 💵 Impuestos (fisco)
* 🛠️ Depreciación (contable, no es salida de efectivo real)
* 📝 Amortización (contable, no es salida de efectivo real)

**EBITDA mide la capacidad operativa de generar efectivo.**

---

### 📉 Cálculo

**Método 1 (Bottom-Up):**
```
Utilidad Neta
+ Intereses
+ Impuestos
+ Depreciación
+ Amortización
= EBITDA
```

**Método 2 (Top-Down):**
```
Ventas Netas
- Costo de Ventas
- Gastos Operativos (sueldos, marketing, alquiler)
= EBITDA
```

---

### 📊 Ejemplo Práctico

```
Ventas:                $1,000,000
Costo de Ventas:       $  400,000
----------------------------------
Margen Bruto:          $  600,000

Gastos Operativos:
  - Sueldos:           $  250,000
  - Marketing:         $   80,000
  - Alquiler:          $   30,000
  - Otros:             $   40,000
----------------------------------
EBITDA:                $  200,000

Depreciación:          $   50,000
Amortización:          $   10,000
----------------------------------
EBIT:                  $  140,000

Intereses:             $   20,000
Impuestos (30%):       $   36,000
----------------------------------
Utilidad Neta:         $   84,000
```

**Margen EBITDA:** (200,000 / 1,000,000) × 100 = **20%**

---

### 🎯 Benchmarks

| Industria | EBITDA Margin % |
|-----------|----------------|
| Software (SaaS) | 20-40% |
| Retail | 5-10% |
| Manufactura | 10-15% |
| Telecomunicaciones | 30-50% |
| E-commerce | 5-15% |

**Un EBITDA alto indica:**
* ✅ Operación eficiente
* ✅ Buena generación de efectivo
* ✅ Capacidad de pagar deuda

````python
import pandas as pd

print("📈 CÁLCULO DE EBITDA")
print("="*70)

# Estado de resultados completo
estado = pd.DataFrame({
    'Concepto': [
        'Ventas_Netas',
        'Costo_Ventas',
        'Margen_Bruto',
        'Gastos_Sueldos',
        'Gastos_Marketing',
        'Gastos_Alquiler',
        'Gastos_Otros',
        'EBITDA',
        'Depreciacion',
        'Amortizacion',
        'EBIT',
        'Intereses',
        'Utilidad_Antes_Impuestos',
        'Impuestos',
        'Utilidad_Neta'
    ],
    'Monto': [
        1000000,   # Ventas
        -400000,   # Costo de Ventas
        0,         # Calculado
        -250000,   # Sueldos
        -80000,    # Marketing
        -30000,    # Alquiler
        -40000,    # Otros
        0,         # Calculado
        -50000,    # Depreciación
        -10000,    # Amortización
        0,         # Calculado
        -20000,    # Intereses
        0,         # Calculado
        0,         # Calculado
        0          # Calculado
    ]
})

print("\n📄 Estado de Resultados (inicial):")
print(estado)

print("\n" + "="*70)
print("\n1️⃣  CALCULAR MARGEN BRUTO")
print("-"*70)

estado.loc[estado['Concepto'] == 'Margen_Bruto', 'Monto'] = (
    estado.loc[estado['Concepto'] == 'Ventas_Netas', 'Monto'].values[0] + 
    estado.loc[estado['Concepto'] == 'Costo_Ventas', 'Monto'].values[0]
)

margen_bruto = estado.loc[estado['Concepto'] == 'Margen_Bruto', 'Monto'].values[0]
print(f"Margen Bruto: ${margen_bruto:,.0f}")

print("\n" + "-"*70)
print("\n2️⃣  CALCULAR EBITDA")
print("-"*70)

gastos_operativos = (
    estado.loc[estado['Concepto'] == 'Gastos_Sueldos', 'Monto'].values[0] +
    estado.loc[estado['Concepto'] == 'Gastos_Marketing', 'Monto'].values[0] +
    estado.loc[estado['Concepto'] == 'Gastos_Alquiler', 'Monto'].values[0] +
    estado.loc[estado['Concepto'] == 'Gastos_Otros', 'Monto'].values[0]
)

ebitda = margen_bruto + gastos_operativos
estado.loc[estado['Concepto'] == 'EBITDA', 'Monto'] = ebitda

print(f"Margen Bruto:      ${margen_bruto:,.0f}")
print(f"Gastos Operativos: ${gastos_operativos:,.0f}")
print(f"EBITDA:            ${ebitda:,.0f}")

ventas = estado.loc[estado['Concepto'] == 'Ventas_Netas', 'Monto'].values[0]
margen_ebitda = (ebitda / ventas * 100)
print(f"\nMargen EBITDA:     {margen_ebitda:.2f}%")

print("\n" + "-"*70)
print("\n3️⃣  CALCULAR EBIT")
print("-"*70)

depreciacion = estado.loc[estado['Concepto'] == 'Depreciacion', 'Monto'].values[0]
amortizacion = estado.loc[estado['Concepto'] == 'Amortizacion', 'Monto'].values[0]

ebit = ebitda + depreciacion + amortizacion
estado.loc[estado['Concepto'] == 'EBIT', 'Monto'] = ebit

print(f"EBITDA:            ${ebitda:,.0f}")
print(f"Depreciación:      ${depreciacion:,.0f}")
print(f"Amortización:      ${amortizacion:,.0f}")
print(f"EBIT:              ${ebit:,.0f}")

print("\n" + "-"*70)
print("\n4️⃣  CALCULAR UTILIDAD NETA")
print("-"*70)

intereses = estado.loc[estado['Concepto'] == 'Intereses', 'Monto'].values[0]
util_antes_impuestos = ebit + intereses
impuestos = util_antes_impuestos * -0.30  # 30% de impuestos
util_neta = util_antes_impuestos + impuestos

estado.loc[estado['Concepto'] == 'Utilidad_Antes_Impuestos', 'Monto'] = util_antes_impuestos
estado.loc[estado['Concepto'] == 'Impuestos', 'Monto'] = impuestos
estado.loc[estado['Concepto'] == 'Utilidad_Neta', 'Monto'] = util_neta

print(f"EBIT:                        ${ebit:,.0f}")
print(f"Intereses:                   ${intereses:,.0f}")
print(f"Utilidad Antes de Impuestos: ${util_antes_impuestos:,.0f}")
print(f"Impuestos (30%):             ${impuestos:,.0f}")
print(f"Utilidad Neta:               ${util_neta:,.0f}")

print("\n" + "="*70)
print("\n📄 ESTADO DE RESULTADOS COMPLETO")
print("-"*70)
print(estado.to_string(index=False))

print("\n" + "="*70)
print("✅ EBITDA y Estado de Resultados completo")
````

````python
import pandas as pd
import numpy as np

print("💼 CASO INTEGRADOR: ANÁLISIS POR LÍNEA DE PRODUCTO")
print("="*70)

# Datos por línea de producto
lineas = pd.DataFrame({
    'Linea': ['Software', 'Hardware', 'Servicios', 'Consultoría'],
    'Ventas': [1000000, 500000, 300000, 450000],
    'Costo_Ventas': [250000, 375000, 180000, 270000],
    'Gastos_Operativos': [400000, 80000, 60000, 120000]
})

print("\n📊 Datos por línea de producto:")
print(lineas)

print("\n" + "="*70)
print("\n🔢 CÁLCULO DE MÉTRICAS")
print("-"*70)

# Calcular métricas
lineas['Margen_Bruto'] = lineas['Ventas'] - lineas['Costo_Ventas']
lineas['Margen_Bruto_%'] = (lineas['Margen_Bruto'] / lineas['Ventas'] * 100).round(2)

lineas['EBITDA'] = lineas['Margen_Bruto'] - lineas['Gastos_Operativos']
lineas['EBITDA_%'] = (lineas['EBITDA'] / lineas['Ventas'] * 100).round(2)

print("\nMétricas por línea:")
print(lineas[['Linea', 'Ventas', 'Margen_Bruto', 'Margen_Bruto_%', 'EBITDA', 'EBITDA_%']])

print("\n" + "-"*70)
print("\n🏆 RANKING POR MARGEN BRUTO %")
print("-"*70)

ranking_margen = lineas.sort_values('Margen_Bruto_%', ascending=False)
print(ranking_margen[['Linea', 'Margen_Bruto_%', 'Margen_Bruto']])

print("\n👉 Línea más rentable por margen: ", ranking_margen.iloc[0]['Linea'])

print("\n" + "-"*70)
print("\n🏆 RANKING POR EBITDA ABSOLUTO")
print("-"*70)

ranking_ebitda = lineas.sort_values('EBITDA', ascending=False)
print(ranking_ebitda[['Linea', 'EBITDA', 'EBITDA_%']])

print("\n👉 Línea que más genera operativamente: ", ranking_ebitda.iloc[0]['Linea'])

print("\n" + "-"*70)
print("\n📊 TOTALES CONSOLIDADOS")
print("-"*70)

totales = pd.DataFrame({
    'Métrica': ['Ventas Totales', 'Margen Bruto Total', 'EBITDA Total'],
    'Monto': [
        lineas['Ventas'].sum(),
        lineas['Margen_Bruto'].sum(),
        lineas['EBITDA'].sum()
    ]
})

totales['%_Sobre_Ventas'] = (totales['Monto'] / totales.iloc[0]['Monto'] * 100).round(2)

print(totales.to_string(index=False))

print("\n" + "-"*70)
print("\n⚠️  ANÁLISIS CRÍTICO")
print("-"*70)

for idx, row in lineas.iterrows():
    linea = row['Linea']
    margen = row['Margen_Bruto_%']
    ebitda_pct = row['EBITDA_%']
    
    print(f"\n{linea}:")
    
    if margen >= 60:
        print(f"  ✅ Excelente margen bruto: {margen}%")
    elif margen >= 40:
        print(f"  🟡 Buen margen bruto: {margen}%")
    else:
        print(f"  ⚠️ Margen bruto bajo: {margen}%")
    
    if ebitda_pct >= 20:
        print(f"  ✅ Excelente EBITDA: {ebitda_pct}%")
    elif ebitda_pct >= 10:
        print(f"  🟡 Buen EBITDA: {ebitda_pct}%")
    elif ebitda_pct >= 0:
        print(f"  ⚠️ EBITDA bajo: {ebitda_pct}%")
    else:
        print(f"  ❌ EBITDA negativo: {ebitda_pct}% - Línea NO rentable")

print("\n" + "="*70)
print("✅ Análisis de rentabilidad por línea completado")
````
