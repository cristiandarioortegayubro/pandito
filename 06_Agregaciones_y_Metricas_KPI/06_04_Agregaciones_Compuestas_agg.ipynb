## 📚 .agg(): Agregaciones Avanzadas

### 🔧 ¿Qué es .agg()?

**`.agg()`** (aggregate) es el método más flexible de Pandas para aplicar **múltiples funciones de agregación** en una sola llamada.

---

### 📊 Niveles de Complejidad

#### 1️⃣ **Nivel Básico: Una función**

```python
df.groupby('Region')['Ventas'].agg('sum')
# Equivalente a:
df.groupby('Region')['Ventas'].sum()
```

---

#### 2️⃣ **Nivel Intermedio: Múltiples funciones**

```python
df.groupby('Region')['Ventas'].agg(['sum', 'mean', 'count'])
```

**Resultado:**
```
Region    sum      mean    count
Norte     10000    2500    4
Sur       15000    3000    5
```

---

#### 3️⃣ **Nivel Avanzado: Funciones personalizadas**

```python
def rango(x):
    return x.max() - x.min()

df.groupby('Region')['Ventas'].agg(['sum', 'mean', rango])
```

---

#### 4️⃣ **Nivel Experto: Named Aggregations**

```python
df.groupby('Region').agg(
    Total=('Ventas', 'sum'),
    Promedio=('Ventas', 'mean'),
    Desviacion=('Ventas', 'std'),
    Maximo=('Ventas', 'max')
)
```

**Ventaja:** Nombres descriptivos en vez de "sum", "mean"

---

### 🎯 Funciones Disponibles

| Función | Descripción | Ejemplo |
|---------|-------------|----------|
| `'sum'` | Suma | Total ventas |
| `'mean'` | Promedio | Ticket promedio |
| `'median'` | Mediana | Valor central |
| `'std'` | Desviación estándar | Volatilidad |
| `'min'` / `'max'` | Mínimo / Máximo | Rango |
| `'count'` | Conteo | Cantidad de registros |
| `'nunique'` | Valores únicos | Clientes distintos |
| `lambda x: ...` | Personalizada | Cualquier lógica |

---

### 💼 Caso de Uso: Dashboard de KPIs

**Objetivo:** Calcular todas las métricas de un dashboard en una línea.

```python
kpis = df.groupby('Sucursal').agg(
    Ventas_Total=('Ventas', 'sum'),
    Ventas_Promedio=('Ventas', 'mean'),
    Num_Transacciones=('Ventas', 'count'),
    Ticket_Max=('Ventas', 'max'),
    Ticket_Min=('Ventas', 'min'),
    Desviacion=('Ventas', 'std'),
    Clientes_Unicos=('Cliente', 'nunique')
).round(2)
```

**Resultado:** Un DataFrame listo para exportar o visualizar.

---

1. **Conciso:** Muchas métricas en una línea
2. **Legible:** Nombres descriptivos
3. **Rápido:** Optimizado por Pandas
4. **Flexible:** Funciones built-in + personalizadas

## 📦 .agg() con datos reales de Los Andes Market

### 🔧 Aplicando .agg() al dataset real

Los datos de `ventas_mensuales_mendoza_h3` permiten construir KPIs profesionales con `.agg()`:

```python
# KPIs por sucursal en una sola llamada
df_ventas.groupby('sucursal_nombre').agg(
    ventas_total=('ventas', 'sum'),
    ventas_promedio=('ventas', 'mean'),
    ventas_max=('ventas', 'max'),
    meses_activos=('ventas', 'count'),
    variabilidad=('ventas', 'std')
)
```

---

### 📊 Named Aggregations: La mejor práctica

```python
# Sintaxis: nombre_salida=('columna', 'funcion')
df.groupby('zona').agg(
    facturacion=('ventas', 'sum'),
    ticket_prom=('ventas', 'mean'),
    sucursales=('sucursal_id', 'nunique'),
    max_ventas=('ventas', 'max'),
    min_ventas=('ventas', 'min')
)
```

**Ventajas:**
* Nombres descriptivos sin renombrar después
* Diferentes funciones por columna en una sola llamada
* Código más legible que `df.agg({'col': 'sum', 'col2': 'mean'})`

---

### 💡 Pipeline de KPIs con datos reales

1. **KPIs por zona** → .agg() con múltiples métricas
2. **Ranking de sucursales** → .agg() + sort_values
3. **Evolución temporal** → groupby(año) + .agg()
4. **Dashboard ejecutivo** → Named aggregations en un solo DataFrame

````python
import pandas as pd
import numpy as np

print("📦 .agg() CON DATOS REALES DE LOS ANDES MARKET")
print("="*70)

if USAR_DATOS_REALES and df_ventas is not None:
    print("\n1️⃣  .agg() MÚLTIPLE: KPIs por zona")
    print("-"*70)

    kpi_zona = df_ventas.groupby('zona').agg(
        facturacion=('ventas', 'sum'),
        ticket_prom=('ventas', 'mean'),
        ventas_max=('ventas', 'max'),
        ventas_min=('ventas', 'min'),
        sucursales=('sucursal_id', 'nunique'),
        registros=('ventas', 'count'),
        desviacion=('ventas', 'std')
    ).round(0).sort_values('facturacion', ascending=False)
    print("\n   KPIs por zona:")
    print(kpi_zona)

    print("\n" + "="*70)
    print("\n2️⃣  NAMED AGGREGATIONS: Ranking de sucursales")
    print("-"*70)

    ranking = df_ventas.groupby('sucursal_nombre').agg(
        ventas_total=('ventas', 'sum'),
        ventas_promedio=('ventas', 'mean'),
        ventas_maximas=('ventas', 'max'),
        meses_activos=('ventas', 'count'),
        cv=((('ventas', lambda x: (x.std() / x.mean() * 100).round(1))))
    ).round(0).sort_values('ventas_total', ascending=False)
    print("\n   Ranking de sucursales con KPIs múltiples:")
    print(ranking)

    print("\n" + "="*70)
    print("\n3️⃣  FUNCIÓN PERSONALIZADA: Rango y percentiles")
    print("-"*70)

    def p90(x):
        return x.quantile(0.90)
    def p10(x):
        return x.quantile(0.10)

    custom_agg = df_ventas.groupby('zona')['ventas'].agg(
        ['sum', 'mean', 'min', 'max', p10, p90]
    ).round(0)
    print("\n   Agregación con percentiles p10 y p90:")
    print(custom_agg)

    print("\n" + "="*70)
    print("\n4️⃣  EVOLUCIÓN TEMPORAL: .agg() por año")
    print("-"*70)

    evolucion = df_ventas.groupby('año').agg(
        facturacion=('ventas', 'sum'),
        sucursales_activas=('sucursal_id', 'nunique'),
        ticket_promedio=('ventas', 'mean'),
        crecimiento=((('ventas', lambda x: x.pct_change().mean() * 100)))
    ).round(2)
    print("\n   Evolución anual de KPIs:")
    print(evolucion)

    print("\n" + "="*70)
    print("\n5️⃣  DASHBOARD EJECUTIVO: Todo en un solo .agg()")
    print("-"*70)

    dashboard = df_ventas.groupby(['zona', 'año']).agg(
        facturacion=('ventas', 'sum'),
        ticket_prom=('ventas', 'mean'),
        sucursales=('sucursal_id', 'nunique'),
        max_venta=('ventas', 'max'),
        meses_con_datos=('ventas', 'count')
    ).round(0)
    print("\n   📋 Dashboard ejecutivo: Zona × Año")
    print(dashboard)
    print("\n   ✅ Un solo .agg() genera un reporte completo para stakeholders")
else:
    print("⚠️  No hay datos reales disponibles")

print("\n" + "="*70)
````

````python
import pandas as pd
import numpy as np

print("📦 .agg() CON MÚLTIPLES FUNCIONES")
print("="*70)

ventas = pd.DataFrame({
    'Región': ['Norte', 'Norte', 'Norte', 'Sur', 'Sur', 'Sur'],
    'Producto': ['Laptop', 'Mouse', 'Teclado', 'Laptop', 'Mouse', 'Monitor'],
    'Ventas': [80000, 1500, 3500, 75000, 1400, 12000],
    'Unidades': [1, 10, 2, 1, 8, 3]
})

print("\n📊 Datos de ventas:")
print(ventas)

print("\n" + "="*70)
print("\n1️⃣  AGREGACIÓN SIMPLE")
print("-"*70)

resumen1 = ventas.groupby('Región')['Ventas'].sum()
print("\nTotal ventas por región:")
print(resumen1)

print("\n" + "-"*70)
print("\n2️⃣  MÚLTIPLES FUNCIONES EN UNA COLUMNA")
print("-"*70)

resumen2 = ventas.groupby('Región')['Ventas'].agg(['sum', 'mean', 'count', 'min', 'max'])
print("\nMúltiples métricas:")
print(resumen2)

print("\n" + "-"*70)
print("\n3️⃣  FUNCIONES DIFERENTES POR COLUMNA")
print("-"*70)

resumen3 = ventas.groupby('Región').agg({
    'Ventas': ['sum', 'mean'],
    'Unidades': ['sum', 'max'],
    'Producto': 'count'  # Cantidad de productos
}).reset_index()

print("\nAgregaciones personalizadas por columna:")
print(resumen3)

print("\n" + "-"*70)
print("\n4️⃣  NAMED AGGREGATIONS (RECOMENDADO)")
print("-"*70)

resumen4 = ventas.groupby('Región').agg(
    Total_Ventas=('Ventas', 'sum'),
    Promedio_Ventas=('Ventas', 'mean'),
    Total_Unidades=('Unidades', 'sum'),
    Productos_Vendidos=('Producto', 'count'),
    Ticket_Max=('Ventas', 'max')
).reset_index()

print("\nCon nombres descriptivos:")
print(resumen4)
print("\n✅ Columnas claras y legibles")

print("\n" + "="*70)
print("✅ Agregaciones múltiples dominadas")
````

````python
import pandas as pd
import numpy as np

print("🔧 FUNCIONES PERSONALIZADAS")
print("="*70)

ventas = pd.DataFrame({
    'Región': ['Norte', 'Norte', 'Norte', 'Sur', 'Sur', 'Sur', 'Centro', 'Centro'],
    'Ventas': [50000, 30000, 80000, 45000, 90000, 55000, 60000, 70000]
})

print("\n📊 Datos de ventas:")
print(ventas)

print("\n" + "="*70)
print("\n1️⃣  FUNCIÓN PERSONALIZADA: RANGO")
print("-"*70)

def rango(serie):
    """Calcula el rango (max - min)"""
    return serie.max() - serie.min()

resumen1 = ventas.groupby('Región').agg(
    Total=('Ventas', 'sum'),
    Promedio=('Ventas', 'mean'),
    Rango=('Ventas', rango)
).reset_index()

print("\nCon función personalizada:")
print(resumen1)

print("\n" + "-"*70)
print("\n2️⃣  FUNCIÓN LAMBDA")
print("-"*70)

resumen2 = ventas.groupby('Región').agg(
    Total=('Ventas', 'sum'),
    Percentil_75=('Ventas', lambda x: x.quantile(0.75)),
    Coef_Variacion=('Ventas', lambda x: x.std() / x.mean() * 100)
).reset_index()

resumen2['Coef_Variacion'] = resumen2['Coef_Variacion'].round(2)

print("\nCon funciones lambda:")
print(resumen2)
print("\n💡 Coef. de Variación: Mide dispersión relativa")

print("\n" + "-"*70)
print("\n3️⃣  MÚLTIPLES FUNCIONES PERSONALIZADAS")
print("-"*70)

def percentil_25(x):
    return x.quantile(0.25)

def percentil_75(x):
    return x.quantile(0.75)

resumen3 = ventas.groupby('Región').agg(
    Min=('Ventas', 'min'),
    P25=('Ventas', percentil_25),
    Mediana=('Ventas', 'median'),
    P75=('Ventas', percentil_75),
    Max=('Ventas', 'max')
).reset_index()

print("\nEstadística descriptiva completa:")
print(resumen3)

print("\n" + "="*70)
print("✅ Funciones personalizadas dominadas")
````

````python
import pandas as pd
import numpy as np

print("📊 DASHBOARD DE KPIs INTEGRADOR")
print("="*70)

# Dataset realista
np.random.seed(42)
fechas = pd.date_range('2024-01-01', periods=30, freq='D')

ventas = pd.DataFrame({
    'Fecha': np.random.choice(fechas, 200),
    'Región': np.random.choice(['Norte', 'Sur', 'Centro'], 200),
    'Producto': np.random.choice(['Laptop', 'Mouse', 'Teclado', 'Monitor'], 200),
    'Vendedor': np.random.choice(['Juan', 'María', 'Pedro', 'Ana'], 200),
    'Ventas': np.random.randint(1000, 100000, 200),
    'Costo': np.random.randint(500, 70000, 200),
    'Unidades': np.random.randint(1, 10, 200)
})

ventas['Margen'] = ventas['Ventas'] - ventas['Costo']

print(f"\n📊 Dataset: {len(ventas)} transacciones")
print(ventas.head())

print("\n" + "="*70)
print("\n📈 CONSTRUYENDO DASHBOARD COMPLETO DE KPIs")
print("-"*70)

print("\n1️⃣  KPIs POR REGIÓN")
print("-"*70)

kpis_region = ventas.groupby('Región').agg(
    Transacciones=('Fecha', 'count'),
    Total_Ventas=('Ventas', 'sum'),
    Total_Costo=('Costo', 'sum'),
    Total_Margen=('Margen', 'sum'),
    Ticket_Promedio=('Ventas', 'mean'),
    Unidades_Vendidas=('Unidades', 'sum')
).reset_index()

kpis_region['Margen_%'] = (kpis_region['Total_Margen'] / kpis_region['Total_Ventas'] * 100).round(2)
kpis_region['Ticket_Promedio'] = kpis_region['Ticket_Promedio'].round(0)

print(kpis_region)

print("\n" + "-"*70)
print("\n2️⃣  KPIs POR PRODUCTO")
print("-"*70)

kpis_producto = ventas.groupby('Producto').agg(
    Transacciones=('Fecha', 'count'),
    Total_Ventas=('Ventas', 'sum'),
    Total_Margen=('Margen', 'sum'),
    Unidades=('Unidades', 'sum')
).reset_index()

kpis_producto['Margen_%'] = (kpis_producto['Total_Margen'] / kpis_producto['Total_Ventas'] * 100).round(2)
kpis_producto = kpis_producto.sort_values('Total_Ventas', ascending=False)

print(kpis_producto)

print("\n" + "-"*70)
print("\n3️⃣  KPIs POR VENDEDOR")
print("-"*70)

kpis_vendedor = ventas.groupby('Vendedor').agg(
    Transacciones=('Fecha', 'count'),
    Total_Ventas=('Ventas', 'sum'),
    Ticket_Promedio=('Ventas', 'mean'),
    Mejor_Venta=('Ventas', 'max')
).reset_index()

kpis_vendedor['Ticket_Promedio'] = kpis_vendedor['Ticket_Promedio'].round(0)
kpis_vendedor = kpis_vendedor.sort_values('Total_Ventas', ascending=False)

print(kpis_vendedor)

print("\n" + "-"*70)
print("\n4️⃣  KPIs GENERALES")
print("-"*70)

kpis_general = pd.DataFrame({
    'Métrica': [
        'Total Ventas',
        'Total Costo',
        'Total Margen',
        'Margen %',
        'Ticket Promedio',
        'Transacciones',
        'Unidades Vendidas'
    ],
    'Valor': [
        f"${ventas['Ventas'].sum():,.0f}",
        f"${ventas['Costo'].sum():,.0f}",
        f"${ventas['Margen'].sum():,.0f}",
        f"{(ventas['Margen'].sum() / ventas['Ventas'].sum() * 100):.2f}%",
        f"${ventas['Ventas'].mean():,.0f}",
        len(ventas),
        ventas['Unidades'].sum()
    ]
})

print(kpis_general.to_string(index=False))

print("\n" + "="*70)
print("✅ Dashboard completo de KPIs generado")
````
