## 💡 Por Qué Importa

**`.agg()` es la navaja suiza de agregaciones:**

* 📦 Múltiples métricas en una línea
* 🎯 Funciones personalizadas
* 📊 Nombres descriptivos
* ⚡ Performance optimizada

**Dominar .agg() = Reportes profesionales rápidos**

````python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

print("📦 AGREGACIONES COMPUESTAS CON .agg()")
print("="*70)
print("• Múltiples funciones de agregación en una llamada")
print("• Funciones personalizadas")
print("• Named aggregations (nombres descriptivos)")
print("\n🔧 Método avanzado para KPIs")
print("="*70)
print("✅ Librerías cargadas")
````

````python
import pandas as pd
import numpy as np

print("📦 .agg() CON MÚLTIPLES FUNCIONES")
print("="*70)

ventas = pd.DataFrame({
    'Región': ['Norte', 'Norte', 'Norte', 'Sur', 'Sur', 'Sur'],
    'Producto': ['Laptop', 'Mouse', 'Teclado', 'Laptop', 'Mouse', 'Monitor'],
    'Ventas': [80000, 1500, 3500, 75000, 1400, 12000],
    'Unidades': [1, 10, 2, 1, 8, 3]
})

print("\n📊 Datos de ventas:")
print(ventas)

print("\n" + "="*70)
print("\n1️⃣  AGREGACIÓN SIMPLE")
print("-"*70)

resumen1 = ventas.groupby('Región')['Ventas'].sum()
print("\nTotal ventas por región:")
print(resumen1)

print("\n" + "-"*70)
print("\n2️⃣  MÚLTIPLES FUNCIONES EN UNA COLUMNA")
print("-"*70)

resumen2 = ventas.groupby('Región')['Ventas'].agg(['sum', 'mean', 'count', 'min', 'max'])
print("\nMúltiples métricas:")
print(resumen2)

print("\n" + "-"*70)
print("\n3️⃣  FUNCIONES DIFERENTES POR COLUMNA")
print("-"*70)

resumen3 = ventas.groupby('Región').agg({
    'Ventas': ['sum', 'mean'],
    'Unidades': ['sum', 'max'],
    'Producto': 'count'  # Cantidad de productos
}).reset_index()

print("\nAgregaciones personalizadas por columna:")
print(resumen3)

print("\n" + "-"*70)
print("\n4️⃣  NAMED AGGREGATIONS (RECOMENDADO)")
print("-"*70)

resumen4 = ventas.groupby('Región').agg(
    Total_Ventas=('Ventas', 'sum'),
    Promedio_Ventas=('Ventas', 'mean'),
    Total_Unidades=('Unidades', 'sum'),
    Productos_Vendidos=('Producto', 'count'),
    Ticket_Max=('Ventas', 'max')
).reset_index()

print("\nCon nombres descriptivos:")
print(resumen4)
print("\n✅ Columnas claras y legibles")

print("\n" + "="*70)
print("✅ Agregaciones múltiples dominadas")
````

````python
import pandas as pd
import numpy as np

print("🔧 FUNCIONES PERSONALIZADAS")
print("="*70)

ventas = pd.DataFrame({
    'Región': ['Norte', 'Norte', 'Norte', 'Sur', 'Sur', 'Sur', 'Centro', 'Centro'],
    'Ventas': [50000, 30000, 80000, 45000, 90000, 55000, 60000, 70000]
})

print("\n📊 Datos de ventas:")
print(ventas)

print("\n" + "="*70)
print("\n1️⃣  FUNCIÓN PERSONALIZADA: RANGO")
print("-"*70)

def rango(serie):
    """Calcula el rango (max - min)"""
    return serie.max() - serie.min()

resumen1 = ventas.groupby('Región').agg(
    Total=('Ventas', 'sum'),
    Promedio=('Ventas', 'mean'),
    Rango=('Ventas', rango)
).reset_index()

print("\nCon función personalizada:")
print(resumen1)

print("\n" + "-"*70)
print("\n2️⃣  FUNCIÓN LAMBDA")
print("-"*70)

resumen2 = ventas.groupby('Región').agg(
    Total=('Ventas', 'sum'),
    Percentil_75=('Ventas', lambda x: x.quantile(0.75)),
    Coef_Variacion=('Ventas', lambda x: x.std() / x.mean() * 100)
).reset_index()

resumen2['Coef_Variacion'] = resumen2['Coef_Variacion'].round(2)

print("\nCon funciones lambda:")
print(resumen2)
print("\n💡 Coef. de Variación: Mide dispersión relativa")

print("\n" + "-"*70)
print("\n3️⃣  MÚLTIPLES FUNCIONES PERSONALIZADAS")
print("-"*70)

def percentil_25(x):
    return x.quantile(0.25)

def percentil_75(x):
    return x.quantile(0.75)

resumen3 = ventas.groupby('Región').agg(
    Min=('Ventas', 'min'),
    P25=('Ventas', percentil_25),
    Mediana=('Ventas', 'median'),
    P75=('Ventas', percentil_75),
    Max=('Ventas', 'max')
).reset_index()

print("\nEstadística descriptiva completa:")
print(resumen3)

print("\n" + "="*70)
print("✅ Funciones personalizadas dominadas")
````

````python
import pandas as pd
import numpy as np

print("📊 DASHBOARD DE KPIs INTEGRADOR")
print("="*70)

# Dataset realista
np.random.seed(42)
fechas = pd.date_range('2024-01-01', periods=30, freq='D')

ventas = pd.DataFrame({
    'Fecha': np.random.choice(fechas, 200),
    'Región': np.random.choice(['Norte', 'Sur', 'Centro'], 200),
    'Producto': np.random.choice(['Laptop', 'Mouse', 'Teclado', 'Monitor'], 200),
    'Vendedor': np.random.choice(['Juan', 'María', 'Pedro', 'Ana'], 200),
    'Ventas': np.random.randint(1000, 100000, 200),
    'Costo': np.random.randint(500, 70000, 200),
    'Unidades': np.random.randint(1, 10, 200)
})

ventas['Margen'] = ventas['Ventas'] - ventas['Costo']

print(f"\n📊 Dataset: {len(ventas)} transacciones")
print(ventas.head())

print("\n" + "="*70)
print("\n📈 CONSTRUYENDO DASHBOARD COMPLETO DE KPIs")
print("-"*70)

print("\n1️⃣  KPIs POR REGIÓN")
print("-"*70)

kpis_region = ventas.groupby('Región').agg(
    Transacciones=('Fecha', 'count'),
    Total_Ventas=('Ventas', 'sum'),
    Total_Costo=('Costo', 'sum'),
    Total_Margen=('Margen', 'sum'),
    Ticket_Promedio=('Ventas', 'mean'),
    Unidades_Vendidas=('Unidades', 'sum')
).reset_index()

kpis_region['Margen_%'] = (kpis_region['Total_Margen'] / kpis_region['Total_Ventas'] * 100).round(2)
kpis_region['Ticket_Promedio'] = kpis_region['Ticket_Promedio'].round(0)

print(kpis_region)

print("\n" + "-"*70)
print("\n2️⃣  KPIs POR PRODUCTO")
print("-"*70)

kpis_producto = ventas.groupby('Producto').agg(
    Transacciones=('Fecha', 'count'),
    Total_Ventas=('Ventas', 'sum'),
    Total_Margen=('Margen', 'sum'),
    Unidades=('Unidades', 'sum')
).reset_index()

kpis_producto['Margen_%'] = (kpis_producto['Total_Margen'] / kpis_producto['Total_Ventas'] * 100).round(2)
kpis_producto = kpis_producto.sort_values('Total_Ventas', ascending=False)

print(kpis_producto)

print("\n" + "-"*70)
print("\n3️⃣  KPIs POR VENDEDOR")
print("-"*70)

kpis_vendedor = ventas.groupby('Vendedor').agg(
    Transacciones=('Fecha', 'count'),
    Total_Ventas=('Ventas', 'sum'),
    Ticket_Promedio=('Ventas', 'mean'),
    Mejor_Venta=('Ventas', 'max')
).reset_index()

kpis_vendedor['Ticket_Promedio'] = kpis_vendedor['Ticket_Promedio'].round(0)
kpis_vendedor = kpis_vendedor.sort_values('Total_Ventas', ascending=False)

print(kpis_vendedor)

print("\n" + "-"*70)
print("\n4️⃣  KPIs GENERALES")
print("-"*70)

kpis_general = pd.DataFrame({
    'Métrica': [
        'Total Ventas',
        'Total Costo',
        'Total Margen',
        'Margen %',
        'Ticket Promedio',
        'Transacciones',
        'Unidades Vendidas'
    ],
    'Valor': [
        f"${ventas['Ventas'].sum():,.0f}",
        f"${ventas['Costo'].sum():,.0f}",
        f"${ventas['Margen'].sum():,.0f}",
        f"{(ventas['Margen'].sum() / ventas['Ventas'].sum() * 100):.2f}%",
        f"${ventas['Ventas'].mean():,.0f}",
        len(ventas),
        ventas['Unidades'].sum()
    ]
})

print(kpis_general.to_string(index=False))

print("\n" + "="*70)
print("✅ Dashboard completo de KPIs generado")
````

### 🛠️ Guía Rápida de .agg()

**Caso 1: Una función, una columna**
```python
df.groupby('cat')['val'].sum()
```

**Caso 2: Múltiples funciones, una columna**
```python
df.groupby('cat')['val'].agg(['sum', 'mean', 'count'])
```

**Caso 3: Funciones diferentes por columna**
```python
df.groupby('cat').agg({
    'ventas': ['sum', 'mean'],
    'unidades': 'sum'
})
```

**Caso 4: Named aggregations (MEJOR)**
```python
df.groupby('cat').agg(
    Total_Ventas=('ventas', 'sum'),
    Promedio=('ventas', 'mean'),
    Total_Unidades=('unidades', 'sum')
)
```
