## Módulo 6: Agregaciones, Operaciones Grupales y Métricas KPI de Negocio

### Objetivos del Módulo:
1. Aplicar el patrón **Split-Apply-Combine** mediante `groupby()`.
2. Calcular indicadores financieros clave (KPIs): Margen Bruto, EBITDA, ARPU, Churn Rate.
3. Implementar agregaciones compuestas y personalizadas con `.agg()`.


```python
import pandas as pd

ventas_kpi = pd.DataFrame({
    'Linea_Producto': ['Software', 'Software', 'Hardware', 'Hardware', 'Servicios', 'Servicios'],
    'Ventas_Brutas': [500000, 750000, 1200000, 950000, 300000, 450000],
    'Costo_Directo': [100000, 150000, 900000, 700000, 180000, 270000]
})

# Agregación por línea de producto
resumen_kpi = ventas_kpi.groupby('Linea_Producto').agg(
    Ventas_Totales=('Ventas_Brutas', 'sum'),
    Costo_Total=('Costo_Directo', 'sum')
)

resumen_kpi['Margen_Bruto'] = resumen_kpi['Ventas_Totales'] - resumen_kpi['Costo_Total']
resumen_kpi['Margen_Porcentaje'] = (resumen_kpi['Margen_Bruto'] / resumen_kpi['Ventas_Totales']) * 100

print("--- Resumen de Rentabilidad por Línea de Producto ---")
print(resumen_kpi)

```
