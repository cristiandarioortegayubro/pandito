## 📚 El Patrón Split-Apply-Combine

### 🎯 Concepto Fundamental

**GroupBy** implementa el patrón **Split-Apply-Combine**:

```
Datos Originales
    ↓
┌─────────────────────────┐
│  1. SPLIT (Dividir)      │
│  Agrupar por categorías    │
└─────────────────────────┘
    ↓
┌─────────────────────────┐
│  2. APPLY (Aplicar)      │
│  Calcular en cada grupo  │
└─────────────────────────┘
    ↓
┌─────────────────────────┐
│  3. COMBINE (Combinar)   │
│  Unir resultados         │
└─────────────────────────┘
    ↓
Resultado Agregado
```

---

### 📊 Ejemplo Conceptual

**Datos:** Ventas por vendedor
```
Vendedor  | Monto
----------|-------
Juan      | 1000
Juan      | 1500
María     | 2000
María     | 1800
Pedro     | 1200
```

**1. SPLIT:** Dividir por vendedor
```
Grupo Juan:  [1000, 1500]
Grupo María: [2000, 1800]
Grupo Pedro: [1200]
```

**2. APPLY:** Calcular suma en cada grupo
```
Juan  → sum([1000, 1500]) = 2500
María → sum([2000, 1800]) = 3800
Pedro → sum([1200])       = 1200
```

**3. COMBINE:** Unir resultados
```
Vendedor | Total_Ventas
---------|-------------
Juan     | 2500
María    | 3800
Pedro    | 1200
```

### 💻 Sintaxis en Pandas

```python
df.groupby('Vendedor')['Monto'].sum()
#      ↑           ↑         ↑
#   Columna    Columna    Función
#   de grupo   a agregar  agregación
```

````python
import pandas as pd

print("🔢 GROUPBY BÁSICO: PRIMER EJEMPLO")
print("="*70)

# Dataset de ventas
ventas = pd.DataFrame({
    'Vendedor': ['Juan', 'Juan', 'María', 'María', 'Pedro', 'Pedro', 'Juan'],
    'Producto': ['Laptop', 'Mouse', 'Laptop', 'Teclado', 'Laptop', 'Mouse', 'Teclado'],
    'Monto': [80000, 1500, 80000, 3500, 80000, 1500, 3500],
    'Cantidad': [1, 5, 1, 2, 1, 3, 1]
})

print("\n📊 Datos originales:")
print(ventas)
print(f"\nTotal filas: {len(ventas)}")

print("\n" + "="*70)
print("\n1️⃣  AGRUPAMIENTO SIMPLE")
print("-"*70)

# GroupBy por Vendedor
ventas_por_vendedor = ventas.groupby('Vendedor')['Monto'].sum()
print("\nVentas totales por vendedor:")
print(ventas_por_vendedor)
print(f"\nTipo: {type(ventas_por_vendedor)}")
print("➡️  Resultado es una Serie (columna)")

print("\n" + "-"*70)
print("\n2️⃣  CONVERTIR A DATAFRAME")
print("-"*70)

# Convertir a DataFrame
ventas_df = ventas.groupby('Vendedor')['Monto'].sum().reset_index()
print("\nVentas como DataFrame:")
print(ventas_df)
print(f"\nTipo: {type(ventas_df)}")

print("\n" + "-"*70)
print("\n3️⃣  AGREGACIONES MÚLTIPLES")
print("-"*70)

# Varias métricas a la vez
resumen = ventas.groupby('Vendedor')['Monto'].agg(['sum', 'mean', 'count'])
print("\nResumen completo por vendedor:")
print(resumen)

print("\n" + "="*70)
print("✅ GroupBy básico dominado")
````

````python
import pandas as pd

print("📈 AGREGACIONES MÚLTIPLES")
print("="*70)

ventas = pd.DataFrame({
    'Región': ['Norte', 'Norte', 'Sur', 'Sur', 'Centro', 'Centro'],
    'Ventas': [150000, 180000, 120000, 140000, 200000, 190000],
    'Costo': [90000, 108000, 72000, 84000, 120000, 114000],
    'Unidades': [50, 60, 40, 45, 70, 65]
})

print("\n📊 Datos de ventas por región:")
print(ventas)

print("\n" + "="*70)
print("\n1️⃣  MÉTODO 1: Agregar columna por columna")
print("-"*70)

resumen1 = pd.DataFrame({
    'Total_Ventas': ventas.groupby('Región')['Ventas'].sum(),
    'Total_Costo': ventas.groupby('Región')['Costo'].sum(),
    'Total_Unidades': ventas.groupby('Región')['Unidades'].sum()
})

print("\nResumen por región:")
print(resumen1)
print("\n⚠️  Método verboso - repite .groupby() 3 veces")

print("\n" + "-"*70)
print("\n2️⃣  MÉTODO 2: .agg() con diccionario (RECOMENDADO)")
print("-"*70)

resumen2 = ventas.groupby('Región').agg({
    'Ventas': 'sum',
    'Costo': 'sum',
    'Unidades': 'sum'
}).reset_index()

print("\nResumen con .agg():")
print(resumen2)
print("\n✅ Más limpio y eficiente")

print("\n" + "-"*70)
print("\n3️⃣  MÉTODO 3: Múltiples funciones por columna")
print("-"*70)

resumen3 = ventas.groupby('Región').agg({
    'Ventas': ['sum', 'mean', 'max'],
    'Unidades': ['sum', 'mean']
}).reset_index()

print("\nMúltiples métricas:")
print(resumen3)
print("\n📊 Columnas jerárquicas (MultiIndex)")

print("\n" + "-"*70)
print("\n4️⃣  APLANAR COLUMNAS JERÁRQUICAS")
print("-"*70)

resumen3.columns = ['_'.join(col).strip('_') for col in resumen3.columns.values]
print("\nColumnas aplanadas:")
print(resumen3)

print("\n" + "="*70)
print("✅ Agregaciones múltiples dominadas")
````

````python
import pandas as pd

print("🔢 GROUPBY CON MÚLTIPLES COLUMNAS")
print("="*70)

ventas = pd.DataFrame({
    'Región': ['Norte', 'Norte', 'Norte', 'Sur', 'Sur', 'Sur'],
    'Producto': ['Laptop', 'Laptop', 'Mouse', 'Laptop', 'Mouse', 'Mouse'],
    'Trimestre': ['Q1', 'Q2', 'Q1', 'Q1', 'Q1', 'Q2'],
    'Monto': [80000, 85000, 1500, 78000, 1400, 1600],
    'Unidades': [1, 1, 10, 1, 8, 12]
})

print("\n📊 Ventas por Región, Producto y Trimestre:")
print(ventas)

print("\n" + "="*70)
print("\n1️⃣  AGRUPAR POR 1 COLUMNA")
print("-"*70)

grupo1 = ventas.groupby('Región')['Monto'].sum().reset_index()
print("\nVentas por Región:")
print(grupo1)

print("\n" + "-"*70)
print("\n2️⃣  AGRUPAR POR 2 COLUMNAS")
print("-"*70)

grupo2 = ventas.groupby(['Región', 'Producto'])['Monto'].sum().reset_index()
print("\nVentas por Región y Producto:")
print(grupo2)
print("\n📊 Ahora vemos el detalle por producto en cada región")

print("\n" + "-"*70)
print("\n3️⃣  AGRUPAR POR 3 COLUMNAS")
print("-"*70)

grupo3 = ventas.groupby(['Región', 'Producto', 'Trimestre']).agg({
    'Monto': 'sum',
    'Unidades': 'sum'
}).reset_index()

print("\nVentas por Región, Producto y Trimestre:")
print(grupo3)
print("\n📊 Máximo detalle - cada combinación es un grupo")

print("\n" + "-"*70)
print("\n4️⃣  ORDENAR RESULTADOS")
print("-"*70)

grupo3_ordenado = grupo3.sort_values(['Región', 'Monto'], ascending=[True, False])
print("\nOrdenado por Región y Monto (descendente):")
print(grupo3_ordenado)

print("\n" + "="*70)
print("✅ Agrupamiento múltiple dominado")
````

````python
import pandas as pd

print("🔍 FILTRADO POST-AGREGACIÓN")
print("="*70)

ventas = pd.DataFrame({
    'Vendedor': ['Juan', 'Juan', 'María', 'María', 'Pedro', 'Carlos', 'Ana'],
    'Monto': [50000, 30000, 120000, 95000, 25000, 180000, 40000],
    'Unidades': [2, 1, 5, 4, 1, 8, 2]
})

print("\n📊 Ventas individuales:")
print(ventas)

print("\n" + "="*70)
print("\n1️⃣  AGREGACIÓN NORMAL")
print("-"*70)

total_por_vendedor = ventas.groupby('Vendedor').agg({
    'Monto': 'sum',
    'Unidades': 'sum'
}).reset_index()

print("\nTotal por vendedor:")
print(total_por_vendedor)

print("\n" + "-"*70)
print("\n2️⃣  FILTRAR VENDEDORES CON VENTAS > $100K")
print("-"*70)

top_vendedores = total_por_vendedor[total_por_vendedor['Monto'] > 100000]
print("\nVendedores con ventas > $100K:")
print(top_vendedores)
print(f"\nResultado: {len(top_vendedores)} de {len(total_por_vendedor)} vendedores califican")

print("\n" + "-"*70)
print("\n3️⃣  TOP 3 VENDEDORES")
print("-"*70)

top3 = total_por_vendedor.sort_values('Monto', ascending=False).head(3)
print("\nTop 3 vendedores por monto:")
print(top3)

print("\n" + "-"*70)
print("\n4️⃣  CALCULAR % DE PARTICIPACIÓN")
print("-"*70)

total_general = total_por_vendedor['Monto'].sum()
total_por_vendedor['Participacion_%'] = (total_por_vendedor['Monto'] / total_general * 100).round(2)

print("\nParticipación de cada vendedor:")
print(total_por_vendedor.sort_values('Monto', ascending=False))
print(f"\nTotal general: ${total_general:,.0f}")

print("\n" + "="*70)
print("✅ Filtrado post-agregación dominado")
````

````python
import pandas as pd
import numpy as np

print("🔄 TRANSFORMACIONES POR GRUPO")
print("="*70)

ventas = pd.DataFrame({
    'Región': ['Norte', 'Norte', 'Norte', 'Sur', 'Sur', 'Sur'],
    'Vendedor': ['Juan', 'María', 'Pedro', 'Ana', 'Carlos', 'Luis'],
    'Monto': [50000, 80000, 60000, 45000, 90000, 55000]
})

print("\n📊 Ventas por vendedor y región:")
print(ventas)

print("\n" + "="*70)
print("\n1️⃣  CALCULAR PROMEDIO POR REGIÓN")
print("-"*70)

promedio_region = ventas.groupby('Región')['Monto'].transform('mean')
ventas['Promedio_Region'] = promedio_region

print("\nCon promedio de región:")
print(ventas)
print("\n📊 transform() agrega el valor a CADA fila del grupo")

print("\n" + "-"*70)
print("\n2️⃣  COMPARAR CON EL PROMEDIO")
print("-"*70)

ventas['Diferencia_vs_Promedio'] = ventas['Monto'] - ventas['Promedio_Region']
ventas['Performance'] = ventas['Diferencia_vs_Promedio'].apply(
    lambda x: '✅ Arriba' if x > 0 else '⚠️ Abajo' if x < 0 else '= Promedio'
)

print("\nPerformance vs promedio regional:")
print(ventas[['Región', 'Vendedor', 'Monto', 'Promedio_Region', 'Diferencia_vs_Promedio', 'Performance']])

print("\n" + "-"*70)
print("\n3️⃣  CALCULAR % DE PARTICIPACIÓN EN LA REGIÓN")
print("-"*70)

total_region = ventas.groupby('Región')['Monto'].transform('sum')
ventas['Participacion_%'] = (ventas['Monto'] / total_region * 100).round(2)

print("\nParticipación de cada vendedor en su región:")
print(ventas[['Región', 'Vendedor', 'Monto', 'Participacion_%']])

print("\n" + "-"*70)
print("\n4️⃣  RANKING DENTRO DE CADA REGIÓN")
print("-"*70)

ventas['Ranking_Region'] = ventas.groupby('Región')['Monto'].rank(ascending=False, method='dense').astype(int)

print("\nRanking de vendedores por región:")
print(ventas[['Región', 'Vendedor', 'Monto', 'Ranking_Region']].sort_values(['Región', 'Ranking_Region']))

print("\n" + "="*70)
print("✅ Transformaciones por grupo dominadas")
````

````python
import pandas as pd
import numpy as np

print("💼 CASO INTEGRADOR: ANÁLISIS DE VENTAS COMPLETO")
print("="*70)

# Dataset realista de ventas
np.random.seed(42)
fechas = pd.date_range('2024-01-01', periods=100, freq='D')

ventas = pd.DataFrame({
    'Fecha': np.random.choice(fechas, 100),
    'Región': np.random.choice(['Norte', 'Sur', 'Centro'], 100),
    'Producto': np.random.choice(['Laptop', 'Mouse', 'Teclado', 'Monitor'], 100),
    'Vendedor': np.random.choice(['Juan', 'María', 'Pedro', 'Ana', 'Carlos'], 100),
    'Monto': np.random.randint(1000, 100000, 100),
    'Unidades': np.random.randint(1, 10, 100)
})

ventas['Trimestre'] = ventas['Fecha'].dt.to_period('Q').astype(str)

print(f"\n📊 Dataset de ventas: {len(ventas)} transacciones")
print(ventas.head(10))

print("\n" + "="*70)
print("\n🎯 OBJETIVO: Responder 5 preguntas de negocio")
print("-"*70)

print("\n\n1️⃣  ¿CUÁL ES LA REGIÓN MÁS RENTABLE?")
print("-"*70)

ventas_por_region = ventas.groupby('Región').agg({
    'Monto': 'sum',
    'Unidades': 'sum',
    'Fecha': 'count'  # Cantidad de transacciones
}).rename(columns={'Fecha': 'Transacciones'}).reset_index()

ventas_por_region['Ticket_Promedio'] = (ventas_por_region['Monto'] / ventas_por_region['Transacciones']).round(0)
ventas_por_region = ventas_por_region.sort_values('Monto', ascending=False)

print(ventas_por_region)
print(f"\n🏆 Región líder: {ventas_por_region.iloc[0]['Región']} con ${ventas_por_region.iloc[0]['Monto']:,.0f}")

print("\n\n2️⃣  ¿QUÉ PRODUCTO ES EL MÁS VENDIDO POR REGIÓN?")
print("-"*70)

producto_region = ventas.groupby(['Región', 'Producto'])['Unidades'].sum().reset_index()
top_producto = producto_region.loc[producto_region.groupby('Región')['Unidades'].idxmax()]

print(top_producto)

print("\n\n3️⃣  ¿QUIÉN ES EL MEJOR VENDEDOR DE CADA REGIÓN?")
print("-"*70)

vendedor_region = ventas.groupby(['Región', 'Vendedor'])['Monto'].sum().reset_index()
top_vendedor = vendedor_region.loc[vendedor_region.groupby('Región')['Monto'].idxmax()]

print(top_vendedor)

print("\n\n4️⃣  ¿CÓMO EVOLUCIONARON LAS VENTAS POR TRIMESTRE?")
print("-"*70)

ventas_trimestre = ventas.groupby('Trimestre')['Monto'].sum().reset_index()
ventas_trimestre = ventas_trimestre.sort_values('Trimestre')

print(ventas_trimestre)

if len(ventas_trimestre) > 1:
    variacion = ((ventas_trimestre.iloc[-1]['Monto'] - ventas_trimestre.iloc[0]['Monto']) / 
                 ventas_trimestre.iloc[0]['Monto'] * 100)
    print(f"\n📈 Variación: {variacion:+.1f}% respecto al primer trimestre")

print("\n\n5️⃣  ¿QUÉ VENDEDORES ESTÁN POR DEBAJO DEL PROMEDIO?")
print("-"*70)

ventas_vendedor = ventas.groupby('Vendedor')['Monto'].sum().reset_index()
promedio_general = ventas_vendedor['Monto'].mean()
ventas_vendedor['Status'] = ventas_vendedor['Monto'].apply(
    lambda x: '✅ Arriba del promedio' if x > promedio_general else '⚠️ Debajo del promedio'
)

print(ventas_vendedor.sort_values('Monto', ascending=False))
print(f"\nPromedio general: ${promedio_general:,.0f}")

bajo_promedio = ventas_vendedor[ventas_vendedor['Monto'] < promedio_general]
print(f"⚠️  {len(bajo_promedio)} vendedores están debajo del promedio")

print("\n" + "="*70)
print("✅ Análisis completo de ventas finalizado")
````

## 🎓 Conclusiones del Notebook 06_01

### ✅ Lo Que Aprendiste

1. **El Patrón Split-Apply-Combine:**
   - Split: Dividir datos por grupos
   - Apply: Aplicar función a cada grupo
   - Combine: Unir resultados

2. **GroupBy Básico:**
   ```python
   df.groupby('columna')['valor'].sum()
   ```

3. **Agregaciones Múltiples:**
   ```python
   df.groupby('col').agg({
       'val1': 'sum',
       'val2': ['mean', 'max']
   })
   ```

4. **Agrupamiento Múltiple:**
   ```python
   df.groupby(['col1', 'col2', 'col3']).agg(...)
   ```

5. **Transformaciones:**
   - `.transform()`: Agrega valor a cada fila
   - `.rank()`: Ranking dentro de grupos

---

### 📊 Funciones de Agregación Clave

| Función | Descripción | Ejemplo |
|---------|-------------|--------|
| `sum()` | Suma total | Ventas totales |
| `mean()` | Promedio | Ticket promedio |
| `count()` | Cantidad | Nº de transacciones |
| `min()` / `max()` | Mínimo / Máximo | Menor/Mayor venta |
| `std()` | Desviación estándar | Variabilidad |
| `median()` | Mediana | Valor central |

---

### 🛠️ Guía Rápida

**¿Cuándo usar GroupBy?**

✅ "Quiero ver ventas **por** vendedor" → `groupby('Vendedor')`  
✅ "Promedio **por** región y producto" → `groupby(['Región', 'Producto'])`  
✅ "Total **por** mes" → `groupby(df['Fecha'].dt.to_period('M'))`

**¿Cuándo usar .agg()?**

✅ Necesitas múltiples métricas  
✅ Diferentes funciones por columna  
✅ Renombrar columnas de resultado

**¿Cuándo usar .transform()?**

✅ Necesitas agregar el valor grupal a cada fila  
✅ Comparar cada fila con su grupo  
✅ Calcular % de participación
