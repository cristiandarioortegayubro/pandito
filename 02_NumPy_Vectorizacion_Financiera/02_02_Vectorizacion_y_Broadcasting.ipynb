## 2️⃣ Comparación: Loops Python vs NumPy Vectorizado

### 🐢 Python Nativo con Bucles (LENTO)

```python
# Calcular retornos de una acción
precios = [100, 105, 103, 108, 110]
retornos = []

for i in range(1, len(precios)):
    retorno = (precios[i] - precios[i-1]) / precios[i-1] * 100
    retornos.append(retorno)

print(retornos)  # [5.0, -1.9, 4.85, 1.85]
```

❌ **Problemas:**
* Bucle explícito (lento)
* Múltiples líneas de código
* No escala bien a millones de datos
* Interpretado línea por línea

---

### ⚡ NumPy Vectorizado (RÁPIDO)

```python
import numpy as np

# Calcular retornos de una acción
precios = np.array([100, 105, 103, 108, 110])
retornos = (precios[1:] - precios[:-1]) / precios[:-1] * 100

print(retornos)  # [5.0, -1.9, 4.85, 1.85]
```

✅ **Ventajas:**
* Sin bucle explícito
* Una sola línea
* Escalable a millones de datos
* Ejecutado en C (compilado)

---

### 📈 Speedup Real

| Tamaño Datos | Python Nativo | NumPy | Speedup |
|--------------|---------------|-------|----------|
| 1,000 elementos | 0.8 ms | 0.01 ms | **80x** |
| 10,000 elementos | 8.2 ms | 0.05 ms | **164x** |
| 100,000 elementos | 85 ms | 0.3 ms | **283x** |
| 1,000,000 elementos | 920 ms | 2.1 ms | **438x** |

👉 **Conclusión:** Cuantos más datos, mayor el speedup!

````python
import numpy as np
import time

print("🔥 BENCHMARKING: Python Loops vs NumPy Vectorizado")
print("="*70)

# Datos de prueba: 100,000 precios
n = 100_000
precios_lista = list(range(100, 100 + n))
precios_array = np.array(precios_lista)

print(f"\n📊 Dataset: {n:,} precios")

print("\n" + "-"*70)

# Método 1: Python nativo con bucle
print("🐢 Método 1: Python nativo con bucle for")

start = time.time()
retornos_python = []
for i in range(1, len(precios_lista)):
    ret = (precios_lista[i] - precios_lista[i-1]) / precios_lista[i-1] * 100
    retornos_python.append(ret)
tiempo_python = time.time() - start

print(f"Tiempo: {tiempo_python*1000:.2f} ms")
print(f"Primeros 5 retornos: {retornos_python[:5]}")

print("\n" + "-"*70)

# Método 2: NumPy vectorizado
print("⚡ Método 2: NumPy vectorizado (SIN bucles)")

start = time.time()
retornos_numpy = (precios_array[1:] - precios_array[:-1]) / precios_array[:-1] * 100
tiempo_numpy = time.time() - start

print(f"Tiempo: {tiempo_numpy*1000:.2f} ms")
print(f"Primeros 5 retornos: {retornos_numpy[:5]}")

print("\n" + "="*70)

# Comparación
speedup = tiempo_python / tiempo_numpy

print("🏆 RESULTADOS:")
print(f"\n  Python nativo:  {tiempo_python*1000:>8.2f} ms")
print(f"  NumPy:          {tiempo_numpy*1000:>8.2f} ms")
print(f"\n  🚀 Speedup: {speedup:.0f}x más rápido!")

if speedup > 100:
    print("\n  💥 NumPy es más de 100x más rápido!")
elif speedup > 50:
    print("\n  🚀 NumPy es extremadamente rápido!")
else:
    print("\n  ✅ NumPy es significativamente más rápido")

print("\n" + "="*70)
print("⚠️  LECCIÓN: SIEMPRE usa NumPy para operaciones numéricas masivas")
````

## 3️⃣ Broadcasting: Operar Arrays de Diferentes Shapes

### 🔄 ¿Qué es Broadcasting?

Broadcasting permite operar arrays de **diferentes tamaños** sin tener que duplicar manualmente los datos. NumPy "estira" automáticamente el array pequeño para que coincida con el grande.

---

### 📐 Reglas de Broadcasting

**NumPy compara shapes de derecha a izquierda:**

1. Si las dimensiones son iguales → ✅ compatible
2. Si una dimensión es 1 → ✅ compatible (se "estira")
3. Si una dimensión no existe → ✅ se agrega dimensión 1
4. De lo contrario → ❌ incompatible

---

### 📊 Ejemplos Visuales

#### **Caso 1: Escalar + Array**
```python
arr = np.array([1, 2, 3])  # shape (3,)
arr + 10                   # shape (3,)
# Broadcasting: 10 se repite 3 veces → [10, 10, 10]
# Resultado: [11, 12, 13]
```

#### **Caso 2: Vector + Matriz (broadcasting por filas)**
```python
matriz = np.array([[1, 2, 3],    # shape (2, 3)
                   [4, 5, 6]])
vector = np.array([10, 20, 30])  # shape (3,)

matriz + vector  # vector se repite en cada fila
# [[11, 22, 33],
#  [14, 25, 36]]
```

#### **Caso 3: Matriz + Columna (broadcasting por columnas)**
```python
matriz = np.array([[1, 2, 3],     # shape (2, 3)
                   [4, 5, 6]])
columna = np.array([[10],          # shape (2, 1)
                    [20]])

matriz + columna  # columna se repite en cada fila
# [[11, 12, 13],
#  [24, 25, 26]]
```

---

### 💼 Casos Financieros

```python
# Aplicar comisión diferente por accción
precios = np.array([[100, 50, 75],   # 3 días x 3 acciones
                    [105, 52, 78],
                    [103, 51, 76]])

comisiones = np.array([0.98, 0.99, 0.97])  # 2%, 1%, 3%

precios_netos = precios * comisiones  # Broadcasting!
```

````python
import numpy as np

print("🔄 BROADCASTING: Operaciones con Arrays de Diferentes Shapes")
print("="*70)

# Ejemplo 1: Escalar + Vector
print("📈 1. Escalar + Vector:")
precios = np.array([100, 105, 103])
print(f"Precios:           {precios}")
print(f"Precios + 10:      {precios + 10}")
print(f"Precios × 1.21:     {precios * 1.21}")

print("\n" + "="*70)

# Ejemplo 2: Vector + Matriz (broadcasting por filas)
print("📉 2. Vector + Matriz (suma por columnas):")

precios_matriz = np.array([
    [100, 50, 75],   # Día 1
    [105, 52, 78],   # Día 2
    [103, 51, 76]    # Día 3
])

ajuste = np.array([5, -2, 3])  # Ajuste por acción

print("\nPrecios (3 días x 3 acciones):")
print(precios_matriz)
print(f"\nAjuste por acción: {ajuste}")
print("\nPrecios + Ajuste (broadcasting):")
print(precios_matriz + ajuste)
print("\n👉 El vector [5, -2, 3] se suma a CADA fila")

print("\n" + "="*70)

# Ejemplo 3: Matriz + Columna (broadcasting por columnas)
print("📈 3. Matriz + Columna (suma por filas):")

multiplicador_dia = np.array([[1.00],   # Día 1: sin cambio
                               [1.05],   # Día 2: +5%
                               [0.98]])  # Día 3: -2%

print("\nMultiplicador por día:")
print(multiplicador_dia)
print("\nPrecios × Multiplicador (broadcasting):")
print(precios_matriz * multiplicador_dia)
print("\n👉 La columna se multiplica por CADA columna de la matriz")

print("\n" + "="*70)

# Ejemplo 4: Caso financiero real - Comisiones diferenciadas
print("💼 4. CASO PRÁCTICO: Comisiones Diferenciadas por Acción")

precios = np.array([
    [100, 50, 75],
    [105, 52, 78],
    [103, 51, 76],
    [108, 53, 79],
    [110, 54, 80]
])

acciones = ['AAPL', 'GOOGL', 'MSFT']
comisiones = np.array([0.98, 0.99, 0.97])  # 2%, 1%, 3%

print(f"\nAcciones: {acciones}")
print(f"Comisiones: {comisiones} (descuentos: 2%, 1%, 3%)")
print("\nPrecios originales (primeros 3 días):")
print(precios[:3])
print("\nPrecios después de comisiones (broadcasting):")
precios_netos = precios * comisiones
print(precios_netos[:3])

print("\nComisión cobrada por acción (primeros 3 días):")
comision_cobrada = precios - precios_netos
print(comision_cobrada[:3])

print(f"\nComisión total cobrada: ${comision_cobrada.sum():.2f}")
````

## 4️⃣ Boolean Masking: Filtrado Condicional Potente

### 🔍 ¿Qué es Boolean Masking?

Usa **arrays booleanos** (True/False) como "máscaras" para filtrar datos. Es extremadamente útil en análisis financiero.

---

### 📊 Cómo Funciona

```python
precios = np.array([100, 105, 98, 110, 95])

# Paso 1: Crear máscara booleana
mascara = precios > 100
print(mascara)  # [False, True, False, True, False]

# Paso 2: Aplicar máscara
precios_altos = precios[mascara]
print(precios_altos)  # [105, 110]

# O en una sola línea:
precios_altos = precios[precios > 100]
```

---

### 🔥 Operadores Lógicos

| Operador | Python | NumPy | Ejemplo |
|----------|--------|-------|----------|
| AND | `and` | `&` | `(precios > 100) & (precios < 110)` |
| OR | `or` | `\|` | `(precios < 95) \| (precios > 110)` |
| NOT | `not` | `~` | `~(precios > 100)` |

⚠️ **Importante:** Usa `&` y `|`, NO `and` y `or`. Siempre pon cada condición entre paréntesis.

---

### 💼 Casos Financieros

```python
# Filtrar días con retornos positivos
retornos = np.array([0.05, -0.02, 0.03, -0.01, 0.04])
retornos_positivos = retornos[retornos > 0]

# Filtrar precios en un rango
precios_rango = precios[(precios >= 100) & (precios <= 110)]

# Contar cuántos días subieron
dias_subida = (retornos > 0).sum()  # True = 1, False = 0

# Reemplazar valores atípicos
precios[precios > 150] = 150  # Cap a $150
```

````python
import numpy as np

print("🔍 BOOLEAN MASKING: Filtrado Condicional Avanzado")
print("="*70)

# Dataset: Precios de AAPL (10 días)
precios = np.array([100, 105, 103, 108, 110, 107, 112, 115, 113, 118])
retornos = (precios[1:] - precios[:-1]) / precios[:-1] * 100

print("💹 Dataset: Precios de AAPL (10 días)")
print(f"Precios:  {precios}")
print(f"Retornos: {retornos.round(2)}")

print("\n" + "="*70)

# Ejemplo 1: Filtrar precios > $110
print("🔹 1. Filtrar precios > $110:")
mascara = precios > 110
print(f"\nMáscara:         {mascara}")
print(f"Precios > 110:   {precios[mascara]}")
print(f"Cantidad:        {mascara.sum()} días")

print("\n" + "="*70)

# Ejemplo 2: Filtrar retornos positivos
print("🔺 2. Filtrar retornos positivos:")
retornos_positivos = retornos[retornos > 0]
print(f"\nRetornos positivos: {retornos_positivos.round(2)}")
print(f"Días de subida:     {len(retornos_positivos)} / {len(retornos)}")
print(f"Probabilidad:       {len(retornos_positivos) / len(retornos) * 100:.1f}%")

print("\n" + "="*70)

# Ejemplo 3: Múltiples condiciones (AND)
print("🔸 3. Precios entre $105 y $115:")
mascara_rango = (precios >= 105) & (precios <= 115)
precios_rango = precios[mascara_rango]
print(f"\nMáscara:         {mascara_rango}")
print(f"Precios en rango: {precios_rango}")
print(f"Cantidad:         {len(precios_rango)} días")

print("\n" + "="*70)

# Ejemplo 4: Condiciones complejas (OR)
print("🔷 4. Precios extremos (< $105 O > $115):")
mascara_extremos = (precios < 105) | (precios > 115)
precios_extremos = precios[mascara_extremos]
print(f"\nMáscara:           {mascara_extremos}")
print(f"Precios extremos:  {precios_extremos}")
print(f"Cantidad:          {len(precios_extremos)} días")

print("\n" + "="*70)

# Ejemplo 5: Negación (NOT)
print("🔶 5. Precios NO mayores a $110:")
mascara_no_altos = ~(precios > 110)
precios_no_altos = precios[mascara_no_altos]
print(f"\nMáscara:          {mascara_no_altos}")
print(f"Precios <= 110:   {precios_no_altos}")

print("\n" + "="*70)

# Ejemplo 6: CASO PRÁCTICO - Alertas de Trading
print("🚨 6. ALERTAS DE TRADING:")
print("-"*70)

# Condiciones de alerta
alerta_compra = retornos < -2.0   # Caída > 2%
alerta_venta = retornos > 3.0     # Subida > 3%

print(f"\n🟢 Días con señal de COMPRA (caída > 2%):  {alerta_compra.sum()}")
if alerta_compra.any():
    indices_compra = np.where(alerta_compra)[0]
    print(f"   Índices: {indices_compra}")
    print(f"   Retornos: {retornos[alerta_compra].round(2)}%")

print(f"\n🔴 Días con señal de VENTA (subida > 3%):   {alerta_venta.sum()}")
if alerta_venta.any():
    indices_venta = np.where(alerta_venta)[0]
    print(f"   Índices: {indices_venta}")
    print(f"   Retornos: {retornos[alerta_venta].round(2)}%")

print("\n" + "="*70)
print("✅ Boolean masking es ESENCIAL para filtrado de datos financieros")
````

## 5️⃣ Fancy Indexing: Selección Avanzada

### 🎯 ¿Qué es Fancy Indexing?

Seleccionar múltiples elementos usando **arrays de índices** en lugar de un rango.

---

### 📊 Ejemplos Básicos

```python
arr = np.array([10, 20, 30, 40, 50, 60])

# Selección con lista de índices
indices = [0, 2, 4]
arr[indices]  # [10, 30, 50]

# Selección con array booleano
mascara = arr > 30
arr[mascara]  # [40, 50, 60]

# Selección con np.where()
indices = np.where(arr > 30)[0]
arr[indices]  # [40, 50, 60]
```

---

### 📉 Funciones Útiles

| Función | Descripción | Ejemplo |
|---------|-------------|----------|
| `np.where(condicion)` | Índices donde la condición es True | `np.where(arr > 10)` |
| `np.argmax()` | Índice del valor máximo | `arr.argmax()` |
| `np.argmin()` | Índice del valor mínimo | `arr.argmin()` |
| `np.argsort()` | Índices que ordenarían el array | `arr.argsort()` |
| `np.nonzero()` | Índices de elementos != 0 | `np.nonzero(arr)` |

---

### 💼 Casos Financieros

```python
# Encontrar días con mayor retorno
retornos = np.array([0.02, -0.01, 0.05, -0.03, 0.04])
mejor_dia = retornos.argmax()
print(f"Mejor día: {mejor_dia} con {retornos[mejor_dia]*100}%")

# Top 3 días con mayor retorno
top_3 = retornos.argsort()[-3:][::-1]
print(f"Top 3: {top_3} → Retornos: {retornos[top_3]}")

# Extraer precios de días específicos
dias_interes = [0, 5, 10, 15, 20]  # Lunes de cada semana
precios_lunes = precios[dias_interes]
```

````python
import numpy as np

print("💼 CASO PRÁCTICO FINAL: Análisis de Volatilidad y Riesgo")
print("="*70)

# Simular precios de 3 acciones durante 30 días
np.random.seed(42)
dias = 30
acciones = ['AAPL', 'GOOGL', 'MSFT']

precios = np.array([
    100 + np.cumsum(np.random.normal(0.5, 2, dias)),  # AAPL
    50 + np.cumsum(np.random.normal(0.3, 1.5, dias)),  # GOOGL
    75 + np.cumsum(np.random.normal(0.4, 1.8, dias))   # MSFT
]).T  # Transponer para tener shape (30, 3)

print(f"📅 Dataset: {dias} días, {len(acciones)} acciones")
print(f"Acciones: {acciones}")
print("\nPrimeros 5 días:")
print(precios[:5])

print("\n" + "="*70)

# 1. Calcular retornos diarios
retornos = (precios[1:] - precios[:-1]) / precios[:-1] * 100

print("📈 1. RETORNOS DIARIOS (%):")
print("-"*70)
print("\nRetornos promedio:")
for i, accion in enumerate(acciones):
    ret_promedio = retornos[:, i].mean()
    print(f"  {accion}: {ret_promedio:+.2f}%")

print("\n" + "="*70)

# 2. Volatilidad (desviación estándar)
print("📉 2. VOLATILIDAD (desv. estándar):")
print("-"*70)
volatilidad = retornos.std(axis=0)

print("\nVolatilidad diaria:")
for i, accion in enumerate(acciones):
    print(f"  {accion}: {volatilidad[i]:.2f}%")

print("\nVolatilidad anualizada (asumiendo 252 días de trading):")
volatilidad_anual = volatilidad * np.sqrt(252)
for i, accion in enumerate(acciones):
    print(f"  {accion}: {volatilidad_anual[i]:.2f}%")

# Clasificación por riesgo
accion_menos_volatil = acciones[volatilidad.argmin()]
accion_mas_volatil = acciones[volatilidad.argmax()]

print(f"\n🟢 Menos volátil (menor riesgo): {accion_menos_volatil}")
print(f"🔴 Más volátil (mayor riesgo):  {accion_mas_volatil}")

print("\n" + "="*70)

# 3. Filtrar días extremos (retornos > 2 desv. est.)
print("🚨 3. DÍAS EXTREMOS (retornos > 2σ):")
print("-"*70)

for i, accion in enumerate(acciones):
    ret_accion = retornos[:, i]
    umbral_superior = 2 * volatilidad[i]
    umbral_inferior = -2 * volatilidad[i]
    
    dias_extremos = (ret_accion > umbral_superior) | (ret_accion < umbral_inferior)
    num_extremos = dias_extremos.sum()
    
    print(f"\n{accion}:")
    print(f"  Días extremos: {num_extremos} / {len(ret_accion)} ({num_extremos/len(ret_accion)*100:.1f}%)")
    
    if num_extremos > 0:
        indices_extremos = np.where(dias_extremos)[0]
        print(f"  Índices: {indices_extremos}")
        print(f"  Retornos: {ret_accion[dias_extremos].round(2)}%")

print("\n" + "="*70)

# 4. Mejor y peor día de cada acción
print("🏆 4. MEJOR Y PEOR DÍA POR ACCIÓN:")
print("-"*70)

for i, accion in enumerate(acciones):
    ret_accion = retornos[:, i]
    mejor_dia = ret_accion.argmax()
    peor_dia = ret_accion.argmin()
    
    print(f"\n{accion}:")
    print(f"  Mejor día: Día {mejor_dia+1} con {ret_accion[mejor_dia]:+.2f}%")
    print(f"  Peor día:  Día {peor_dia+1} con {ret_accion[peor_dia]:+.2f}%")
    print(f"  Rango:     {ret_accion[mejor_dia] - ret_accion[peor_dia]:.2f}%")

print("\n" + "="*70)

# 5. Ratio retorno/riesgo (Sharpe ratio simplificado)
print("📈 5. RATIO RETORNO/RIESGO:")
print("-"*70)

retorno_promedio = retornos.mean(axis=0)
ratio = retorno_promedio / volatilidad

print("\nRatio retorno/riesgo (mayor = mejor):")
for i, accion in enumerate(acciones):
    print(f"  {accion}: {ratio[i]:.3f}")

mejor_ratio = acciones[ratio.argmax()]
print(f"\n🎯 Mejor ratio retorno/riesgo: {mejor_ratio}")

print("\n" + "="*70)
print("✅ Análisis de volatilidad y riesgo completado")
print("\n💡 Tip: En finanzas, mayor retorno suele significar mayor riesgo.")
print("   Busca el mejor balance retorno/riesgo para tu tolerancia.")
````
