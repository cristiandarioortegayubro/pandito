## 💰 Valor Presente Neto (VAN) y Tasa Interna de Retorno (TIR)

### VAN (NPV)

El VAN descuenta flujos futuros al presente:

$$VAN = \sum_{t=0}^{n} \frac{F_t}{(1+r)^t}$$

* **VAN > 0** → Proyecto genera valor
* **VAN < 0** → Proyecto destruye valor

### TIR (IRR)

Tasa que hace VAN = 0. Compara con tasa requerida.

**Implementación NumPy:**
```python
flujos = np.array([-100000, 30000, 35000, 40000, 45000])
tasa = 0.10
periodos = np.arange(len(flujos))
van = np.sum(flujos / (1 + tasa) ** periodos)
```

````python
import numpy as np

print("💰 MODELOS FINANCIEROS COMPLETOS")
print("="*70)

# 1. VAN
print("\n1️⃣ VALOR PRESENTE NETO (VAN):")
flujos = np.array([-100000, 30000, 35000, 40000, 45000])
tasa = 0.10
periodos = np.arange(len(flujos))
van = np.sum(flujos / (1 + tasa) ** periodos)
print(f"Flujos: {flujos}")
print(f"Tasa: {tasa*100}%")
print(f"VAN = ${van:,.2f}")
if van > 0:
    print("🟢 Proyecto viable - Genera valor")

# 2. TIR (Búsqueda binaria simplificada)
print("\n2️⃣ TASA INTERNA DE RETORNO (TIR):")
def calcular_van(flujos, tasa):
    periodos = np.arange(len(flujos))
    return np.sum(flujos / (1 + tasa) ** periodos)

# Búsqueda simple
tasa_min, tasa_max = 0.001, 1.0
for _ in range(100):
    tasa_media = (tasa_min + tasa_max) / 2
    van_media = calcular_van(flujos, tasa_media)
    if abs(van_media) < 0.01:
        break
    if van_media > 0:
        tasa_min = tasa_media
    else:
        tasa_max = tasa_media

tir = tasa_media
print(f"TIR = {tir*100:.2f}%")
print(f"Verificación: VAN(TIR) = ${calcular_van(flujos, tir):.2f} ≈ 0")

# 3. Portafolio
print("\n3️⃣ ANÁLISIS DE PORTAFOLIO:")
np.random.seed(42)
retornos = np.random.normal([0.0005, 0.0003, 0.0004], [0.02, 0.015, 0.018], (252, 3))
acciones = ['AAPL', 'GOOGL', 'MSFT']
pesos = np.array([0.40, 0.35, 0.25])

retorno_anual = retornos.mean(axis=0) * 252
volatilidad_anual = retornos.std(axis=0) * np.sqrt(252)

print(f"\nAcciones: {acciones}")
print(f"Pesos: {pesos}")
print(f"\nRetornos anuales:")
for i, a in enumerate(acciones):
    print(f"  {a}: {retorno_anual[i]*100:.2f}%")

retorno_portafolio = np.dot(pesos, retorno_anual)
cov_matrix = np.cov(retornos, rowvar=False) * 252
volatilidad_portafolio = np.sqrt(np.dot(pesos, np.dot(cov_matrix, pesos)))

print(f"\nPortafolio:")
print(f"  Retorno esperado: {retorno_portafolio*100:.2f}%")
print(f"  Volatilidad:      {volatilidad_portafolio*100:.2f}%")

# 4. Sharpe
tasa_libre = 0.03
sharpe = (retorno_portafolio - tasa_libre) / volatilidad_portafolio
print(f"  🎯 Sharpe Ratio:    {sharpe:.3f}")

if sharpe > 1.0:
    print("  🟢 Excelente retorno/riesgo")
else:
    print("  🟡 Buen retorno/riesgo")

print("\n" + "="*70)
print("✅ Modelos financieros implementados con NumPy vectorizado")
````
