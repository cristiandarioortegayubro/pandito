## 📊 Fundamentos del Valor del Dinero en el Tiempo

### 💵 Principio Central

**Un peso hoy vale MÁS que un peso mañana**

¿Por qué?
1. **Costo de oportunidad**: Ese peso podría estar generando rendimientos
2. **Inflación**: El poder adquisitivo disminuye con el tiempo
3. **Riesgo**: Incertidumbre sobre si recibirás ese peso futuro

---

### 🔢 Fórmulas Fundamentales

#### Valor Futuro (VF)

Cuánto vale hoy una cantidad **después de N períodos** con interés compuesto:

$$VF = VP \times (1 + r)^n$$

**Ejemplo:**
* Invierto $10,000 hoy
* Tasa: 8% anual
* Período: 5 años
* $VF = 10,000 \times (1.08)^5 = $14,693$

```python
VP = 10000
r = 0.08
n = 5
VF = VP * (1 + r) ** n
print(f"Valor futuro: ${VF:,.2f}")  # $14,693.28
```

---

#### Valor Presente (VP)

Cuánto vale **hoy** una cantidad futura:

$$VP = \frac{VF}{(1 + r)^n}$$

**Ejemplo:**
* Recibiré $15,000 en 5 años
* Tasa de descuento: 10% anual
* ¿Cuánto vale hoy? $VP = 15,000 / (1.10)^5 = $9,313$

```python
VF = 15000
r = 0.10
n = 5
VP = VF / (1 + r) ** n
print(f"Valor presente: ${VP:,.2f}")  # $9,313.82
```

---

### 💼 Aplicación: Flujos de Caja Múltiples

En la vida real, los proyectos generan **múltiples flujos** en diferentes períodos.

**Ejemplo - Proyecto de inversión:**

| Período | Flujo de Caja |
|---------|---------------|
| 0 (hoy) | -$100,000 (inversión) |
| 1       | +$30,000 |
| 2       | +$35,000 |
| 3       | +$40,000 |
| 4       | +$45,000 |

**Pregunta:** ¿Este proyecto es viable con tasa de descuento 10%?

**Respuesta:** Calculamos el VAN (Valor Actual Neto)...

````python
import numpy as np

print("💵 IMPLEMENTACIÓN DE VAN (NPV) CON NumPy")
print("="*70)

print("\n🎯 PROYECTO: Apertura de Nueva Sucursal")
print("\nFlujos de caja esperados:")

# Definir flujos
flujos = np.array([-100000, 30000, 35000, 40000, 45000])
periodos = np.arange(len(flujos))
tasa = 0.10

print(f"\nPeríodo 0 (hoy): ${flujos[0]:>10,.0f}  (Inversión inicial)")
for i in range(1, len(flujos)):
    print(f"Período {i}:       ${flujos[i]:>10,.0f}  (Ingreso año {i})")

print(f"\nTasa de descuento: {tasa*100}% anual")

print("\n" + "-"*70)
print("\n⚡ Cálculo Vectorizado (SIN bucles):")

# Método 1: Vectorizado puro
factores_descuento = (1 + tasa) ** periodos
flujos_descontados = flujos / factores_descuento
van = flujos_descontados.sum()

print(f"\nFactores de descuento: {factores_descuento}")
print(f"\nFlujos descontados:")
for i in range(len(flujos)):
    print(f"  Período {i}: ${flujos[i]:>10,.0f} / {factores_descuento[i]:.4f} = ${flujos_descontados[i]:>10,.2f}")

print(f"\n{'='*70}")
print(f"VAN TOTAL = ${van:,.2f}")
print(f"{'='*70}")

if van > 0:
    print("\n🟢 DECISIÓN: ACEPTAR el proyecto")
    print(f"   El proyecto genera ${van:,.2f} de valor presente neto")
else:
    print("\n🔴 DECISIÓN: RECHAZAR el proyecto")
    print(f"   El proyecto destruye ${abs(van):,.2f} de valor")

print("\n" + "-"*70)

# Método 2: Usando np.npv (función integrada)
print("\n✅ Verificación con np.npv():")
van_numpy = np.npv(tasa, flujos)
print(f"VAN (np.npv) = ${van_numpy:,.2f}")
print(f"\n✓ Coincide con cálculo manual")

print("\n" + "="*70)
print("✅ VAN calculado exitosamente")
````

## 🎯 Análisis de Sensibilidad del VAN

### 📊 ¿Cómo afecta la tasa de descuento al VAN?

La tasa de descuento representa el **costo de oportunidad** del capital:
* Tasa más alta → VAN más bajo (flujos futuros valen menos)
* Tasa más baja → VAN más alto (flujos futuros valen más)

---

### 🔬 Ejemplo: Impacto de Tasas Diferentes

**Proyecto:** Inversión $100,000, flujos $30K-$35K-$40K-$45K

| Tasa | VAN | Decisión |
|------|-----|----------|
| 5%   | $26,583 | 🟢 ACEPTAR |
| 10%  | $13,492 | 🟢 ACEPTAR |
| 15%  | $2,461  | 🟡 MARGINAL |
| 20%  | -$6,788 | 🔴 RECHAZAR |

---

### 💡 Interpretación Financiera

**¿Qué tasa usar?**

1. **WACC** (Weighted Average Cost of Capital): Costo promedio ponderado del capital
   * Promedio entre costo de deuda y costo de equity
   * Ejemplo: 10% anual

2. **Tasa libre de riesgo + Prima de riesgo**
   * T-Bonds USA (3%) + Riesgo del proyecto (7%) = 10%

3. **Tasa de rendimiento requerida por inversionistas**
   * Depende del perfil de riesgo del proyecto

---

### ⚡ Vectorización: Calcular VAN para MÚltiples Tasas

Con NumPy podemos calcular VAN para **cientos de tasas simultáneamente**:

```python
tasas = np.linspace(0.01, 0.30, 30)  # 30 tasas de 1% a 30%
periodos = np.arange(len(flujos))  # [0, 1, 2, 3, 4]

# Broadcasting mágico: (30, 1) ** (1, 5) → (30, 5)
factores = (1 + tasas[:, np.newaxis]) ** periodos

# (30, 5) / (30, 5) → (30, 5)
flujos_desc = flujos / factores

# Sum por fila → (30,)
vans = flujos_desc.sum(axis=1)
```

💥 **Sin bucles, sin iteraciones, todo vectorizado**

````python
import numpy as np
import matplotlib.pyplot as plt

print("📊 ANÁLISIS DE SENSIBILIDAD: VAN vs Tasa de Descuento")
print("="*70)

# Flujos del proyecto
flujos = np.array([-100000, 30000, 35000, 40000, 45000])
periodos = np.arange(len(flujos))

print("\n📈 Calculando VAN para 50 tasas diferentes (1% a 30%)...")

# Generar 50 tasas de 1% a 30%
tasas = np.linspace(0.01, 0.30, 50)

# Método vectorizado con broadcasting
factores = (1 + tasas[:, np.newaxis]) ** periodos  # Shape (50, 5)
flujos_desc = flujos / factores  # Broadcasting automático
vans = flujos_desc.sum(axis=1)  # Shape (50,)

print(f"\n✅ Calculados {len(tasas)} VAN en una sola operación vectorizada")

print("\n" + "-"*70)
print("\nRESULTADOS CLAVE:")
print(f"\nTasa  5%:  VAN = ${np.interp(0.05, tasas, vans):>10,.2f}  🟢 Muy atractivo")
print(f"Tasa 10%:  VAN = ${np.interp(0.10, tasas, vans):>10,.2f}  🟢 Atractivo")
print(f"Tasa 15%:  VAN = ${np.interp(0.15, tasas, vans):>10,.2f}  🟡 Marginal")
print(f"Tasa 20%:  VAN = ${np.interp(0.20, tasas, vans):>10,.2f}  🔴 Rechazar")

# Encontrar TIR (tasa donde VAN = 0)
tir_idx = np.argmin(np.abs(vans))
tir_aprox = tasas[tir_idx]
print(f"\n🎯 TIR aproximada: {tir_aprox*100:.2f}% (donde VAN ≈ 0)")

print("\n" + "-"*70)

# Visualización
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(tasas * 100, vans, linewidth=2.5, color='steelblue', label='VAN')
ax.axhline(0, color='black', linestyle='--', linewidth=1, alpha=0.7)
ax.axvline(tir_aprox * 100, color='red', linestyle=':', linewidth=2, label=f'TIR ≈ {tir_aprox*100:.1f}%')
ax.fill_between(tasas * 100, 0, vans, where=(vans > 0), color='green', alpha=0.2, label='Zona viable (VAN > 0)')
ax.fill_between(tasas * 100, 0, vans, where=(vans <= 0), color='red', alpha=0.2, label='Zona no viable (VAN < 0)')

ax.set_xlabel('Tasa de Descuento (%)', fontsize=12, fontweight='bold')
ax.set_ylabel('VAN ($)', fontsize=12, fontweight='bold')
ax.set_title('Análisis de Sensibilidad: VAN vs Tasa de Descuento', fontsize=14, fontweight='bold')
ax.legend(loc='upper right', fontsize=10)
ax.grid(alpha=0.3)
ax.axhline(0, color='black', linewidth=1)

plt.tight_layout()
display(fig)
plt.close()

print("\n💡 INTERPRETACIÓN:")
print("  • A mayor tasa de descuento, menor VAN")
print("  • TIR es la tasa que hace VAN = 0")
print("  • Si costo de capital < TIR → Proyecto viable")
print("  • Si costo de capital > TIR → Proyecto NO viable")

print("\n" + "="*70)
print("✅ Análisis de sensibilidad completado")
````

## 🎯 Tasa Interna de Retorno (TIR / IRR)

### 📊 ¿Qué es la TIR?

La TIR es la **tasa de descuento que hace el VAN = 0**

$$VAN = \sum_{t=0}^{n} \frac{F_t}{(1+TIR)^t} = 0$$

---

### 💡 Interpretación

**Regla de decisión:**
* Si **TIR > Tasa requerida** → 🟢 ACEPTAR proyecto
* Si **TIR < Tasa requerida** → 🔴 RECHAZAR proyecto
* Si **TIR = Tasa requerida** → 🟡 INDIFERENTE

**Ejemplo:**
* TIR del proyecto = 18%
* Costo de capital (WACC) = 10%
* **Decisión:** ACEPTAR (18% > 10%)
* **Interpretación:** El proyecto genera 18% de retorno, superior al 10% requerido

---

### ⚠️ Limitaciones de la TIR

1. **Múltiples TIR**
   * Si los flujos cambian de signo varias veces
   * Puede haber 0, 1 o múltiples TIR
   * Ejemplo: [-100, +250, -150] tiene 2 TIR

2. **Proyectos mutuamente excluyentes**
   * TIR puede dar resultado contradictorio con VAN
   * VAN es más confiable en estos casos

3. **Supuesto de reinversión**
   * TIR asume que flujos intermedios se reinvierten a la TIR
   * VAN asume reinversión a la tasa de descuento (más realista)

---

### 🔧 Métodos de Cálculo

#### 1. Método Analítico
No existe fórmula cerrada. Se resuelve numéricamente.

#### 2. Método Numérico (Newton-Raphson)
Iteración hasta convergencia:

$$TIR_{n+1} = TIR_n - \frac{VAN(TIR_n)}{VAN'(TIR_n)}$$

#### 3. Búsqueda Binaria (más simple)
```python
tir_min, tir_max = 0.0, 1.0
while abs(tir_max - tir_min) > 0.0001:
    tir_media = (tir_min + tir_max) / 2
    van = calcular_van(flujos, tir_media)
    if van > 0:
        tir_min = tir_media  # Aumentar tasa
    else:
        tir_max = tir_media  # Disminuir tasa
```

#### 4. Función NumPy (más rápido)
```python
tir = np.irr(flujos)  # Usa Newton-Raphson optimizado
```

````python
import numpy as np

print("🔧 IMPLEMENTACIÓN TIR (IRR) DESDE CERO")
print("="*70)

# Función auxiliar: calcular VAN
def calcular_van(flujos, tasa):
    periodos = np.arange(len(flujos))
    return np.sum(flujos / (1 + tasa) ** periodos)

# Proyecto
flujos = np.array([-100000, 30000, 35000, 40000, 45000])

print("\n🎯 PROYECTO: Apertura de Nueva Sucursal")
print(f"Flujos: {flujos}")

print("\n" + "-"*70)
print("\n💻 Método 1: Búsqueda Binaria (implementación manual)")

tir_min, tir_max = 0.001, 1.0
iteraciones = 0
max_iter = 100

while abs(tir_max - tir_min) > 0.00001 and iteraciones < max_iter:
    tir_media = (tir_min + tir_max) / 2
    van_media = calcular_van(flujos, tir_media)
    
    if abs(van_media) < 0.01:
        break
    
    if van_media > 0:
        tir_min = tir_media  # VAN positivo → aumentar tasa
    else:
        tir_max = tir_media  # VAN negativo → disminuir tasa
    
    iteraciones += 1

tir_manual = tir_media

print(f"\nTIR calculada: {tir_manual*100:.4f}%")
print(f"Iteraciones: {iteraciones}")
print(f"VAN(TIR): ${calcular_van(flujos, tir_manual):.2f} ≈ 0 ✅")

print("\n" + "-"*70)
print("\n⚡ Método 2: NumPy Finance (optimizado)")

# NumPy deprecó np.irr, pero podemos usar np.irr del paquete numpy_financial
# o scipy.optimize
from scipy.optimize import newton

def van_func(tasa, flujos):
    periodos = np.arange(len(flujos))
    return np.sum(flujos / (1 + tasa) ** periodos)

# Newton-Raphson
tir_scipy = newton(van_func, 0.1, args=(flujos,))

print(f"\nTIR (scipy.optimize): {tir_scipy*100:.4f}%")
print(f"VAN(TIR): ${calcular_van(flujos, tir_scipy):.2f} ≈ 0 ✅")

print("\n" + "-"*70)
print("\n📊 COMPARACIÓN CON COSTO DE CAPITAL")

wacc = 0.10  # Costo de capital 10%

print(f"\nTIR del proyecto:     {tir_manual*100:.2f}%")
print(f"Costo de capital (WACC): {wacc*100:.2f}%")
print(f"Spread:               {(tir_manual - wacc)*100:.2f} puntos porcentuales")

if tir_manual > wacc:
    print("\n🟢 DECISIÓN: ACEPTAR el proyecto")
    print(f"   El proyecto genera {(tir_manual - wacc)*100:.2f}pp más que el costo de capital")
    print(f"   Crea valor para los accionistas")
else:
    print("\n🔴 DECISIÓN: RECHAZAR el proyecto")
    print(f"   El proyecto NO cubre el costo de capital")
    print(f"   Destruye valor para los accionistas")

print("\n" + "="*70)
print("✅ TIR calculada exitosamente con múltiples métodos")
````

## 🏦 Tabla de Amortización de Préstamos

### 💵 Sistema de Amortización Francés

El sistema más común en préstamos personales e hipotecas:
* **Cuota constante** (igual todos los meses)
* Compuesta por: **Cuota = Interés + Capital Amortizado**
* Al principio: más interés, menos capital
* Al final: menos interés, más capital

---

### 📊 Fórmulas

#### Cuota Mensual Constante

$$Cuota = Principal \times \frac{r \times (1+r)^n}{(1+r)^n - 1}$$

Donde:
* $Principal$ = Monto del préstamo
* $r$ = Tasa de interés **mensual** (anual / 12)
* $n$ = Número de cuotas

---

#### Para Cada Período:

**Interés del período:**
$$Interés_t = Saldo_{t-1} \times r$$

**Capital amortizado:**
$$Capital\_Amortizado_t = Cuota - Interés_t$$

**Nuevo saldo:**
$$Saldo_t = Saldo_{t-1} - Capital\_Amortizado_t$$

---

### 💼 Ejemplo Real: Préstamo Hipotecario

**Parámetros:**
* Monto: $200,000
* Tasa anual: 6%
* Plazo: 20 años (240 meses)

**Calculemos:**
```python
P = 200000
r_anual = 0.06
r_mensual = r_anual / 12  # 0.005
n = 20 * 12  # 240 meses

cuota = P * (r_mensual * (1+r_mensual)**n) / ((1+r_mensual)**n - 1)
print(f"Cuota mensual: ${cuota:,.2f}")  # ~$1,432.86
```

**Total pagado:**
* Cuotas: 240 × $1,432.86 = **$343,886**
* Interés total: $343,886 - $200,000 = **$143,886** (72% del principal!)

---

### ⚡ Vectorización de Tabla de Amortización

Con NumPy podemos generar la tabla COMPLETA sin bucles:

```python
# Saldos vectorizados
saldos = np.zeros(n+1)
saldos[0] = P

for i in range(n):
    interes = saldos[i] * r_mensual
    capital_amortizado = cuota - interes
    saldos[i+1] = saldos[i] - capital_amortizado
```

⚠️ Este cálculo tiene dependencia secuencial (cada saldo depende del anterior), 
por lo que es difícil vectorizarlo completamente. **Excepción a la regla de vectorización.**

````python
import numpy as np
import pandas as pd

print("🏦 TABLA DE AMORTIZACIÓN: Sistema Francés")
print("="*70)

# Parámetros del préstamo
P = 200000  # Principal
tasa_anual = 0.06
r = tasa_anual / 12  # Tasa mensual
n = 20 * 12  # 240 meses (20 años)

print("\n💰 PARÁMETROS DEL PRÉSTAMO:")
print(f"\nMonto:         ${P:>12,.0f}")
print(f"Tasa anual:    {tasa_anual*100:>12.2f}%")
print(f"Tasa mensual:  {r*100:>12.4f}%")
print(f"Plazo:         {n//12:>12} años ({n} meses)")

print("\n" + "-"*70)

# Cálculo de la cuota mensual
cuota = P * (r * (1+r)**n) / ((1+r)**n - 1)

print(f"\n💵 CUOTA MENSUAL: ${cuota:,.2f}")

print("\n" + "-"*70)
print("\n📊 Generando tabla de amortización...")

# Inicializar arrays
periodos = np.arange(n + 1)
saldos = np.zeros(n + 1)
intereses = np.zeros(n)
capital_amortizado = np.zeros(n)
cuotas = np.zeros(n)

saldos[0] = P

# Calcular tabla (dependencia secuencial, necesita bucle)
for i in range(n):
    intereses[i] = saldos[i] * r
    capital_amortizado[i] = cuota - intereses[i]
    saldos[i+1] = saldos[i] - capital_amortizado[i]
    cuotas[i] = cuota

print("✅ Tabla generada")

print("\n" + "-"*70)
print("\n📑 PRIMERAS 6 CUOTAS:")
print("\n{:<8} {:<12} {:<12} {:<12} {:<12}".format(
    "Mes", "Cuota", "Interés", "Capital", "Saldo"))
print("-" * 70)
for i in range(6):
    print(f"{i+1:<8} ${cuotas[i]:>10,.2f} ${intereses[i]:>10,.2f} ${capital_amortizado[i]:>10,.2f} ${saldos[i+1]:>10,.2f}")

print("\n...")
print("\n📑 ÚLTIMAS 6 CUOTAS:")
print("\n{:<8} {:<12} {:<12} {:<12} {:<12}".format(
    "Mes", "Cuota", "Interés", "Capital", "Saldo"))
print("-" * 70)
for i in range(n-6, n):
    print(f"{i+1:<8} ${cuotas[i]:>10,.2f} ${intereses[i]:>10,.2f} ${capital_amortizado[i]:>10,.2f} ${saldos[i+1]:>10,.2f}")

print("\n" + "="*70)

# Totales
total_pagado = cuotas.sum()
total_interes = intereses.sum()
total_capital = capital_amortizado.sum()

print("\n📊 RESUMEN FINANCIERO:")
print(f"\nTotal pagado:      ${total_pagado:>12,.2f}")
print(f"Total interés:     ${total_interes:>12,.2f}  ({total_interes/P*100:.1f}% del principal)")
print(f"Total capital:     ${total_capital:>12,.2f}")
print(f"Principal original: ${P:>12,.2f}")

print("\n💡 OBSERVACIONES:")
print(f"  • Primera cuota: ${intereses[0]:,.2f} interés, ${capital_amortizado[0]:,.2f} capital ({intereses[0]/cuota*100:.1f}% interés)")
print(f"  • Última cuota:  ${intereses[-1]:,.2f} interés, ${capital_amortizado[-1]:,.2f} capital ({capital_amortizado[-1]/cuota*100:.1f}% capital)")
print(f"  • El interés total ({total_interes/P*100:.1f}%) es {total_interes/P:.2f}x el monto prestado")

print("\n" + "="*70)
print("✅ Tabla de amortización completa generada")
````

## 💼 Análisis de Portafolios de Inversiones

### 👥 Teoría Moderna de Portafolios (Markowitz)

Desarrollada por Harry Markowitz (Premio Nobel 1990):
* **Diversificación** reduce riesgo sin sacrificar retorno
* Combinar activos **no perfectamente correlacionados**
* Objetivo: **Maximizar retorno** para un nivel de riesgo dado

---

### 📊 Métricas Clave

#### 1. Retorno Esperado del Portafolio

$$E[R_p] = \sum_{i=1}^{n} w_i \times E[R_i]$$

Donde:
* $w_i$ = Peso del activo $i$ en el portafolio
* $E[R_i]$ = Retorno esperado del activo $i$

**Con NumPy:**
```python
retorno_portafolio = np.dot(pesos, retornos_esperados)
```

---

#### 2. Riesgo (Volatilidad) del Portafolio

$$\sigma_p = \sqrt{w^T \Sigma w}$$

Donde:
* $w$ = Vector de pesos
* $\Sigma$ = Matriz de covarianza
* $w^T$ = Transpuesta de $w$

**Con NumPy:**
```python
volatilidad = np.sqrt(np.dot(pesos, np.dot(cov_matrix, pesos)))
```

---

### 🔗 Matriz de Covarianza

Mide cómo se mueven los activos entre sí:

| | AAPL | GOOGL | MSFT |
|---------|------|-------|------|
| AAPL | Var(AAPL) | Cov(AAPL,GOOGL) | Cov(AAPL,MSFT) |
| GOOGL | Cov(GOOGL,AAPL) | Var(GOOGL) | Cov(GOOGL,MSFT) |
| MSFT | Cov(MSFT,AAPL) | Cov(MSFT,GOOGL) | Var(MSFT) |

* **Diagonal:** Varianza de cada activo (riesgo individual)
* **Fuera de diagonal:** Covarianza entre pares (cómo se mueven juntos)

**Con NumPy:**
```python
cov_matrix = np.cov(retornos, rowvar=False)
# rowvar=False: cada columna es una variable (activo)
```

### 💡 Beneficio de la Diversificación

**Ejemplo: 2 Activos**

| Escenario | Peso A | Peso B | Retorno A | Retorno B | Correlación |
|-----------|--------|--------|-----------|-----------|-------------|
| 1 | 100% | 0% | 10% | - | - |
| 2 | 0% | 100% | - | 8% | - |
| 3 | 50% | 50% | 10% | 8% | 0.3 |

**Resultados:**
* Escenario 1: Retorno 10%, Riesgo (std) 20%
* Escenario 2: Retorno 8%, Riesgo (std) 15%
* Escenario 3: Retorno 9%, Riesgo (std) **12%** ⬅️ MENOR que ambos!

🎯 **Conclusión:** La diversificación reduce riesgo cuando correlación < 1

````python
import numpy as np

print("📊 ANÁLISIS DE PORTAFOLIO: Retorno y Riesgo")
print("="*70)

# Simular retornos diarios de 3 acciones (252 días de trading)
np.random.seed(42)
acciones = ['AAPL', 'GOOGL', 'MSFT']
n_dias = 252

# Retornos diarios simulados (media y std diferentes por acción)
retornos = np.array([
    np.random.normal(0.0008, 0.020, n_dias),  # AAPL: 0.08% diario, 2% std
    np.random.normal(0.0006, 0.018, n_dias),  # GOOGL: 0.06% diario, 1.8% std
    np.random.normal(0.0007, 0.015, n_dias),  # MSFT: 0.07% diario, 1.5% std
]).T  # Shape (252, 3) - cada columna es una acción

print("\n📋 Dataset: 252 días de trading (1 año)")
print(f"Shape: {retornos.shape} (días × acciones)")

print("\n" + "-"*70)
print("\n📊 MÉTRICAS INDIVIDUALES (anualizadas):")

# Anualizar retornos y volatilidades
retorno_anual = retornos.mean(axis=0) * 252
volatilidad_anual = retornos.std(axis=0) * np.sqrt(252)

print("\n{:<10} {:<15} {:<15}".format("Acción", "Retorno Anual", "Volatilidad"))
print("-" * 45)
for i, acc in enumerate(acciones):
    print(f"{acc:<10} {retorno_anual[i]*100:>13.2f}%  {volatilidad_anual[i]*100:>13.2f}%")

print("\n" + "="*70)
print("\n🎯 CONSTRUCCIÓN DE PORTAFOLIO")

# Definir pesos del portafolio
pesos = np.array([0.40, 0.35, 0.25])  # 40% AAPL, 35% GOOGL, 25% MSFT

print(f"\nPesos: {dict(zip(acciones, pesos))}")
print(f"Suma de pesos: {pesos.sum()} (debe ser 1.0 ✅)")

print("\n" + "-"*70)
print("\n⚡ Cálculo de Retorno del Portafolio (vectorizado):")

# Retorno esperado del portafolio
retorno_portafolio = np.dot(pesos, retorno_anual)

print(f"\nRetorno portafolio = w^T × r")
print(f"                   = [{pesos[0]:.2f}, {pesos[1]:.2f}, {pesos[2]:.2f}] × [{retorno_anual[0]:.4f}, {retorno_anual[1]:.4f}, {retorno_anual[2]:.4f}]")
print(f"                   = {retorno_portafolio:.4f}")
print(f"\n💹 Retorno esperado: {retorno_portafolio*100:.2f}% anual")

print("\n" + "-"*70)
print("\n🔗 Matriz de Covarianza (anualizada):")

# Matriz de covarianza
cov_matrix = np.cov(retornos, rowvar=False) * 252  # Anualizar

print("\n{:<10}".format(""), end="")
for acc in acciones:
    print(f"{acc:>12}", end="")
print()

for i, acc in enumerate(acciones):
    print(f"{acc:<10}", end="")
    for j in range(len(acciones)):
        print(f"{cov_matrix[i, j]:>12.6f}", end="")
    print()

print("\n💡 Diagonal: Varianza de cada activo")
print("💡 Fuera de diagonal: Covarianza entre pares")

print("\n" + "-"*70)
print("\n⚡ Cálculo de Riesgo del Portafolio (vectorizado):")

# Volatilidad del portafolio
volatilidad_portafolio = np.sqrt(np.dot(pesos, np.dot(cov_matrix, pesos)))

print(f"\nVolatilidad portafolio = sqrt(w^T × Σ × w)")
print(f"\n📉 Volatilidad: {volatilidad_portafolio*100:.2f}% anual")

print("\n" + "="*70)
print("\n📊 COMPARACIÓN: Portafolio vs Activos Individuales")

print("\n{:<15} {:<15} {:<15}".format("Inversión", "Retorno", "Volatilidad"))
print("-" * 50)
for i, acc in enumerate(acciones):
    print(f"{acc+' (100%)'}:<15} {retorno_anual[i]*100:>13.2f}%  {volatilidad_anual[i]*100:>13.2f}%")

print(f"{'-'*50}")
print(f"{'PORTAFOLIO':<15} {retorno_portafolio*100:>13.2f}%  {volatilidad_portafolio*100:>13.2f}%  ⬅️")

print("\n💡 BENEFICIO DE LA DIVERSIFICACIÓN:")
volatilidad_promedio = volatilidad_anual.mean()
reduccion_riesgo = (volatilidad_promedio - volatilidad_portafolio) / volatilidad_promedio * 100

print(f"  • Volatilidad promedio individual: {volatilidad_promedio*100:.2f}%")
print(f"  • Volatilidad del portafolio:      {volatilidad_portafolio*100:.2f}%")
print(f"  • Reducción de riesgo:             {reduccion_riesgo:.1f}%")

print("\n" + "="*70)
print("✅ Análisis de portafolio completado")
````

## 🏆 Índice de Sharpe: Retorno Ajustado por Riesgo

### 📊 ¿Qué es el Índice de Sharpe?

Mide el **exceso de retorno por unidad de riesgo**:

$$Sharpe = \frac{E[R_p] - R_f}{\sigma_p}$$

Donde:
* $E[R_p]$ = Retorno esperado del portafolio
* $R_f$ = Tasa libre de riesgo (T-Bonds, CETES, etc.)
* $\sigma_p$ = Volatilidad del portafolio

---

### 💡 Interpretación

**¿Qué significa?**
* Sharpe > 1.0: 🟢 **Excelente** - Buen retorno por riesgo asumido
* Sharpe 0.5-1.0: 🟡 **Bueno** - Retorno aceptable
* Sharpe < 0.5: 🔴 **Malo** - Bajo retorno por riesgo
* Sharpe < 0: 🔴 **Pésimo** - No supera tasa libre de riesgo

**Ejemplo:**
* Portafolio A: Retorno 12%, Volatilidad 15%, RF 3% → Sharpe = (12%-3%)/15% = **0.60**
* Portafolio B: Retorno 18%, Volatilidad 25%, RF 3% → Sharpe = (18%-3%)/25% = **0.60**
* **Conclusión:** Ambos tienen el mismo retorno ajustado por riesgo

---

### 🔍 Comparación de Inversiones

| Inversión | Retorno | Volatilidad | Sharpe (RF=3%) | Ranking |
|----------|---------|-------------|----------------|----------|
| Bonos | 5% | 2% | 1.00 | 1º |
| Portafolio Conservador | 8% | 8% | 0.63 | 3º |
| Portafolio Balanceado | 12% | 12% | 0.75 | 2º |
| Portafolio Agresivo | 18% | 22% | 0.68 | 3º |
| Crypto | 50% | 80% | 0.59 | 4º |

🏆 **Ganador:** Bonos (Sharpe más alto)

⚠️ **Sorpresa:** Crypto tiene alto retorno pero Sharpe bajo (demasiado volátil)

---

### ⚡ Cálculo con NumPy

```python
retorno_portafolio = np.dot(pesos, retornos_esperados)
volatilidad_portafolio = np.sqrt(np.dot(pesos, np.dot(cov_matrix, pesos)))
tasa_libre_riesgo = 0.03

sharpe = (retorno_portafolio - tasa_libre_riesgo) / volatilidad_portafolio
```

### 💼 Aplicación Práctica

**¿Cómo usarlo?**
1. Calcular Sharpe de múltiples portafolios
2. Elegir el de **mayor Sharpe** (mejor retorno/riesgo)
3. Comparar con benchmarks (S&P 500, IPC, etc.)
4. Optimizar pesos para maximizar Sharpe

**Limitaciones:**
* Asume distribuciones normales (no siempre cierto)
* Sensible a la elección de $R_f$
* No captura riesgos de cola (eventos extremos)

````python
import numpy as np

print("🏆 ÍNDICE DE SHARPE: Retorno Ajustado por Riesgo")
print("="*70)

# Usar los mismos datos del análisis anterior
np.random.seed(42)
acciones = ['AAPL', 'GOOGL', 'MSFT']
n_dias = 252

retornos = np.array([
    np.random.normal(0.0008, 0.020, n_dias),
    np.random.normal(0.0006, 0.018, n_dias),
    np.random.normal(0.0007, 0.015, n_dias),
]).T

retorno_anual = retornos.mean(axis=0) * 252
volatilidad_anual = retornos.std(axis=0) * np.sqrt(252)
cov_matrix = np.cov(retornos, rowvar=False) * 252

pesos = np.array([0.40, 0.35, 0.25])
retorno_portafolio = np.dot(pesos, retorno_anual)
volatilidad_portafolio = np.sqrt(np.dot(pesos, np.dot(cov_matrix, pesos)))

print("\n📊 PARÁMETROS:")
print(f"\nRetorno portafolio:      {retorno_portafolio*100:>6.2f}%")
print(f"Volatilidad portafolio:  {volatilidad_portafolio*100:>6.2f}%")

print("\n" + "-"*70)

# Tasa libre de riesgo (T-Bonds USA 10 años)
tasa_libre_riesgo = 0.03  # 3% anual

print(f"\n💵 Tasa libre de riesgo (RF): {tasa_libre_riesgo*100:.2f}%")
print("   (T-Bonds USA 10 años)")

print("\n" + "-"*70)
print("\n⚡ Cálculo del Índice de Sharpe:")

exceso_retorno = retorno_portafolio - tasa_libre_riesgo
sharpe = exceso_retorno / volatilidad_portafolio

print(f"\nExceso de retorno = {retorno_portafolio*100:.2f}% - {tasa_libre_riesgo*100:.2f}% = {exceso_retorno*100:.2f}%")
print(f"\nSharpe = Exceso / Volatilidad")
print(f"       = {exceso_retorno*100:.2f}% / {volatilidad_portafolio*100:.2f}%")
print(f"       = {sharpe:.4f}")

print(f"\n{'='*70}")
print(f"🎯 ÍNDICE DE SHARPE = {sharpe:.3f}")
print(f"{'='*70}")

# Interpretación
print("\n📊 INTERPRETACIÓN:")
if sharpe > 1.0:
    print("\n🟢 EXCELENTE retorno ajustado por riesgo")
    print("   El portafolio genera buen exceso de retorno por unidad de riesgo")
elif sharpe > 0.5:
    print("\n🟡 BUENO retorno ajustado por riesgo")
    print("   El portafolio tiene retorno aceptable dado su nivel de riesgo")
else:
    print("\n🔴 BAJO retorno ajustado por riesgo")
    print("   El portafolio NO compensa adecuadamente por el riesgo asumido")

print("\n" + "="*70)
print("\n📊 COMPARACIÓN: Portafolio vs Activos Individuales")

print("\n{:<15} {:<10} {:<12} {:<10}".format("Inversión", "Retorno", "Volatilidad", "Sharpe"))
print("-" * 55)

# Sharpe de cada activo individual
for i, acc in enumerate(acciones):
    sharpe_individual = (retorno_anual[i] - tasa_libre_riesgo) / volatilidad_anual[i]
    print(f"{acc+' (100%)'}:<15} {retorno_anual[i]*100:>8.2f}%  {volatilidad_anual[i]*100:>10.2f}%  {sharpe_individual:>8.3f}")

print(f"{'-'*55}")
print(f"{'PORTAFOLIO':<15} {retorno_portafolio*100:>8.2f}%  {volatilidad_portafolio*100:>10.2f}%  {sharpe:>8.3f}  ⬅️")

# Comparación con benchmarks típicos
print("\n" + "="*70)
print("\n🏁 COMPARACIÓN CON BENCHMARKS TÍPICOS:")

benchmarks = {
    'S&P 500': (0.10, 0.15),  # 10% retorno, 15% volatilidad
    'Bonos': (0.05, 0.05),     # 5% retorno, 5% volatilidad
    'Crypto (BTC)': (0.40, 0.70)  # 40% retorno, 70% volatilidad
}

print("\n{:<15} {:<10} {:<12} {:<10}".format("Benchmark", "Retorno", "Volatilidad", "Sharpe"))
print("-" * 55)

for nombre, (ret, vol) in benchmarks.items():
    sharpe_bench = (ret - tasa_libre_riesgo) / vol
    print(f"{nombre:<15} {ret*100:>8.2f}%  {vol*100:>10.2f}%  {sharpe_bench:>8.3f}")

print(f"{'-'*55}")
print(f"{'TU PORTAFOLIO':<15} {retorno_portafolio*100:>8.2f}%  {volatilidad_portafolio*100:>10.2f}%  {sharpe:>8.3f}")

print("\n💡 OBSERVACIONES:")
print("  • Bonos: Alto Sharpe (bajo riesgo, retorno estable)")
print("  • S&P 500: Sharpe moderado (equilibrio retorno/riesgo)")
print("  • Crypto: Sharpe bajo (alto retorno pero volatilidad extrema)")
print(f"  • Tu portafolio: Sharpe {sharpe:.3f} - Posición competitiva")

print("\n" + "="*70)
print("✅ Índice de Sharpe calculado y comparado exitosamente")
````

## 💰 Valor Presente Neto (VAN) y Tasa Interna de Retorno (TIR)

### VAN (NPV)

El VAN descuenta flujos futuros al presente:

$$VAN = \sum_{t=0}^{n} \frac{F_t}{(1+r)^t}$$

* **VAN > 0** → Proyecto genera valor
* **VAN < 0** → Proyecto destruye valor

### TIR (IRR)

Tasa que hace VAN = 0. Compara con tasa requerida.

**Implementación NumPy:**
```python
flujos = np.array([-100000, 30000, 35000, 40000, 45000])
tasa = 0.10
periodos = np.arange(len(flujos))
van = np.sum(flujos / (1 + tasa) ** periodos)
```

````python
import numpy as np

print("💰 MODELOS FINANCIEROS COMPLETOS")
print("="*70)

# 1. VAN
print("\n1️⃣ VALOR PRESENTE NETO (VAN):")
flujos = np.array([-100000, 30000, 35000, 40000, 45000])
tasa = 0.10
periodos = np.arange(len(flujos))
van = np.sum(flujos / (1 + tasa) ** periodos)
print(f"Flujos: {flujos}")
print(f"Tasa: {tasa*100}%")
print(f"VAN = ${van:,.2f}")
if van > 0:
    print("🟢 Proyecto viable - Genera valor")

# 2. TIR (Búsqueda binaria simplificada)
print("\n2️⃣ TASA INTERNA DE RETORNO (TIR):")
def calcular_van(flujos, tasa):
    periodos = np.arange(len(flujos))
    return np.sum(flujos / (1 + tasa) ** periodos)

# Búsqueda simple
tasa_min, tasa_max = 0.001, 1.0
for _ in range(100):
    tasa_media = (tasa_min + tasa_max) / 2
    van_media = calcular_van(flujos, tasa_media)
    if abs(van_media) < 0.01:
        break
    if van_media > 0:
        tasa_min = tasa_media
    else:
        tasa_max = tasa_media

tir = tasa_media
print(f"TIR = {tir*100:.2f}%")
print(f"Verificación: VAN(TIR) = ${calcular_van(flujos, tir):.2f} ≈ 0")

# 3. Portafolio
print("\n3️⃣ ANÁLISIS DE PORTAFOLIO:")
np.random.seed(42)
retornos = np.random.normal([0.0005, 0.0003, 0.0004], [0.02, 0.015, 0.018], (252, 3))
acciones = ['AAPL', 'GOOGL', 'MSFT']
pesos = np.array([0.40, 0.35, 0.25])

retorno_anual = retornos.mean(axis=0) * 252
volatilidad_anual = retornos.std(axis=0) * np.sqrt(252)

print(f"\nAcciones: {acciones}")
print(f"Pesos: {pesos}")
print(f"\nRetornos anuales:")
for i, a in enumerate(acciones):
    print(f"  {a}: {retorno_anual[i]*100:.2f}%")

retorno_portafolio = np.dot(pesos, retorno_anual)
cov_matrix = np.cov(retornos, rowvar=False) * 252
volatilidad_portafolio = np.sqrt(np.dot(pesos, np.dot(cov_matrix, pesos)))

print(f"\nPortafolio:")
print(f"  Retorno esperado: {retorno_portafolio*100:.2f}%")
print(f"  Volatilidad:      {volatilidad_portafolio*100:.2f}%")

# 4. Sharpe
tasa_libre = 0.03
sharpe = (retorno_portafolio - tasa_libre) / volatilidad_portafolio
print(f"  🎯 Sharpe Ratio:    {sharpe:.3f}")

if sharpe > 1.0:
    print("  🟢 Excelente retorno/riesgo")
else:
    print("  🟡 Buen retorno/riesgo")

print("\n" + "="*70)
print("✅ Modelos financieros implementados con NumPy vectorizado")
````
