3. Realizar **Simulaciones de Montecarlo** para proyectar flujos de fondos y evaluar el riesgo de inversión (VAN/TIR).


```python
import numpy as np

# Ejemplo: Simulación Montecarlo de Flujo de Fondos Proyectado (1,000 escenarios)
np.random.seed(42)
escenarios = 1000
periodos = 5

# Promedio de ingreso esperado: $100,000 con desviación estándar de $15,000
flujos = np.random.normal(loc=100000, scale=15000, size=(escenarios, periodos))
tasa_descuento = 0.12

# Cálculo de VAN vectorizado
descuentos = (1 + tasa_descuento) ** np.arange(1, periodos + 1)
van = np.sum(flujos / descuentos, axis=1) - 300000 # Inversión inicial de $300,000

print(f"VAN Promedio Esperado: ${van.mean():,.2f}")
print(f"Probabilidad de VAN Positivo: {(van > 0).mean() * 100:.1f}%")

```
