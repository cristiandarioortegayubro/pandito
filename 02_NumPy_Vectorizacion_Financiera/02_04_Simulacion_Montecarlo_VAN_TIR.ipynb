```python
import numpy as np
import matplotlib.pyplot as plt

print("🎲 SIMULACIÓN MONTE CARLO")
print("="*70)

print("\n🔍 ¿Qué es Monte Carlo?")
print("""Una técnica de simulación que genera MILES de escenarios aleatorios 
para modelar incertidumbre y cuantificar riesgo.""")

print("\n🎯 ¿Cuándo usarlo?")
print("  • Proyectos con incertidumbre en ingresos/costos")
print("  • Evaluación de riesgo de inversiones")
print("  • Pricing de derivados financieros")
print("  • Planificación de jubilación")
print("  • Optimización de portafolios")

print("\n📊 En este notebook:")
print("  1. Simular 10,000 escenarios de flujos de caja")
print("  2. Calcular distribución de VAN")
print("  3. Analizar percentiles P10, P50, P90")
print("  4. Calcular probabilidad de viabilidad P(VAN>0)")
print("  5. Visualizar resultados con histogramas")

print("\n✅ Librerías cargadas correctamente")
```

```python
import numpy as np
import matplotlib.pyplot as plt

print("🎲 SIMULACIÓN MONTE CARLO: 10,000 Escenarios")
print("="*70)

# Parámetros
inversion = 100000
flujos_base = np.array([30000, 35000, 40000, 45000])
desv_std = 5000
tasa = 0.10
n_sim = 10000

print(f"Inversión: ${inversion:,}")
print(f"Flujos base: {flujos_base}")
print(f"Incertidumbre: ±${desv_std:,}")
print(f"Escenarios: {n_sim:,}")

# Simulación
np.random.seed(42)
flujos_sim = np.random.normal(flujos_base, desv_std, (n_sim, len(flujos_base)))
periodos = np.arange(1, len(flujos_base) + 1)
van_sim = -inversion + np.sum(flujos_sim / (1 + tasa) ** periodos, axis=1)

print(f"\n📉 RESULTADOS:")
print(f"VAN Promedio:  ${van_sim.mean():,.2f}")
print(f"Desv. Estándar: ${van_sim.std():,.2f}")

# Percentiles
p10, p50, p90 = np.percentile(van_sim, [10, 50, 90])
print(f"\nP10 (pesimista): ${p10:,.2f}")
print(f"P50 (mediana):   ${p50:,.2f}")
print(f"P90 (optimista): ${p90:,.2f}")

# Probabilidad
prob_exito = (van_sim > 0).mean()
print(f"\n🎯 P(VAN > 0) = {prob_exito*100:.1f}%")

if prob_exito >= 0.70:
    print("🟢 Proyecto viable - Alta probabilidad de éxito")
else:
    print("🔴 Proyecto riesgoso - Baja probabilidad de éxito")

# Visualización
fig, ax = plt.subplots(figsize=(10, 5))
ax.hist(van_sim, bins=50, color='steelblue', alpha=0.7, edgecolor='black')
ax.axvline(van_sim.mean(), color='red', linestyle='--', linewidth=2, label=f'Promedio: ${van_sim.mean():,.0f}')
ax.axvline(p10, color='orange', linestyle=':', linewidth=2, label=f'P10: ${p10:,.0f}')
ax.axvline(p90, color='green', linestyle=':', linewidth=2, label=f'P90: ${p90:,.0f}')
ax.axvline(0, color='black', linestyle='-', linewidth=1)
ax.set_xlabel('VAN ($)', fontsize=12)
ax.set_ylabel('Frecuencia', fontsize=12)
ax.set_title('Distribución de VAN - 10,000 Escenarios Monte Carlo', fontsize=14, fontweight='bold')
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
display(fig)
plt.close()

print("\n" + "="*70)
print("✅ Simulación completada")
```
