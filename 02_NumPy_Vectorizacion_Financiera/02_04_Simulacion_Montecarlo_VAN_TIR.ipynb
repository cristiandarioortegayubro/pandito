## 🎲 Fundamentos de la Simulación Monte Carlo

### 📚 Historia y Origen

**Nombre:** Proviene del casino Monte Carlo en Mónaco (juegos de azar, aleatoriedad)

**Desarrollo:** Años 1940s por Stanislaw Ulam y John von Neumann durante el Proyecto Manhattan:
* Necesitaban simular difusión de neutrones en bombas atómicas
* Cálculos analíticos eran imposibles
* Usaron aleatoriedad para aproximar resultados

---

### 🔬 ¿Qué es la Simulación Monte Carlo?

**Definición:**
Técnica computacional que usa **muestreo aleatorio repetido** para:
1. Modelar sistemas con incertidumbre
2. Estimar distribuciones de probabilidad
3. Cuantificar riesgo

**Proceso básico:**
```
1. Identificar variables inciertas (ingresos, costos, tasa, etc.)
2. Definir distribuciones de probabilidad para cada variable
3. Generar miles de escenarios aleatorios
4. Calcular métrica de interés (VAN, TIR, etc.) en cada escenario
5. Analizar distribución de resultados
```

---

### 💼 Aplicaciones en Finanzas

1. **Evaluación de proyectos**
   * Flujos de caja inciertos
   * Probabilidad de viabilidad
   * Rango de VAN posibles

2. **Pricing de derivados**
   * Opciones europeas/americanas
   * Modelos Black-Scholes
   * Productos estructurados

3. **Gestión de riesgo**
   * Value at Risk (VaR)
   * Conditional Value at Risk (CVaR)
   * Stress testing

4. **Planificación de jubilación**
   * Probabilidad de fondos suficientes
   * Estrategias de retiro

---

### ⚡ Por Qué Usar Monte Carlo

**Ventajas:**
* ✅ Modela incertidumbre realista
* ✅ No requiere supuestos de normalidad
* ✅ Genera distribuciones completas (no solo promedios)
* ✅ Cuantifica probabilidades de eventos
* ✅ Incorpora correlaciones entre variables

**Desventajas:**
* ❌ Computacionalmente intensivo (miles de simulaciones)
* ❌ Requiere definir distribuciones correctamente
* ❌ Basura entra, basura sale (GIGO)

---

### 📊 Ejemplo: VAN Determinista vs Monte Carlo

**Enfoque determinista (tradicional):**
```python
VAN = -100,000 + 30,000/1.10 + 35,000/1.10^2 + ...
VAN = $13,492
Decisión: ACEPTAR (VAN > 0)
```

**Enfoque Monte Carlo:**
```python
Simular 10,000 escenarios con flujos inciertos
Resultado: Distribución de VAN
  - P10 (pesimista): $-5,000
  - P50 (mediana):    $12,000
  - P90 (optimista):  $30,000
  - P(VAN > 0) = 75%
Decisión: ACEPTAR con 75% de confianza
```

💡 **Monte Carlo da INFORMACIÓN MUCHO MÁS RICA**

## 📊 Distribuciones de Probabilidad para Simulación

### 📊 Distribución Normal (Gaussiana)

**Más común en finanzas**

$$X \sim N(\mu, \sigma^2)$$

**Características:**
* Simétrica alrededor de la media
* 68% de valores entre $\mu \pm \sigma$
* 95% entre $\mu \pm 2\sigma$
* 99.7% entre $\mu \pm 3\sigma$

**Cuándo usarla:**
* Retornos de acciones (aprox.)
* Errores de forecasting
* Demanda con variabilidad estable

**Con NumPy:**
```python
retornos = np.random.normal(mu=0.10, sigma=0.15, size=10000)
# 10,000 retornos con media 10% y std 15%
```

---

### 📏 Distribución Uniforme

**Todos los valores igualmente probables**

$$X \sim U(a, b)$$

**Características:**
* Probabilidad constante entre $a$ y $b$
* Media: $(a+b)/2$
* No hay valores más probables

**Cuándo usarla:**
* Incertidumbre total (sin información)
* Tasas de descuento en rango
* Variables sin patrón histórico

**Con NumPy:**
```python
tasa = np.random.uniform(low=0.08, high=0.12, size=10000)
# Tasas uniformemente entre 8% y 12%
```

---

### 📐 Distribución Triangular

**Tres puntos: mínimo, moda, máximo**

**Características:**
* Define escenario pesimista, base y optimista
* Más realista que uniforme
* Fácil de estimar para expertos

**Cuándo usarla:**
* Proyectos nuevos con estimaciones de expertos
* Costos con mejor guía ("entre $50K y $100K, probablemente $70K")

**Con NumPy:**
```python
costos = np.random.triangular(left=50000, mode=70000, right=100000, size=10000)
# Costos con escenarios pesi/base/opti
```

---

### 📊 Distribución Lognormal

**Para variables que NO pueden ser negativas**

**Características:**
* Siempre positiva (ideal para precios, ingresos)
* Asimétrica (cola derecha larga)
* $\log(X) \sim N(\mu, \sigma^2)$

**Cuándo usarla:**
* Precios de acciones
* Ingresos/ventas
* Cualquier variable estrictamente positiva

**Con NumPy:**
```python
precios = np.random.lognormal(mean=4.6, sigma=0.2, size=10000)
# Precios lognormales (siempre > 0)
```

---

### 🎯 ¿Cómo Elegir la Distribución?

| Variable | Distribución Recomendada | Razón |
|----------|---------------------------|--------|
| Flujos de caja | Normal o Triangular | Pueden ser +/-, variabilidad conocida |
| Precios | Lognormal | Siempre positivos, crec. exponencial |
| Tasas de interés | Normal (truncada) o Uniforme | Rango conocido, no pueden ser muy negativas |
| Costos | Triangular | Expertos dan pesi/base/opti |
| Demanda | Normal (truncada) o Poisson | Variabilidad histórica |
| Tiempo hasta evento | Exponencial | Procesos de llegada |

````python
import numpy as np
import matplotlib.pyplot as plt

print("🎲 GENERACIÓN DE NÚMEROS ALEATORIOS CON NumPy")
print("="*70)

print("\n🔢 np.random: Módulo para generación aleatoria")
print("\n⚠️  IMPORTANTE: Siempre usar np.random.seed() para reproducibilidad")

np.random.seed(42)
print("\nnp.random.seed(42)  # Garantiza mismos resultados cada vez")

print("\n" + "-"*70)
print("\n1️⃣  DISTRIBUCIÓN NORMAL")

# Normal
flujos_normal = np.random.normal(loc=30000, scale=5000, size=10)
print(f"\nnp.random.normal(loc=30000, scale=5000, size=10)")
print(f"Media: 30,000 | Std: 5,000 | Cantidad: 10")
print(f"\nResultados: {flujos_normal.astype(int)}")
print(f"Promedio real: ${flujos_normal.mean():,.0f}")
print(f"Std real: ${flujos_normal.std():,.0f}")

print("\n" + "-"*70)
print("\n2️⃣  DISTRIBUCIÓN UNIFORME")

# Uniforme
tasas_uniforme = np.random.uniform(low=0.08, high=0.12, size=10)
print(f"\nnp.random.uniform(low=0.08, high=0.12, size=10)")
print(f"Mínimo: 8% | Máximo: 12% | Cantidad: 10")
print(f"\nResultados: {tasas_uniforme}")
print(f"Promedio real: {tasas_uniforme.mean()*100:.2f}% (debería ≈ 10%)")

print("\n" + "-"*70)
print("\n3️⃣  DISTRIBUCIÓN TRIANGULAR")

# Triangular
costos_triangular = np.random.triangular(left=20000, mode=30000, right=50000, size=10)
print(f"\nnp.random.triangular(left=20000, mode=30000, right=50000, size=10)")
print(f"Pesimista: $20K | Base: $30K | Optimista: $50K")
print(f"\nResultados: {costos_triangular.astype(int)}")
print(f"Promedio real: ${costos_triangular.mean():,.0f}")

print("\n" + "="*70)
print("\n📊 VISUALIZACIÓN: 10,000 muestras de cada distribución")

np.random.seed(42)
n_samples = 10000

data_normal = np.random.normal(30000, 5000, n_samples)
data_uniforme = np.random.uniform(20000, 40000, n_samples)
data_triangular = np.random.triangular(20000, 30000, 40000, n_samples)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Normal
axes[0].hist(data_normal, bins=50, color='steelblue', alpha=0.7, edgecolor='black')
axes[0].axvline(data_normal.mean(), color='red', linestyle='--', linewidth=2, label=f'Media: ${data_normal.mean():,.0f}')
axes[0].set_title('Distribución Normal', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Valor ($)')
axes[0].set_ylabel('Frecuencia')
axes[0].legend()
axes[0].grid(alpha=0.3)

# Uniforme
axes[1].hist(data_uniforme, bins=50, color='orange', alpha=0.7, edgecolor='black')
axes[1].axvline(data_uniforme.mean(), color='red', linestyle='--', linewidth=2, label=f'Media: ${data_uniforme.mean():,.0f}')
axes[1].set_title('Distribución Uniforme', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Valor ($)')
axes[1].set_ylabel('Frecuencia')
axes[1].legend()
axes[1].grid(alpha=0.3)

# Triangular
axes[2].hist(data_triangular, bins=50, color='green', alpha=0.7, edgecolor='black')
axes[2].axvline(data_triangular.mean(), color='red', linestyle='--', linewidth=2, label=f'Media: ${data_triangular.mean():,.0f}')
axes[2].set_title('Distribución Triangular', fontsize=12, fontweight='bold')
axes[2].set_xlabel('Valor ($)')
axes[2].set_ylabel('Frecuencia')
axes[2].legend()
axes[2].grid(alpha=0.3)

plt.tight_layout()
display(fig)
plt.close()

print("\n💡 OBSERVACIONES:")
print("  • Normal: Simétrica, valores centrales más probables")
print("  • Uniforme: Todos los valores igualmente probables")
print("  • Triangular: Sesgo hacia el modo (valor más probable)")

print("\n" + "="*70)
print("✅ Generación de números aleatorios completada")
````

## 📍 Ejemplo Paso a Paso: Simulación Simple

### 🎯 Problema: Evaluación de Proyecto con Incertidumbre

**Escenario:**
* Inversión inicial: $100,000
* Flujos esperados: $40,000/año durante 3 años
* Incertidumbre: Flujos pueden variar ±$10,000 (distribución normal)
* Tasa de descuento: 10%

**Pregunta:** ¿Cuál es la probabilidad de que el proyecto sea viable (VAN > 0)?

---

### 📋 Enfoque Determinista (Tradicional)

**Un solo escenario:**

```python
flujos = np.array([-100000, 40000, 40000, 40000])
tasa = 0.10
periodos = np.array([0, 1, 2, 3])

VAN = np.sum(flujos / (1 + tasa) ** periodos)
VAN = $-100,000 + $36,364 + $33,058 + $30,053
VAN = -$525

Decisión: RECHAZAR (VAN < 0)
```

❌ **Problema:** Ignora la incertidumbre, decisiones basadas en un solo escenario

---

### 🎲 Enfoque Monte Carlo (Estocástico)

**Proceso:**

1. **Generar 10,000 escenarios aleatorios** de flujos de caja
   ```python
   np.random.seed(42)
   flujos_sim = np.random.normal(40000, 10000, (10000, 3))
   # 10,000 filas (escenarios) × 3 columnas (años)
   ```

2. **Calcular VAN para cada escenario**
   ```python
   periodos = np.array([1, 2, 3])
   van_sim = -100000 + np.sum(flujos_sim / (1.10)**periodos, axis=1)
   # Resultado: array de 10,000 VAN
   ```

3. **Analizar distribución de resultados**
   ```python
   van_promedio = van_sim.mean()  # VAN esperado
   van_std = van_sim.std()        # Volatilidad del VAN
   P_exito = (van_sim > 0).mean() # Probabilidad de viabilidad
   ```

4. **Percentiles (escenarios)**
   ```python
   P10 = np.percentile(van_sim, 10)  # Pesimista
   P50 = np.percentile(van_sim, 50)  # Mediana
   P90 = np.percentile(van_sim, 90)  # Optimista
   ```

---

### 📊 Resultados Típicos

| Métrica | Valor |
|---------|-------|
| VAN Promedio | $-500 |
| Desv. Estándar | $20,000 |
| P10 (pesimista) | -$25,000 |
| P50 (mediana) | -$500 |
| P90 (optimista) | $24,000 |
| **P(VAN > 0)** | **49%** |

**Decisión informada:**
* El proyecto tiene 49% de probabilidad de éxito
* En el peor escenario (P10), perdemos $25K
* En el mejor escenario (P90), ganamos $24K
* **Recomendación:** RECHAZAR (probabilidad < 50%)

✅ **Ventaja:** Decisión basada en **toda la distribución**, no un solo valor

````python
import numpy as np
import matplotlib.pyplot as plt

print("📍 SIMULACIÓN MONTE CARLO: Ejemplo Simple Paso a Paso")
print("="*70)

print("\n🎯 PROYECTO: Inversión con flujos inciertos")
print("\nParámetros:")
print("  • Inversión: $100,000")
print("  • Flujos esperados: $40,000/año (3 años)")
print("  • Incertidumbre: ±$10,000 (normal)")
print("  • Tasa: 10%")

print("\n" + "-"*70)
print("\n1️⃣  ENFOQUE DETERMINISTA (escenario único):")

inversion = 100000
flujos_det = np.array([40000, 40000, 40000])
tasa = 0.10
periodos = np.array([1, 2, 3])

van_det = -inversion + np.sum(flujos_det / (1 + tasa) ** periodos)

print(f"\nFlujos: {flujos_det}")
print(f"VAN determinista: ${van_det:,.2f}")

if van_det > 0:
    print("🟢 Decisión: ACEPTAR")
else:
    print("🔴 Decisión: RECHAZAR")

print("\n⚠️  Problema: Ignora la incertidumbre, decisión binaria")

print("\n" + "="*70)
print("\n2️⃣  ENFOQUE MONTE CARLO (10,000 escenarios):")

# Simulación
np.random.seed(42)
n_sim = 10000
flujos_esperados = 40000
incertidumbre = 10000

print(f"\n🎲 Generando {n_sim:,} escenarios aleatorios...")

# Generar flujos aleatorios para cada año
flujos_sim = np.random.normal(flujos_esperados, incertidumbre, (n_sim, 3))

print(f"Shape de flujos simulados: {flujos_sim.shape}")
print(f"  • {n_sim:,} filas (escenarios)")
print(f"  • 3 columnas (años)")

print(f"\nEjemplos de flujos simulados (primeros 5 escenarios):")
for i in range(5):
    print(f"  Escenario {i+1}: ${flujos_sim[i, 0]:>7,.0f}  ${flujos_sim[i, 1]:>7,.0f}  ${flujos_sim[i, 2]:>7,.0f}")

# Calcular VAN para cada escenario
print(f"\n⚡ Calculando VAN de {n_sim:,} escenarios (vectorizado)...")
van_sim = -inversion + np.sum(flujos_sim / (1 + tasa) ** periodos, axis=1)

print(f"\n✅ {n_sim:,} VAN calculados")
print(f"\nEjemplos de VAN (primeros 5):")
for i in range(5):
    print(f"  Escenario {i+1}: ${van_sim[i]:>10,.2f}")

print("\n" + "-"*70)
print("\n📊 ESTADÍSTICAS DE LA DISTRIBUCIÓN:")

van_promedio = van_sim.mean()
van_std = van_sim.std()
van_min = van_sim.min()
van_max = van_sim.max()

print(f"\nVAN Promedio:  ${van_promedio:>12,.2f}")
print(f"Desv. Estándar: ${van_std:>12,.2f}")
print(f"Mínimo:         ${van_min:>12,.2f}")
print(f"Máximo:         ${van_max:>12,.2f}")

# Percentiles
p10, p50, p90 = np.percentile(van_sim, [10, 50, 90])

print(f"\nPercentiles:")
print(f"  P10 (pesimista): ${p10:>12,.2f}  (10% de escenarios peores)")
print(f"  P50 (mediana):   ${p50:>12,.2f}  (escenario típico)")
print(f"  P90 (optimista): ${p90:>12,.2f}  (10% de escenarios mejores)")

# Probabilidad de éxito
prob_exito = (van_sim > 0).mean()

print(f"\n🎯 PROBABILIDAD DE VIABILIDAD:")
print(f"\nP(VAN > 0) = {prob_exito*100:.1f}%")
print(f"\nDe {n_sim:,} escenarios:")
print(f"  • {(van_sim > 0).sum():,} tienen VAN positivo (🟢 viable)")
print(f"  • {(van_sim <= 0).sum():,} tienen VAN negativo (🔴 no viable)")

print("\n" + "="*70)
print("\n📈 VISUALIZACIÓN:")

fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(van_sim, bins=50, color='steelblue', alpha=0.7, edgecolor='black')
ax.axvline(van_promedio, color='red', linestyle='--', linewidth=2.5, label=f'Promedio: ${van_promedio:,.0f}')
ax.axvline(0, color='black', linestyle='-', linewidth=2, label='VAN = 0 (punto de equilibrio)')
ax.axvline(p10, color='orange', linestyle=':', linewidth=2, label=f'P10: ${p10:,.0f}')
ax.axvline(p90, color='green', linestyle=':', linewidth=2, label=f'P90: ${p90:,.0f}')

# Sombrear regiones
ax.axvspan(van_sim.min(), 0, alpha=0.15, color='red', label='Zona no viable (VAN < 0)')
ax.axvspan(0, van_sim.max(), alpha=0.15, color='green', label='Zona viable (VAN > 0)')

ax.set_xlabel('VAN ($)', fontsize=12, fontweight='bold')
ax.set_ylabel('Frecuencia', fontsize=12, fontweight='bold')
ax.set_title(f'Distribución de VAN - {n_sim:,} Escenarios Monte Carlo', fontsize=14, fontweight='bold')
ax.legend(loc='upper right', fontsize=9)
ax.grid(alpha=0.3)

plt.tight_layout()
display(fig)
plt.close()

print("\n" + "="*70)
print("\n📊 DECISIÓN INFORMADA:")

if prob_exito >= 0.70:
    print("\n🟢 RECOMENDACIÓN: ACEPTAR el proyecto")
    print(f"   • Alta probabilidad de éxito ({prob_exito*100:.1f}%)")
    print(f"   • Escenario típico (P50): ${p50:,.0f}")
elif prob_exito >= 0.50:
    print("\n🟡 RECOMENDACIóN: CONSIDERAR CON CAUTELA")
    print(f"   • Probabilidad moderada ({prob_exito*100:.1f}%)")
    print(f"   • Analizar mitigación de riesgos")
else:
    print("\n🔴 RECOMENDACIÓN: RECHAZAR el proyecto")
    print(f"   • Baja probabilidad de éxito ({prob_exito*100:.1f}%)")
    print(f"   • Riesgo elevado de pérdida")

print(f"\n💡 Comparación con enfoque determinista:")
print(f"   • Determinista: VAN = ${van_det:,.0f} (decisión binaria)")
print(f"   • Monte Carlo: Rango ${p10:,.0f} a ${p90:,.0f} con P={prob_exito*100:.0f}%")
print(f"   • Monte Carlo da MUCHA MÁS información para decidir")

print("\n" + "="*70)
print("✅ Simulación simple completada")
````

## 📊 Interpretación de Percentiles

### 🎯 ¿Qué son los Percentiles?

**Definición:** Un percentil P_X indica que X% de los valores están por debajo de ese punto.

---

### 📈 Percentiles Comunes en Análisis de Riesgo

| Percentil | Nombre | Interpretación | Uso |
|-----------|--------|----------------|-----|
| **P10** | Pesimista | Solo 10% de escenarios son peores | **Peor caso realista** |
| **P25** | Q1 | 25% de escenarios peores | Primer cuartil |
| **P50** | Mediana | 50% arriba, 50% abajo | **Escenario típico** (mejor que promedio si hay outliers) |
| **P75** | Q3 | 75% de escenarios peores | Tercer cuartil |
| **P90** | Optimista | 90% de escenarios peores | **Mejor caso realista** |
| **P95** | Muy optimista | Solo 5% mejores | Caso muy favorable |
| **P99** | Extremo | Solo 1% mejores | Casi imposible |

---

### 💡 ¿Por Qué P10/P50/P90?

**Estándar de la industria:**
* **P10:** Escenario pesimista para análisis de riesgo
  - "Si todo sale mal, ¿qué tan mal?"
  - Útil para calcular capital de contingencia
  
* **P50:** Mediana, el resultado más típico
  - Más robusto que el promedio (no afectado por outliers extremos)
  - "¿Qué resultado es más probable?"
  
* **P90:** Escenario optimista pero realista
  - "Si todo sale bien, ¿qué tan bien?"
  - No es el máximo (evita valores extremos poco probables)

---

### 🎲 Ejemplo: Proyecto con VAN Simulado

```python
P10 = -$5,000   # Solo 10% de escenarios son peores
P50 = $12,000   # Resultado típico (mediana)
P90 = $30,000   # Solo 10% de escenarios son mejores
```

**Interpretación para inversores:**
* **Caso base:** Esperar VAN cercano a $12K (P50)
* **Downside:** En el peor escenario realista, pierdo $5K (P10)
* **Upside:** En el mejor escenario realista, gano $30K (P90)
* **Asimetría:** Riesgo de $5K vs potencial de $30K → favorable

---

### 📉 Value at Risk (VaR)

**Concepto relacionado:**
* **VaR al 90%** = -P10 (pérdida máxima en el 10% peor de casos)
* **VaR al 95%** = -P5
* **VaR al 99%** = -P1

**Ejemplo:**
* Si P10 del VAN = -$5,000
* VaR al 90% = $5,000
* Interpretación: "Hay 90% de probabilidad de NO perder más de $5,000"

---

### 🎯 Reglas de Decisión con Percentiles

**Proyectos de inversión:**

1. **Evaluación de viabilidad:**
   - Si **P50 > 0** → Proyecto típicamente rentable
   - Si **P10 > 0** → Proyecto rentable incluso en escenario pesimista (bajo riesgo)
   - Si **P10 < 0 pero P50 > 0** → Rentable pero riesgoso
   - Si **P50 < 0** → Proyecto no rentable en escenario típico

2. **Evaluación de riesgo/retorno:**
   ```python
   Upside = P90 - P50
   Downside = P50 - P10
   Ratio = Upside / Downside
   
   Si Ratio > 1.5 → Asimetría favorable (más upside que downside)
   ```

3. **Capital de contingencia:**
   ```python
   Contingencia = P50 - P10
   # Reservar esta cantidad para cubrir escenarios pesimistas
   ```

---

### 💼 Comunicación a Stakeholders

**En vez de:**
> "El VAN esperado es $12,000"

**Mejor:**
> "El proyecto tiene un VAN mediano de $12,000. En un escenario pesimista (P10), el VAN es -$5,000. En un escenario optimista (P90), el VAN es $30,000. La probabilidad de que el proyecto sea rentable (VAN > 0) es del 75%."

## 🔬 Análisis de Sensibilidad con Monte Carlo

### 🎯 ¿Qué es el Análisis de Sensibilidad?

Evaluar **cómo cambia el resultado** cuando varían los parámetros de entrada.

**Pregunta clave:** ¿Qué variables tienen mayor impacto en el VAN?

---

### 📈 Tipos de Sensibilidad

1. **Sensibilidad Univariada (clásica)**
   - Variar UN parámetro a la vez
   - Mantener los demás constantes
   - Ejemplo: VAN vs tasa de descuento (ya lo vimos en notebook 02_03)

2. **Sensibilidad Multivariada (Monte Carlo)**
   - Variar TODOS los parámetros simultáneamente
   - Captura interacciones entre variables
   - Más realista: en la vida real, todo varía a la vez

---

### 🔬 Escenarios Comparativos

Comparar distribuciones de VAN bajo diferentes supuestos:

**Ejemplo: Impacto de Reducir Incertidumbre**

| Escenario | Desv. Estándar Flujos | P(VAN>0) | P50 | Riesgo |
|-----------|-------------------------|----------|-----|--------|
| Base | ±$5,000 | 75% | $13K | Moderado |
| Optimista | ±$3,000 | 85% | $13K | Bajo |
| Pesimista | ±$8,000 | 65% | $13K | Alto |

💡 **Insight:** Reducir incertidumbre aumenta probabilidad de éxito SIN cambiar VAN esperado

---

### 📊 Tornado Diagram (Diagrama de Tornado)

**Herramienta visual** para identificar qué variables importan más:

1. Variar cada parámetro entre P10 y P90
2. Calcular impacto en VAN
3. Ordenar de mayor a menor impacto

**Resultado:**
```
                VAN: $5K --- $15K --- $25K
Tasa descuento: |========|         (mayor impacto)
Flujos Año 1:   |======|           (alto impacto)
Flujos Año 2:   |====|             (impacto medio)
Flujos Año 4:   |==|               (bajo impacto)
```

**Interpretación:**
* Tasa de descuento es la variable más crítica
* Flujos del Año 1 también muy importantes
* Flujos del Año 4 tienen poco impacto (descuento fuerte)

🎯 **Acción:** Enfocar esfuerzos en mejorar estimación de variables críticas

---

### 🛡️ Estrategias de Mitigación de Riesgo

**Basadas en análisis de sensibilidad:**

1. **Cobertura (Hedging)**
   - Si tasa de interés es crítica → Contratar tasa fija
   
2. **Diversificación**
   - Si un flujo es muy volátil → Diversificar fuentes de ingresos
   
3. **Contratos a precio fijo**
   - Si costos son volátiles → Negociar contratos de largo plazo
   
4. **Opciones reales**
   - Diseñar proyecto con flexibilidad para expandir/contraer según escenarios

---

### 📊 Comparación: Antes vs Después de Mitigación

**Ejemplo:**

| Métrica | Antes | Después Mitigación | Mejora |
|---------|-------|----------------------|--------|
| P(VAN>0) | 65% | 85% | +20pp |
| P10 | -$10K | -$2K | +$8K |
| Desv. Est. | $20K | $12K | -40% |

✅ **Mitigación reduce riesgo downside sin sacrificar upside**

````python
import numpy as np
import matplotlib.pyplot as plt

print("🔬 COMPARACIÓN DE ESCENARIOS: Impacto de Incertidumbre")
print("="*70)

print("\n🎯 PREGUNTA: ¿Cómo afecta el nivel de incertidumbre a la viabilidad?")

# Parámetros comunes
inversion = 100000
flujos_base = np.array([30000, 35000, 40000, 45000])
tasa = 0.10
n_sim = 10000
periodos = np.arange(1, len(flujos_base) + 1)

# Tres escenarios de incertidumbre
escenarios = {
    'Baja Incertidumbre': 3000,
    'Incertidumbre Base': 5000,
    'Alta Incertidumbre': 8000
}

print("\n📊 ESCENARIOS A SIMULAR:")
for nombre, std in escenarios.items():
    print(f"  • {nombre:25s}: Desv. Estándar = ±${std:,}")

print("\n" + "="*70)
print("\n🎲 Ejecutando 10,000 simulaciones por escenario...")

# Almacenar resultados
resultados = {}

np.random.seed(42)
for nombre, desv_std in escenarios.items():
    flujos_sim = np.random.normal(flujos_base, desv_std, (n_sim, len(flujos_base)))
    van_sim = -inversion + np.sum(flujos_sim / (1 + tasa) ** periodos, axis=1)
    
    resultados[nombre] = {
        'van': van_sim,
        'promedio': van_sim.mean(),
        'std': van_sim.std(),
        'p10': np.percentile(van_sim, 10),
        'p50': np.percentile(van_sim, 50),
        'p90': np.percentile(van_sim, 90),
        'prob_exito': (van_sim > 0).mean()
    }

print("✅ Simulaciones completadas")

print("\n" + "="*70)
print("\n📊 COMPARACIÓN DE RESULTADOS:")

print("\n{:<25} {:>12} {:>12} {:>12} {:>10}".format(
    "Escenario", "P10", "P50 (Mediana)", "P90", "P(VAN>0)"
))
print("-" * 75)

for nombre, res in resultados.items():
    print(f"{nombre:<25} ${res['p10']:>10,.0f} ${res['p50']:>10,.0f} ${res['p90']:>10,.0f}   {res['prob_exito']*100:>6.1f}%")

print("\n📊 OBSERVACIONES:")

prob_base = resultados['Incertidumbre Base']['prob_exito']
prob_baja = resultados['Baja Incertidumbre']['prob_exito']
prob_alta = resultados['Alta Incertidumbre']['prob_exito']

print(f"\n1. P50 es similar en todos los escenarios (~${resultados['Incertidumbre Base']['p50']:,.0f})")
print(f"   ➜ La mediana NO cambia con la incertidumbre")

print(f"\n2. P(VAN>0) cambia significativamente:")
print(f"   • Baja incert.: {prob_baja*100:.1f}% (+{(prob_baja-prob_base)*100:.1f}pp vs base)")
print(f"   • Base:         {prob_base*100:.1f}%")
print(f"   • Alta incert.: {prob_alta*100:.1f}% ({(prob_alta-prob_base)*100:.1f}pp vs base)")
print(f"   ➜ REDUCIR INCERTIDUMBRE aumenta probabilidad de éxito")

p10_base = resultados['Incertidumbre Base']['p10']
p10_baja = resultados['Baja Incertidumbre']['p10']
p10_alta = resultados['Alta Incertidumbre']['p10']

print(f"\n3. Riesgo downside (P10):")
print(f"   • Baja incert.: ${p10_baja:>10,.0f}  (mejor P10)")
print(f"   • Base:         ${p10_base:>10,.0f}")
print(f"   • Alta incert.: ${p10_alta:>10,.0f}  (peor P10)")
print(f"   ➜ MAYOR incertidumbre = MAYOR riesgo de pérdida")

print("\n" + "="*70)
print("\n📈 VISUALIZACIÓN: Distribución de VAN por Escenario")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Histogramas superpuestos
colores = ['green', 'steelblue', 'red']
alphas = [0.5, 0.5, 0.5]

for (nombre, res), color, alpha in zip(resultados.items(), colores, alphas):
    ax1.hist(res['van'], bins=50, color=color, alpha=alpha, edgecolor='black', label=nombre)

ax1.axvline(0, color='black', linestyle='-', linewidth=2, label='VAN = 0')
ax1.set_xlabel('VAN ($)', fontsize=12, fontweight='bold')
ax1.set_ylabel('Frecuencia', fontsize=12, fontweight='bold')
ax1.set_title('Distribución de VAN por Nivel de Incertidumbre', fontsize=13, fontweight='bold')
ax1.legend(loc='upper right')
ax1.grid(alpha=0.3)

# Box plots comparativos
data_boxplot = [resultados[nombre]['van'] for nombre in escenarios.keys()]
labels_boxplot = ['Baja\nIncert.', 'Base', 'Alta\nIncert.']

bp = ax2.boxplot(data_boxplot, labels=labels_boxplot, patch_artist=True,
                  boxprops=dict(facecolor='lightblue', color='blue'),
                  medianprops=dict(color='red', linewidth=2),
                  whiskerprops=dict(color='blue', linewidth=1.5),
                  capprops=dict(color='blue', linewidth=1.5))

ax2.axhline(0, color='black', linestyle='-', linewidth=2, label='VAN = 0')
ax2.set_ylabel('VAN ($)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Escenario', fontsize=12, fontweight='bold')
ax2.set_title('Box Plot Comparativo', fontsize=13, fontweight='bold')
ax2.legend()
ax2.grid(alpha=0.3, axis='y')

plt.tight_layout()
display(fig)
plt.close()

print("\n" + "="*70)
print("\n💡 INSIGHTS CLAVE:")

print("\n1. 🎯 VALOR ESPERADO (P50) es ROBUSTO")
print("   • No cambia significativamente con nivel de incertidumbre")
print("   • Depende de flujos base, no de volatilidad")

print("\n2. 📉 PROBABILIDAD DE ÉXITO es SENSIBLE")
print("   • Reducir incertidumbre aumenta P(VAN>0)")
print("   • Estrategia: Contratos a precio fijo, coberturas, diversificación")

print("\n3. 🛡️ RIESGO DOWNSIDE es CONTROLABLE")
print("   • Gestionar incertidumbre mejora P10 (peor caso)")
print("   • Proyectos más predecibles = menos riesgo de pérdidas grandes")

print("\n" + "="*70)
print("\n🎯 RECOMENDACIÓN ESTRATÉGICA:")

if prob_baja >= 0.80 and p10_baja > 0:
    print("\n🟢 Si se puede REDUCIR incertidumbre:")
    print(f"   • P(VAN>0) sube a {prob_baja*100:.0f}%")
    print(f"   • P10 mejora a ${p10_baja:,.0f}")
    print(f"   ➜ INVERTIR en reducción de incertidumbre (estudios, contratos)")
elif prob_alta < 0.60:
    print("\n🔴 Si incertidumbre es ALTA:")
    print(f"   • P(VAN>0) baja a {prob_alta*100:.0f}%")
    print(f"   • Riesgo elevado de pérdida")
    print(f"   ➜ POSPONER proyecto hasta reducir incertidumbre")
else:
    print("\n🟡 Con incertidumbre BASE:")
    print(f"   • P(VAN>0) = {prob_base*100:.0f}% (aceptable)")
    print(f"   • Monitorear riesgos activamente")

print("\n" + "="*70)
print("✅ Comparación de escenarios completada")
````

## 🔗 Correlaciones entre Variables

### 🎯 ¿Por Qué Importan las Correlaciones?

En la vida real, las variables **NO son independientes**:

**Ejemplo:**
* Si la economía va bien → Ventas altas Y costos altos (ambos suben juntos)
* Si hay recesión → Ventas bajas Y costos bajos (ambos bajan juntos)

⚠️ **Ignorar correlaciones** puede:
* Subestimar riesgo (si variables se mueven juntas)
* Sobreestimar beneficios de diversificación

---

### 📊 Tipos de Correlación

| Correlación | Valor | Significado | Ejemplo |
|--------------|-------|-------------|--------|
| **Positiva fuerte** | +0.7 a +1.0 | Suben/bajan juntas | Ventas y costos variables |
| **Positiva moderada** | +0.3 a +0.7 | Tendencia a moverse juntas | PIB e ingresos |
| **No correlacionadas** | -0.3 a +0.3 | Independientes | Clima y tasas de interés |
| **Negativa moderada** | -0.7 a -0.3 | Tendencia opuesta | Tasas interés y valuación bonos |
| **Negativa fuerte** | -1.0 a -0.7 | Opuestas | Cobertura perfecta |

---

### 💻 Implementación con NumPy

**Variables independientes (incorrecto si hay correlación):**
```python
# Genera flujos Año 1 y Año 2 INDEPENDIENTES
flujo1 = np.random.normal(30000, 5000, 10000)
flujo2 = np.random.normal(35000, 5000, 10000)
# Problema: Si Año 1 va bien, no afecta Año 2 (poco realista)
```

**Variables correlacionadas (correcto):**
```python
# Crear matriz de covarianza
media = [30000, 35000]
std = [5000, 6000]
corr = 0.7  # Correlación positiva

cov_matrix = np.array([
    [std[0]**2, corr*std[0]*std[1]],
    [corr*std[0]*std[1], std[1]**2]
])

# Generar flujos correlacionados
flujos = np.random.multivariate_normal(media, cov_matrix, 10000)
# Ahora si Año 1 va bien, Año 2 tiende a ir bien también
```

---

### 📉 Impacto en el Análisis de Riesgo

**Ejemplo: Flujos de 2 años con $30K esperados, std $5K**

| Supuesto | VAN Std | P(VAN>0) | Interpretación |
|----------|---------|----------|----------------|
| **Independientes** (corr=0) | $8,000 | 78% | Subestima riesgo |
| **Correlación positiva** (corr=+0.7) | $12,000 | 70% | Realista: años buenos/malos juntos |
| **Correlación perfecta** (corr=+1.0) | $14,000 | 65% | Riesgo máximo |

💡 **Insight:** Correlación positiva AUMENTA riesgo (más volatilidad del VAN)

---

### 🎯 Correlaciones Típicas en Finanzas

**Flujos de caja de un proyecto:**
* Años consecutivos: +0.5 a +0.8 (momentum de negocio)
* Ingresos y costos variables: +0.6 a +0.9 (ambos dependen de volumen)
* Flujos de diferentes proyectos mismo sector: +0.4 a +0.7

**Factores macroeconómicos:**
* PIB y ventas: +0.7 a +0.9
* Inflación y costos: +0.6 a +0.8
* Tipo de cambio y exportaciones: -0.5 a -0.8 (dólar alto, menos exportaciones)

---

### 🛡️ Diversificación y Correlación

**Principio clave:** Diversificar es más efectivo con activos de **baja correlación**

**Ejemplo: 2 proyectos con VAN esperado $10K, std $20K**

| Correlación | Std del Portafolio | Beneficio Diversificación |
|--------------|--------------------|--------------------------|
| +1.0 | $20K | 0% (sin beneficio) |
| +0.5 | $17K | 15% reducción riesgo |
| 0.0 | $14K | 30% reducción riesgo |
| -0.5 | $11K | 45% reducción riesgo |
| -1.0 | $0 | 100% (cobertura perfecta) |

✅ **Buscar proyectos con correlación baja/negativa para diversificar riesgo**

---

### 📊 Cómo Estimar Correlaciones

1. **Datos históricos**
   ```python
   correlacion = np.corrcoef(flujos_historicos_1, flujos_historicos_2)[0,1]
   ```

2. **Opinión de expertos**
   * Preguntar: "Si el Año 1 va 20% mejor, ¿cómo afecta al Año 2?"
   * Traducir respuesta a correlación

3. **Modelos económicos**
   * Correlación con PIB, inflación, etc.
   * Usar correlaciones publicadas de variables macro

4. **Sensibilidad**
   * Probar múltiples escenarios de correlación (0, 0.5, 0.8)
   * Ver cómo cambia el resultado

````python
import numpy as np
import matplotlib.pyplot as plt

print("🔗 SIMULACIÓN MONTE CARLO CON CORRELACIONES")
print("="*70)

print("\n🎯 COMPARACIÓN: Flujos Independientes vs Correlacionados")

# Parámetros
inversion = 100000
flujos_esperados = np.array([30000, 35000, 40000, 45000])
desv_std = np.array([5000, 5500, 6000, 6500])
tasa = 0.10
n_sim = 10000
periodos = np.arange(1, len(flujos_esperados) + 1)

print("\nFlujos esperados: ", flujos_esperados)
print("Desviaciones est.: ", desv_std)

print("\n" + "="*70)
print("\n1️⃣  ESCENARIO 1: Flujos INDEPENDIENTES (correlación = 0)")

np.random.seed(42)
flujos_indep = np.random.normal(flujos_esperados, desv_std, (n_sim, 4))
van_indep = -inversion + np.sum(flujos_indep / (1 + tasa) ** periodos, axis=1)

print(f"\n🎲 {n_sim:,} simulaciones generadas (cada año independiente)")
print(f"\nResultados:")
print(f"  VAN Promedio:  ${van_indep.mean():>12,.0f}")
print(f"  Desv. Estándar: ${van_indep.std():>12,.0f}")
print(f"  P10:           ${np.percentile(van_indep, 10):>12,.0f}")
print(f"  P50:           ${np.percentile(van_indep, 50):>12,.0f}")
print(f"  P90:           ${np.percentile(van_indep, 90):>12,.0f}")
print(f"  P(VAN > 0):    {(van_indep > 0).mean()*100:>12.1f}%")

print("\n" + "="*70)
print("\n2️⃣  ESCENARIO 2: Flujos CORRELACIONADOS (correlación = +0.7)")

print("\n🔗 Creando matriz de covarianza con correlación +0.7...")

# Crear matriz de correlación
corr = 0.7
corr_matrix = np.full((4, 4), corr)
np.fill_diagonal(corr_matrix, 1.0)

print("\nMatriz de correlación:")
print(corr_matrix)

# Convertir a matriz de covarianza
cov_matrix = np.outer(desv_std, desv_std) * corr_matrix

print("\n🎲 Generando flujos con np.random.multivariate_normal()...")

np.random.seed(42)
flujos_corr = np.random.multivariate_normal(flujos_esperados, cov_matrix, n_sim)
van_corr = -inversion + np.sum(flujos_corr / (1 + tasa) ** periodos, axis=1)

print(f"\n🎲 {n_sim:,} simulaciones generadas (flujos correlacionados)")
print(f"\nResultados:")
print(f"  VAN Promedio:  ${van_corr.mean():>12,.0f}")
print(f"  Desv. Estándar: ${van_corr.std():>12,.0f}  ⬆️ (mayor que independiente)")
print(f"  P10:           ${np.percentile(van_corr, 10):>12,.0f}")
print(f"  P50:           ${np.percentile(van_corr, 50):>12,.0f}")
print(f"  P90:           ${np.percentile(van_corr, 90):>12,.0f}")
print(f"  P(VAN > 0):    {(van_corr > 0).mean()*100:>12.1f}%  ⬇️ (menor que independiente)")

print("\n" + "="*70)
print("\n📊 COMPARACIÓN DIRECTA:")

print("\n{:<25} {:>15} {:>15} {:>12}".format(
    "Métrica", "Independientes", "Correlacionados", "Diferencia"
))
print("-" * 75)

metricas = [
    ("VAN Promedio", van_indep.mean(), van_corr.mean()),
    ("Desv. Estándar", van_indep.std(), van_corr.std()),
    ("P10 (pesimista)", np.percentile(van_indep, 10), np.percentile(van_corr, 10)),
    ("P50 (mediana)", np.percentile(van_indep, 50), np.percentile(van_corr, 50)),
    ("P90 (optimista)", np.percentile(van_indep, 90), np.percentile(van_corr, 90)),
]

for metrica, val_indep, val_corr in metricas:
    dif = val_corr - val_indep
    print(f"{metrica:<25} ${val_indep:>12,.0f} ${val_corr:>12,.0f}   ${dif:>9,.0f}")

prob_indep = (van_indep > 0).mean()
prob_corr = (van_corr > 0).mean()
dif_prob = (prob_corr - prob_indep) * 100

print(f"{'P(VAN > 0)':<25} {prob_indep*100:>12.1f}% {prob_corr*100:>12.1f}%   {dif_prob:>9.1f}pp")

print("\n💡 OBSERVACIONES:")
print(f"\n1. VAN Promedio (P50) es SIMILAR en ambos casos")
print(f"   • Independientes: ${np.percentile(van_indep, 50):,.0f}")
print(f"   • Correlacionados: ${np.percentile(van_corr, 50):,.0f}")
print(f"   ➜ Correlación NO afecta el valor esperado")

print(f"\n2. Desv. Estándar es MAYOR con correlación")
print(f"   • Independientes: ${van_indep.std():,.0f}")
print(f"   • Correlacionados: ${van_corr.std():,.0f} (+{((van_corr.std()/van_indep.std()-1)*100):.0f}%)")
print(f"   ➜ Correlación positiva AUMENTA riesgo")

print(f"\n3. P(VAN > 0) es MENOR con correlación")
print(f"   • Independientes: {prob_indep*100:.1f}%")
print(f"   • Correlacionados: {prob_corr*100:.1f}% ({dif_prob:.1f}pp)")
print(f"   ➜ Ignorar correlaciones SUBESTIMA el riesgo")

print("\n" + "="*70)
print("\n📈 VISUALIZACIÓN: Distribución con y sin Correlación")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Histogramas
axes[0].hist(van_indep, bins=50, color='blue', alpha=0.6, edgecolor='black', label='Independientes')
axes[0].hist(van_corr, bins=50, color='red', alpha=0.6, edgecolor='black', label='Correlacionados (r=0.7)')
axes[0].axvline(0, color='black', linestyle='-', linewidth=2)
axes[0].set_xlabel('VAN ($)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Frecuencia', fontsize=12, fontweight='bold')
axes[0].set_title('Distribución de VAN: Independientes vs Correlacionados', fontsize=12, fontweight='bold')
axes[0].legend()
axes[0].grid(alpha=0.3)

# Box plots comparativos
box_data = [van_indep, van_corr]
box_labels = ['Independientes\n(r=0)', 'Correlacionados\n(r=0.7)']

bp = axes[1].boxplot(box_data, labels=box_labels, patch_artist=True,
                      boxprops=dict(facecolor='lightblue', color='blue'),
                      medianprops=dict(color='red', linewidth=2.5),
                      whiskerprops=dict(color='blue', linewidth=1.5),
                      capprops=dict(color='blue', linewidth=1.5))

axes[1].axhline(0, color='black', linestyle='-', linewidth=2, label='VAN = 0')
axes[1].set_ylabel('VAN ($)', fontsize=12, fontweight='bold')
axes[1].set_title('Box Plot Comparativo', fontsize=12, fontweight='bold')
axes[1].legend()
axes[1].grid(alpha=0.3, axis='y')

plt.tight_layout()
display(fig)
plt.close()

print("\n" + "="*70)
print("\n🎯 IMPLICACIONES PARA LA PRÁCTICA:")

print("\n⚠️  NO asumir independencia automáticamente:")
print("   1. Investigar si hay factores comunes (economía, industria, etc.)")
print("   2. Estimar correlaciones con datos históricos o expertos")
print("   3. Usar np.random.multivariate_normal() para correlaciones")

print("\n✅ Incluir correlaciones hace el análisis MÁS REALISTA:")
print("   • Captura cómo variables se mueven juntas")
print("   • Refleja mejor el riesgo real del proyecto")
print("   • Evita sorpresas desagradables")

print("\n🔗 En proyectos reales:")
print("   • Flujos de años consecutivos suelen tener r = 0.5 a 0.8")
print("   • Ingresos y costos variables: r = 0.6 a 0.9")
print("   • Proyectos en misma industria: r = 0.4 a 0.7")

print("\n" + "="*70)
print("✅ Análisis con correlaciones completado")
````
