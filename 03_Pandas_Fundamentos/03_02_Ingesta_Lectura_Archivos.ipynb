## 📄 pd.read_csv() - Archivos CSV

### 📚 Definición

**CSV** (Comma-Separated Values) es el formato más común para intercambio de datos.

```
Fecha,Sucursal,Ventas,Estado
2024-01-01,Centro,15000,Cobrado
2024-01-02,Norte,22000,Pendiente
```

---

### 🔑 Parámetros Importantes

| Parámetro | Descripción | Ejemplo |
|-----------|-------------|--------|
| `filepath` | Ruta del archivo | `'datos.csv'` |
| `sep` | Separador | `','` (default), `';'`, `'\t'` |
| `encoding` | Codificación | `'utf-8'`, `'latin-1'`, `'cp1252'` |
| `header` | Fila de encabezados | `0` (default), `None` |
| `index_col` | Columna como índice | `0`, `'Fecha'` |
| `usecols` | Columnas a leer | `['col1', 'col2']` |
| `dtype` | Tipos de datos | `{'col1': str, 'col2': int}` |
| `parse_dates` | Parsear como fechas | `['Fecha']` |
| `na_values` | Valores nulos | `['NA', 'N/A', '']` |
| `skiprows` | Saltar filas | `2` (saltar 2 primeras) |
| `nrows` | Leer n filas | `1000` |

---

### ⚠️ Problemas Comunes

**1. Encoding incorrecto:**
```python
# Error: UnicodeDecodeError
df = pd.read_csv('datos.csv', encoding='latin-1')  # Probar diferentes encodings
```

**2. Separador incorrecto:**
```python
# Si usa punto y coma
df = pd.read_csv('datos.csv', sep=';')
```

**3. Decimal con coma:**
```python
# En Europa: 1.234,56
df = pd.read_csv('datos.csv', decimal=',')
```

````python
import pandas as pd
from io import StringIO

print("📄 LECTURA DE ARCHIVOS CSV")
print("="*70)

# Simulación de CSV con StringIO (sin archivo físico)
csv_data = """Fecha,Sucursal,Ventas,Costos,Estado
2024-01-01,Centro,15000,10000,Cobrado
2024-01-02,Norte,22000,15000,Pendiente
2024-01-03,Sur,9800,7000,Cobrado
2024-01-04,Este,12000,8000,Cobrado
2024-01-05,Centro,28000,19000,Pendiente"""

print("\n1️⃣  LECTURA BÁSICA:")
df = pd.read_csv(StringIO(csv_data))
print(df)
print(f"\nShape: {df.shape}")
print(f"Tipos: \n{df.dtypes}")

print("\n" + "-"*70)
print("\n2️⃣  CON FECHA COMO ÍNDICE:")
df_idx = pd.read_csv(StringIO(csv_data), index_col='Fecha', parse_dates=['Fecha'])
print(df_idx)
print(f"\nÍndice: {df_idx.index}")

print("\n" + "-"*70)
print("\n3️⃣  SELECCIÓN DE COLUMNAS:")
df_subset = pd.read_csv(StringIO(csv_data), usecols=['Fecha', 'Sucursal', 'Ventas'])
print(df_subset)

print("\n" + "-"*70)
print("\n4️⃣  CON TIPOS ESPECÍFICOS:")
df_tipos = pd.read_csv(StringIO(csv_data), dtype={'Ventas': float, 'Costos': float})
print(f"\nTipos después de especificar:\n{df_tipos.dtypes}")

print("\n" + "-"*70)
print("\n5️⃣  CSV CON PUNTO Y COMA (Europa):")
csv_europeo = """Fecha;Sucursal;Ventas
01/01/2024;Centro;15000
02/01/2024;Norte;22000"""

df_euro = pd.read_csv(StringIO(csv_europeo), sep=';')
print(df_euro)

print("\n" + "="*70)
print("✅ Lectura CSV completada")
````

## 🗜️ pd.read_parquet() - Formato Columnar

### 📚 ¿Qué es Parquet?

**Apache Parquet** es un formato de almacenamiento columnar:

**Ventajas:**
* ✅ **Compresión:** 10-100x más pequeño que CSV
* ✅ **Rápido:** Lectura optimizada (solo columnas necesarias)
* ✅ **Tipos preservados:** Mantiene tipos de datos
* ✅ **Compatible:** Spark, Hive, AWS Athena, BigQuery

**Desventajas:**
* ❌ No es legible por humanos (binario)
* ❌ Requiere librería adicional

---

### 🔑 Uso

```python
# Lectura
df = pd.read_parquet('datos.parquet')

# Escritura
df.to_parquet('output.parquet', compression='snappy')
```

---

### 📊 Comparación de Formatos

| Formato | Tamaño | Velocidad Lectura | Uso |
|---------|--------|-------------------|-----|
| **CSV** | 🔴 Grande | 🟡 Media | Intercambio, legible |
| **Excel** | 🟡 Medio | 🟡 Media | Informes, presentaciones |
| **Parquet** | 🟢 Pequeño | 🟢 Rápida | Big Data, pipelines |
| **JSON** | 🔴 Grande | 🔴 Lenta | APIs, semi-estructurado |

---

### 💡 Cuándo Usar Parquet

✅ Datos grandes (> 100 MB)  
✅ Pipelines de datos automatizados  
✅ Integración con Spark/Databricks  
✅ Almacenamiento a largo plazo  

❌ Intercambio con no-técnicos → Usar CSV  
❌ Datos pequeños (< 10 MB) → CSV es suficiente

````python
import pandas as pd
import numpy as np

print("🗜️ FORMATO PARQUET")
print("="*70)

# Crear DataFrame de ejemplo
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=1000),
    'Sucursal': np.random.choice(['Centro', 'Norte', 'Sur', 'Este'], 1000),
    'Ventas': np.random.randint(5000, 50000, 1000),
    'Costos': np.random.randint(3000, 30000, 1000)
})

print("\n📊 DataFrame de ejemplo (1000 filas):")
print(df.head())
print(f"\nShape: {df.shape}")

print("\n" + "-"*70)
print("\n💾 GUARDANDO EN DIFERENTES FORMATOS:")

# Guardar CSV
df.to_csv('/tmp/datos.csv', index=False)

# Guardar Parquet
df.to_parquet('/tmp/datos.parquet', compression='snappy')

print("\n✅ Archivos guardados en /tmp/")

print("\n" + "-"*70)
print("\n📏 COMPARACIÓN DE TAMAÑOS:")

import os

tamaño_csv = os.path.getsize('/tmp/datos.csv')
tamaño_parquet = os.path.getsize('/tmp/datos.parquet')

print(f"\nCSV:      {tamaño_csv:,} bytes ({tamaño_csv/1024:.1f} KB)")
print(f"Parquet:  {tamaño_parquet:,} bytes ({tamaño_parquet/1024:.1f} KB)")
print(f"\n🎯 Reducción: {(1 - tamaño_parquet/tamaño_csv)*100:.1f}%")
print(f"   Parquet es {tamaño_csv/tamaño_parquet:.1f}x más pequeño")

print("\n" + "-"*70)
print("\n⚡ VELOCIDAD DE LECTURA:")

import time

# Lectura CSV
start = time.time()
df_csv = pd.read_csv('/tmp/datos.csv')
tiempo_csv = time.time() - start

# Lectura Parquet
start = time.time()
df_parquet = pd.read_parquet('/tmp/datos.parquet')
tiempo_parquet = time.time() - start

print(f"\nCSV:      {tiempo_csv*1000:.2f} ms")
print(f"Parquet:  {tiempo_parquet*1000:.2f} ms")
print(f"\n🎯 Parquet es {tiempo_csv/tiempo_parquet:.1f}x más rápido")

print("\n" + "="*70)
print("✅ Comparación Parquet vs CSV completada")
````

## ⚠️ Manejo de Errores Comunes

### 🔴 Error 1: UnicodeDecodeError

**Problema:** Encoding incorrecto

```python
# ❌ ERROR
df = pd.read_csv('datos.csv')
# UnicodeDecodeError: 'utf-8' codec can't decode...
```

**Solución:** Probar diferentes encodings

```python
# ✅ SOLUCIÓN
for encoding in ['utf-8', 'latin-1', 'cp1252', 'iso-8859-1']:
    try:
        df = pd.read_csv('datos.csv', encoding=encoding)
        print(f"✅ Funciona con: {encoding}")
        break
    except UnicodeDecodeError:
        print(f"❌ Falló con: {encoding}")
```

---

### 🔴 Error 2: ParserError (separador incorrecto)

**Problema:** Pandas asume `,` pero el archivo usa `;`

```python
# ❌ ERROR
df = pd.read_csv('datos.csv')
# ParserError: Error tokenizing data
```

**Solución:**

```python
# ✅ SOLUCIÓN
df = pd.read_csv('datos.csv', sep=';')  # Especificar separador
# o
df = pd.read_csv('datos.csv', sep=None, engine='python')  # Auto-detectar
```

---

### 🔴 Error 3: FileNotFoundError

**Problema:** Ruta incorrecta

```python
# ❌ ERROR
df = pd.read_csv('datos.csv')
# FileNotFoundError: [Errno 2] No such file or directory
```

**Solución:**

```python
import os

# Verificar si existe
if os.path.exists('datos.csv'):
    df = pd.read_csv('datos.csv')
else:
    print("❌ Archivo no encontrado")
    print(f"Directorio actual: {os.getcwd()}")
```

---

### 🔴 Error 4: DtypeWarning (tipos mixtos)

**Problema:** Columna con tipos mezclados

**Solución:**

```python
# Especificar tipos explícitamente
df = pd.read_csv('datos.csv', dtype={'col1': str, 'col2': float})

# O forzar todo como string y convertir después
df = pd.read_csv('datos.csv', dtype=str)
df['col_numerica'] = pd.to_numeric(df['col_numerica'], errors='coerce')
```

````python
import pandas as pd
from io import StringIO

print("⚠️ MANEJO DE ERRORES COMUNES")
print("="*70)

print("\n1️⃣  SIMULACIÓN: CSV con separador punto y coma")

csv_punto_coma = """Fecha;Sucursal;Ventas
01/01/2024;Centro;15000
02/01/2024;Norte;22000"""

print("\n❌ Intentando con separador incorrecto (coma):")
try:
    df = pd.read_csv(StringIO(csv_punto_coma), sep=',')
    print(df)
except Exception as e:
    print(f"Error: {type(e).__name__}")
    print("El DataFrame se crea pero con una sola columna")

print("\n✅ Usando separador correcto (punto y coma):")
df = pd.read_csv(StringIO(csv_punto_coma), sep=';')
print(df)

print("\n" + "-"*70)
print("\n2️⃣  SIMULACIÓN: Valores nulos y tipos mixtos")

csv_mixto = """Fecha,Monto,Estado
2024-01-01,15000,OK
2024-01-02,N/A,OK
2024-01-03,18500,Pendiente
2024-01-04,ERROR,OK"""

print("\n⚠️  Lectura básica (tipos automáticos):")
df_mixto = pd.read_csv(StringIO(csv_mixto))
print(df_mixto)
print(f"\nTipos: \n{df_mixto.dtypes}")
print("\n⚠️  Monto es 'object' (string) por valores no numéricos")

print("\n✅ Conversión robusta con errors='coerce':")
df_mixto['Monto_limpio'] = pd.to_numeric(df_mixto['Monto'], errors='coerce')
print(df_mixto)
print("\n💡 Los valores no numéricos se convierten a NaN")

print("\n" + "-"*70)
print("\n3️⃣  DETECCIÓN DE VALORES NULOS:")
print(f"\nValores nulos por columna:\n{df_mixto.isnull().sum()}")
print(f"\nFilas con nulos: {df_mixto.isnull().any(axis=1).sum()}")

print("\n" + "="*70)
print("✅ Manejo de errores demostrado")
````

## 💡 Optimizaciones para Archivos Grandes

### 🚀 Técnicas de Optimización

#### 1️⃣  Leer solo columnas necesarias

```python
# ❌ MAL: Leer todo
df = pd.read_csv('datos_grandes.csv')  # 50 columnas, 10 GB

# ✅ BIEN: Solo columnas necesarias
df = pd.read_csv('datos_grandes.csv', usecols=['fecha', 'ventas', 'sucursal'])
```

🎯 **Ahorro:** 90% menos memoria

---

#### 2️⃣  Leer en chunks (trozos)

```python
# Para archivos que no caben en memoria
chunksize = 10000
for chunk in pd.read_csv('datos_grandes.csv', chunksize=chunksize):
    # Procesar cada chunk
    resultado = chunk[chunk['ventas'] > 10000]
    resultado.to_csv('filtrado.csv', mode='a', header=False)
```

---

#### 3️⃣  Especificar tipos de datos

```python
# ❌ MAL: Pandas infiere tipos (lento)
df = pd.read_csv('datos.csv')

# ✅ BIEN: Especificar tipos
dtypes = {
    'sucursal': 'category',  # Strings repetidos → category
    'ventas': 'float32',     # float64 por defecto es excesivo
    'fecha': 'str'           # Parsear después si es necesario
}
df = pd.read_csv('datos.csv', dtype=dtypes)
```

🎯 **Ahorro:** 50-70% menos memoria

---

#### 4️⃣  Usar Parquet en vez de CSV

```python
# Una vez: Convertir CSV a Parquet
df = pd.read_csv('datos_grandes.csv')
df.to_parquet('datos.parquet')

# Siempre después: Leer Parquet
df = pd.read_parquet('datos.parquet')  # 5-10x más rápido
```

---

### 📈 Comparación de Performance

| Técnica | Velocidad | Memoria | Uso |
|---------|-----------|---------|-----|
| CSV completo | 🔴 Lento | 🔴 Alta | Archivos pequeños |
| CSV + usecols | 🟡 Medio | 🟢 Baja | Subset de columnas |
| CSV + chunks | 🟡 Medio | 🟢 Baja | No cabe en RAM |
| CSV + dtypes | 🟡 Medio | 🟢 Baja | Optimizar memoria |
| Parquet | 🟢 Rápido | 🟢 Baja | **Recomendado** |

````python
import pandas as pd
import numpy as np
import time

print("💡 OPTIMIZACIONES")
print("="*70)

# Crear archivo grande de ejemplo
print("\n🔨 Creando archivo de prueba (50,000 filas)...")
df_grande = pd.DataFrame({
    'fecha': pd.date_range('2020-01-01', periods=50000),
    'sucursal': np.random.choice(['Centro', 'Norte', 'Sur', 'Este', 'Oeste'], 50000),
    'producto': np.random.choice(['Laptop', 'Monitor', 'Teclado', 'Mouse'], 50000),
    'ventas': np.random.randint(1000, 50000, 50000),
    'costos': np.random.randint(500, 30000, 50000),
    'cantidad': np.random.randint(1, 100, 50000)
})

df_grande.to_csv('/tmp/datos_grandes.csv', index=False)
print("✅ Archivo creado")

print("\n" + "-"*70)
print("\n1️⃣  LECTURA COMPLETA vs COLUMNAS SELECTAS:")

# Lectura completa
start = time.time()
df_completo = pd.read_csv('/tmp/datos_grandes.csv')
tiempo_completo = time.time() - start
memoria_completo = df_completo.memory_usage(deep=True).sum() / 1024**2

print(f"\n📄 Completo: {tiempo_completo*1000:.0f} ms, {memoria_completo:.2f} MB")

# Solo columnas necesarias
start = time.time()
df_subset = pd.read_csv('/tmp/datos_grandes.csv', usecols=['fecha', 'sucursal', 'ventas'])
tiempo_subset = time.time() - start
memoria_subset = df_subset.memory_usage(deep=True).sum() / 1024**2

print(f"🎯 Subset:   {tiempo_subset*1000:.0f} ms, {memoria_subset:.2f} MB")
print(f"\n📉 Ahorro: {(1-tiempo_subset/tiempo_completo)*100:.1f}% tiempo, {(1-memoria_subset/memoria_completo)*100:.1f}% memoria")

print("\n" + "-"*70)
print("\n2️⃣  OPTIMIZACIÓN DE TIPOS DE DATOS:")

# Sin optimización
df_sin_opt = pd.read_csv('/tmp/datos_grandes.csv')
memoria_sin_opt = df_sin_opt.memory_usage(deep=True).sum() / 1024**2

print(f"\nSin optimización: {memoria_sin_opt:.2f} MB")
print(f"Tipos:\n{df_sin_opt.dtypes}")

# Con optimización
dtypes = {
    'sucursal': 'category',
    'producto': 'category',
    'ventas': 'float32',
    'costos': 'float32',
    'cantidad': 'int16'
}
df_opt = pd.read_csv('/tmp/datos_grandes.csv', dtype=dtypes, parse_dates=['fecha'])
memoria_opt = df_opt.memory_usage(deep=True).sum() / 1024**2

print(f"\nCon optimización: {memoria_opt:.2f} MB")
print(f"\n📉 Ahorro: {(1-memoria_opt/memoria_sin_opt)*100:.1f}% memoria")

print("\n" + "="*70)
print("✅ Optimizaciones demostradas")
````
