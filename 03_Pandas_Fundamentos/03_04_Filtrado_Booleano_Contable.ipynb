## 🕵️ Filtrado básico con condiciones

### 📊 Cómo funciona

**Paso 1:** Crear condición booleana
```python
condicion = df['Ventas'] > 20000
# Resultado: Series de True/False
```

**Paso 2:** Aplicar filtro con .loc
```python
df_filtrado = df.loc[condicion]
# o directamente
df_filtrado = df[df['Ventas'] > 20000]
```

---

### 📊 Ejemplos básicos

```python
# Ventas mayores a $20,000
df[df['Ventas'] > 20000]

# Estado igual a 'Cobrado'
df[df['Estado'] == 'Cobrado']

# Sucursal diferente de 'Centro'
df[df['Sucursal'] != 'Centro']

# Ventas entre $10,000 y $25,000
df[(df['Ventas'] >= 10000) & (df['Ventas'] <= 25000)]
```

---

### ⚠️ Errores comunes

❌ **MAL:** `df[df['Ventas'] > 20000 & df['Estado'] == 'Cobrado']`
* Falta paréntesis

✅ **BIEN:** `df[(df['Ventas'] > 20000) & (df['Estado'] == 'Cobrado')]`
* Paréntesis en cada condición

❌ **MAL:** `df[df['Ventas'] > 20000 and df['Estado'] == 'Cobrado']`
* Usa `and` en vez de `&`

✅ **BIEN:** Usa `&` para DataFrames (no `and`)

## 🔗 Múltiples condiciones (AND / OR)

### 📊 Operadores lógicos

| Operador | Significado | Ejemplo |
|----------|-------------|--------|
| `&` | AND (y) | `(cond1) & (cond2)` |
| `\|` | OR (o) | `(cond1) \| (cond2)` |
| `~` | NOT (negación) | `~(condicion)` |

---

### 📊 Ejemplos

#### AND (&) - Ambas condiciones deben cumplirse

```python
# Ventas altas Y estado cobrado
df[(df['Ventas'] > 20000) & (df['Estado'] == 'Cobrado')]

# Ventas entre $10K y $25K
df[(df['Ventas'] >= 10000) & (df['Ventas'] <= 25000)]

# Sucursal Centro Y ventas > $15K
df[(df['Sucursal'] == 'Centro') & (df['Ventas'] > 15000)]
```

---

#### OR (|) - Al menos una condición debe cumplirse

```python
# Ventas muy altas O ventas muy bajas
df[(df['Ventas'] > 25000) | (df['Ventas'] < 10000)]

# Sucursal Centro O Norte
df[(df['Sucursal'] == 'Centro') | (df['Sucursal'] == 'Norte')]

# Estado Pendiente O ventas > $20K
df[(df['Estado'] == 'Pendiente') | (df['Ventas'] > 20000)]
```

---

#### NOT (~) - Negación

```python
# Todo EXCEPTO Sucursal Centro
df[~(df['Sucursal'] == 'Centro')]
# Equivalente a:
df[df['Sucursal'] != 'Centro']

# Todo EXCEPTO ventas entre $10K y $20K
df[~((df['Ventas'] >= 10000) & (df['Ventas'] <= 20000))]
```

---

### ⚠️ Regla de Oro

🚨 **Siempre usar paréntesis en CADA condición**

```python
# ❌ MAL (falla)
df[df['Ventas'] > 10000 & df['Estado'] == 'Cobrado']

# ✅ BIEN
df[(df['Ventas'] > 10000) & (df['Estado'] == 'Cobrado')]
```

## 🛠️ Métodos auxiliares de filtrado

### 📊 .isin() - Pertenencia a lista

**¿Está el valor en esta lista?**

```python
# Sucursal es Centro, Norte o Sur
df[df['Sucursal'].isin(['Centro', 'Norte', 'Sur'])]

# Estado es Cobrado o Verificado
df[df['Estado'].isin(['Cobrado', 'Verificado'])]

# Ventas es 10000, 20000 o 30000
df[df['Ventas'].isin([10000, 20000, 30000])]
```

**Equivalente a múltiples OR:**
```python
# Más fácil:
df[df['Sucursal'].isin(['Centro', 'Norte'])]

# Que esto:
df[(df['Sucursal'] == 'Centro') | (df['Sucursal'] == 'Norte')]
```

---

### 📊 .between() - Rango inclusivo

**¿Está entre estos valores?**

```python
# Ventas entre $10,000 y $25,000 (inclusivo)
df[df['Ventas'].between(10000, 25000)]

# Fecha entre 2024-01-01 y 2024-01-31
df[df['Fecha'].between('2024-01-01', '2024-01-31')]
```

**Equivalente a:**
```python
# Más fácil:
df[df['Ventas'].between(10000, 25000)]

# Que esto:
df[(df['Ventas'] >= 10000) & (df['Ventas'] <= 25000)]
```

---

### 📊 .str.contains() - Búsqueda de texto

**¿Contiene este substring?**

```python
# Sucursal contiene 'Nor'
df[df['Sucursal'].str.contains('Nor')]

# Estado contiene 'Pend' (sensible a mayúsculas)
df[df['Estado'].str.contains('Pend', case=False)]  # case=False ignora mayúsculas

# Empieza con 'Ce'
df[df['Sucursal'].str.startswith('Ce')]

# Termina con 'ro'
df[df['Sucursal'].str.endswith('ro')]
```

---

### 📊 .query() - Sintaxis SQL-like

**Filtrado con sintaxis más legible:**

```python
# En vez de:
df[(df['Ventas'] > 20000) & (df['Estado'] == 'Cobrado')]

# Puedes usar:
df.query('Ventas > 20000 and Estado == "Cobrado"')

# Más ejemplos:
df.query('Sucursal in ["Centro", "Norte"]')
df.query('10000 <= Ventas <= 25000')
df.query('Ventas > Costos * 2')
```

💡 **Ventaja:** Más legible, especialmente con condiciones complejas

```python
df[(cond1) | (cond2)]  # OR
```

🚨 **Siempre paréntesis:** `(df['col'] > 10) & (df['otra'] < 5)`

🛠️ **Métodos auxiliares:**
* `.isin([lista])` → Pertenencia
* `.between(a, b)` → Rango
* `.query('...')` → Sintaxis legible
