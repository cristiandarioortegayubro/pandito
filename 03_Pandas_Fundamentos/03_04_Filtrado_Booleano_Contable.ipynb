## 🕵️ Filtrado Básico con Condiciones

### 📊 Cómo Funciona

**Paso 1:** Crear condición booleana
```python
condicion = df['Ventas'] > 20000
# Resultado: Series de True/False
```

**Paso 2:** Aplicar filtro con .loc
```python
df_filtrado = df.loc[condicion]
# o directamente
df_filtrado = df[df['Ventas'] > 20000]
```

---

### 📊 Ejemplos Básicos

```python
# Ventas mayores a $20,000
df[df['Ventas'] > 20000]

# Estado igual a 'Cobrado'
df[df['Estado'] == 'Cobrado']

# Sucursal diferente de 'Centro'
df[df['Sucursal'] != 'Centro']

# Ventas entre $10,000 y $25,000
df[(df['Ventas'] >= 10000) & (df['Ventas'] <= 25000)]
```

---

### ⚠️ Errores Comunes

❌ **MAL:** `df[df['Ventas'] > 20000 & df['Estado'] == 'Cobrado']`
* Falta paréntesis

✅ **BIEN:** `df[(df['Ventas'] > 20000) & (df['Estado'] == 'Cobrado')]`
* Paréntesis en cada condición

❌ **MAL:** `df[df['Ventas'] > 20000 and df['Estado'] == 'Cobrado']`
* Usa `and` en vez de `&`

✅ **BIEN:** Usa `&` para DataFrames (no `and`)

````python
import pandas as pd
import numpy as np

print("🕵️ FILTRADO BÁSICO")
print("="*70)

# Crear DataFrame contable
np.random.seed(42)
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=10),
    'Sucursal': np.random.choice(['Centro', 'Norte', 'Sur', 'Este'], 10),
    'Ventas': [15000, 22000, 9800, 31000, 12000, 28000, 17500, 8500, 25000, 19000],
    'Costos': [10000, 15000, 7000, 21000, 8000, 19000, 11000, 6000, 17000, 13000],
    'Estado': ['Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente',
               'Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente']
})

print("\n📊 DataFrame original:")
print(df)

print("\n" + "="*70)
print("\n1️⃣  FILTRO: Ventas mayores a $20,000")
filtro1 = df[df['Ventas'] > 20000]
print(filtro1)
print(f"\n📊 Encontradas: {len(filtro1)} transacciones")

print("\n" + "="*70)
print("\n2️⃣  FILTRO: Estado = 'Cobrado'")
filtro2 = df[df['Estado'] == 'Cobrado']
print(filtro2[['Fecha', 'Sucursal', 'Ventas', 'Estado']])
print(f"\n📊 Cobradas: {len(filtro2)} de {len(df)} ({len(filtro2)/len(df)*100:.0f}%)")

print("\n" + "="*70)
print("\n3️⃣  FILTRO: Sucursal != 'Centro'")
filtro3 = df[df['Sucursal'] != 'Centro']
print(filtro3[['Fecha', 'Sucursal', 'Ventas']])

print("\n" + "="*70)
print("\n4️⃣  VERIFICACIÓN: ¿Qué devuelve una condición?")
condicion = df['Ventas'] > 20000
print(f"\nTipo: {type(condicion)}")
print(f"\nPrimeros valores:")
print(condicion.head())
print("\n💡 Es una Series de True/False (booleanos)")

print("\n" + "="*70)
print("✅ Filtrado básico completado")
````

## 🔗 Múltiples Condiciones (AND / OR)

### 📊 Operadores Lógicos

| Operador | Significado | Ejemplo |
|----------|-------------|--------|
| `&` | AND (y) | `(cond1) & (cond2)` |
| `\|` | OR (o) | `(cond1) \| (cond2)` |
| `~` | NOT (negación) | `~(condicion)` |

---

### 📊 Ejemplos

#### AND (&) - Ambas condiciones deben cumplirse

```python
# Ventas altas Y estado cobrado
df[(df['Ventas'] > 20000) & (df['Estado'] == 'Cobrado')]

# Ventas entre $10K y $25K
df[(df['Ventas'] >= 10000) & (df['Ventas'] <= 25000)]

# Sucursal Centro Y ventas > $15K
df[(df['Sucursal'] == 'Centro') & (df['Ventas'] > 15000)]
```

---

#### OR (|) - Al menos una condición debe cumplirse

```python
# Ventas muy altas O ventas muy bajas
df[(df['Ventas'] > 25000) | (df['Ventas'] < 10000)]

# Sucursal Centro O Norte
df[(df['Sucursal'] == 'Centro') | (df['Sucursal'] == 'Norte')]

# Estado Pendiente O ventas > $20K
df[(df['Estado'] == 'Pendiente') | (df['Ventas'] > 20000)]
```

---

#### NOT (~) - Negación

```python
# Todo EXCEPTO Sucursal Centro
df[~(df['Sucursal'] == 'Centro')]
# Equivalente a:
df[df['Sucursal'] != 'Centro']

# Todo EXCEPTO ventas entre $10K y $20K
df[~((df['Ventas'] >= 10000) & (df['Ventas'] <= 20000))]
```

---

### ⚠️ Regla de Oro

🚨 **Siempre usar paréntesis en CADA condición**

```python
# ❌ MAL (falla)
df[df['Ventas'] > 10000 & df['Estado'] == 'Cobrado']

# ✅ BIEN
df[(df['Ventas'] > 10000) & (df['Estado'] == 'Cobrado')]
```

````python
import pandas as pd
import numpy as np

print("🔗 MÚLTIPLES CONDICIONES")
print("="*70)

# Recrear DataFrame
np.random.seed(42)
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=10),
    'Sucursal': np.random.choice(['Centro', 'Norte', 'Sur', 'Este'], 10),
    'Ventas': [15000, 22000, 9800, 31000, 12000, 28000, 17500, 8500, 25000, 19000],
    'Costos': [10000, 15000, 7000, 21000, 8000, 19000, 11000, 6000, 17000, 13000],
    'Estado': ['Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente',
               'Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente']
})
df['Ganancia'] = df['Ventas'] - df['Costos']

print("\n📊 DataFrame:")
print(df[['Sucursal', 'Ventas', 'Estado', 'Ganancia']])

print("\n" + "="*70)
print("\n1️⃣  AND (&): Ventas > $20K Y Estado = 'Cobrado'")
filtro_and = df[(df['Ventas'] > 20000) & (df['Estado'] == 'Cobrado')]
print(filtro_and[['Sucursal', 'Ventas', 'Estado']])
print(f"\n📊 {len(filtro_and)} transacciones cumplen AMBAS condiciones")

print("\n" + "="*70)
print("\n2️⃣  AND (&): Ventas entre $15K y $25K")
filtro_rango = df[(df['Ventas'] >= 15000) & (df['Ventas'] <= 25000)]
print(filtro_rango[['Sucursal', 'Ventas']])

print("\n" + "="*70)
print("\n3️⃣  OR (|): Ventas muy altas (>$25K) O muy bajas (<$10K)")
filtro_or = df[(df['Ventas'] > 25000) | (df['Ventas'] < 10000)]
print(filtro_or[['Sucursal', 'Ventas']])
print(f"\n📊 {len(filtro_or)} transacciones cumplen AL MENOS una condición")

print("\n" + "="*70)
print("\n4️⃣  OR (|): Sucursal Centro O Norte")
filtro_sucursales = df[(df['Sucursal'] == 'Centro') | (df['Sucursal'] == 'Norte')]
print(filtro_sucursales[['Sucursal', 'Ventas']])

print("\n" + "="*70)
print("\n5️⃣  NOT (~): Todo EXCEPTO Sucursal Centro")
filtro_not = df[~(df['Sucursal'] == 'Centro')]
print(filtro_not[['Sucursal', 'Ventas']])

print("\n" + "="*70)
print("\n6️⃣  COMBINACIÓN COMPLEJA: (Ventas > $20K Y Cobrado) O Ganancia > $10K")
filtro_complejo = df[((df['Ventas'] > 20000) & (df['Estado'] == 'Cobrado')) | (df['Ganancia'] > 10000)]
print(filtro_complejo[['Sucursal', 'Ventas', 'Estado', 'Ganancia']])

print("\n" + "="*70)
print("✅ Múltiples condiciones dominadas")
````

## 🛠️ Métodos Auxiliares de Filtrado

### 📊 .isin() - Pertenencia a lista

**¿Está el valor en esta lista?**

```python
# Sucursal es Centro, Norte o Sur
df[df['Sucursal'].isin(['Centro', 'Norte', 'Sur'])]

# Estado es Cobrado o Verificado
df[df['Estado'].isin(['Cobrado', 'Verificado'])]

# Ventas es 10000, 20000 o 30000
df[df['Ventas'].isin([10000, 20000, 30000])]
```

**Equivalente a múltiples OR:**
```python
# Más fácil:
df[df['Sucursal'].isin(['Centro', 'Norte'])]

# Que esto:
df[(df['Sucursal'] == 'Centro') | (df['Sucursal'] == 'Norte')]
```

---

### 📊 .between() - Rango inclusivo

**¿Está entre estos valores?**

```python
# Ventas entre $10,000 y $25,000 (inclusivo)
df[df['Ventas'].between(10000, 25000)]

# Fecha entre 2024-01-01 y 2024-01-31
df[df['Fecha'].between('2024-01-01', '2024-01-31')]
```

**Equivalente a:**
```python
# Más fácil:
df[df['Ventas'].between(10000, 25000)]

# Que esto:
df[(df['Ventas'] >= 10000) & (df['Ventas'] <= 25000)]
```

---

### 📊 .str.contains() - Búsqueda de texto

**¿Contiene este substring?**

```python
# Sucursal contiene 'Nor'
df[df['Sucursal'].str.contains('Nor')]

# Estado contiene 'Pend' (sensible a mayúsculas)
df[df['Estado'].str.contains('Pend', case=False)]  # case=False ignora mayúsculas

# Empieza con 'Ce'
df[df['Sucursal'].str.startswith('Ce')]

# Termina con 'ro'
df[df['Sucursal'].str.endswith('ro')]
```

---

### 📊 .query() - Sintaxis SQL-like

**Filtrado con sintaxis más legible:**

```python
# En vez de:
df[(df['Ventas'] > 20000) & (df['Estado'] == 'Cobrado')]

# Puedes usar:
df.query('Ventas > 20000 and Estado == "Cobrado"')

# Más ejemplos:
df.query('Sucursal in ["Centro", "Norte"]')
df.query('10000 <= Ventas <= 25000')
df.query('Ventas > Costos * 2')
```

💡 **Ventaja:** Más legible, especialmente con condiciones complejas

````python
import pandas as pd
import numpy as np

print("🛠️ MÉTODOS AUXILIARES")
print("="*70)

# DataFrame
np.random.seed(42)
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=10),
    'Sucursal': ['Centro', 'Norte', 'Sur', 'Centro', 'Este', 'Norte', 'Centro', 'Sur', 'Este', 'Norte'],
    'Ventas': [15000, 22000, 9800, 31000, 12000, 28000, 17500, 8500, 25000, 19000],
    'Estado': ['Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente',
               'Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente']
})

print("\n📊 DataFrame:")
print(df)

print("\n" + "="*70)
print("\n1️⃣  .isin() - Sucursal en ['Centro', 'Norte']")
filtro_isin = df[df['Sucursal'].isin(['Centro', 'Norte'])]
print(filtro_isin[['Sucursal', 'Ventas']])
print(f"\n💡 Más fácil que: (df['Sucursal'] == 'Centro') | (df['Sucursal'] == 'Norte')")

print("\n" + "="*70)
print("\n2️⃣  .between() - Ventas entre $15K y $25K")
filtro_between = df[df['Ventas'].between(15000, 25000)]
print(filtro_between[['Sucursal', 'Ventas']])
print(f"\n💡 Más fácil que: (df['Ventas'] >= 15000) & (df['Ventas'] <= 25000)")

print("\n" + "="*70)
print("\n3️⃣  .str.contains() - Sucursal contiene 'Nor'")
filtro_contains = df[df['Sucursal'].str.contains('Nor')]
print(filtro_contains[['Sucursal', 'Ventas']])

print("\n" + "="*70)
print("\n4️⃣  .query() - Sintaxis SQL-like")
filtro_query = df.query('Ventas > 20000 and Estado == "Cobrado"')
print(filtro_query[['Sucursal', 'Ventas', 'Estado']])
print(f"\n💡 Más legible que condiciones con & y paréntesis")

print("\n" + "="*70)
print("\n5️⃣  .query() con rango")
filtro_query_rango = df.query('15000 <= Ventas <= 25000')
print(filtro_query_rango[['Sucursal', 'Ventas']])

print("\n" + "="*70)
print("\n6️⃣  COMPARACIÓN: Mismo filtro con 3 métodos")

print("\nCondición: Sucursal in ['Centro', 'Norte'] AND Ventas > $20K")

print("\nMétodo 1 (tradicional):")
r1 = df[(df['Sucursal'].isin(['Centro', 'Norte'])) & (df['Ventas'] > 20000)]
print(r1[['Sucursal', 'Ventas']])

print("\nMétodo 2 (.query):")
r2 = df.query('Sucursal in ["Centro", "Norte"] and Ventas > 20000')
print(r2[['Sucursal', 'Ventas']])

print("\n✅ Resultado idéntico, elige el que prefieras")

print("\n" + "="*70)
print("✅ Métodos auxiliares completados")
````

### 🎯 Caso 5: Segmentación de Clientes

**Problema:** Clasificar ventas por tamaño

```python
df.loc[df['Ventas'] < 10000, 'Categoria'] = 'Baja'
df.loc[df['Ventas'].between(10000, 25000), 'Categoria'] = 'Media'
df.loc[df['Ventas'] > 25000, 'Categoria'] = 'Alta'
```

````python
import pandas as pd
import numpy as np

print("💼 CASOS CONTABLES REALES")
print("="*70)

# DataFrame con datos más realistas
np.random.seed(42)
df = pd.DataFrame({
    'Fecha': pd.date_range('2023-12-01', periods=20),
    'Sucursal': np.random.choice(['Centro', 'Norte', 'Sur', 'Este'], 20),
    'Ventas': [15300, 22000, 9800, 31000, 12000, 28000, 17500, 8500, 25000, 19000,
               30000, 14500, 21000, 10000, 27500, 18000, 23000, 11000, 26000, 20000],
    'Costos': [10000, 15000, 7000, 21000, 8000, 19000, 11000, 6000, 17000, 13000,
               20000, 9500, 14000, 7000, 18500, 12000, 15500, 7500, 17500, 13500],
    'Estado': ['Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente',
               'Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente',
               'Cobrado', 'Pendiente', 'Cobrado', 'Pendiente', 'Cobrado',
               'Pendiente', 'Cobrado', 'Pendiente', 'Cobrado', 'Pendiente']
})

df['Ganancia'] = df['Ventas'] - df['Costos']
df['Margen_%'] = (df['Ganancia'] / df['Ventas'] * 100).round(1)

print("\n📊 Dataset (20 transacciones):")
print(df.head())

print("\n" + "="*70)
print("\n1️⃣  CASO: Cuentas por Cobrar Vencidas (>30 días)")

fecha_actual = pd.Timestamp('2024-01-15')
dias_transcurridos = (fecha_actual - df['Fecha']).dt.days

vencidas = df[
    (df['Estado'] == 'Pendiente') & 
    (dias_transcurridos > 30)
].copy()
vencidas['Dias_Vencidos'] = (fecha_actual - vencidas['Fecha']).dt.days

print(f"\n⚠️  {len(vencidas)} transacciones vencidas:")
print(vencidas[['Fecha', 'Sucursal', 'Ventas', 'Dias_Vencidos']])
print(f"\n💰 Total vencido: ${vencidas['Ventas'].sum():,.0f}")

print("\n" + "="*70)
print("\n2️⃣  CASO: Transacciones Anómalas (margen muy bajo <15% o alto >60%)")

anomalas = df[
    (df['Margen_%'] < 15) | 
    (df['Margen_%'] > 60)
]

print(f"\n⚠️  {len(anomalas)} transacciones anómalas:")
print(anomalas[['Fecha', 'Sucursal', 'Ventas', 'Costos', 'Margen_%']])

print("\n" + "="*70)
print("\n3️⃣  CASO: Alerta Ventas Bajas (30% bajo promedio)")

promedio_ventas = df['Ventas'].mean()
umbral = promedio_ventas * 0.7

bajas = df[df['Ventas'] < umbral]

print(f"\n📊 Promedio: ${promedio_ventas:,.0f}")
print(f"⚠️  Umbral (70%): ${umbral:,.0f}")
print(f"\n🚨 {len(bajas)} sucursales con ventas bajas:")
print(bajas[['Fecha', 'Sucursal', 'Ventas']])

print("\n" + "="*70)
print("\n4️⃣  CASO: Auditoría - Transacciones Redondas (múltiplos de 1000)")

redondas = df[df['Ventas'] % 1000 == 0]

print(f"\n🔍 {len(redondas)} transacciones sospechosas (montos redondos exactos):")
print(redondas[['Fecha', 'Sucursal', 'Ventas']])
print("\n💡 Revisar: Posible manipulación o aproximación")

print("\n" + "="*70)
print("\n5️⃣  CASO: Segmentación de Ventas")

df_seg = df.copy()
df_seg.loc[df_seg['Ventas'] < 15000, 'Categoria'] = 'Baja'
df_seg.loc[df_seg['Ventas'].between(15000, 25000), 'Categoria'] = 'Media'
df_seg.loc[df_seg['Ventas'] > 25000, 'Categoria'] = 'Alta'

print("\nDistribución por categoría:")
print(df_seg['Categoria'].value_counts())
print("\nVentas promedio por categoría:")
print(df_seg.groupby('Categoria')['Ventas'].mean().round(0))

print("\n" + "="*70)
print("✅ Casos contables aplicados exitosamente")
````
