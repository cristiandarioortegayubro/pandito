## 🐼 ¿Qué es Pandas?

### 📚 Definición

**Pandas** (Panel Data) es la librería más importante de Python para análisis de datos estructurados.

**Creada por:** Wes McKinney en 2008 (ex analista cuantitativo)  
**Libro:** *Python for Data Analysis* (O'Reilly)

---

### 🎯 ¿Por Qué Usar Pandas?

**Pandas es el "Excel de Python"**, pero mucho más potente:

| Característica | Excel | Pandas |
|-----------------|-------|--------|
| Tamaño máximo | ~1 millón filas | Miles de millones |
| Automatización | Macros/VBA | Scripts Python |
| Reproducibilidad | Manual | 100% reproducible |
| Integración | Limitada | SQL, APIs, Cloud |
| Performance | Lenta con datos grandes | Optimizada (C) |
| Visualización | Gráficos básicos | matplotlib, seaborn |
| ML/AI | No nativo | scikit-learn, TensorFlow |

---

### 💼 Casos de Uso en Finanzas

1. **Contabilidad:**
   * Conciliaciones bancarias automatizadas
   * Análisis de libro mayor
   * Reportes mensuales programados

2. **Finanzas Corporativas:**
   * Flujos de caja proyectados
   * Análisis de rentabilidad por producto
   * Consolidación de subsidiarias

3. **Auditoría:**
   * Detección de transacciones anómalas
   * Muestreo estadístico
   * Tests de Benford

4. **Inversiones:**
   * Análisis de portafolios
   * Backtesting de estrategias
   * Retornos históricos

## 📊 Series: Vectores Unidimensionales con Índice

### 📝 Definición

Una **Series** es un array unidimensional etiquetado (como una columna de Excel con nombres de fila).

**Estructura:**
```
Índice    Valor
-------    -----
A          100
B          200  
C          150
```

---

### 🔑 Componentes de una Series

1. **Valores (.values):** Array NumPy subyacente
2. **Índice (.index):** Etiquetas para cada valor
3. **Nombre (.name):** Nombre opcional de la serie

---

### 💼 Caso Contable: Ventas Mensuales

```python
ventas_enero = pd.Series(
    [15000, 22000, 18500, 31000],
    index=['Sucursal_Centro', 'Sucursal_Norte', 'Sucursal_Sur', 'Sucursal_Este'],
    name='Ventas_Enero_2024'
)
```

**Ventaja sobre NumPy:** Los índices permiten acceso por nombre, no solo por posición.

```python
ventas_enero['Sucursal_Norte']  # Más legible que ventas[1]
```

````python
import pandas as pd
import numpy as np

print("📊 SERIES: Vectores con Índice")
print("="*70)

print("\n1️⃣  CREACIÓN desde lista:")
ventas = pd.Series([15000, 22000, 18500, 31000])
print(ventas)
print(f"\nTipo: {type(ventas)}")
print(f"Índice automático: {ventas.index.tolist()}")

print("\n" + "-"*70)
print("\n2️⃣  CREACIÓN con índice personalizado:")
ventas_etiquetadas = pd.Series(
    [15000, 22000, 18500, 31000],
    index=['Centro', 'Norte', 'Sur', 'Este'],
    name='Ventas_Enero'
)
print(ventas_etiquetadas)

print("\n" + "-"*70)
print("\n3️⃣  ACCESO por etiqueta:")
print(f"\nVentas Sucursal Norte: ${ventas_etiquetadas['Norte']:,.0f}")
print(f"Ventas Sucursal Centro: ${ventas_etiquetadas['Centro']:,.0f}")

print("\n" + "-"*70)
print("\n4️⃣  OPERACIONES vectorizadas:")
ventas_con_iva = ventas_etiquetadas * 1.21
print("\nVentas + IVA (21%):")
print(ventas_con_iva)

print("\n" + "-"*70)
print("\n5️⃣  ESTADÍSTICAS:")
print(f"\nTotal: ${ventas_etiquetadas.sum():,.0f}")
print(f"Promedio: ${ventas_etiquetadas.mean():,.0f}")
print(f"Máximo: ${ventas_etiquetadas.max():,.0f} ({ventas_etiquetadas.idxmax()})")
print(f"Mínimo: ${ventas_etiquetadas.min():,.0f} ({ventas_etiquetadas.idxmin()})")

print("\n" + "="*70)
print("✅ Series creada y manipulada exitosamente")
````

## 📋 DataFrames: Tablas Bidimensionales

### 📝 Definición

Un **DataFrame** es una tabla bidimensional con filas y columnas etiquetadas (como una hoja de Excel).

**Estructura:**
```
       Producto    Ventas  Costo
Enero  Laptop      15000   10000
Febrero Monitor     8000    5000
Marzo  Teclado     3000    1500
```

---

### 🔑 Componentes de un DataFrame

1. **Columnas:** Cada columna es una Series
2. **Índice (.index):** Etiquetas de filas
3. **Nombres de columnas (.columns):** Etiquetas de columnas
4. **Valores (.values):** Array NumPy 2D subyacente

---

### 💼 DataFrame = Colección de Series

**Conceptualmente:**
```python
df['Ventas']  # Devuelve una Series
df['Costo']   # Devuelve otra Series
```

Todas las columnas comparten el mismo índice → **alineación automática**

---

### 🎯 Métodos de Creación

1. **Desde diccionario:** `pd.DataFrame({'col1': [...], 'col2': [...]})`
2. **Desde lista de diccionarios:** `pd.DataFrame([{'a': 1, 'b': 2}, ...])`
3. **Desde array NumPy:** `pd.DataFrame(array, columns=[...])`
4. **Desde archivo:** `pd.read_csv()`, `pd.read_excel()`, etc.

````python
import pandas as pd
import numpy as np

print("📋 DATAFRAMES: Creación")
print("="*70)

print("\n1️⃣  DESDE DICCIONARIO (más común):")
datos_ventas = {
    'Fecha': pd.date_range('2024-01-01', periods=5),
    'Sucursal': ['Centro', 'Norte', 'Centro', 'Sur', 'Norte'],
    'Ventas': [15000, 22000, 18500, 9800, 31000],
    'Estado': ['Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente']
}

df_ventas = pd.DataFrame(datos_ventas)
print(df_ventas)
print(f"\nTipo: {type(df_ventas)}")
print(f"Shape (filas, columnas): {df_ventas.shape}")

print("\n" + "-"*70)
print("\n2️⃣  DESDE LISTA DE DICCIONARIOS:")
transacciones = [
    {'cuenta': 'Caja', 'debe': 10000, 'haber': 0},
    {'cuenta': 'Ventas', 'debe': 0, 'haber': 10000},
    {'cuenta': 'Bancos', 'debe': 5000, 'haber': 0}
]
df_asientos = pd.DataFrame(transacciones)
print(df_asientos)

print("\n" + "-"*70)
print("\n3️⃣  DESDE ARRAY NumPy:")
array_datos = np.array([
    [100, 200, 300],
    [400, 500, 600],
    [700, 800, 900]
])
df_numpy = pd.DataFrame(array_datos, 
                        columns=['A', 'B', 'C'],
                        index=['Fila1', 'Fila2', 'Fila3'])
print(df_numpy)

print("\n" + "="*70)
print("✅ DataFrames creados exitosamente")
````

## 🔍 Inspección de DataFrames

### 🛠️ Métodos Esenciales

| Método | Descripción | Uso |
|---------|--------------|-----|
| `.head(n)` | Primeras n filas | Vista rápida de datos |
| `.tail(n)` | Últimas n filas | Verificar datos finales |
| `.info()` | Resumen de estructura | Tipos, nulls, memoria |
| `.describe()` | Estadísticas descriptivas | Media, std, percentiles |
| `.shape` | (filas, columnas) | Tamaño del dataset |
| `.columns` | Nombres de columnas | Listar columnas |
| `.index` | Índice de filas | Listar índices |
| `.dtypes` | Tipos de datos por columna | Verificar tipos |

---

### 📊 .describe() - Estadísticas Automáticas

**Para columnas numéricas:**
* count: Cantidad de valores no nulos
* mean: Promedio
* std: Desviación estándar
* min: Mínimo
* 25%, 50%, 75%: Percentiles (cuartiles)
* max: Máximo

**Para columnas de texto (con `include='object'`):**
* count, unique, top, freq

---

### ⚠️ .info() - Diagnóstico Rápido

Revela:
* Cantidad de entradas (filas)
* Columnas y sus tipos
* Valores nulos por columna (¡crítico!)
* Uso de memoria

**Ejemplo:**
```
<class 'pandas.core.frame.DataFrame'>
RangeIndex: 5 entries, 0 to 4
Data columns (total 4 columns):
Fecha      5 non-null datetime64[ns]
Sucursal   5 non-null object
Ventas     5 non-null int64
Estado     5 non-null object
```

````python
import pandas as pd
import numpy as np

print("🔍 INSPECCIÓN DE DATAFRAMES")
print("="*70)

# Recrear datos de ventas
datos_ventas = {
    'Fecha': pd.date_range('2024-01-01', periods=10),
    'Sucursal': ['Centro', 'Norte', 'Centro', 'Sur', 'Norte', 
                 'Centro', 'Este', 'Sur', 'Norte', 'Centro'],
    'Ventas': [15000, 22000, 18500, 9800, 31000, 17500, 12000, 23000, 19500, 28000],
    'Costos': [10000, 15000, 12000, 7000, 21000, 11000, 8000, 16000, 13000, 19000],
    'Estado': ['Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente',
               'Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente']
}

df = pd.DataFrame(datos_ventas)

print("\n1️⃣  .head() - Primeras 5 filas:")
print(df.head())

print("\n" + "-"*70)
print("\n2️⃣  .tail(3) - Últimas 3 filas:")
print(df.tail(3))

print("\n" + "-"*70)
print("\n3️⃣  .info() - Resumen estructural:")
df.info()

print("\n" + "-"*70)
print("\n4️⃣  .describe() - Estadísticas numéricas:")
print(df.describe())

print("\n" + "-"*70)
print("\n5️⃣  .shape, .columns, .dtypes:")
print(f"\nShape: {df.shape} (10 filas × 5 columnas)")
print(f"\nColumnas: {df.columns.tolist()}")
print(f"\nTipos de datos:\n{df.dtypes}")

print("\n" + "-"*70)
print("\n6️⃣  Verificación de nulos:")
print(f"\nValores nulos por columna:\n{df.isnull().sum()}")
print(f"\nTotal de nulos: {df.isnull().sum().sum()}")

print("\n" + "="*70)
print("✅ Inspección completada")
````

## 🎯 Selección Básica de Datos

### 📊 Selección de Columnas

**Una columna (devuelve Series):**
```python
df['Ventas']       # Notación corchete (preferida)
df.Ventas          # Notación punto (solo si nombre válido)
```

**Múltiples columnas (devuelve DataFrame):**
```python
df[['Ventas', 'Costos']]  # Doble corchete
```

---

### 📋 Selección de Filas

**Por posición (numérica):**
```python
df.iloc[0]       # Primera fila (Series)
df.iloc[0:3]     # Primeras 3 filas (DataFrame)
```

**Por etiqueta (índice):**
```python
df.loc[0]        # Fila con índice 0
df.loc[0:3]      # Filas 0, 1, 2, 3 (inclusivo)
```

---

### 🔗 Selección Combinada

```python
# Filas 0-2, columnas Ventas y Costos
df.loc[0:2, ['Ventas', 'Costos']]

# Primeras 3 filas, columnas 1 y 2
df.iloc[0:3, 1:3]
```

**⚡ En notebooks siguientes:** `.loc` con filtrado booleano

````python
import pandas as pd

print("🎯 SELECCIÓN DE DATOS")
print("="*70)

# Recrear DataFrame
datos = {
    'Fecha': pd.date_range('2024-01-01', periods=5),
    'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Centro'],
    'Ventas': [15000, 22000, 9800, 12000, 28000],
    'Costos': [10000, 15000, 7000, 8000, 19000]
}
df = pd.DataFrame(datos)

print("\nDataFrame original:")
print(df)

print("\n" + "-"*70)
print("\n1️⃣  COLUMNA INDIVIDUAL (devuelve Series):")
ventas = df['Ventas']
print(ventas)
print(f"\nTipo: {type(ventas)}")

print("\n" + "-"*70)
print("\n2️⃣  MÚLTIPLES COLUMNAS (devuelve DataFrame):")
ventas_costos = df[['Ventas', 'Costos']]
print(ventas_costos)
print(f"\nTipo: {type(ventas_costos)}")

print("\n" + "-"*70)
print("\n3️⃣  PRIMERAS 3 FILAS:")
print(df.head(3))

print("\n" + "-"*70)
print("\n4️⃣  FILA POR POSICIÓN (.iloc):")
print("\nFila 0:")
print(df.iloc[0])
print("\nFilas 1 a 3:")
print(df.iloc[1:4])

print("\n" + "-"*70)
print("\n5️⃣  COMBINACIÓN: Filas específicas + Columnas específicas")
print("\nFilas 0-2, columnas Sucursal y Ventas:")
print(df.loc[0:2, ['Sucursal', 'Ventas']])

print("\n" + "="*70)
print("✅ Selección de datos completada")
````

## ⚡ Operaciones Vectorizadas en Columnas

### 📊 Aritmética

Las operaciones son **elemento por elemento** (vectorizadas):

```python
df['Ganancia'] = df['Ventas'] - df['Costos']
df['Margen_%'] = (df['Ganancia'] / df['Ventas']) * 100
df['Ventas_IVA'] = df['Ventas'] * 1.21
```

⚡ **Más rápido que loops:** Pandas usa NumPy optimizado en C

---

### 📊 Funciones de Agregación

| Método | Descripción |
|---------|-------------|
| `.sum()` | Suma total |
| `.mean()` | Promedio |
| `.median()` | Mediana |
| `.min()` / `.max()` | Mínimo / Máximo |
| `.std()` | Desviación estándar |
| `.count()` | Cantidad de valores no nulos |
| `.nunique()` | Cantidad de valores únicos |

**Ejemplo:**
```python
total_ventas = df['Ventas'].sum()
promedio_ventas = df['Ventas'].mean()
```

---

### 🔗 Alineación Automática de Índices

**Magia de Pandas:** Operaciones se alinean por índice

```python
s1 = pd.Series([1, 2, 3], index=['A', 'B', 'C'])
s2 = pd.Series([4, 5], index=['A', 'C'])
resultado = s1 + s2
# A    5  (1+4)
# B    NaN (no hay 'B' en s2)
# C    8  (3+5)
```

⚠️ Si no hay coincidencia de índice → **NaN**

````python
import pandas as pd

print("⚡ OPERACIONES EN COLUMNAS")
print("="*70)

# DataFrame
datos = {
    'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste'],
    'Ventas': [15000, 22000, 9800, 12000, 28000],
    'Costos': [10000, 15000, 7000, 8000, 19000]
}
df = pd.DataFrame(datos)

print("\nDataFrame original:")
print(df)

print("\n" + "-"*70)
print("\n1️⃣  CREAR COLUMNA CALCULADA:")
df['Ganancia'] = df['Ventas'] - df['Costos']
df['Margen_%'] = (df['Ganancia'] / df['Ventas']) * 100
print(df)

print("\n" + "-"*70)
print("\n2️⃣  AGREGAR IVA:")
df['Ventas_con_IVA'] = df['Ventas'] * 1.21
print(df[['Sucursal', 'Ventas', 'Ventas_con_IVA']])

print("\n" + "-"*70)
print("\n3️⃣  ESTADÍSTICAS DE COLUMNA:")
print(f"\nTotal Ventas: ${df['Ventas'].sum():,.0f}")
print(f"Promedio Ventas: ${df['Ventas'].mean():,.0f}")
print(f"Máximo Ventas: ${df['Ventas'].max():,.0f}")
print(f"Mínimo Ventas: ${df['Ventas'].min():,.0f}")
print(f"Desv. Estándar: ${df['Ventas'].std():,.0f}")

print("\n" + "-"*70)
print("\n4️⃣  ALINEACIÓN AUTOMÁTICA DE ÍNDICES:")
s1 = pd.Series([100, 200, 300], index=['A', 'B', 'C'], name='Serie1')
s2 = pd.Series([10, 30], index=['A', 'C'], name='Serie2')
resultado = s1 + s2

print("\nSerie 1:")
print(s1)
print("\nSerie 2:")
print(s2)
print("\nResultado (s1 + s2):")
print(resultado)
print("\n⚠️  Nota: 'B' es NaN porque no existe en Serie2")

print("\n" + "="*70)
print("✅ Operaciones vectorizadas completadas")
````
