```python
import pandas as pd

# Creación de DataFrame de ventas contables
datos_ventas = {
    'Fecha': pd.date_range(start='2024-01-01', periods=5, freq='D'),
    'Sucursal': ['Centro', 'Norte', 'Centro', 'Sur', 'Norte'],
    'Monto_Venta': [15000.0, 22000.5, 18500.0, 9800.0, 31000.0],
    'Estado_Cobro': ['Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente']
}

df_ventas = pd.DataFrame(datos_ventas)
print("--- DataFrame de Ventas ---")
print(df_ventas)

# Filtrado booleano: Ventas pendientes de cobro mayores a $20,000
pendientes_altas = df_ventas.loc[(df_ventas['Estado_Cobro'] == 'Pendiente') & (df_ventas['Monto_Venta'] > 20000)]
print("
--- Ventas Pendientes Altas ---")
print(pendientes_altas)

```
