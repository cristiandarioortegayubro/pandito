## 🔄 GroupBy: Split-Apply-Combine

### 📚 El Patrón Fundamental

**GroupBy** es la operación más poderosa de Pandas. Sigue el patrón **Split-Apply-Combine**:

1. **Split (Dividir):** Separar datos en grupos según criterio
2. **Apply (Aplicar):** Aplicar función a cada grupo
3. **Combine (Combinar):** Unir resultados

**Sintaxis:**
```python
df.groupby('columna')['otra_columna'].funcion()
```

---

### 🎯 Funciones de Agregación Comunes

| Función | Descripción | Ejemplo |
|---------|-------------|--------|
| `.sum()` | Suma total | Ventas totales por sucursal |
| `.mean()` | Promedio | Ticket promedio por cliente |
| `.count()` | Cantidad de registros | Transacciones por mes |
| `.min()` / `.max()` | Mínimo / Máximo | Venta mínima/máxima |
| `.std()` | Desviación estándar | Variabilidad de ventas |
| `.agg([...])` | Múltiples funciones | sum, mean, count a la vez |

---

### 💼 Caso Contable: Ventas por Sucursal

```python
# Ventas totales por sucursal
df.groupby('Sucursal')['Ventas'].sum()

# Múltiples métricas a la vez
df.groupby('Sucursal')['Ventas'].agg(['sum', 'mean', 'count'])

# Agrupar por múltiples columnas
df.groupby(['Sucursal', 'Mes'])['Ventas'].sum()
```

---

### 🔑 Diferencia con SQL

**SQL:**
```sql
SELECT Sucursal, SUM(Ventas)
FROM df
GROUP BY Sucursal
```

**Pandas:**
```python
df.groupby('Sucursal')['Ventas'].sum()
```

✅ Pandas es más conciso y flexible

````python
import pandas as pd
import numpy as np

print("🔄 GROUPBY: Split-Apply-Combine")
print("="*70)

# Dataset de ventas
np.random.seed(42)
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=20),
    'Sucursal': np.random.choice(['Centro', 'Norte', 'Sur', 'Este'], 20),
    'Producto': np.random.choice(['Laptop', 'Monitor', 'Teclado'], 20),
    'Ventas': np.random.randint(5000, 50000, 20),
    'Cantidad': np.random.randint(1, 10, 20)
})

print("\n📊 Dataset (20 transacciones):")
print(df.head())

print("\n" + "="*70)
print("\n1️⃣  VENTAS TOTALES POR SUCURSAL:")
ventas_sucursal = df.groupby('Sucursal')['Ventas'].sum()
print(ventas_sucursal)
print(f"\nTipo: {type(ventas_sucursal)} (Series)")

print("\n" + "-"*70)
print("\n2️⃣  MÚLTIPLES MÉTRICAS A LA VEZ (.agg):")
metricas = df.groupby('Sucursal')['Ventas'].agg(['sum', 'mean', 'count'])
print(metricas)
print(f"\nTipo: {type(metricas)} (DataFrame)")

print("\n" + "-"*70)
print("\n3️⃣  AGRUPAR POR MÚLTIPLES COLUMNAS:")
ventas_sucursal_producto = df.groupby(['Sucursal', 'Producto'])['Ventas'].sum()
print(ventas_sucursal_producto)
print("\n💡 Índice jerárquico (MultiIndex)")

print("\n" + "-"*70)
print("\n4️⃣  MÚLTIPLES COLUMNAS + MÚLTIPLES FUNCIONES:")
resumen = df.groupby('Sucursal').agg({
    'Ventas': ['sum', 'mean'],
    'Cantidad': ['sum', 'mean']
})
print(resumen)

print("\n" + "-"*70)
print("\n5️⃣  RESETEAR ÍNDICE (convertir a columnas):")
ventas_reset = df.groupby('Sucursal')['Ventas'].sum().reset_index()
print(ventas_reset)
print("\n💡 Ahora 'Sucursal' es columna, no índice")

print("\n" + "-"*70)
print("\n6️⃣  ENCONTRAR MÁXIMOS POR GRUPO:")
max_venta = df.groupby('Sucursal')['Ventas'].max()
print("\nVenta máxima por sucursal:")
print(max_venta)

print("\n" + "="*70)
print("✅ GroupBy dominado")
````

## 📊 Ordenamiento: .sort_values() y .sort_index()

### 🔑 Dos Tipos de Ordenamiento

**1. Por valores (.sort_values):**
```python
df.sort_values('Ventas')  # Ascendente
df.sort_values('Ventas', ascending=False)  # Descendente
```

**2. Por índice (.sort_index):**
```python
df.sort_index()  # Ordenar por índice de filas
df.sort_index(axis=1)  # Ordenar columnas alfabéticamente
```

---

### 📊 Ordenamiento Múltiple

**Por múltiples columnas:**
```python
df.sort_values(['Sucursal', 'Ventas'], ascending=[True, False])
```

**Orden:**
1. Primero por Sucursal (A→Z)
2. Dentro de cada sucursal, por Ventas (mayor→menor)

---

### 💼 Casos de Uso Contables

**Top N transacciones:**
```python
# Top 10 ventas más altas
df.sort_values('Ventas', ascending=False).head(10)
```

**Ranking de sucursales:**
```python
df.groupby('Sucursal')['Ventas'].sum().sort_values(ascending=False)
```

**Cronológico inverso:**
```python
df.sort_values('Fecha', ascending=False)  # Más reciente primero
```

---

### ⚠️ Importante: inplace

```python
# ❌ NO modifica el original
df.sort_values('Ventas')

# ✅ Asigna el resultado
df_ordenado = df.sort_values('Ventas')

# ✅ O modifica in-place
df.sort_values('Ventas', inplace=True)
```

````python
import pandas as pd
import numpy as np

print("📊 ORDENAMIENTO")
print("="*70)

# Dataset
np.random.seed(42)
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=10),
    'Sucursal': ['Centro', 'Norte', 'Centro', 'Sur', 'Norte', 'Centro', 'Este', 'Sur', 'Norte', 'Centro'],
    'Ventas': [15000, 22000, 18500, 9800, 31000, 17500, 12000, 23000, 19500, 28000],
    'Estado': ['Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente',
               'Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente']
})

print("\n📋 DataFrame original:")
print(df)

print("\n" + "="*70)
print("\n1️⃣  ORDENAR POR VENTAS (menor a mayor):")
df_asc = df.sort_values('Ventas')
print(df_asc[['Sucursal', 'Ventas', 'Estado']])

print("\n" + "-"*70)
print("\n2️⃣  TOP 5 VENTAS MÁS ALTAS:")
top5 = df.sort_values('Ventas', ascending=False).head(5)
print(top5[['Sucursal', 'Ventas']])

print("\n" + "-"*70)
print("\n3️⃣  ORDENAR POR MÚLTIPLES COLUMNAS:")
df_multi = df.sort_values(['Sucursal', 'Ventas'], ascending=[True, False])
print(df_multi[['Sucursal', 'Ventas']])
print("\n💡 Primero por Sucursal (A→Z), luego por Ventas (mayor→menor)")

print("\n" + "-"*70)
print("\n4️⃣  CRONOLÓGICO INVERSO (más reciente primero):")
df_reciente = df.sort_values('Fecha', ascending=False)
print(df_reciente[['Fecha', 'Sucursal', 'Ventas']].head())

print("\n" + "-"*70)
print("\n5️⃣  RANKING DE SUCURSALES (ventas totales):")
ranking = df.groupby('Sucursal')['Ventas'].sum().sort_values(ascending=False)
print(ranking)
print(f"\n🥇 Ganadora: {ranking.idxmax()} con ${ranking.max():,.0f}")

print("\n" + "-"*70)
print("\n6️⃣  ORDENAR POR ÍNDICE:")
df_idx = df.set_index('Fecha').sort_index(ascending=False)
print(df_idx[['Sucursal', 'Ventas']].head())

print("\n" + "="*70)
print("✅ Ordenamiento completado")
````

## 🔍 Manejo de Valores Nulos (NaN)

### 📚 ¿Qué son los NaN?

**NaN** (Not a Number) = Valor faltante o indefinido

**Causas comunes:**
* Datos no capturados
* Errores de integración
* Operaciones inválidas (división por 0)
* Alineación de índices sin match

---

### 🔎 Detectar Nulos

| Método | Descripción |
|--------|-------------|
| `.isnull()` / `.isna()` | Devuelve True donde hay nulo |
| `.notnull()` / `.notna()` | Devuelve True donde NO hay nulo |
| `.isnull().sum()` | Cantidad de nulos por columna |
| `.isnull().sum().sum()` | Total de nulos en DataFrame |

---

### 🛠️ Manejar Nulos

**1. Eliminar:**
```python
df.dropna()  # Eliminar filas con CUALQUIER nulo
df.dropna(how='all')  # Solo si TODAS las columnas son nulas
df.dropna(subset=['Ventas'])  # Solo si 'Ventas' es nulo
```

**2. Rellenar:**
```python
df.fillna(0)  # Reemplazar por 0
df['Ventas'].fillna(df['Ventas'].mean())  # Por promedio
df.fillna(method='ffill')  # Forward fill (propagar anterior)
df.fillna(method='bfill')  # Backward fill (propagar siguiente)
```

**3. Interpolar:**
```python
df['Ventas'].interpolate()  # Interpolación lineal
```

---

### ⚠️ Buenas Prácticas

✅ **Siempre inspeccionar nulos** con `.info()` o `.isnull().sum()`  
✅ **Entender el motivo** antes de eliminar/rellenar  
✅ **Documentar decisiones** de imputación  
⚠️ **No asumir:** Nulo ≠ Cero (significados diferentes)

````python
import pandas as pd
import numpy as np

print("🔍 MANEJO DE VALORES NULOS")
print("="*70)

# Dataset con nulos
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=10),
    'Sucursal': ['Centro', 'Norte', None, 'Sur', 'Norte', 'Centro', None, 'Sur', 'Norte', 'Centro'],
    'Ventas': [15000, 22000, 18500, np.nan, 31000, 17500, 12000, np.nan, 19500, 28000],
    'Estado': ['Cobrado', None, 'Cobrado', 'Cobrado', None,
               'Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', None]
})

print("\n📋 DataFrame con nulos:")
print(df)

print("\n" + "="*70)
print("\n1️⃣  DETECTAR NULOS:")
print("\nCantidad de nulos por columna:")
print(df.isnull().sum())
print(f"\nTotal de nulos: {df.isnull().sum().sum()}")
print(f"Porcentaje de nulos: {df.isnull().sum().sum() / df.size * 100:.1f}%")

print("\n" + "-"*70)
print("\n2️⃣  VISUALIZAR FILAS CON NULOS:")
filas_con_nulos = df[df.isnull().any(axis=1)]
print(filas_con_nulos)

print("\n" + "-"*70)
print("\n3️⃣  ELIMINAR FILAS CON NULOS:")
df_sin_nulos = df.dropna()
print(df_sin_nulos)
print(f"\nFilas eliminadas: {len(df) - len(df_sin_nulos)}")

print("\n" + "-"*70)
print("\n4️⃣  RELLENAR CON VALOR CONSTANTE:")
df_fill_0 = df.copy()
df_fill_0['Ventas'] = df_fill_0['Ventas'].fillna(0)
print(df_fill_0[['Sucursal', 'Ventas']])

print("\n" + "-"*70)
print("\n5️⃣  RELLENAR CON PROMEDIO:")
df_fill_mean = df.copy()
promedio_ventas = df['Ventas'].mean()
df_fill_mean['Ventas'] = df_fill_mean['Ventas'].fillna(promedio_ventas)
print(f"\nPromedio de Ventas: ${promedio_ventas:,.0f}")
print(df_fill_mean[['Sucursal', 'Ventas']])

print("\n" + "-"*70)
print("\n6️⃣  RELLENAR TEXTO CON VALOR DEFAULT:")
df_fill_text = df.copy()
df_fill_text['Sucursal'] = df_fill_text['Sucursal'].fillna('Sin Asignar')
df_fill_text['Estado'] = df_fill_text['Estado'].fillna('Pendiente')
print(df_fill_text)

print("\n" + "-"*70)
print("\n7️⃣  FORWARD FILL (propagar valor anterior):")
df_ffill = df.copy()
df_ffill['Estado'] = df_ffill['Estado'].fillna(method='ffill')
print(df_ffill[['Fecha', 'Estado']])
print("\n💡 Los nulos heredan el valor anterior")

print("\n" + "="*70)
print("✅ Manejo de nulos completado")
````

## ⚙️ .apply() - Aplicar Funciones Personalizadas

### 📚 ¿Qué es .apply()?

**.apply()** permite aplicar una función a cada elemento, fila o columna del DataFrame.

**Sintaxis:**
```python
df['columna'].apply(funcion)  # A cada elemento de la columna
df.apply(funcion, axis=0)     # A cada columna
df.apply(funcion, axis=1)     # A cada fila
```

---

### 🎯 Casos de Uso

**1. Transformaciones simples:**
```python
# Redondear a miles
df['Ventas'].apply(lambda x: round(x, -3))

# Capitalizar texto
df['Sucursal'].apply(lambda x: x.upper())
```

**2. Lógica condicional:**
```python
# Categorizar ventas
def categorizar(ventas):
    if ventas < 10000:
        return 'Baja'
    elif ventas < 25000:
        return 'Media'
    else:
        return 'Alta'

df['Categoria'] = df['Ventas'].apply(categorizar)
```

**3. Operaciones fila por fila:**
```python
# Calcular descuento según condición
def calcular_descuento(fila):
    if fila['Estado'] == 'Cobrado' and fila['Ventas'] > 20000:
        return 0.10
    else:
        return 0

df['Descuento_%'] = df.apply(calcular_descuento, axis=1)
```

---

### ⚡ Alternativas Más Rápidas

**Cuando sea posible, evita .apply() y usa operaciones vectorizadas:**

```python
# ❌ Lento con .apply()
df['Doble'] = df['Ventas'].apply(lambda x: x * 2)

# ✅ Rápido (vectorizado)
df['Doble'] = df['Ventas'] * 2
```

**Pero .apply() es útil para lógica compleja no vectorizable**

````python
import pandas as pd
import numpy as np

print("⚙️ .apply() - FUNCIONES PERSONALIZADAS")
print("="*70)

# Dataset
np.random.seed(42)
df = pd.DataFrame({
    'Sucursal': ['centro', 'norte', 'sur', 'este', 'centro'],
    'Ventas': [15300, 22000, 9800, 31000, 28000],
    'Costos': [10000, 15000, 7000, 21000, 19000],
    'Estado': ['Cobrado', 'Pendiente', 'Cobrado', 'Cobrado', 'Pendiente']
})

print("\n📋 DataFrame original:")
print(df)

print("\n" + "="*70)
print("\n1️⃣  TRANSFORMACIÓN SIMPLE (capitalizar):")
df['Sucursal_Cap'] = df['Sucursal'].apply(lambda x: x.capitalize())
print(df[['Sucursal', 'Sucursal_Cap']])

print("\n" + "-"*70)
print("\n2️⃣  CATEGORIZAR VENTAS (función personalizada):")

def categorizar_ventas(ventas):
    if ventas < 15000:
        return 'Baja'
    elif ventas < 25000:
        return 'Media'
    else:
        return 'Alta'

df['Categoria'] = df['Ventas'].apply(categorizar_ventas)
print(df[['Sucursal', 'Ventas', 'Categoria']])

print("\n" + "-"*70)
print("\n3️⃣  REDONDEAR A MILES:")
df['Ventas_Redondeadas'] = df['Ventas'].apply(lambda x: round(x, -3))
print(df[['Ventas', 'Ventas_Redondeadas']])

print("\n" + "-"*70)
print("\n4️⃣  OPERACIÓN FILA POR FILA (axis=1):")

def calcular_descuento(fila):
    if fila['Estado'] == 'Cobrado' and fila['Ventas'] > 20000:
        return 10.0
    elif fila['Estado'] == 'Cobrado':
        return 5.0
    else:
        return 0.0

df['Descuento_%'] = df.apply(calcular_descuento, axis=1)
print(df[['Sucursal', 'Ventas', 'Estado', 'Descuento_%']])

print("\n" + "-"*70)
print("\n5️⃣  CALCULAR COLUMNA DERIVADA:")

def calcular_margen(fila):
    ganancia = fila['Ventas'] - fila['Costos']
    margen = (ganancia / fila['Ventas']) * 100
    return round(margen, 1)

df['Margen_%'] = df.apply(calcular_margen, axis=1)
print(df[['Sucursal', 'Ventas', 'Costos', 'Margen_%']])

print("\n" + "-"*70)
print("\n6️⃣  COMPARACIÓN: .apply() vs vectorizado")

import time

# .apply()
start = time.time()
df['Doble_apply'] = df['Ventas'].apply(lambda x: x * 2)
tiempo_apply = time.time() - start

# Vectorizado
start = time.time()
df['Doble_vect'] = df['Ventas'] * 2
tiempo_vect = time.time() - start

print(f"\n⏱️  .apply(): {tiempo_apply*1000:.4f} ms")
print(f"⏱️  Vectorizado: {tiempo_vect*1000:.4f} ms")
print(f"\n💡 Vectorizado es {tiempo_apply/tiempo_vect:.0f}x más rápido")

print("\n" + "="*70)
print("✅ .apply() dominado")
````

## 🔄 Pivot Tables: Tablas Dinámicas

### 📚 ¿Qué son las Pivot Tables?

**Pivot** = Reestructurar datos (de largo a ancho, o viceversa)

Convierten datos transaccionales en formato de reporte.

---

### 🎯 .pivot_table()

**Sintaxis:**
```python
df.pivot_table(
    values='columna_valores',    # Qué agregar
    index='columna_filas',        # Filas
    columns='columna_columnas',   # Columnas
    aggfunc='sum'                 # Función de agregación
)
```

**Ejemplo:**
```python
# Ventas por Sucursal (filas) y Mes (columnas)
df.pivot_table(
    values='Ventas',
    index='Sucursal',
    columns='Mes',
    aggfunc='sum'
)
```

**Resultado:**
```
Mes       Enero  Febrero  Marzo
Sucursal                        
Centro    50000   45000   52000
Norte     62000   58000   61000
```

---

### 📊 Funciones de Agregación

* `'sum'`: Sumar
* `'mean'`: Promedio
* `'count'`: Contar
* `'min'` / `'max'`: Mínimo / Máximo
* `['sum', 'mean']`: Múltiples a la vez

---

### 🔄 .melt() - Inverso de Pivot

**De ancho a largo:**

```python
# Antes (ancho)
Sucursal  Enero  Febrero
Centro    50000  45000

# Después (largo)
Sucursal  Mes      Ventas
Centro    Enero    50000
Centro    Febrero  45000
```

**Código:**
```python
df.melt(id_vars=['Sucursal'], var_name='Mes', value_name='Ventas')
```

````python
import pandas as pd
import numpy as np

print("🔄 PIVOT TABLES")
print("="*70)

# Dataset
np.random.seed(42)
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=30),
    'Sucursal': np.random.choice(['Centro', 'Norte', 'Sur'], 30),
    'Producto': np.random.choice(['Laptop', 'Monitor', 'Teclado'], 30),
    'Ventas': np.random.randint(5000, 50000, 30)
})

df['Mes'] = df['Fecha'].dt.month_name()

print("\n📋 Dataset (30 transacciones):")
print(df.head(10))

print("\n" + "="*70)
print("\n1️⃣  PIVOT: Ventas por Sucursal (filas) y Producto (columnas):")
pivot_1 = df.pivot_table(
    values='Ventas',
    index='Sucursal',
    columns='Producto',
    aggfunc='sum'
)
print(pivot_1)

print("\n" + "-"*70)
print("\n2️⃣  PIVOT: Múltiples funciones de agregación:")
pivot_2 = df.pivot_table(
    values='Ventas',
    index='Sucursal',
    columns='Producto',
    aggfunc=['sum', 'mean', 'count']
)
print(pivot_2)

print("\n" + "-"*70)
print("\n3️⃣  PIVOT: Con totales (margins):")
pivot_3 = df.pivot_table(
    values='Ventas',
    index='Sucursal',
    columns='Producto',
    aggfunc='sum',
    margins=True,
    margins_name='Total'
)
print(pivot_3)

print("\n" + "-"*70)
print("\n4️⃣  PIVOT: Rellenar nulos con 0:")
pivot_4 = df.pivot_table(
    values='Ventas',
    index='Sucursal',
    columns='Producto',
    aggfunc='sum',
    fill_value=0
)
print(pivot_4)

print("\n" + "-"*70)
print("\n5️⃣  MELT: De ancho a largo (inverso)")

# Crear tabla ancha
df_ancho = pd.DataFrame({
    'Sucursal': ['Centro', 'Norte', 'Sur'],
    'Enero': [50000, 62000, 45000],
    'Febrero': [55000, 58000, 48000]
})

print("\nTabla ancha (formato reporte):")
print(df_ancho)

df_largo = df_ancho.melt(
    id_vars=['Sucursal'],
    var_name='Mes',
    value_name='Ventas'
)

print("\nTabla larga (formato transaccional):")
print(df_largo)

print("\n" + "="*70)
print("✅ Pivot tables completadas")
````

## 💼 Caso Integrador: Análisis Completo de Ventas

### 📈 Entregables

3. Alertas de bajo desempeño
4. Visualización de top productos

````python
import pandas as pd
import numpy as np

print("💼 CASO INTEGRADOR: Reporte Ejecutivo")
print("="*70)

# Dataset realista
np.random.seed(42)
n = 100
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=n),
    'Sucursal': np.random.choice(['Centro', 'Norte', 'Sur', 'Este', 'Oeste'], n),
    'Producto': np.random.choice(['Laptop', 'Monitor', 'Teclado', 'Mouse', 'Webcam'], n),
    'Ventas': np.random.randint(5000, 50000, n),
    'Costos': np.random.randint(3000, 35000, n),
    'Estado': np.random.choice(['Cobrado', 'Pendiente'], n, p=[0.7, 0.3])
})

df['Mes'] = df['Fecha'].dt.month
df['Ganancia'] = df['Ventas'] - df['Costos']
df['Margen_%'] = (df['Ganancia'] / df['Ventas'] * 100).round(1)

print("\n📊 Dataset (100 transacciones):")
print(df.head())
print(f"\nShape: {df.shape}")

print("\n" + "="*70)
print("\n1️⃣  VENTAS TOTALES POR SUCURSAL:")
ventas_sucursal = df.groupby('Sucursal')['Ventas'].agg(['sum', 'mean', 'count'])
ventas_sucursal.columns = ['Total', 'Promedio', 'Transacciones']
ventas_sucursal = ventas_sucursal.sort_values('Total', ascending=False)
print(ventas_sucursal)

print("\n" + "-"*70)
print("\n2️⃣  TOP 5 PRODUCTOS MÁS VENDIDOS:")
top_productos = df.groupby('Producto')['Ventas'].sum().sort_values(ascending=False).head(5)
print(top_productos)
print(f"\n🥇 Producto estrella: {top_productos.idxmax()} (${top_productos.max():,.0f})")

print("\n" + "-"*70)
print("\n3️⃣  ALERTA: Sucursales con bajo desempeño (<70% promedio):")
promedio_general = df.groupby('Sucursal')['Ventas'].sum().mean()
umbral = promedio_general * 0.7

sucursales_bajas = ventas_sucursal[ventas_sucursal['Total'] < umbral]
print(f"\nPromedio general: ${promedio_general:,.0f}")
print(f"Umbral (70%): ${umbral:,.0f}")
print(f"\n⚠️  {len(sucursales_bajas)} sucursales bajo umbral:")
if len(sucursales_bajas) > 0:
    print(sucursales_bajas[['Total']])
else:
    print("✅ Todas las sucursales sobre el umbral")

print("\n" + "-"*70)
print("\n4️⃣  TASA DE COBRO POR SUCURSAL:")
tasa_cobro = df.groupby(['Sucursal', 'Estado'])['Ventas'].sum().unstack(fill_value=0)
tasa_cobro['Total'] = tasa_cobro.sum(axis=1)
tasa_cobro['Tasa_Cobro_%'] = (tasa_cobro['Cobrado'] / tasa_cobro['Total'] * 100).round(1)
print(tasa_cobro[['Cobrado', 'Pendiente', 'Tasa_Cobro_%']])

print("\n" + "-"*70)
print("\n5️⃣  ANÁLISIS DE MÁRGENES POR PRODUCTO:")

def categorizar_margen(margen):
    if margen < 20:
        return 'Bajo'
    elif margen < 40:
        return 'Medio'
    else:
        return 'Alto'

df['Cat_Margen'] = df['Margen_%'].apply(categorizar_margen)

margen_producto = df.groupby('Producto').agg({
    'Margen_%': 'mean',
    'Ventas': 'sum'
}).round(1)
margen_producto = margen_producto.sort_values('Margen_%', ascending=False)
print(margen_producto)

print("\n" + "-"*70)
print("\n6️⃣  TABLA PIVOT: Ventas por Sucursal × Producto:")
pivot_reporte = df.pivot_table(
    values='Ventas',
    index='Sucursal',
    columns='Producto',
    aggfunc='sum',
    margins=True,
    margins_name='Total'
)
print(pivot_reporte)

print("\n" + "="*70)
print("\n📈 RESUMEN EJECUTIVO:")
print(f"\n💰 Ventas totales: ${df['Ventas'].sum():,.0f}")
print(f"💵 Ganancia total: ${df['Ganancia'].sum():,.0f}")
print(f"📊 Margen promedio: {df['Margen_%'].mean():.1f}%")
print(f"✅ Transacciones cobradas: {(df['Estado'] == 'Cobrado').sum()} ({(df['Estado'] == 'Cobrado').sum()/len(df)*100:.0f}%)")
print(f"⏳ Transacciones pendientes: {(df['Estado'] == 'Pendiente').sum()} ({(df['Estado'] == 'Pendiente').sum()/len(df)*100:.0f}%)")

print("\n" + "="*70)
print("✅ Reporte ejecutivo completado")
````
