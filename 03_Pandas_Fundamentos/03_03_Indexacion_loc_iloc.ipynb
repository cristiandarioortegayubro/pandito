## 🔢 .iloc - Indexación por Posición (Integer Location)

### 📚 Definición

**.iloc** selecciona por **posición numérica** (0-indexed, como listas de Python).

---

### 🔑 Sintaxis

```python
df.iloc[fila, columna]  # Ambos son enteros o slices
```

---

### 📊 Ejemplos

| Código | Descripción | Devuelve |
|--------|-------------|----------|
| `df.iloc[0]` | Primera fila | Series |
| `df.iloc[2]` | Tercera fila | Series |
| `df.iloc[-1]` | Última fila | Series |
| `df.iloc[0:3]` | Filas 0, 1, 2 | DataFrame |
| `df.iloc[[0, 2, 4]]` | Filas 0, 2, 4 | DataFrame |
| `df.iloc[:, 0]` | Primera columna | Series |
| `df.iloc[:, 0:2]` | Primeras 2 columnas | DataFrame |
| `df.iloc[1:4, 0:2]` | Filas 1-3, columnas 0-1 | DataFrame |

---

### ⚠️ Importante

❗ **Slicing con .iloc es EXCLUSIVO del último**

```python
df.iloc[0:3]  # Filas 0, 1, 2 (NO incluye 3)
```

(Igual que Python: `lista[0:3]`)

---

### 💼 Caso de Uso

**Cuándo usar .iloc:**
* Primeras/últimas N filas
* Posiciones específicas conocidas
* Bucles numéricos
* No te importan las etiquetas

````python
import pandas as pd

print("🔢 .iloc - INDEXACIÓN POR POSICIÓN")
print("="*70)

# DataFrame de ejemplo
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=6),
    'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste', 'Centro'],
    'Ventas': [15000, 22000, 9800, 12000, 28000, 17500],
    'Costos': [10000, 15000, 7000, 8000, 19000, 11000]
})

print("\n📊 DataFrame original:")
print(df)
print(f"\nShape: {df.shape} (6 filas × 4 columnas)")

print("\n" + "-"*70)
print("\n1️⃣  PRIMERA FILA (posición 0):")
print(df.iloc[0])
print(f"\nTipo: {type(df.iloc[0])} (Series)")

print("\n" + "-"*70)
print("\n2️⃣  TERCERA FILA (posición 2):")
print(df.iloc[2])

print("\n" + "-"*70)
print("\n3️⃣  ÚLTIMA FILA (posición -1):")
print(df.iloc[-1])

print("\n" + "-"*70)
print("\n4️⃣  PRIMERAS 3 FILAS (0:3):")
print(df.iloc[0:3])
print("\n⚠️  Nota: 0:3 incluye 0,1,2 (NO incluye 3)")

print("\n" + "-"*70)
print("\n5️⃣  FILAS ESPECÍFICAS [0, 2, 4]:")
print(df.iloc[[0, 2, 4]])

print("\n" + "-"*70)
print("\n6️⃣  FILA 1, COLUMNAS 1 Y 2:")
print(df.iloc[1, [1, 2]])

print("\n7️⃣  FILAS 0-2, COLUMNAS 1-3:")
print(df.iloc[0:3, 1:4])

print("\n" + "="*70)
print("✅ .iloc dominado")
````

## 🏷️ .loc - Indexación por Etiqueta (Label Location)

### 📚 Definición

**.loc** selecciona por **nombres de índice y columnas** (etiquetas).

---

### 🔑 Sintaxis

```python
df.loc[etiqueta_fila, etiqueta_columna]  # Ambos son nombres
```

---

### 📊 Ejemplos

| Código | Descripción | Devuelve |
|--------|-------------|----------|
| `df.loc[0]` | Fila con índice 0 | Series |
| `df.loc['A']` | Fila con índice 'A' | Series |
| `df.loc[0:3]` | Filas 0, 1, 2, **3** | DataFrame |
| `df.loc[[0, 2]]` | Filas 0 y 2 | DataFrame |
| `df.loc[:, 'Ventas']` | Columna 'Ventas' | Series |
| `df.loc[:, ['Ventas', 'Costos']]` | Columnas Ventas y Costos | DataFrame |
| `df.loc[0:2, 'Ventas':'Costos']` | Filas 0-2, cols Ventas a Costos | DataFrame |

---

### ⚠️ Diferencia Crítica

❗ **Slicing con .loc es INCLUSIVO**

```python
df.loc[0:3]  # Filas 0, 1, 2, 3 (incluye 3)
```

🎯 **Diferente a .iloc:**
```python
df.iloc[0:3]  # Filas 0, 1, 2 (NO incluye 3)
df.loc[0:3]   # Filas 0, 1, 2, 3 (incluye 3)
```

---

### 💼 Caso de Uso

**Cuándo usar .loc:**
* Tienes índice personalizado (fechas, IDs, nombres)
* Quieres columnas específicas por nombre
* Filtrado booleano (próximo notebook)
* Código legible y auto-documentado

````python
import pandas as pd

print("🏷️ .loc - INDEXACIÓN POR ETIQUETA")
print("="*70)

# DataFrame con índice personalizado
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=5),
    'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste'],
    'Ventas': [15000, 22000, 9800, 12000, 28000],
    'Costos': [10000, 15000, 7000, 8000, 19000]
})

# Establecer Sucursal como índice
df.set_index('Sucursal', inplace=True)

print("\n📊 DataFrame con índice personalizado (Sucursal):")
print(df)
print(f"\nÍndice: {df.index.tolist()}")

print("\n" + "-"*70)
print("\n1️⃣  FILA CON ETIQUETA 'Centro':")
print(df.loc['Centro'])

print("\n" + "-"*70)
print("\n2️⃣  COLUMNA 'Ventas' (todas las filas):")
print(df.loc[:, 'Ventas'])
print(f"\nTipo: {type(df.loc[:, 'Ventas'])} (Series)")

print("\n" + "-"*70)
print("\n3️⃣  MÚLTIPLES COLUMNAS:")
print(df.loc[:, ['Ventas', 'Costos']])

print("\n" + "-"*70)
print("\n4️⃣  FILAS 'Centro' y 'Norte', columna 'Ventas':")
print(df.loc[['Centro', 'Norte'], 'Ventas'])

print("\n" + "-"*70)
print("\n5️⃣  SLICING INCLUSIVO ('Centro' a 'Sur'):")
print(df.loc['Centro':'Sur'])
print("\n⚠️  Nota: Incluye 'Centro', 'Norte' Y 'Sur' (inclusivo)")

print("\n" + "-"*70)
print("\n6️⃣  VALOR ESPECÍFICO (Centro, Ventas):")
valor = df.loc['Centro', 'Ventas']
print(f"\nVentas en Centro: ${valor:,.0f}")

print("\n" + "="*70)
print("✅ .loc dominado")
````

## 🔪 Comparación .loc vs .iloc

### 📊 Tabla Comparativa

| Característica | .iloc | .loc |
|----------------|-------|------|
| **Tipo de selector** | Posición (enteros) | Etiquetas (nombres) |
| **Indexación** | 0, 1, 2... | 'A', 0, '2024-01-01'... |
| **Slicing** | Exclusivo (0:3 → 0,1,2) | Inclusivo (0:3 → 0,1,2,3) |
| **Uso común** | Primeras/últimas filas | Columnas por nombre |
| **Legibilidad** | 🟡 Medio | 🟢 Alta |
| **Performance** | 🟢 Rápido | 🟢 Rápido |

---

### 📊 Ejemplos Lado a Lado

```python
# Obtener primera fila
df.iloc[0]        # Por posición
df.loc[0]         # Por etiqueta (si índice es 0)

# Obtener primeras 3 filas
df.iloc[0:3]      # Filas 0, 1, 2 (exclusivo)
df.loc[0:2]       # Filas 0, 1, 2 (inclusivo)

# Obtener columna
df.iloc[:, 2]     # Tercera columna (posición)
df.loc[:, 'Ventas']  # Columna 'Ventas' (nombre)

# Celda específica
df.iloc[1, 2]     # Fila 1, columna 2
df.loc[0, 'Ventas']  # Fila con índice 0, columna 'Ventas'
```

---

### 🎯 ¿Cuándo Usar Cada Uno?

🔢 **Usa .iloc cuando:**
* Posición es lo que importa ("primeros 10")
* Bucles numéricos
* Código genérico (no sabes nombres de columnas)

🏷️ **Usa .loc cuando:**
* Tienes etiquetas significativas
* Quieres código legible
* Filtrado booleano (próximo notebook)
* **👑 Recomendado en general**

---

### 💡 Buena Práctica

✅ **Preferir .loc siempre que sea posible**

```python
# 🟢 BIEN (legible)
df_subset = df.loc[:, ['Fecha', 'Ventas', 'Costos']]

# 🟡 OK pero menos claro
df_subset = df.iloc[:, [0, 2, 3]]
```

````python
import pandas as pd

print("🔪 COMPARACIÓN .loc vs .iloc")
print("="*70)

# DataFrame de ejemplo
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=5),
    'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste'],
    'Ventas': [15000, 22000, 9800, 12000, 28000],
    'Costos': [10000, 15000, 7000, 8000, 19000]
})

print("\n📊 DataFrame original:")
print(df)

print("\n" + "="*70)
print("\n1️⃣  PRIMERA FILA - Ambos métodos")
print("\n.iloc[0]:")
print(df.iloc[0])
print("\n.loc[0]:")
print(df.loc[0])
print("\n✅ Resultado idéntico (cuando índice es 0, 1, 2...)")

print("\n" + "="*70)
print("\n2️⃣  SLICING - LA GRAN DIFERENCIA")
print("\n.iloc[0:3] (Exclusivo):")
print(df.iloc[0:3])
print("\n.loc[0:3] (Inclusivo):")
print(df.loc[0:3])
print("\n⚠️  .iloc[0:3] = 3 filas (0,1,2)")
print("⚠️  .loc[0:3]  = 4 filas (0,1,2,3)")

print("\n" + "="*70)
print("\n3️⃣  COLUMNA - Por posición vs nombre")
print("\n.iloc[:, 2] (tercera columna por posición):")
print(df.iloc[:, 2])
print("\n.loc[:, 'Ventas'] (columna por nombre):")
print(df.loc[:, 'Ventas'])
print("\n✅ .loc es más legible")

print("\n" + "="*70)
print("\n4️⃣  VALOR ESPECÍFICO")
valor_iloc = df.iloc[1, 2]
valor_loc = df.loc[1, 'Ventas']
print(f"\n.iloc[1, 2]:      ${valor_iloc:,.0f}")
print(f".loc[1, 'Ventas']: ${valor_loc:,.0f}")
print("\n✅ Mismo resultado, .loc más claro")

print("\n" + "="*70)
print("✅ Comparación completada")
````

## ✏️ Asignación de Valores con .loc e .iloc

### 📝 Modificar Datos

Puedes usar .loc e .iloc para **asignar nuevos valores**.

---

### 🔑 Sintaxis

```python
# Asignar valor único
df.loc[fila, columna] = nuevo_valor
df.iloc[fila, columna] = nuevo_valor

# Asignar a múltiples celdas
df.loc[filas, columnas] = nuevo_valor
df.loc[condicion, columna] = nuevo_valor  # Filtrado booleano
```

---

### 📊 Ejemplos

```python
# Cambiar valor específico
df.loc[0, 'Ventas'] = 20000

# Cambiar columna completa
df.loc[:, 'Estado'] = 'Cobrado'

# Cambiar múltiples valores
df.loc[0:2, 'Estado'] = 'Pendiente'

# Crear columna nueva
df.loc[:, 'Ganancia'] = df['Ventas'] - df['Costos']

# Asignación condicional (filtrado booleano)
df.loc[df['Ventas'] > 20000, 'Estado'] = 'Alta'
```

---

### ⚠️ Advertencia: SettingWithCopyWarning

❌ **EVITAR:**
```python
df[df['Ventas'] > 10000]['Estado'] = 'Alta'  # Puede fallar
```

✅ **CORRECTO:**
```python
df.loc[df['Ventas'] > 10000, 'Estado'] = 'Alta'  # Siempre funciona
```

**Regla de oro:** Usa siempre .loc para asignaciones

````python
import pandas as pd
import numpy as np

print("✏️ ASIGNACIÓN DE VALORES")
print("="*70)

# DataFrame de ejemplo
df = pd.DataFrame({
    'Fecha': pd.date_range('2024-01-01', periods=5),
    'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste'],
    'Ventas': [15000, 22000, 9800, 12000, 28000],
    'Costos': [10000, 15000, 7000, 8000, 19000],
    'Estado': ['Pendiente'] * 5
})

print("\n📊 DataFrame original:")
print(df)

print("\n" + "-"*70)
print("\n1️⃣  MODIFICAR VALOR ESPECÍFICO:")
df.loc[0, 'Estado'] = 'Cobrado'
print("\nDespués de: df.loc[0, 'Estado'] = 'Cobrado'")
print(df)

print("\n" + "-"*70)
print("\n2️⃣  CREAR COLUMNA CALCULADA:")
df.loc[:, 'Ganancia'] = df['Ventas'] - df['Costos']
df.loc[:, 'Margen_%'] = (df['Ganancia'] / df['Ventas'] * 100).round(1)
print("\nDespués de crear Ganancia y Margen_%:")
print(df)

print("\n" + "-"*70)
print("\n3️⃣  ASIGNACIÓN CONDICIONAL (importante para próximo notebook):")
print("\nMarcar como 'Cobrado' donde Ventas > 15000:")
df.loc[df['Ventas'] > 15000, 'Estado'] = 'Cobrado'
print(df[['Sucursal', 'Ventas', 'Estado']])

print("\n" + "-"*70)
print("\n4️⃣  MODIFICAR RANGO DE FILAS:")
df_copy = df.copy()
df_copy.loc[0:2, 'Estado'] = 'Verificado'
print("\nDespués de: df.loc[0:2, 'Estado'] = 'Verificado'")
print(df_copy[['Sucursal', 'Estado']])

print("\n" + "="*70)
print("✅ Asignación de valores dominada")
````

### 📊 Guía Rápida: .loc vs .iloc

| Tarea | .iloc | .loc |
|-------|-------|------|
| Primera fila | `.iloc[0]` | `.loc[0]` |
| Última fila | `.iloc[-1]` | `.loc[df.index[-1]]` |
| Primeras 3 filas | `.iloc[0:3]` | `.loc[0:2]` |
| Columna por nombre | `.iloc[:, 2]` | `.loc[:, 'Ventas']` |
| Valor específico | `.iloc[1, 2]` | `.loc[1, 'Ventas']` |
| **Filtrado booleano** | ❌ No soportado | ✅ `.loc[condicion]` |
