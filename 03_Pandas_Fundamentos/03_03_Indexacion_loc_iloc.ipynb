## 🔢 .iloc - Indexación por Posición (Integer Location)

### 📚 Definición

**.iloc** selecciona por **posición numérica** (0-indexed, como listas de Python).

---

### 🔑 Sintaxis

```python
df.iloc[fila, columna]  # Ambos son enteros o slices
```

---

### 📊 Ejemplos

| Código | Descripción | Devuelve |
|--------|-------------|----------|
| `df.iloc[0]` | Primera fila | Series |
| `df.iloc[2]` | Tercera fila | Series |
| `df.iloc[-1]` | Última fila | Series |
| `df.iloc[0:3]` | Filas 0, 1, 2 | DataFrame |
| `df.iloc[[0, 2, 4]]` | Filas 0, 2, 4 | DataFrame |
| `df.iloc[:, 0]` | Primera columna | Series |
| `df.iloc[:, 0:2]` | Primeras 2 columnas | DataFrame |
| `df.iloc[1:4, 0:2]` | Filas 1-3, columnas 0-1 | DataFrame |

---

### ⚠️ Importante

❗ **Slicing con .iloc es EXCLUSIVO del último**

```python
df.iloc[0:3]  # Filas 0, 1, 2 (NO incluye 3)
```

(Igual que Python: `lista[0:3]`)

---

### 💼 Caso de uso

**Cuándo usar .iloc:**
* Primeras/últimas N filas
* Posiciones específicas conocidas
* Bucles numéricos
* No te importan las etiquetas

````python
import pandas as pd

print("🔢 .iloc - INDEXACIÓN POR POSICIÓN")
print("="*70)

if USAR_DATOS_REALES and df_andes is not None:
    print("\n🏪 USANDO DATOS REALES: Los Andes Market")
    print("-"*70)
    
    df = df_andes.copy()
    
    print("\n📊 DataFrame completo:")
    print(f"Shape: {df.shape} ({df.shape[0]} filas × {df.shape[1]} columnas)")
    print(f"Columnas: {list(df.columns)}")
    
    print("\n" + "-"*70)
    print("\n1️⃣  PRIMERA FILA (posición 0):")
    print(df.iloc[0])
    print(f"\nTipo: {type(df.iloc[0])} (Series)")
    print(f"\nInterpretación: Primer registro de ventas - {df.iloc[0]['sucursal_nombre']}")
    
    print("\n" + "-"*70)
    print("\n2️⃣  DÉCIMA FILA (posición 9):")
    print(df.iloc[9])
    print(f"\nSucursal: {df.iloc[9]['sucursal_nombre']} | Ventas: ${df.iloc[9]['ventas']:,.2f}")
    
    print("\n" + "-"*70)
    print("\n3️⃣  ÚLTIMA FILA (posición -1):")
    fila_ultima = df.iloc[-1]
    print(fila_ultima)
    print(f"\nÚltimo registro: {fila_ultima['fecha'].strftime('%Y-%m-%d')} - {fila_ultima['sucursal_nombre']}")
    
    print("\n" + "-"*70)
    print("\n4️⃣  PRIMERAS 10 FILAS (0:10):")
    print(df.iloc[0:10][['fecha', 'sucursal_nombre', 'ventas', 'zona']])
    print("\n⚠️  Nota: 0:10 incluye filas 0-9 (NO incluye 10)")
    
    print("\n" + "-"*70)
    print("\n5️⃣  FILAS ESPECÍFICAS [0, 50, 100, 150, 200]:")
    print(df.iloc[[0, 50, 100, 150, 200]][['fecha', 'sucursal_nombre', 'ventas']])
    
    print("\n" + "-"*70)
    print("\n6️⃣  FILA 10, COLUMNAS 1, 2 Y 6 (sucursal_id, sucursal_nombre, ventas):")
    resultado = df.iloc[10, [1, 2, 6]]
    print(resultado)
    print(f"\nInterpretación: Registro {resultado.name} - {resultado['sucursal_nombre']}: ${resultado['ventas']:,.2f}")
    
    print("\n" + "-"*70)
    print("\n7️⃣  PRIMERAS 15 FILAS, COLUMNAS 0-4 (fecha, sucursal_id, sucursal_nombre, lat, lon):")
    print(df.iloc[0:15, 0:5])
    
    print("\n" + "-"*70)
    print("\n8️⃣  ÚLTIMAS 5 FILAS, COLUMNAS DE VENTAS Y UBICACIÓN:")
    print(df.iloc[-5:, [0, 2, 3, 4, 6]])
    
else:
    print("\n📋 EJEMPLO SINTÉTICO:")
    print("-"*70)
    
    # DataFrame de ejemplo
    df = pd.DataFrame({
        'Fecha': pd.date_range('2024-01-01', periods=6),
        'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste', 'Centro'],
        'Ventas': [15000, 22000, 9800, 12000, 28000, 17500],
        'Costos': [10000, 15000, 7000, 8000, 19000, 11000]
    })
    
    print("\n📊 DataFrame original:")
    print(df)
    print(f"\nShape: {df.shape} (6 filas × 4 columnas)")
    
    print("\n" + "-"*70)
    print("\n1️⃣  PRIMERA FILA (posición 0):")
    print(df.iloc[0])
    print(f"\nTipo: {type(df.iloc[0])} (Series)")
    
    print("\n" + "-"*70)
    print("\n2️⃣  TERCERA FILA (posición 2):")
    print(df.iloc[2])
    
    print("\n" + "-"*70)
    print("\n3️⃣  ÚLTIMA FILA (posición -1):")
    print(df.iloc[-1])
    
    print("\n" + "-"*70)
    print("\n4️⃣  PRIMERAS 3 FILAS (0:3):")
    print(df.iloc[0:3])
    print("\n⚠️  Nota: 0:3 incluye 0,1,2 (NO incluye 3)")
    
    print("\n" + "-"*70)
    print("\n5️⃣  FILAS ESPECÍFICAS [0, 2, 4]:")
    print(df.iloc[[0, 2, 4]])
    
    print("\n" + "-"*70)
    print("\n6️⃣  FILA 1, COLUMNAS 1 Y 2:")
    print(df.iloc[1, [1, 2]])
    
    print("\n7️⃣  FILAS 0-2, COLUMNAS 1-3:")
    print(df.iloc[0:3, 1:4])

print("\n" + "="*70)
print("✅ .iloc dominado")
````

## 🏷️ .loc - Indexación por Etiqueta (Label Location)

### 📚 Definición

**.loc** selecciona por **nombres de índice y columnas** (etiquetas).

---

### 🔑 Sintaxis

```python
df.loc[etiqueta_fila, etiqueta_columna]  # Ambos son nombres
```

---

### 📊 Ejemplos

| Código | Descripción | Devuelve |
|--------|-------------|----------|
| `df.loc[0]` | Fila con índice 0 | Series |
| `df.loc['A']` | Fila con índice 'A' | Series |
| `df.loc[0:3]` | Filas 0, 1, 2, **3** | DataFrame |
| `df.loc[[0, 2]]` | Filas 0 y 2 | DataFrame |
| `df.loc[:, 'Ventas']` | Columna 'Ventas' | Series |
| `df.loc[:, ['Ventas', 'Costos']]` | Columnas Ventas y Costos | DataFrame |
| `df.loc[0:2, 'Ventas':'Costos']` | Filas 0-2, cols Ventas a Costos | DataFrame |

---

### ⚠️ Diferencia crítica

❗ **Slicing con .loc es INCLUSIVO**

```python
df.loc[0:3]  # Filas 0, 1, 2, 3 (incluye 3)
```

🎯 **Diferente a .iloc:**
```python
df.iloc[0:3]  # Filas 0, 1, 2 (NO incluye 3)
df.loc[0:3]   # Filas 0, 1, 2, 3 (incluye 3)
```

---

### 💼 Caso de uso

**Cuándo usar .loc:**
* Tienes índice personalizado (fechas, IDs, nombres)
* Quieres columnas específicas por nombre
* Filtrado booleano (próximo notebook)
* Código legible y auto-documentado

````python
import pandas as pd

print("🏷️ .loc - INDEXACIÓN POR ETIQUETA")
print("="*70)

if USAR_DATOS_REALES and df_andes_indexed is not None:
    print("\n🏪 USANDO DATOS REALES: Los Andes Market (con índice de fecha)")
    print("-"*70)
    
    df = df_andes_indexed.copy()
    
    print("\n📊 DataFrame con índice de FECHA:")
    print(f"Shape: {df.shape}")
    print(f"Índice: {df.index.min()} a {df.index.max()}")
    print(f"\nPrimeras 5 filas:")
    print(df.head()[['sucursal_nombre', 'ventas', 'zona']])
    
    print("\n" + "-"*70)
    print("\n1️⃣  SELECCIONAR POR FECHA ESPECÍFICA (2019-01-01):")
    fecha_especifica = df.loc['2019-01-01']
    print(fecha_especifica[['sucursal_nombre', 'ventas', 'zona']])
    print(f"\nTotal registros en esa fecha: {len(fecha_especifica) if hasattr(fecha_especifica, '__len__') else 1}")
    
    print("\n" + "-"*70)
    print("\n2️⃣  COLUMNA 'ventas' (todas las fechas):")
    serie_ventas = df.loc[:, 'ventas']
    print(f"\nPrimeras 10 ventas:")
    print(serie_ventas.head(10))
    print(f"\nTipo: {type(serie_ventas)} (Series)")
    print(f"Total registros: {len(serie_ventas)}")
    
    print("\n" + "-"*70)
    print("\n3️⃣  MÚLTIPLES COLUMNAS:")
    columnas_mult = df.loc[:, ['sucursal_nombre', 'ventas', 'zona']]
    print(f"\nPrimeras 10 filas con columnas seleccionadas:")
    print(columnas_mult.head(10))
    
    print("\n" + "-"*70)
    print("\n4️⃣  RANGO DE FECHAS (Enero 2019):")
    enero_2019 = df.loc['2019-01-01':'2019-01-31', ['sucursal_nombre', 'ventas']]
    print(enero_2019)
    print(f"\nTotal registros en enero 2019: {len(enero_2019)}")
    
    print("\n" + "-"*70)
    print("\n5️⃣  SLICING INCLUSIVO DE FECHAS (primeros 3 meses de 2019):")
    primeros_meses = df.loc['2019-01':'2019-03', ['sucursal_nombre', 'ventas']]
    print(f"\nPrimeras 15 filas:")
    print(primeros_meses.head(15))
    print(f"\n⚠️  Nota: Incluye TODAS las fechas desde '2019-01' hasta '2019-03' (inclusivo)")
    print(f"Total registros: {len(primeros_meses)}")
    
    print("\n" + "-"*70)
    print("\n6️⃣  VALOR ESPECÍFICO - Primera fecha, columna 'ventas':")
    primera_fecha = df.index[0]
    primer_valor = df.loc[primera_fecha, 'ventas']
    if hasattr(primer_valor, '__len__'):  # Es una Serie
        print(f"\nFecha: {primera_fecha}")
        print(f"Ventas (todas las sucursales):")
        print(primer_valor)
    else:
        print(f"\nFecha: {primera_fecha}")
        print(f"Ventas: ${primer_valor:,.2f}")
    
    print("\n" + "-"*70)
    print("\n7️⃣  SELECCIÓN AVANZADA - Rango de fechas + columnas específicas:")
    seleccion = df.loc['2020-01':'2020-03', ['sucursal_id', 'sucursal_nombre', 'ventas', 'lat', 'lon']]
    print(f"\nPrimeras 10 filas de Q1 2020:")
    print(seleccion.head(10))
    
else:
    print("\n📋 EJEMPLO SINTÉTICO:")
    print("-"*70)
    
    # DataFrame con índice personalizado
    df = pd.DataFrame({
        'Fecha': pd.date_range('2024-01-01', periods=5),
        'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste'],
        'Ventas': [15000, 22000, 9800, 12000, 28000],
        'Costos': [10000, 15000, 7000, 8000, 19000]
    })
    
    # Establecer Sucursal como índice
    df.set_index('Sucursal', inplace=True)
    
    print("\n📊 DataFrame con índice personalizado (Sucursal):")
    print(df)
    print(f"\nÍndice: {df.index.tolist()}")
    
    print("\n" + "-"*70)
    print("\n1️⃣  FILA CON ETIQUETA 'Centro':")
    print(df.loc['Centro'])
    
    print("\n" + "-"*70)
    print("\n2️⃣  COLUMNA 'Ventas' (todas las filas):")
    print(df.loc[:, 'Ventas'])
    print(f"\nTipo: {type(df.loc[:, 'Ventas'])} (Series)")
    
    print("\n" + "-"*70)
    print("\n3️⃣  MÚLTIPLES COLUMNAS:")
    print(df.loc[:, ['Ventas', 'Costos']])
    
    print("\n" + "-"*70)
    print("\n4️⃣  FILAS 'Centro' y 'Norte', columna 'Ventas':")
    print(df.loc[['Centro', 'Norte'], 'Ventas'])
    
    print("\n" + "-"*70)
    print("\n5️⃣  SLICING INCLUSIVO ('Centro' a 'Sur'):")
    print(df.loc['Centro':'Sur'])
    print("\n⚠️  Nota: Incluye 'Centro', 'Norte' Y 'Sur' (inclusivo)")
    
    print("\n" + "-"*70)
    print("\n6️⃣  VALOR ESPECÍFICO (Centro, Ventas):")
    valor = df.loc['Centro', 'Ventas']
    print(f"\nVentas en Centro: ${valor:,.0f}")

print("\n" + "="*70)
print("✅ .loc dominado")
````

## 🔪 Comparación .loc vs .iloc

### 📊 Tabla comparativa

| Característica | .iloc | .loc |
|----------------|-------|------|
| **Tipo de selector** | Posición (enteros) | Etiquetas (nombres) |
| **Indexación** | 0, 1, 2... | 'A', 0, '2024-01-01'... |
| **Slicing** | Exclusivo (0:3 → 0,1,2) | Inclusivo (0:3 → 0,1,2,3) |
| **Uso común** | Primeras/últimas filas | Columnas por nombre |
| **Legibilidad** | 🟡 Medio | 🟢 Alta |
| **Performance** | 🟢 Rápido | 🟢 Rápido |

---

### 📊 Ejemplos lado a lado

```python
# Obtener primera fila
df.iloc[0]        # Por posición
df.loc[0]         # Por etiqueta (si índice es 0)

# Obtener primeras 3 filas
df.iloc[0:3]      # Filas 0, 1, 2 (exclusivo)
df.loc[0:2]       # Filas 0, 1, 2 (inclusivo)

# Obtener columna
df.iloc[:, 2]     # Tercera columna (posición)
df.loc[:, 'Ventas']  # Columna 'Ventas' (nombre)

# Celda específica
df.iloc[1, 2]     # Fila 1, columna 2
df.loc[0, 'Ventas']  # Fila con índice 0, columna 'Ventas'
```

---

### 🎯 ¿Cuándo usar cada uno?

🔢 **Usa .iloc cuando:**
* Posición es lo que importa ("primeros 10")
* Bucles numéricos
* Código genérico (no sabes nombres de columnas)

🏷️ **Usa .loc cuando:**
* Tienes etiquetas significativas
* Quieres código legible
* Filtrado booleano (próximo notebook)
* **👑 Recomendado en general**

---

### 💡 Buenas prácticas

✅ **Preferir .loc siempre que sea posible**

```python
# 🟢 BIEN (legible)
df_subset = df.loc[:, ['Fecha', 'Ventas', 'Costos']]

# 🟡 OK pero menos claro
df_subset = df.iloc[:, [0, 2, 3]]
```

````python
import pandas as pd

print("🔪 COMPARACIÓN .loc vs .iloc")
print("="*70)

if USAR_DATOS_REALES and df_andes is not None:
    print("\n🏪 COMPARANDO CON DATOS REALES: Los Andes Market")
    print("-"*70)
    
    df = df_andes.copy()
    
    print("\n📊 Dataset completo:")
    print(f"Shape: {df.shape}")
    print(f"Primeras 3 filas (para referencia):")
    print(df.head(3)[['fecha', 'sucursal_nombre', 'ventas']])
    
    print("\n" + "="*70)
    print("\n1️⃣  PRIMERA FILA - Ambos métodos")
    print("\n.iloc[0] (por posición):")
    print(df.iloc[0][['fecha', 'sucursal_id', 'sucursal_nombre', 'ventas', 'zona']])
    print("\n.loc[0] (por etiqueta de índice):")
    print(df.loc[0][['fecha', 'sucursal_id', 'sucursal_nombre', 'ventas', 'zona']])
    print("\n✅ Resultado idéntico (cuando índice es 0, 1, 2...)")
    
    print("\n" + "="*70)
    print("\n2️⃣  SLICING - LA GRAN DIFERENCIA")
    print("\n.iloc[0:10] (Exclusivo - posiciones 0 a 9):")
    print(df.iloc[0:10][['fecha', 'sucursal_nombre', 'ventas']])
    print(f"\nTotal filas: {len(df.iloc[0:10])}")
    
    print("\n.loc[0:10] (Inclusivo - índices 0 a 10):")
    print(df.loc[0:10][['fecha', 'sucursal_nombre', 'ventas']])
    print(f"\nTotal filas: {len(df.loc[0:10])}")
    
    print("\n⚠️  .iloc[0:10] = 10 filas (0 a 9, exclusivo)")
    print("⚠️  .loc[0:10]  = 11 filas (0 a 10, inclusivo)")
    
    print("\n" + "="*70)
    print("\n3️⃣  COLUMNA - Por posición vs nombre")
    print("\n.iloc[:, 6] (séptima columna por posición = 'ventas'):")
    serie_iloc = df.iloc[:, 6]
    print(serie_iloc.head(10))
    print(f"Columna: {serie_iloc.name}")
    
    print("\n.loc[:, 'ventas'] (columna por nombre):")
    serie_loc = df.loc[:, 'ventas']
    print(serie_loc.head(10))
    print(f"Columna: {serie_loc.name}")
    
    print("\n✅ .loc es más legible y mantenible")
    print("   Si cambia el orden de columnas, .loc sigue funcionando")
    
    print("\n" + "="*70)
    print("\n4️⃣  VALOR ESPECÍFICO")
    # Fila 10, columna de ventas
    valor_iloc = df.iloc[10, 6]  # Posición 6 es 'ventas'
    valor_loc = df.loc[10, 'ventas']  # Por nombre
    
    print(f"\n.iloc[10, 6] (fila 10, columna posición 6):  ${valor_iloc:,.2f}")
    print(f".loc[10, 'ventas'] (fila 10, columna 'ventas'): ${valor_loc:,.2f}")
    print("\n✅ Mismo resultado, .loc es autoexplicativo")
    
    print("\n" + "="*70)
    print("\n5️⃣  SELECCIÓN MÚLTIPLE - Filas y Columnas")
    print("\n.iloc[0:5, [0, 2, 6]] (primeras 5 filas, columnas por posición):")
    print(df.iloc[0:5, [0, 2, 6]])
    
    print("\n.loc[0:4, ['fecha', 'sucursal_nombre', 'ventas']] (por nombres):")
    print(df.loc[0:4, ['fecha', 'sucursal_nombre', 'ventas']])
    
    print("\n✅ .loc: código legible, fácil de mantener")
    
else:
    print("\n📋 EJEMPLO SINTÉTICO:")
    print("-"*70)
    
    # DataFrame de ejemplo
    df = pd.DataFrame({
        'Fecha': pd.date_range('2024-01-01', periods=5),
        'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste'],
        'Ventas': [15000, 22000, 9800, 12000, 28000],
        'Costos': [10000, 15000, 7000, 8000, 19000]
    })
    
    print("\n📊 DataFrame original:")
    print(df)
    
    print("\n" + "="*70)
    print("\n1️⃣  PRIMERA FILA - Ambos métodos")
    print("\n.iloc[0]:")
    print(df.iloc[0])
    print("\n.loc[0]:")
    print(df.loc[0])
    print("\n✅ Resultado idéntico (cuando índice es 0, 1, 2...)")
    
    print("\n" + "="*70)
    print("\n2️⃣  SLICING - LA GRAN DIFERENCIA")
    print("\n.iloc[0:3] (Exclusivo):")
    print(df.iloc[0:3])
    print("\n.loc[0:3] (Inclusivo):")
    print(df.loc[0:3])
    print("\n⚠️  .iloc[0:3] = 3 filas (0,1,2)")
    print("⚠️  .loc[0:3]  = 4 filas (0,1,2,3)")
    
    print("\n" + "="*70)
    print("\n3️⃣  COLUMNA - Por posición vs nombre")
    print("\n.iloc[:, 2] (tercera columna por posición):")
    print(df.iloc[:, 2])
    print("\n.loc[:, 'Ventas'] (columna por nombre):")
    print(df.loc[:, 'Ventas'])
    print("\n✅ .loc es más legible")
    
    print("\n" + "="*70)
    print("\n4️⃣  VALOR ESPECÍFICO")
    valor_iloc = df.iloc[1, 2]
    valor_loc = df.loc[1, 'Ventas']
    print(f"\n.iloc[1, 2]:      ${valor_iloc:,.0f}")
    print(f".loc[1, 'Ventas']: ${valor_loc:,.0f}")
    print("\n✅ Mismo resultado, .loc más claro")

print("\n" + "="*70)
print("✅ Comparación completada")
````

## ✏️ Asignación de valores con .loc e .iloc

### 📝 Modificar datos

Puedes usar .loc e .iloc para **asignar nuevos valores**.

---

### 🔑 Sintaxis

```python
# Asignar valor único
df.loc[fila, columna] = nuevo_valor
df.iloc[fila, columna] = nuevo_valor

# Asignar a múltiples celdas
df.loc[filas, columnas] = nuevo_valor
df.loc[condicion, columna] = nuevo_valor  # Filtrado booleano
```

---

### 📊 Ejemplos

```python
# Cambiar valor específico
df.loc[0, 'Ventas'] = 20000

# Cambiar columna completa
df.loc[:, 'Estado'] = 'Cobrado'

# Cambiar múltiples valores
df.loc[0:2, 'Estado'] = 'Pendiente'

# Crear columna nueva
df.loc[:, 'Ganancia'] = df['Ventas'] - df['Costos']

# Asignación condicional (filtrado booleano)
df.loc[df['Ventas'] > 20000, 'Estado'] = 'Alta'
```

---

### ⚠️ Advertencia: SettingWithCopyWarning

❌ **EVITAR:**
```python
df[df['Ventas'] > 10000]['Estado'] = 'Alta'  # Puede fallar
```

✅ **CORRECTO:**
```python
df.loc[df['Ventas'] > 10000, 'Estado'] = 'Alta'  # Siempre funciona
```

**Regla de oro:** Usa siempre .loc para asignaciones

````python
import pandas as pd
import numpy as np

print("✏️ ASIGNACIÓN DE VALORES")
print("="*70)

if USAR_DATOS_REALES and df_andes is not None:
    print("\n🏪 USANDO DATOS REALES: Los Andes Market")
    print("-"*70)
    
    # Trabajar con un subset para no modificar todos los datos
    df = df_andes.head(20).copy()  # Primeras 20 filas para ejemplos
    df['estado_pago'] = 'Pendiente'  # Columna nueva para ejemplos
    
    print("\n📊 Dataset inicial (primeras 20 filas):")
    print(df[['fecha', 'sucursal_nombre', 'ventas', 'estado_pago']])
    
    print("\n" + "-"*70)
    print("\n1️⃣  MODIFICAR VALOR ESPECÍFICO:")
    print("\nCambiar estado de la primera transacción:")
    df.loc[0, 'estado_pago'] = 'Cobrado'
    print("\nDespués de: df.loc[0, 'estado_pago'] = 'Cobrado'")
    print(df[['fecha', 'sucursal_nombre', 'ventas', 'estado_pago']].head(5))
    
    print("\n" + "-"*70)
    print("\n2️⃣  CREAR COLUMNAS CALCULADAS:")
    # Calcular ventas en miles
    df.loc[:, 'ventas_miles'] = (df['ventas'] / 1000).round(1)
    # Calcular IVA
    df.loc[:, 'iva_21'] = (df['ventas'] * 0.21).round(2)
    # Total con IVA
    df.loc[:, 'total_con_iva'] = df['ventas'] + df['iva_21']
    
    print("\nNuevas columnas calculadas:")
    print(df[['sucursal_nombre', 'ventas', 'ventas_miles', 'iva_21', 'total_con_iva']].head(8))
    
    print("\n" + "-"*70)
    print("\n3️⃣  ASIGNACIÓN CONDICIONAL (Filtrado Booleano):")
    print("\nMarcar como 'Cobrado' donde ventas > $150,000:")
    df.loc[df['ventas'] > 150000, 'estado_pago'] = 'Cobrado'
    print("\nMarcar como 'Verificado' donde ventas entre $100,000 y $150,000:")
    df.loc[(df['ventas'] >= 100000) & (df['ventas'] <= 150000), 'estado_pago'] = 'Verificado'
    
    print("\nResultado después de asignaciones condicionales:")
    print(df[['sucursal_nombre', 'ventas', 'estado_pago']].head(15))
    
    print("\nResumen de estados:")
    print(df['estado_pago'].value_counts())
    
    print("\n" + "-"*70)
    print("\n4️⃣  MODIFICAR RANGO DE FILAS (slicing inclusivo):")
    df_copia = df.copy()
    df_copia.loc[0:4, 'estado_pago'] = 'Auditado'
    print("\nDespués de: df.loc[0:4, 'estado_pago'] = 'Auditado'")
    print(df_copia[['fecha', 'sucursal_nombre', 'ventas', 'estado_pago']].head(10))
    print("\n⚠️  Nota: .loc[0:4] modifica filas 0, 1, 2, 3 Y 4 (inclusivo)")
    
    print("\n" + "-"*70)
    print("\n5️⃣  CREAR COLUMNA CATEGÓRICA:")
    # Categorizar ventas
    df.loc[df['ventas'] < 100000, 'categoria'] = 'Bajo'
    df.loc[(df['ventas'] >= 100000) & (df['ventas'] < 150000), 'categoria'] = 'Medio'
    df.loc[df['ventas'] >= 150000, 'categoria'] = 'Alto'
    
    print("\nCategorización de ventas:")
    print(df[['sucursal_nombre', 'ventas', 'categoria']].head(15))
    print("\nDistribución por categoría:")
    print(df['categoria'].value_counts())
    
    print("\n" + "-"*70)
    print("\n6️⃣  MODIFICAR MÚLTIPLES COLUMNAS A LA VEZ:")
    # Ajustar todas las ventas de una sucursal específica
    mask_suc = df['sucursal_id'] == 'SUC001'
    df.loc[mask_suc, ['estado_pago', 'categoria']] = ['Premium', 'VIP']
    
    print("\nDespués de marcar SUC001 como Premium/VIP:")
    print(df[df['sucursal_id'] == 'SUC001'][['sucursal_nombre', 'ventas', 'estado_pago', 'categoria']])
    
else:
    print("\n📋 EJEMPLO SINTÉTICO:")
    print("-"*70)
    
    # DataFrame de ejemplo
    df = pd.DataFrame({
        'Fecha': pd.date_range('2024-01-01', periods=5),
        'Sucursal': ['Centro', 'Norte', 'Sur', 'Este', 'Oeste'],
        'Ventas': [15000, 22000, 9800, 12000, 28000],
        'Costos': [10000, 15000, 7000, 8000, 19000],
        'Estado': ['Pendiente'] * 5
    })
    
    print("\n📊 DataFrame original:")
    print(df)
    
    print("\n" + "-"*70)
    print("\n1️⃣  MODIFICAR VALOR ESPECÍFICO:")
    df.loc[0, 'Estado'] = 'Cobrado'
    print("\nDespués de: df.loc[0, 'Estado'] = 'Cobrado'")
    print(df)
    
    print("\n" + "-"*70)
    print("\n2️⃣  CREAR COLUMNA CALCULADA:")
    df.loc[:, 'Ganancia'] = df['Ventas'] - df['Costos']
    df.loc[:, 'Margen_%'] = (df['Ganancia'] / df['Ventas'] * 100).round(1)
    print("\nDespués de crear Ganancia y Margen_%:")
    print(df)
    
    print("\n" + "-"*70)
    print("\n3️⃣  ASIGNACIÓN CONDICIONAL (importante para próximo notebook):")
    print("\nMarcar como 'Cobrado' donde Ventas > 15000:")
    df.loc[df['Ventas'] > 15000, 'Estado'] = 'Cobrado'
    print(df[['Sucursal', 'Ventas', 'Estado']])
    
    print("\n" + "-"*70)
    print("\n4️⃣  MODIFICAR RANGO DE FILAS:")
    df_copy = df.copy()
    df_copy.loc[0:2, 'Estado'] = 'Verificado'
    print("\nDespués de: df.loc[0:2, 'Estado'] = 'Verificado'")
    print(df_copy[['Sucursal', 'Estado']])

print("\n" + "="*70)
print("✅ Asignación de valores dominada")
````

### 📊 Guía rápida: .loc vs .iloc

| Tarea | .iloc | .loc |
|-------|-------|------|
| Primera fila | `.iloc[0]` | `.loc[0]` |
| Última fila | `.iloc[-1]` | `.loc[df.index[-1]]` |
| Primeras 3 filas | `.iloc[0:3]` | `.loc[0:2]` |
| Columna por nombre | `.iloc[:, 2]` | `.loc[:, 'Ventas']` |
| Valor específico | `.iloc[1, 2]` | `.loc[1, 'Ventas']` |
| **Filtrado booleano** | ❌ No soportado | ✅ `.loc[condicion]` |
