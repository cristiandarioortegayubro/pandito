## 📚 MultiIndex: Índices Jerárquicos

### 🏛️ ¿Qué es un MultiIndex?

Un **MultiIndex** (o índice jerárquico) permite tener **múltiples niveles de índices** en un DataFrame.

**Ejemplo:**
```
Año   Mes   Sucursal    Ventas
2023   1     Centro      15000
       1     Norte       22000
       2     Centro      18000
       2     Norte       24000
```

Aquí, el índice tiene **3 niveles**: Año, Mes, Sucursal

---

### 📈 ¿Por qué usar MultiIndex?

**Ventajas:**

1. **Jerarquías naturales:** Reflejan la estructura real del negocio
   * Empresa → División → Departamento → Cuenta
   * Año → Trimestre → Mes

2. **Agregaciones rápidas por nivel:**
   ```python
   df_multi.groupby(level=0).sum()  # Suma por año
   df_multi.groupby(level=[0,1]).sum()  # Suma por año y mes
   ```

3. **Navegación intuitiva:**
   ```python
   df_multi.loc[2023]  # Todo el año 2023
   df_multi.loc[(2023, 1)]  # Enero 2023
   df_multi.loc[(2023, 1, 'Centro')]  # Enero 2023, Centro
   ```

---

### 💼 Caso de Uso: Estados Financieros P&L

**Estado de Pérdidas y Ganancias (P&L) tiene jerarquía:**

```
Categoría        Subcategoría      Cuenta              Monto
Ingresos         Ventas           Venta Productos     500,000
                                  Venta Servicios     200,000
                 Otros Ingresos   Intereses            10,000
Costos           Costo Ventas     Materia Prima       300,000
                                  Mano de Obra        100,000
Gastos           Administrativos  Sueldos              50,000
                                  Alquileres           20,000
```

**Con MultiIndex:**
* Drill-down automático (ver detalles de Ingresos)
* Agregaciones por nivel (total Ingresos, total Costos)
* Reportes consolidados fáciles

## 🏛️ Creación y navegación de MultiIndex con datos reales

### 📊 Nuestro dataset jerárquico

Los datos de **Los Andes Market** tienen una jerarquía natural:

```
Año → Mes → Sucursal → Ventas
2020  1    SUC001     169804
      1    SUC002      96629
      2    SUC001     166149
      ...
```

---

### 🛠️ Crear MultiIndex con set_index()

```python
df_multi = df.set_index(['año', 'mes', 'sucursal_id'])
df_multi = df_multi.sort_index()  # Siempre ordenar después de crear MultiIndex
```

**Resultado:** El índice tiene 3 niveles. Pandas muestra la jerarquía con sangría.

---

### 🔍 Navegación con .loc[]

```python
df_multi.loc[2020]                  # Todos los datos de 2020
df_multi.loc[(2020, 1)]             # Enero 2020
df_multi.loc[(2020, 1, 'SUC001')]   # Enero 2020, sucursal SUC001
```

**Regla:** De lo general a lo particular. No puedes saltar niveles.

---

### ⚠️ Errores comunes

* ❌ `df_multi.loc[1]` — salta el nivel 'año', error
* ❌ `df_multi.loc['SUC001']` — salta niveles, error
* ✅ `df_multi.loc[(2020, 1, 'SUC001')]` — acceso correcto por jerarquía

````python
import pandas as pd

print("🏛️ CREACIÓN Y NAVEGACIÓN DE MULTIINDEX")
print("="*70)

if USAR_DATOS_REALES and df is not None:
    print("\n1️⃣  CREAR MULTIINDEX")
    print("-"*70)

    df_multi = df.set_index(['año', 'mes', 'sucursal_id'])
    df_multi = df_multi.sort_index()
    print(f"\n   MultiIndex creado: {df_multi.index.names}")
    print(f"   Niveles: {len(df_multi.index.names)} (año → mes → sucursal_id)")
    print(f"   Shape: {df_multi.shape}")
    print("\n   Primeras 10 filas:")
    print(df_multi[['sucursal_nombre', 'zona', 'ventas']].head(10))

    print("\n" + "="*70)
    print("\n2️⃣  NAVEGACIÓN POR NIVELES")
    print("-"*70)

    # Nivel 1: Todo un año
    print("\n   📅 df_multi.loc[2019] — Todos los datos de 2019:")
    datos_2019 = df_multi.loc[2019]
    print(f"      Registros: {len(datos_2019)}")
    print(datos_2019[['sucursal_nombre', 'ventas']].head(6))

    # Nivel 2: Un mes específico
    print("\n   📆 df_multi.loc[(2019, 1)] — Enero 2019:")
    datos_ene2019 = df_multi.loc[(2019, 1)]
    print(f"      Registros: {len(datos_ene2019)}")
    print(datos_ene2019[['sucursal_nombre', 'ventas']].head(6))

    # Nivel 3: Una sucursal concreta
    print("\n   🏪 df_multi.loc[(2019, 1, 'SUC001')] — Enero 2019, SUC001:")
    datos_suc = df_multi.loc[(2019, 1, 'SUC001')]
    print(datos_suc[['sucursal_nombre', 'ventas']])

    print("\n" + "="*70)
    print("\n3️⃣  SLICING CON RANGOS")
    print("-"*70)

    # Rango de meses en un año
    print("\n   df_multi.loc[(2020, slice(1, 3))] — Q1 2020:")
    q1_2020 = df_multi.loc[(2020, slice(1, 3))]
    print(f"      Registros: {len(q1_2020)}")
    print(q1_2020[['sucursal_nombre', 'ventas']].head(10))

    # Rango de años
    print("\n   df_multi.loc[2020:2021] — Años 2020 y 2021:")
    rango_años = df_multi.loc[2020:2021]
    print(f"      Registros: {len(rango_años)}")
    print("      Agrupado por año:")
    print(rango_años.groupby(level=0).size())
else:
    print("⚠️  No hay datos reales disponibles")

print("\n" + "="*70)
````

## 📊 Agregaciones por nivel con datos reales

### 🎯 groupby(level=): El poder del MultiIndex

Con un DataFrame normal, agrupar por año requiere: `df.groupby('año').sum()`
Con MultiIndex, puedes agrupar **directamente por nivel**:

```python
df_multi.groupby(level=0).sum()       # Sumar por año (nivel 0)
df_multi.groupby(level=1).sum()       # Sumar por mes (nivel 1)
df_multi.groupby(level=[0, 1]).sum() # Sumar por año y mes
```

---

### 📈 Múltiples agregaciones

```python
df_multi.groupby(level=0).agg({
    'ventas': ['sum', 'mean', 'count', 'max']
})
```

**Ventaja:** Sin `reset_index()`, las agregaciones operan directamente sobre los niveles.

---

### 💼 Caso de uso: Reporte anual por sucursal

```python
# Total de ventas por año (nivel 0)
ventas_anuales = df_multi.groupby(level=0)['ventas'].sum()

# Total de ventas por sucursal (nivel 2)
ventas_sucursal = df_multi.groupby(level=2)['ventas'].sum()

# Comparativa: año × mes (niveles 0 y 1)
ventas_año_mes = df_multi.groupby(level=[0, 1])['ventas'].sum()
```

````python
import pandas as pd

print("📊 AGREGACIONES POR NIVEL CON MULTIINDEX")
print("="*70)

if USAR_DATOS_REALES and df is not None:
    print("\n1️⃣  AGREGACIÓN POR AÑO (level=0)")
    print("-"*70)

    ventas_anuales = df_multi.groupby(level=0)['ventas'].agg(['sum', 'mean', 'count', 'max'])
    print("\n   Ventas anuales totales:")
    print(ventas_anuales.round(0))

    print("\n" + "="*70)
    print("\n2️⃣  AGREGACIÓN POR MES (level=1)")
    print("-"*70)

    ventas_mensuales = df_multi.groupby(level=1)['ventas'].agg(['sum', 'mean', 'max'])
    print("\n   Ventas promedio por mes (todos los años):")
    print(ventas_mensuales.round(0))

    print("\n" + "="*70)
    print("\n3️⃣  AGREGACIÓN POR AÑO Y MES (level=[0, 1])")
    print("-"*70)

    ventas_año_mes = df_multi.groupby(level=[0, 1])['ventas'].sum()
    print("\n   Ventas por año y mes:")
    print(ventas_año_mes.round(0).head(15))

    print("\n" + "="*70)
    print("\n4️⃣  AGREGACIÓN POR SUCURSAL (level=2)")
    print("-"*70)

    ventas_sucursal = df_multi.groupby(level=2)['ventas'].agg(['sum', 'mean', 'count'])
    print("\n   Ventas por sucursal:")
    print(ventas_sucursal.round(0))

    print("\n" + "="*70)
    print("\n5️⃣  AGREGACIÓN MULTIPLE: sum + mean + max por año")
    print("-"*70)

    multi_agg = df_multi.groupby(level=0).agg({
        'ventas': ['sum', 'mean', 'max', 'min'],
        'sucursal_nombre': 'count'
    })
    print("\n   Agregación multiple por año:")
    print(multi_agg.round(0))
    print("\n   💡 groupby(level=) evita reset_index(), más eficiente")
else:
    print("⚠️  No hay datos reales disponibles")

print("\n" + "="*70)
````

## 💰 Estado P&L con MultiIndex: Reportes financieros jerárquicos

### 🏛️ Estructura de un Estado de Pérdidas y Ganancias

Un P&L tiene una jerarquía natural que MultiIndex representa perfectamente:

```
Categoría        Subcategoría      Monto
Ingresos         Ventas Productos   500,000
                 Ventas Servicios  200,000
                 Otros Ingresos    10,000
Costos           COGS              300,000
                 Gastos Operativos 100,000
Gastos           Administración     50,000
                 Marketing         30,000
```

---

### 📊 Simulación de P&L con datos de Los Andes Market

Usaremos las sucursales como "divisiones" y las zonas como "categorías":

```python
# Estructura P&L: Zona → Sucursal → Ventas
pl = df.set_index(['zona', 'sucursal_nombre'])[['ventas']]

# Total por zona (grupo de mayor nivel)
pl.groupby(level=0).sum()

# Drill-down: detalle por sucursal dentro de cada zona
pl.loc['Corredor Comercial']
```

---

### 💡 Beneficios para reportes financieros

* **Drill-down:** Ver el total y luego profundizar en cada sucursal
* **Comparativas:** Cruzar zona × año con unstack()
* **Rankings:** Ordenar sucursales por ventas dentro de cada zona
* **Exportación:** Formato jerárquico legible para stakeholders

````python
import pandas as pd

print("💰 ESTADO P&L CON MULTIINDEX — DATOS REALES")
print("="*70)

if USAR_DATOS_REALES and df is not None:
    print("\n1️⃣  P&L POR ZONA Y SUCURSAL")
    print("-"*70)

    pl = df.set_index(['zona', 'sucursal_nombre'])[['ventas']]
    pl = pl.sort_index()
    print("\n   Estructura P&L: Zona → Sucursal")
    print(pl.head(15).round(0))

    print("\n" + "="*70)
    print("\n2️⃣  TOTAL POR ZONA (agregación nivel 0)")
    print("-"*70)

    total_zona = pl.groupby(level=0).sum()
    print("\n   Ventas totales por zona:")
    print(total_zona.round(0))
    print("\n   Ranking de zonas:")
    print(total_zona.sort_values('ventas', ascending=False).round(0))

    print("\n" + "="*70)
    print("\n3️⃣  DRILL-DOWN: Sucursales de una zona específica")
    print("-"*70)

    zona_top = total_zona.idxmax()[0]
    print(f"\n   Detalle de sucursales en '{zona_top}':")
    print(pl.loc[zona_top].sort_values('ventas', ascending=False).round(0))

    print("\n" + "="*70)
    print("\n4️⃣  P&L MULTINIVEL: Zona → Año → Sucursal")
    print("-"*70)

    pl_multi = df.set_index(['zona', 'año', 'sucursal_nombre'])[['ventas']]
    pl_multi = pl_multi.sort_index()
    print("\n   P&L con 3 niveles:")
    print(pl_multi.head(15).round(0))

    # Total por zona y año
    print("\n   Total por zona y año:")
    pl_multi.groupby(level=[0, 1]).sum().round(0)

    print("\n" + "="*70)
    print("\n5️⃣  MATRIZ ZONA × AÑO (unstack)")
    print("-"*70)

    matriz = pl_multi.groupby(level=[0, 1])['ventas'].sum().unstack(level=1)
    print("\n   Matriz de ventas: Zona (filas) × Año (columnas):")
    print(matriz.round(0))
    print("\n   ✅ Formato ideal para reporte ejecutivo")

    print("\n" + "="*70)
    print("\n6️⃣  P&L COMPLETO: Margen y % por zona")
    print("-"*70)

    resumen = pl.groupby(level=0).agg({
        'ventas': ['sum', 'mean', 'count', 'max']
    })
    resumen.columns = ['Total', 'Promedio', 'Sucursales', 'Máximo']
    resumen['% del Total'] = (resumen['Total'] / resumen['Total'].sum() * 100).round(1)
    resumen = resumen.sort_values('Total', ascending=False)
    print("\n   Reporte P&L ejecutivo por zona:")
    print(resumen.round(0))
else:
    print("⚠️  No hay datos reales disponibles")

print("\n" + "="*70)
````

## 🔄 Operaciones avanzadas: swaplevel, unstack, sort

### 🔄 swaplevel(): Reordenar niveles

Intercambia dos niveles del MultiIndex sin cambiar los datos:

```python
df_multi.swaplevel('año', 'mes')
# Antes: (año, mes, sucursal) → Después: (mes, año, sucursal)
```

---

### 📐 unstack(): De filas a columnas

Convierte un nivel del índice en columnas (como pivot_table pero más directo):

```python
df_multi['ventas'].unstack(level='sucursal_nombre')
# Cada sucursal se convierte en una columna
```

---

### 📊 Casos de uso avanzados

* `swaplevel()` para reorganizar la jerarquía según la pregunta de negocio
* `unstack()` para crear matrices comparativas
* `sort_index()` para ordenar dentro del MultiIndex
* Combinar con `pct_change()` para variaciones interanuales

````python
import pandas as pd

print("🔄 OPERACIONES AVANZADAS CON MULTIINDEX")
print("="*70)

if USAR_DATOS_REALES and df is not None:
    print("\n1️⃣  SWAPLEVEL: Intercambiar niveles")
    print("-"*70)

    df_swap = df_multi.swaplevel('año', 'mes')
    print("\n   Antes: (año, mes, sucursal_id)")
    print("   Después de swaplevel: (mes, año, sucursal_id)")
    print(df_swap[['sucursal_nombre', 'ventas']].head(10).round(0))

    print("\n" + "="*70)
    print("\n2️⃣  UNSTACK: Sucursal de filas a columnas")
    print("-"*70)

    ventas_unstacked = df_multi['ventas'].unstack(level='sucursal_id')
    print("\n   Ventas por año-mes (filas) × sucursal (columnas):")
    print(ventas_unstacked.round(0).head(12))
    print(f"\n   Shape: {ventas_unstacked.shape} (año×mes × sucursales)")

    print("\n" + "="*70)
    print("\n3️⃣  UNSTACK POR AÑO")
    print("-"*70)

    df_año_suc = df.set_index(['sucursal_nombre', 'año'])['ventas']
    matriz_año = df_año_suc.unstack(level='año')
    print("\n   Matriz: Sucursal (filas) × Año (columnas):")
    print(matriz_año.round(0))

    print("\n" + "="*70)
    print("\n4️⃣  VARIACIÓN INTERANUAL POR SUCURSAL")
    print("-"*70)

    variacion = matriz_año.pct_change(axis=1) * 100
    print("\n   Variación % año vs año anterior:")
    print(variacion.round(1))
    print("\n   💡 NaN en el primer año (no hay año anterior para comparar)")

    print("\n" + "="*70)
    print("\n5️⃣  SORT_INDEX: Ordenar jerarquía")
    print("-"*70)

    df_sorted = df_multi.sort_index(level=0, ascending=False)
    print("\n   MultiIndex ordenado por año descendente:")
    print(df_sorted[['sucursal_nombre', 'ventas']].head(10).round(0))

    print("\n" + "="*70)
    print("\n6️⃣  RESUMEN P&L CONSOLIDADO FINAL")
    print("-"*70)

    # P&L completo: zona → sucursal, con totales y rankings
    pl_final = df.pivot_table(
        values='ventas',
        index=['zona', 'sucursal_nombre'],
        columns='año',
        aggfunc='sum',
        fill_value=0,
        margins=True,
        margins_name='TOTAL'
    )
    print("\n   📋 P&L Consolidado: Zona → Sucursal × Año")
    print(pl_final.round(0))
    print("\n   ✅ Reporte ejecutivo listo para stakeholders")
else:
    print("⚠️  No hay datos reales disponibles")

print("\n" + "="*70)
````

### 🎯 Reglas de Oro

👉 **Regla #1: set_index con orden jerárquico lógico**
```python
# MALO: orden arbitrario, navegación confusa
df.set_index(['sucursal', 'año', 'mes'])

# BUENO: de lo general a lo particular
df.set_index(['año', 'mes', 'sucursal_id'])
# .loc[2023] → .loc[(2023, 1)] → .loc[(2023, 1, 'SUC001')]
```

👉 **Regla #2: groupby(level=) para agregaciones rápidas**
```python
# MALO: reset_index + groupby innecesario
df_multi.reset_index().groupby('año').sum()

# BUENO: aggregate directamente por nivel
df_multi.groupby(level=0).sum()
df_multi.groupby(level=[0, 1]).sum()
```

👉 **Regla #3: Siempre sort_index después de crear MultiIndex**
```python
df_multi = df.set_index(['año', 'mes', 'sucursal_id'])
# Sin sort, .loc puede fallar o ser lento
df_multi = df_multi.sort_index()
```

---

### 📊 Guía de Decisión

| Situación | Método |
|-----------|--------|
| Crear índice multinivel | `df.set_index(['col1', 'col2', ...])` |
| Acceder a un nivel | `df_multi.loc[nivel1]` |
| Acceder a múltiples niveles | `df_multi.loc[(nivel1, nivel2)]` |
| Agregar por un nivel | `df_multi.groupby(level=0).sum()` |
