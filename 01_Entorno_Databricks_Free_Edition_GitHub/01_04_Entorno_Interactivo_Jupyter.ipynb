## 📓 1. Anatomía de un notebook

### 📌 1.1 ¿Qué es un notebook?

Documento interactivo que combina:
* **Código ejecutable** (Python, SQL, Scala, R)
* **Texto formateado** (Markdown)
* **Resultados** (tablas, gráficos, outputs)

**Ventajas vs scripts tradicionales:**
* ✅ Ejecución celda por celda (iteración rápida)
* ✅ Visualización inline de resultados
* ✅ Documentación integrada
* ✅ Colaboración en tiempo real

---

### 🧱 1.2 Tipos de celdas

**A) Celdas de Código**
* Ejecutan código Python/SQL/etc.
* Variables persisten entre celdas
* Output se muestra debajo

**B) Celdas Markdown**
* Texto formateado con Markdown
* No ejecutables (solo renderización)
* Usar `%md` al inicio

**C) Comandos Mágicos**
* Instrucciones especiales precedidas por `%`
* Ejemplos: `%sql`, `%sh`, `%pip`

---

### 🔄 1.3 Ejecución de celdas

**Orden de ejecución:**
* Las celdas se ejecutan en el orden que TÚ elijas
* No necesariamente secuencial
* Número entre corchetes `[3]` indica orden

**Estado de variables:**
```python
# Celda 1
x = 10

# Celda 2 (ejecutada después)
print(x)  # 10 (x existe en memoria)

# Si ejecutas Celda 2 ANTES de Celda 1 → Error: x no definida
```

**Importante:** Las variables persisten en memoria hasta que reinicies el kernel o las redefinas.

````python
# 🎯 Ejercicio 1: Experimentar con Estado de Variables en Notebooks

print("=" * 60)
print("📓 EJERCICIO 1: ANATOMÍA Y ESTADO DE VARIABLES")
print("=" * 60)

# Demostración 1: Variables persisten entre celdas
print("\n1️⃣ Definir variables que usaremos en celdas posteriores:")

# Variables de negocio
empresa = "Mi Empresa S.A."
ventas_mensuales = [45000, 52000, 48000, 61000, 55000, 67000]
presupuesto_anual = 650000

print(f"✅ empresa = '{empresa}'")
print(f"✅ ventas_mensuales = {ventas_mensuales}")
print(f"✅ presupuesto_anual = ${presupuesto_anual:,}")

# Demostración 2: Tipos de celdas
print("\n2️⃣ Tipos de celdas en este notebook:")
print("   📝 Celdas Markdown: Para documentación y teoría")
print("   💻 Celdas Python: Para código ejecutable (como esta)")
print("   ✨ Comandos Mágicos: %md, %sql, %sh, %fs, %pip")

# Demostración 3: Cálculos con variables persistentes
print("\n3️⃣ Calcular métricas usando las variables:")
total_ventas = sum(ventas_mensuales)
promedio_mensual = total_ventas / len(ventas_mensuales)
porcentaje_objetivo = (total_ventas / presupuesto_anual) * 100

print(f"Total ventas (6 meses): ${total_ventas:,}")
print(f"Promedio mensual: ${promedio_mensual:,.2f}")
print(f"% del objetivo anual: {porcentaje_objetivo:.1f}%")

# Demostración 4: Variables disponibles para celdas futuras
print("\n4️⃣ Estas variables ahora están en memoria:")
print(f"   • empresa, ventas_mensuales, presupuesto_anual")
print(f"   • total_ventas, promedio_mensual, porcentaje_objetivo")
print(f"   ✅ Puedes usarlas en cualquier celda posterior")

print("\n" + "=" * 60)
print("💡 Tip: Usa Shift+Enter para ejecutar y avanzar a siguiente celda")
print("=" * 60)
````

## ✨ 2. Comandos mágicos

Los comandos mágicos son instrucciones especiales que empiezan con `%` y permiten ejecutar operaciones especiales en Databricks.

---

### 📝 2.1 %md - Markdown

Permite escribir texto formateado:

```
%md
# Título
## Subtítulo
**Negrita** *Itálica* `código`
```

---

### 🔷 2.2 %sql - SQL Queries

Ejecuta consultas SQL directamente:

```sql
%sql
SELECT 
  product_name,
  SUM(sales) as total_sales
FROM sales_data
GROUP BY product_name
ORDER BY total_sales DESC
LIMIT 10
```
**Resultados automáticamente en tabla interactiva**

---

### 🖥️ 2.3 %sh - Shell Commands

Ejecuta comandos de terminal:

```bash
%sh
ls /dbfs/FileStore/
pwd
df -h
echo "Hola desde shell"
```

---

### 📁 2.4 %fs - Databricks File System

Interactúa con DBFS (sistema de archivos de Databricks):

```python
%fs ls /FileStore/
%fs head /FileStore/data.csv
%fs rm /FileStore/temp.txt
%fs mkdirs /FileStore/mi_carpeta
```

---

### 📦 2.5 %pip - Install Packages

Instala paquetes Python:

```python
%pip install seaborn scikit-learn
%pip install --upgrade pandas
%pip list | grep pandas
```
⚠️ **Importante:** Puede requerir reiniciar kernel después de instalar

---

### 🕒 2.6 %time - Measure Execution Time

Mide tiempo de ejecución:

```python
%time
result = sum(range(1000000))
```

Muestra tiempo de CPU y tiempo total.

````python
# 🎯 Ejercicio 2: Practicar Comandos Mágicos

print("=" * 60)
print("✨ EJERCICIO 2: COMANDOS MÁGICOS")
print("=" * 60)

# Comando 1: Usar %time para medir rendimiento
print("\n1️⃣ Comando %time (medir tiempo de ejecución):")
print("   Ejemplo: Calcular suma de 1 millón de números")
print("   Sintaxis: %time result = sum(range(1000000))")
print("   💡 Tip: Ideal para optimizar código lento")

# Comando 2: %sh para comandos shell
print("\n2️⃣ Comando %sh (ejecutar comandos shell):")
print("   Ejemplos:")
print("   %sh pwd          # Directorio actual")
print("   %sh ls -la       # Listar archivos")
print("   %sh date         # Fecha y hora del sistema")
print("   %sh df -h        # Espacio en disco")

# Comando 3: %fs para sistema de archivos
print("\n3️⃣ Comando %fs (Databricks File System):")
print("   Ejemplos:")
print("   %fs ls /FileStore/                  # Listar archivos")
print("   %fs head /FileStore/data.csv        # Ver primeras líneas")
print("   %fs mkdirs /FileStore/mi_proyecto   # Crear carpeta")

# Comando 4: %pip para instalar librerías
print("\n4️⃣ Comando %pip (instalar paquetes Python):")
print("   Ejemplos:")
print("   %pip install seaborn          # Instalar seaborn")
print("   %pip install --upgrade pandas # Actualizar pandas")
print("   %pip list | grep numpy        # Ver versión de numpy")
print("   ⚠️  Puede requerir reinicio de kernel")

# Comando 5: %md para documentación
print("\n5️⃣ Comando %md (escribir Markdown):")
print("   Ejemplo de celda Markdown:")
print("   %md")
print("   # Mi Análisis")
print("   ## Resultados")
print("   **Importante:** Los datos muestran...")

# Demostración práctica con %time
print("\n" + "=" * 60)
print("🚀 DEMOSTRACIÓN: Medir tiempo de cálculo")
print("=" * 60)

import time

# Simular cálculo pesado
start = time.time()
resultado = sum(i**2 for i in range(100000))
end = time.time()

print(f"\nResultado: {resultado:,}")
print(f"Tiempo de ejecución: {(end - start)*1000:.2f} ms")

print("\n💡 Tip: En producción, usa %time para identificar cuellos de botella")
````

### 🧮 Fórmulas Financieras

#### Retorno de Inversión (ROI)

Fórmula:

$$
ROI = \frac{Ganancia - Inversi\acute{o}n}{Inversi\acute{o}n} \times 100\%
$$

**Ejemplo:**
* Inversión inicial: $10,000
* Ganancia obtenida: $13,500
* ROI = $\frac{13500 - 10000}{10000} \times 100\% = 35\%$

#### Tasa de Crecimiento

Fórmula para crecimiento compuesto:

$$
Valor_{final} = Valor_{inicial} \times (1 + r)^n
$$

Donde:
* $r$ = tasa de crecimiento
* $n$ = número de períodos

````python
# 🎯 Ejercicio 6: Técnicas de Debugging

print("=" * 60)
print("🐞 EJERCICIO 6: DEBUGGING Y MANEJO DE ERRORES")
print("=" * 60)

# =====================================
# DEMOSTRACIÓN 1: Debugging con print()
# =====================================

print("\n1️⃣ Técnica: print() estratégico")
print("=" * 60)

def procesar_ventas(ventas, umbral_minimo=1000):
    """
    Procesa lista de ventas y calcula estadísticas.
    Incluye prints de debugging para seguir el flujo.
    """
    print(f"  [DEBUG] Entrada: {len(ventas)} ventas, umbral: ${umbral_minimo}")
    
    # Filtrar ventas válidas
    ventas_validas = [v for v in ventas if v >= umbral_minimo]
    print(f"  [DEBUG] Ventas válidas: {len(ventas_validas)} (filtradas: {len(ventas) - len(ventas_validas)})")
    
    if not ventas_validas:
        print(f"  [DEBUG] Lista vacía, retornando 0")
        return 0
    
    total = sum(ventas_validas)
    promedio = total / len(ventas_validas)
    
    print(f"  [DEBUG] Total: ${total:,}, Promedio: ${promedio:,.2f}")
    return promedio

ventas_ejemplo = [1500, 800, 2200, 500, 3000, 1200]
resultado = procesar_ventas(ventas_ejemplo, umbral_minimo=1000)
print(f"\n✅ Resultado final: ${resultado:,.2f}")

# =====================================
# DEMOSTRACIÓN 2: type() y validación
# =====================================

print("\n\n2️⃣ Técnica: type() para verificar tipos")
print("=" * 60)

def calcular_total_seguro(valores):
    """
    Suma valores con validación de tipos.
    """
    print(f"  Tipo de entrada: {type(valores)}")
    
    if not isinstance(valores, (list, tuple)):
        print(f"  ⚠️ Error: Se esperaba lista/tupla, recibió {type(valores).__name__}")
        return None
    
    # Verificar que todos los elementos sean números
    for i, valor in enumerate(valores):
        if not isinstance(valor, (int, float)):
            print(f"  ⚠️ Error: Elemento {i} no es número (es {type(valor).__name__})")
            return None
    
    total = sum(valores)
    print(f"  ✅ Total calculado: {total}")
    return total

# Pruebas
print("\nPrueba 1: Lista válida")
calcular_total_seguro([10, 20, 30])

print("\nPrueba 2: Lista con string")
calcular_total_seguro([10, "20", 30])

print("\nPrueba 3: String en lugar de lista")
calcular_total_seguro("123")

# =====================================
# DEMOSTRACIÓN 3: assert para validaciones
# =====================================

print("\n\n3️⃣ Técnica: assert para validaciones")
print("=" * 60)

def calcular_comision(ventas, tasa_comision=0.05):
    """
    Calcula comisión con validaciones estrictas.
    """
    # Validaciones
    assert isinstance(ventas, (int, float)), f"Ventas debe ser número, recibido: {type(ventas).__name__}"
    assert ventas >= 0, f"Ventas no pueden ser negativas: {ventas}"
    assert 0 <= tasa_comision <= 1, f"Tasa debe estar entre 0 y 1: {tasa_comision}"
    
    comision = ventas * tasa_comision
    return comision

# Pruebas con manejo de errores
pruebas = [
    (5000, 0.05, "Caso normal"),
    (-1000, 0.05, "Ventas negativas"),
    (5000, 1.5, "Tasa inválida"),
    ("5000", 0.05, "Tipo incorrecto")
]

for ventas, tasa, descripcion in pruebas:
    print(f"\nPrueba: {descripcion}")
    try:
        resultado = calcular_comision(ventas, tasa)
        print(f"  ✅ Comisión: ${resultado:,.2f}")
    except AssertionError as e:
        print(f"  ⚠️ Error de validación: {e}")
    except Exception as e:
        print(f"  ⚠️ Error inesperado: {type(e).__name__}: {e}")

# =====================================
# DEMOSTRACIÓN 4: try-except robusto
# =====================================

print("\n\n4️⃣ Técnica: try-except para manejo robusto")
print("=" * 60)

def dividir_robusto(numerador, denominador):
    """
    División con manejo completo de errores.
    """
    try:
        resultado = numerador / denominador
        return {"exito": True, "resultado": resultado, "error": None}
    except ZeroDivisionError:
        return {"exito": False, "resultado": None, "error": "División por cero"}
    except TypeError:
        return {"exito": False, "resultado": None, "error": "Tipos incompatibles"}
    except Exception as e:
        return {"exito": False, "resultado": None, "error": f"Error inesperado: {e}"}

# Pruebas
casos = [
    (10, 2, "Caso normal"),
    (10, 0, "División por cero"),
    (10, "2", "Tipo incorrecto"),
    (None, 5, "Numerador None")
]

for num, den, desc in casos:
    print(f"\n{desc}: {num} / {den}")
    resultado = dividir_robusto(num, den)
    
    if resultado["exito"]:
        print(f"  ✅ Resultado: {resultado['resultado']}")
    else:
        print(f"  ⚠️ Error: {resultado['error']}")

# =====================================
# DEMOSTRACIÓN 5: dir() e inspección
# =====================================

print("\n\n5️⃣ Técnica: dir() para explorar objetos")
print("=" * 60)

lista_ejemplo = [1, 2, 3]
print(f"\nObjeto: lista = {lista_ejemplo}")
print(f"Tipo: {type(lista_ejemplo)}")

# Métodos públicos disponibles
metodos_publicos = [m for m in dir(lista_ejemplo) if not m.startswith('_')]
print(f"\nMétodos disponibles ({len(metodos_publicos)}):")
for i, metodo in enumerate(metodos_publicos, 1):
    print(f"  {i:2d}. {metodo}")

print("\n" + "=" * 60)
print("💡 Usa estas técnicas para debuggear tu código eficientemente")
print("=" * 60)
````

Funciones reutilizables para análisis de ventas.

    Args:

    elif compras_anuales > 20000:

    elif compras_anuales > 5000:
        return "Silver"
    else:
        return "Standard"
