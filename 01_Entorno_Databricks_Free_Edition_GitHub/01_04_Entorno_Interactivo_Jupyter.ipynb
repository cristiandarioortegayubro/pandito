## 📓 1. Anatomía de un Notebook

### 📌 1.1 ¿Qué es un Notebook?

Documento interactivo que combina:
* **Código ejecutable** (Python, SQL, Scala, R)
* **Texto formateado** (Markdown)
* **Resultados** (tablas, gráficos, outputs)

**Ventajas vs scripts tradicionales:**
* ✅ Ejecución celda por celda (iteración rápida)
* ✅ Visualización inline de resultados
* ✅ Documentación integrada
* ✅ Colaboración en tiempo real

---

### 🧱 1.2 Tipos de Celdas

**A) Celdas de Código**
* Ejecutan código Python/SQL/etc.
* Variables persisten entre celdas
* Output se muestra debajo

**B) Celdas Markdown**
* Texto formateado con Markdown
* No ejecutables (solo renderización)
* Usar `%md` al inicio

**C) Comandos Mágicos**
* Instrucciones especiales precedidas por `%`
* Ejemplos: `%sql`, `%sh`, `%pip`

---

### 🔄 1.3 Ejecución de Celdas

**Orden de ejecución:**
* Las celdas se ejecutan en el orden que TÚ elijas
* No necesariamente secuencial
* Número entre corchetes `[3]` indica orden

**Estado de variables:**
```python
# Celda 1
x = 10

# Celda 2 (ejecutada después)
print(x)  # 10 (x existe en memoria)

# Si ejecutas Celda 2 ANTES de Celda 1 → Error: x no definida
```

## 📈 5. Visualizaciones Inline

### matplotlib Básico
```python
import matplotlib.pyplot as plt
import numpy as np

# Datos
ventas = [1200, 1350, 1100, 1450, 1600, 2000, 1800]
dias = ['Lun', 'Mar', 'Mié', 'Jue', 'Vie', 'Sáb', 'Dom']

# Gráfico de barras
plt.figure(figsize=(10, 5))
plt.bar(dias, ventas, color='steelblue')
plt.title('Ventas Semanales', fontsize=16)
plt.xlabel('Día')
plt.ylabel('Ventas ($)')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### Múltiples Subplots
```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# Subplot 1: Líneas
ax1.plot(dias, ventas, marker='o')
ax1.set_title('Tendencia')

# Subplot 2: Torta
ax2.pie(ventas, labels=dias, autopct='%1.1f%%')
ax2.set_title('Distribución')

plt.tight_layout()
plt.show()
```

## 🐞 6. Debugging en Notebooks

### 📍 Errores Comunes

**A) NameError: variable no definida**
```python
print(x)  # Error si x no existe

# Solución: ejecutar celda donde se define x primero
```

**B) IndexError: índice fuera de rango**
```python
lista = [1, 2, 3]
print(lista[5])  # Error

# Solución: verificar len(lista)
```

**C) KeyError: clave no existe**
```python
diccionario = {"a": 1}
print(diccionario["b"])  # Error

# Solución: usar get()
print(diccionario.get("b", "default"))
```

---

### 🔍 Técnicas de Debugging

**1. print() estratégico**
```python
def procesar_datos(datos):
    print(f"Recibido: {len(datos)} elementos")  # Debug
    resultado = [x * 2 for x in datos]
    print(f"Resultado: {resultado}")  # Debug
    return resultado
```

**2. type() para verificar tipos**
```python
x = "123"
print(type(x))  # <class 'str'>
x = int(x)
print(type(x))  # <class 'int'>
```

**3. dir() para ver atributos**
```python
lista = [1, 2, 3]
print(dir(lista))  # Ver todos los métodos disponibles
```

**4. assert para validaciones**
```python
def calcular_promedio(notas):
    assert len(notas) > 0, "Lista vacía"
    assert all(0 <= n <= 100 for n in notas), "Notas fuera de rango"
    return sum(notas) / len(notas)
```
