### 🎯 Ejercicio 1: Sistema de comisiones

**Problema:** Una empresa paga comisiones escalonadas:
* 0-5000: 3%
* 5001-10000: 5%
* 10001-20000: 7%
* > 20000: 10%

**Tarea:** Implementar función que calcule comisión según ventas.

````python
# 🎯 Ejercicio 1: Sistema de Comisiones Escalonadas

print("💰 EJERCICIO 1: SISTEMA DE COMISIONES")
print("=" * 60)

ventas_vendedores = [
    {"nombre": "Ana", "ventas": 4500},
    {"nombre": "Carlos", "ventas": 8200},
    {"nombre": "Beatriz", "ventas": 15000},
    {"nombre": "Diego", "ventas": 25000}
]

print("\n📊 Cálculo de Comisiones:")
total_comisiones = 0

for vendedor in ventas_vendedores:
    nombre = vendedor["nombre"]
    ventas = vendedor["ventas"]
    
    # Sistema escalonado
    if ventas <= 5000:
        comision = ventas * 0.03
        tasa_desc = "3%"
    elif ventas <= 10000:
        comision = 5000 * 0.03 + (ventas - 5000) * 0.05
        tasa_desc = "3-5%"
    elif ventas <= 20000:
        comision = 5000 * 0.03 + 5000 * 0.05 + (ventas - 10000) * 0.07
        tasa_desc = "3-5-7%"
    else:
        comision = 5000 * 0.03 + 5000 * 0.05 + 10000 * 0.07 + (ventas - 20000) * 0.10
        tasa_desc = "3-5-7-10%"
    
    total_comisiones += comision
    print(f"{nombre}: ${ventas:,} → Comisión: ${comision:,.2f} ({tasa_desc})")

print(f"\nTotal comisiones: ${total_comisiones:,.2f}")
print("\n✅ Ejercicio 1 completado")
````

## 🔄 2. Bucles y List Comprehensions

### 🔁 2.1 Bucle FOR - Iteración sobre secuencias

**Sintaxis:**
```python
for elemento in secuencia:
    # Código a ejecutar
```

**Casos de uso:**

#### A) Iterar sobre listas
```python
productos = ["Laptop", "Mouse", "Teclado"]
for producto in productos:
    print(producto)
```

#### B) Enumerate (índice + elemento)
```python
productos = ["Laptop", "Mouse", "Teclado"]
for i, producto in enumerate(productos, 1):
    print(f"{i}. {producto}")
```

#### C) Iterar sobre diccionarios
```python
precios = {"Laptop": 1250, "Mouse": 25, "Teclado": 75}

# Solo claves
for producto in precios:
    print(producto)

# Clave + valor
for producto, precio in precios.items():
    print(f"{producto}: ${precio}")
```

#### D) Range (secuencia de números)
```python
for i in range(5):  # 0, 1, 2, 3, 4
    print(i)

for i in range(1, 6):  # 1, 2, 3, 4, 5
    print(i)

for i in range(0, 10, 2):  # 0, 2, 4, 6, 8
    print(i)
```

**Sintaxis:**
```python

# ✗ MAL (bucle infinito)

while x < 10:
    print(x)
    x += 1  # Incrementar para eventualmente salir
```

### 📊 2.4 List Comprehensions (sintaxis pythonic)

**Definición:** Sintaxis compacta para crear listas transformando/filtrando otra secuencia.

**Sintaxis básica:**
```python
[expresión for elemento in secuencia]
```

**Sintaxis con filtro:**
```python
[expresión for elemento in secuencia if condición]
```

#### A) Método tradicional vs List Comprehension

**Tradicional:**
```python
cuadrados = []
for x in range(5):
    cuadrados.append(x**2)
# [0, 1, 4, 9, 16]
```

**List Comprehension:**
```python
cuadrados = [x**2 for x in range(5)]
# [0, 1, 4, 9, 16]
```

#### B) Con transformación
```python
# Convertir precios a formato con IVA
precios = [100, 150, 200]
precios_con_iva = [p * 1.21 for p in precios]
# [121.0, 181.5, 242.0]
```

#### C) Con filtro
```python
# Solo ventas mayores a $1000
ventas = [1200, 450, 800, 2500, 300]
ventas_grandes = [v for v in ventas if v > 1000]
# [1200, 2500]
```

#### D) Transformación + filtro
```python
# Aplicar 15% descuento solo a precios > $75
precios = [100, 150, 80, 200, 50]
precios_descuento = [p * 0.85 for p in precios if p > 75]
# [85.0, 127.5, 68.0, 170.0]
```

#### E) List comprehensions anidadas
```python
# Crear matriz 3x3
matriz = [[i*j for j in range(3)] for i in range(3)]
# [[0, 0, 0], [0, 1, 2], [0, 2, 4]]
```

---

### 🎯 Cuándo usar cada uno

* **FOR:** Iteraciones generales, múltiples operaciones por elemento
* **WHILE:** Repetir hasta condición (no sabes cuántas iteraciones)
* **LIST COMPREHENSION:** Crear nuevas listas de forma concisa (1 línea)

---

### 💡 Tips de rendimiento

* ✅ List comprehensions son ~2x más rápidas que bucles for tradicionales
* ✅ Usa `enumerate()` en lugar de `range(len(lista))`
* ✅ Usa `zip()` para combinar listas en lugar de índices manuales
* ⚠️ No abuses de list comprehensions anidadas (prioriza legibilidad)

````python
# 🎯 Ejercicio 2: Bucles y List Comprehensions en Acción

print("💰 EJERCICIO 2: BUCLES Y LIST COMPREHENSIONS")
print("=" * 60)

print("\n🔁 FOR Loop (Iteración sobre secuencias):")
productos = ["Laptop", "Mouse", "Teclado"]
for i, producto in enumerate(productos, 1):
    print(f"{i}. {producto}")

print("\n🔃 WHILE Loop (Repetir hasta condición):")
cuenta = 100
tasa_interes = 0.05
obj_objetivo = 150
años = 0

print(f"Capital inicial: ${cuenta}")
while cuenta < obj_objetivo:
    años += 1
    interes = cuenta * tasa_interes
    cuenta += interes
    print(f"Año {años}: ${cuenta:.2f}")

print(f"\nSe alcanzó el objetivo de ${obj_objetivo} en {años} años")

print("\n🧩 Iteración con ZIP (combinar listas):")
productos = ["Laptop", "Mouse", "Teclado"]
precios = [1250, 25, 75]
stock = [10, 50, 30]

for prod, prec, stk in zip(productos, precios, stock):
    valor_inv = prec * stk
    print(f"{prod}: ${prec} x {stk} = ${valor_inv:,}")

print("\n" + "=" * 60)
print("📊 LIST COMPREHENSIONS (sintaxis pythonic)")
print("=" * 60)

# Tradicional
print("\n1️⃣ Método tradicional:")
cuadrados = []
for x in range(5):
    cuadrados.append(x**2)
print(f"Cuadrados: {cuadrados}")

# Pythonic
print("\n2️⃣ List Comprehension:")
cuadrados = [x**2 for x in range(5)]
print(f"Cuadrados: {cuadrados}")

# Filtrado
print("\n3️⃣ List Comprehension con filtro:")
ventas = [1200, 450, 800, 2500, 300]
ventas_grandes = [v for v in ventas if v > 1000]
print(f"Ventas originales: {ventas}")
print(f"Ventas > $1000: {ventas_grandes}")

# Transformación con filtro
print("\n4️⃣ Transformación + filtro:")
precios = [100, 150, 80, 200, 50]
precios_descuento = [p * 0.85 for p in precios if p > 75]
print(f"Precios originales: {precios}")
print(f"Precios > $75 con 15% descuento: {precios_descuento}")

print("\n✅ Bucles y List Comprehensions completados")
````

## 🛠️ 3. Funciones en Python

### 📌 3.1 ¿Qué es una Función?

**Definición:** Bloque de código reutilizable que realiza una tarea específica.

**Ventajas:**
* ✅ **Reutilización:** Escribe una vez, usa muchas veces
* ✅ **Organización:** Código modular más fácil de mantener
* ✅ **Testing:** Probar funciones individualmente
* ✅ **Documentación:** Nombre descriptivo y docstring

**Sintaxis:**
```python
def nombre_funcion(parametro1, parametro2):
    """
    Docstring: Describe qué hace la función.
    Args:
        parametro1: Descripción
        parametro2: Descripción
    Returns:
        Qué retorna la función
    """
    # Código de la función
    resultado = parametro1 + parametro2
    return resultado
```

def calcular_roi(ganancia, inversion):
    """Return on Investment."""

def valor_presente_neto(flujos, tasa_descuento):
    """VPN de flujos de caja futuros."""

````python
# 🎯 Ejercicio 3: Funciones en Acción

print("=" * 60)
print("🎯 EJERCICIO 3: APLICANDO FUNCIONES")
print("=" * 60)

# 1. Función con IVA
def calcular_precio_con_iva(precio_base, tasa_iva=0.21):
    """
    Calcula precio final incluyendo IVA.
    
    Args:
        precio_base (float): Precio sin IVA
        tasa_iva (float): Tasa de IVA (default: 0.21 = 21%)
    
    Returns:
        float: Precio con IVA incluido
    """
    return precio_base * (1 + tasa_iva)

print("\n1️⃣ Prueba: Calcular precio con IVA")
precio1 = calcular_precio_con_iva(100)  # Usa default 21%
precio2 = calcular_precio_con_iva(100, 0.10)  # Tasa 10%
print(f"Precio base: $100")
print(f"Con IVA 21% (default): ${precio1:.2f}")
print(f"Con IVA 10%: ${precio2:.2f}")

# 2. Función con múltiples retornos
def calcular_comision(ventas):
    """
    Calcula comisión escalonada.
    
    Returns:
        tuple: (comision, tasa_efectiva, categoria)
    """
    if ventas <= 5000:
        comision = ventas * 0.03
        categoria = "Principiante"
    elif ventas <= 10000:
        comision = 5000 * 0.03 + (ventas - 5000) * 0.05
        categoria = "Intermedio"
    else:
        comision = 5000 * 0.03 + 5000 * 0.05 + (ventas - 10000) * 0.08
        categoria = "Avanzado"
    
    tasa_efectiva = (comision / ventas) * 100
    return comision, tasa_efectiva, categoria

print("\n2️⃣ Prueba: Comisiones con múltiples retornos")
com, tasa, cat = calcular_comision(8000)
print(f"Ventas: $8,000")
print(f"Comisión: ${com:.2f}")
print(f"Tasa efectiva: {tasa:.2f}%")
print(f"Categoría: {cat}")

# 3. Función de validación
def validar_stock(cantidad_solicitada, stock_disponible):
    """
    Valida si hay stock suficiente.
    
    Returns:
        tuple: (bool, mensaje)
    """
    if cantidad_solicitada <= 0:
        return False, "Cantidad debe ser mayor a 0"
    
    if cantidad_solicitada > stock_disponible:
        return False, f"Stock insuficiente. Disponible: {stock_disponible}"
    
    return True, "Stock suficiente"

print("\n3️⃣ Prueba: Validación de stock")
casos = [(15, 20), (25, 20), (-5, 20)]
for cant, stock in casos:
    valido, mensaje = validar_stock(cant, stock)
    print(f"Solicitar {cant} (stock: {stock}): {'\u2705' if valido else '\u274c'} {mensaje}")

# 4. *args - Número variable de argumentos
def suma_total(*numeros):
    """Suma cualquier cantidad de números."""
    return sum(numeros)

print("\n4️⃣ Prueba: *args (argumentos variables)")
print(f"suma_total(1, 2, 3) = {suma_total(1, 2, 3)}")
print(f"suma_total(10, 20, 30, 40, 50) = {suma_total(10, 20, 30, 40, 50)}")

# 5. **kwargs - Argumentos con nombre
def crear_cliente(**info):
    """Crea diccionario de cliente con info dinámica."""
    return info

print("\n5️⃣ Prueba: **kwargs (argumentos con nombre)")
cliente1 = crear_cliente(nombre="Ana", email="ana@email.com", edad=30)
cliente2 = crear_cliente(nombre="Carlos", empresa="TechCorp", telefono="555-1234")
print(f"Cliente 1: {cliente1}")
print(f"Cliente 2: {cliente2}")

# 6. Lambda functions
print("\n6️⃣ Prueba: Lambda functions")

# Ordenar por segundo elemento
ventas = [("Ana", 5000), ("Carlos", 7000), ("Beatriz", 6500)]
ventas_ordenadas = sorted(ventas, key=lambda x: x[1], reverse=True)
print(f"Ventas ordenadas: {ventas_ordenadas}")

# Aplicar descuento
precios = [100, 150, 200]
precios_descuento = list(map(lambda p: p * 0.85, precios))
print(f"Precios con 15% descuento: {precios_descuento}")

# Filtrar
ventas_lista = [1200, 800, 1500, 600, 2000]
ventas_grandes = list(filter(lambda v: v > 1000, ventas_lista))
print(f"Ventas > $1000: {ventas_grandes}")

print("\n\u2705 Ejercicio 3 completado")
````

## 📚 4. Biblioteca de Funciones de Negocio

### 🎯 4.1 ¿Qué es una Biblioteca de Funciones?

**Definición:** Conjunto organizado de funciones reutilizables que resuelven problemas comunes de un dominio específico (finanzas, ventas, inventario, etc.)

**Ventajas:**
* ✅ **Consistencia:** Todos usan la misma lógica de negocio
* ✅ **Mantenimiento:** Corregir un bug una vez, beneficia a todos
* ✅ **Productividad:** No reinventar la rueda
* ✅ **Testing:** Probar funciones de forma aislada
* ✅ **Documentación:** Centralizar conocimiento del negocio

---

### 💼 4.2 Estructura de una Biblioteca

```python
# ventas.py - Biblioteca de funciones de ventas

def calcular_precio_con_descuento(precio, descuento=0):
    """
    Calcula precio final aplicando descuento porcentual.
    
    Args:
        precio (float): Precio original
        descuento (float): Porcentaje de descuento (0-100)
    
    Returns:
        float: Precio final con descuento aplicado
    
    Examples:
        >>> calcular_precio_con_descuento(1000, 15)
        850.0
    """
    if descuento < 0 or descuento > 100:
        raise ValueError("Descuento debe estar entre 0 y 100")
    return precio * (1 - descuento / 100)

def categorizar_cliente(compras_anuales):
    """
    Categoriza cliente según volumen de compras.
    
    Returns:
        tuple: (categoría, descuento_vip)
    """
    if compras_anuales > 50000:
        return "Premium", 25
    elif compras_anuales > 20000:
        return "Gold", 15
    elif compras_anuales > 5000:
        return "Silver", 5
    else:
        return "Standard", 0

def validar_transaccion(monto, saldo_disponible):
    """
    Valida si una transacción puede procesarse.
    
    Returns:
        tuple: (es_valida, mensaje)
    """
    if monto <= 0:
        return False, "Monto debe ser positivo"
    if monto > saldo_disponible:
        return False, f"Saldo insuficiente (disponible: ${saldo_disponible:,.2f})"
    return True, "Transacción válida"
```

```python
def procesar_ventas_del_dia(ventas, umbral_minimo=1000):
    """
    Procesa lista de ventas y retorna estadísticas.
    
    Args:
        ventas (list): Lista de montos de ventas
        umbral_minimo (float): Ventas menores se excluyen
    
    Returns:
        dict: Estadísticas completas
    """
    # List comprehension para filtrar
    ventas_validas = [v for v in ventas if v >= umbral_minimo]
    
    # Validación edge case
    if not ventas_validas:
        return {"total": 0, "promedio": 0, "cantidad": 0}
    
    # Lambda para ordenar
    top_3 = sorted(ventas_validas, key=lambda x: x, reverse=True)[:3]
    
    # Calcular estadísticas
    return {
        "total": sum(ventas_validas),
        "promedio": sum(ventas_validas) / len(ventas_validas),
        "cantidad": len(ventas_validas),
        "top_3": top_3,
        "mejor_venta": max(ventas_validas),
        "peor_venta": min(ventas_validas)
    }
```

````python
# 🎯 Ejercicio 4: Biblioteca de Funciones de Negocio

print("=" * 60)
print("🎯 4. EJERCICIO FINAL: BIBLIOTECA DE FUNCIONES")
print("=" * 60)

# 1. Definir funciones
def calcular_precio_con_descuento(precio, descuento=0):
    """
    Calcula precio final con descuento.
    
    Args:
        precio (float): Precio original
        descuento (float): % de descuento (0-100)
    
    Returns:
        float: Precio final con descuento
    """
    return precio * (1 - descuento / 100)

def categorizar_cliente(compras_anuales):
    """
    Categoriza cliente según volumen de compras.
    
    Returns:
        tuple: (categoría, descuento_vip)
    """
    if compras_anuales > 50000:
        return "Premium", 25
    elif compras_anuales > 20000:
        return "Gold", 15
    elif compras_anuales > 5000:
        return "Silver", 5
    else:
        return "Standard", 0

def calcular_comision_completa(*ventas_mensuales):
    """
    Calcula comisión anual con cualquier número de meses.
    
    Returns:
        dict: Estadísticas de comisiones
    """
    total_ventas = sum(ventas_mensuales)
    mejor_mes = max(ventas_mensuales)
    peor_mes = min(ventas_mensuales)
    promedio = total_ventas / len(ventas_mensuales)
    
    # Comisión escalonada
    if total_ventas < 50000:
        comision = total_ventas * 0.03
    elif total_ventas < 100000:
        comision = 50000 * 0.03 + (total_ventas - 50000) * 0.05
    else:
        comision = 50000 * 0.03 + 50000 * 0.05 + (total_ventas - 100000) * 0.08
    
    return {
        "total_ventas": total_ventas,
        "comision": comision,
        "mejor_mes": mejor_mes,
        "peor_mes": peor_mes,
        "promedio_mes": promedio,
        "tasa_efectiva": (comision / total_ventas * 100) if total_ventas > 0 else 0
    }

# 2. Usar las funciones
print("\n1️⃣ PRUEBA: Calcular Precio con Descuento")
precio_original = 1250
descuento = 15
precio_final = calcular_precio_con_descuento(precio_original, descuento)
print(f"Precio original: ${precio_original}")
print(f"Descuento: {descuento}%")
print(f"Precio final: ${precio_final:.2f}")

print("\n2️⃣ PRUEBA: Categorizar Cliente")
clientes_prueba = [45000, 75000, 3000, 25000]
for i, compras in enumerate(clientes_prueba, 1):
    categoria, desc_vip = categorizar_cliente(compras)
    print(f"Cliente {i}: ${compras:,} → {categoria} (Descuento VIP: {desc_vip}%)")

print("\n3️⃣ PRUEBA: Comisiones Anuales")
ventas_vendedor = [5000, 6500, 7200, 8500, 9000, 10000, 8800, 9500, 11000, 12000, 13500, 15000]
estadisticas = calcular_comision_completa(*ventas_vendedor)

print(f"Total ventas anual: ${estadisticas['total_ventas']:,}")
print(f"Comisión total: ${estadisticas['comision']:,.2f}")
print(f"Tasa efectiva: {estadisticas['tasa_efectiva']:.2f}%")
print(f"Mejor mes: ${estadisticas['mejor_mes']:,}")
print(f"Peor mes: ${estadisticas['peor_mes']:,}")
print(f"Promedio mensual: ${estadisticas['promedio_mes']:,.2f}")

print("\n4️⃣ PRUEBA: Lambda Functions")
productos = [
    {"nombre": "Laptop", "precio": 1250, "stock": 10},
    {"nombre": "Mouse", "precio": 25, "stock": 50},
    {"nombre": "Monitor", "precio": 350, "stock": 15}
]

# Ordenar por precio descendente
productos_ordenados = sorted(productos, key=lambda p: p["precio"], reverse=True)
print("Productos ordenados por precio:")
for p in productos_ordenados:
    print(f"  {p['nombre']}: ${p['precio']}")

# Filtrar productos con stock < 20
stock_bajo = list(filter(lambda p: p["stock"] < 20, productos))
print(f"\nProductos con stock bajo: {[p['nombre'] for p in stock_bajo]}")

# Aplicar descuento 10% a todos
precios_descuento = list(map(lambda p: {**p, "precio_descuento": p["precio"] * 0.9}, productos))
print("\nPrecios con 10% descuento:")
for p in precios_descuento:
    print(f"  {p['nombre']}: ${p['precio']} → ${p['precio_descuento']:.2f}")

print("\n\n" + "=" * 60)
print("🏆 DESAFÍO INTEGRADOR FINAL")
print("=" * 60)
print("""\n🎯 Objetivo: Sistema Completo de Análisis de Ventas

Crea un programa que:
1. Defina función para calcular total de venta con descuento e IVA
2. Use condicionales para categorizar clientes (Standard/Silver/Gold/Premium)
3. Itere sobre lista de ventas y calcule métricas
4. Use list comprehensions para filtrar ventas > umbral
5. Implemente funciones lambda para ordenar resultados
6. Retorne reporte completo con estadísticas

Requisitos:
- Docstrings en todas las funciones
- Manejo de casos edge (división por cero, listas vacías)
- Código limpio y modular

¡Intenta resolver antes de ver solución!
""")

print("\n✅ Ejercicio 4 completado")
print("✅ Notebook 01_03 finalizado exitosamente")
````

# Función: clasificar sucursal por nivel de ventas

````python
import pandas as pd

print("🏪 PRÁCTICA: CONTROL Y FUNCIONES CON DATOS REALES")
print("="*60)

try:
    df = spark.table("pandito_ds.default.ventas_mensuales_mendoza_h3").toPandas()
    df['fecha'] = pd.to_datetime(df['fecha'])
    df['año'] = df['fecha'].dt.year
    
    print("\n1️⃣  CONDICIONALES: Clasificar sucursales por ventas")
    print("-"*60)
    
    def clasificar_sucursal(ventas_promedio):
        """Clasifica sucursal según promedio de ventas mensuales."""
        if ventas_promedio > 150000:
            return "Alta", 0.10
        elif ventas_promedio > 80000:
            return "Media", 0.05
        else:
            return "Baja", 0.0
    
    resumen = df.groupby(['sucursal_nombre', 'zona'])['ventas'].mean().round(0)
    for (sucursal, zona), promedio in resumen.items():
        nivel, bonus = clasificar_sucursal(promedio)
        print(f"   {sucursal:<30} {zona:<22} ${promedio:>10,.0f}  →  {nivel} (bonus {bonus*100:.0f}%)")
    
    print("\n2️⃣  BUCLES: Top 5 meses con más ventas")
    print("-"*60)
    ventas_mensuales = df.groupby('fecha')['ventas'].sum().sort_values(ascending=False)
    for i, (fecha, ventas) in enumerate(ventas_mensuales.head(5).items(), 1):
        print(f"   {i}. {fecha.strftime('%Y-%m')}: ${ventas:,.0f}")
    
    print("\n3️⃣  LIST COMPREHENSIONS: Filtrar y transformar")
    print("-"*60)
    ventas_lista = df['ventas'].tolist()
    ventas_altas = [v for v in ventas_lista if v > 150000]
    print(f"   Ventas > $150,000: {len(ventas_altas)} de {len(ventas_lista)} registros")
    print(f"   Porcentaje: {len(ventas_altas)/len(ventas_lista)*100:.1f}%")
    
    # comprehension con transformación
    ventas_miles = [round(v / 1000) for v in ventas_lista[:10]]
    print(f"   Primeras 10 ventas en miles: {ventas_miles}")
    
    print("\n4️⃣  FUNCIONES: KPIs reutilizables de Los Andes Market")
    print("-"*60)
    
    def calcular_kpis(df_ventas):
        """Calcula KPIs principales de un DataFrame de ventas."""
        return {
            'total': df_ventas['ventas'].sum(),
            'promedio': df_ventas['ventas'].mean(),
            'maximo': df_ventas['ventas'].max(),
            'minimo': df_ventas['ventas'].min(),
            'registros': len(df_ventas),
            'sucursales': df_ventas['sucursal_id'].nunique(),
        }
    
    kpis = calcular_kpis(df)
    print("   KPIs generales:")
    for k, v in kpis.items():
        if k in ('total', 'promedio', 'maximo', 'minimo'):
            print(f"      {k}: ${v:,.0f}")
        else:
            print(f"      {k}: {v}")
    
    # KPIs por año usando bucle + función
    print("\n   KPIs por año:")
    for año in sorted(df['año'].unique()):
        df_año = df[df['año'] == año]
        k = calcular_kpis(df_año)
        print(f"      {año}: ${k['total']:>12,.0f} | {k['registros']} registros | promedio ${k['promedio']:>10,.0f}")
    
    print("\n5️⃣  LAMBDA + SORTED: Ranking de sucursales")
    print("-"*60)
    ventas_por_sucursal = df.groupby('sucursal_nombre')['ventas'].sum().to_dict()
    ranking = sorted(ventas_por_sucursal.items(), key=lambda x: x[1], reverse=True)
    for i, (sucursal, ventas) in enumerate(ranking, 1):
        print(f"   {i}. {sucursal}: ${ventas:,.0f}")
    
    # Lambda para clasificación rápida
    categorizar = lambda v: "Alta" if v > 150000 else ("Media" if v > 80000 else "Baja")
    print("\n   Clasificación con lambda:")
    for sucursal, ventas in ranking:
        print(f"      {sucursal}: {categorizar(ventas)}")
    
    print("\n" + "="*60)
    print("✅ Práctica con datos reales completada")
except Exception as e:
    print(f"⚠️  No se pudo cargar datos reales: {e}")
    print("   Ejecuta primero 02_05_Preparacion_Datos_Empresariales")

print("\n" + "="*60)
````
