## 🏛️ 2. Arquitectura de Databricks: los 3 componentes clave

### 💻 2.1 Primer componente: Workspace (Espacio de Trabajo)

**Definición:** Es tu "escritorio virtual" en la nube donde organizas:

📓 **Notebooks** - Documentos interactivos con código + texto + gráficos.   
📁 **Carpetas** - Organiza proyectos (como en tu computadora).   
📊 **Dashboards** - Reportes visuales dinámicos.   
🔧 **Jobs** - Tareas programadas (ej: "ejecutar reporte cada lunes 8 AM").   
💾 **Repos** - Repositorios de Git sincronizados.   

**Analogía:** Es como Google Drive, pero para análisis de datos.

---

### ⚡ 2.2 Segundo componente: Clusters (Cómputo)

**Definición:** Son las "computadoras virtuales" que ejecutan tu código.

#### Tipos de cómputos:

**A) All-Purpose Clusters (Solo en la versión paga)**

* Tú configuras: cantidad de RAM, CPUs, tipo de máquina
* Se mantiene activo todo el tiempo que quieras
* Costo: ~$0.30-$5.00 por hora (según specs)

**B) Serverless Compute (⭐ Lo que usaremos nosotros - GRATIS)**

✅ Databricks lo gestiona automáticamente.  
✅ Se activa al ejecutar una celda.  
✅ Se apaga automáticamente tras inactividad.  
✅ **Sin costo** en Free Edition.  
⚠️ Limitación: Solo Python, SQL y comandos shell (no R ni Scala).  

**💡 Tip:** En Free Edition, **no ves el botón de "Start Cluster"** porque el compute serverless se inicia solo.

---

### 📓 2.3 Tercer componente: Notebooks (Cuadernos interactivos)

**Definición:** Documento que combina:

1. **Código ejecutable** (Python, SQL, sh)
2. **Texto formateado** (markdown) - para explicaciones
3. **Resultados** (tablas, gráficos, números)

#### Estructura de un notebook:

```
+-------------------------------------------+
| Celda 1: Markdown (Título y explicación)  |
+-------------------------------------------+
| Celda 2: Python (importar bibliotecas)    |
+-------------------------------------------+
| Celda 3: SQL (consultar base de datos)    |
+-------------------------------------------+
| Celda 4: Python (procesar datos)          |
+-------------------------------------------+
| Celda 5: Visualización (gráfico)          |
+-------------------------------------------+
```

**Ventajas vs. Excel:**

| Característica | Excel | Databricks notebook |
|----------------|-------|-----------|
| Límite de filas | 1,048,576 | Miles de millones |
| Lenguajes | Fórmulas de Excel | Python, SQL, R, Scala |
| Versionamiento | Manual (Archivo_v1.xlsx) | Git (automático) |
| Colaboración | Enviar por email | Tiempo real |
| Automatización | Macros (limitadas) | Jobs programados |
| Documentación | Hojas separadas | Markdown integrado |

## 🔧 4. Comandos shell

### 🐧 Shell: La línea de comandos de Linux

**¿Qué son los comandos shell?**

Comandos de terminal de Linux/Unix que permiten:

📁 **Explorar archivos** - `ls`, `pwd`, `cd`.  
📄 **Ver contenido** - `cat`, `head`, `tail`.

---

### ⚡ Magic commands en Databricks

Databricks usa **"magic commands"** para cambiar de lenguaje dentro de un notebook:

```python
%python    # Ejecutar Python (default)
%sql       # Ejecutar SQL
%md        # Escribir Markdown
%sh        # Ejecutar Shell commands
%r         # Ejecutar R (si está disponible)
%scala     # Ejecutar Scala (si está disponible)
```

**Ejemplo de uso:**
```python
%sh ls /dbfs/FileStore/
```

---

### 🔒 Limitaciones en serverless compute

En Databricks **Serverless** (Free Edition), los comandos shell están **restringidos por seguridad**:

**✅ Disponibles:**
* `ls`, `pwd`, `cat`, `echo` - Navegación básica
* `python`, `pip` - Gestión de Python
* `wget`, `curl` - Descarga de archivos

**❌ No disponibles:**
* `apt-get`, `yum` - Instalación de paquetes del sistema
* `sudo` - Permisos de administrador
* Acceso directo a Docker o recursos del host

---

### 💡 Alternativas a comandos shell

En lugar de shell, usa:

| En vez de... | Usa en Python |
|--------------|---------------|
| `%sh ls /dbfs/FileStore/` | `dbutils.fs.ls("/FileStore/")` |
| `%sh cat file.txt` | `with open('file.txt') as f: print(f.read())` |
| `%sh wget <url>` | `import requests; requests.get(url)` |

**⚡ Ejecuta la celda siguiente para ver comandos shell disponibles**

````python
# ⚙️ Ejercicio 2: Explorar el sistema de archivos con comandos shell

print("🔧 COMANDOS SHELL EN DATABRICKS")
print("=" * 60)

print("\n💡 En Databricks puedes ejecutar comandos shell con %sh magic")
print("\nEjemplos comunes:")
print("""  
%sh ls /dbfs/FileStore/              # Listar archivos
%sh pwd                              # Directorio actual  
%sh df -h                            # Espacio en disco
%sh cat /dbfs/FileStore/archivo.txt  # Ver contenido
%sh wget <url> -P /dbfs/FileStore/   # Descargar archivo
""")

print("\n⚠️ Nota: En Serverless, algunos comandos shell pueden estar limitados")
print("\n✅ Comandos funcionando en Serverless:")
print("   - ls, pwd, cat, echo")
print("   - python, pip")
print("\n❌ No disponibles en Serverless:")
print("   - apt-get, yum (gestores de paquetes del sistema)")
print("   - sudo (permisos de administrador)")
````

## 📂 5. Sistemas de almacenamiento en Databricks

### 🗂️ DBFS (Databricks File System)

**Definición:** Sistema de archivos distribuido que aparece como carpetas en Databricks.

**Rutas comunes:**
```
/FileStore/           → Archivos subidos por usuario
/databricks-datasets/ → Datasets de ejemplo (incluidos)
/user/hive/warehouse/ → Tablas de bases de datos
/tmp/                 → Archivos temporales
```

**Ejemplo de uso:**

```python
# Leer archivo CSV desde DBFS
df = spark.read.csv("/FileStore/ventas.csv", header=True)

# Listar contenido de carpeta
dbutils.fs.ls("/FileStore/")

# Subir archivo
dbutils.fs.cp("file:/local/path.csv", "/FileStore/path.csv")
```

---

### 🏭 Unity Catalog vs. DBFS (Databricks File System)

| Aspecto | DBFS | Unity Catalog |
|---------|------|---------------|
| **Disponibilidad** | ✅ Community Edition | ❌ Solo versión paga |
| **Permisos** | Básicos (público) | Granulares (por usuario/grupo) |
| **Organización** | Carpetas simples | Catalogs → Schemas → Tables |
| **Gobernanza** | Limitada | Avanzada (auditoría, linaje) |
| **Ideal para** | Aprendizaje, prototipos | Producción empresarial |

**Para este libro:** Usaremos **DBFS** exclusivamente, suficiente para aprender.

---

### 💡 Tips importantes para tener en cuenta

**✅ Buenas prácticas:**

* Usa **Delta Lake** siempre que sea posible (transacciones ACID, versionamiento)
* Organiza archivos en carpetas lógicas (`/FileStore/proyectos/retail/`)
* Nombra tablas descriptivamente (`ventas_2024_limpio`, no `tabla1`)

**❌ Evita:**

* Archivos CSV gigantes (>1 GB) sin particionamiento
* Nombres de tabla con espacios o caracteres especiales
* Guardar datos sensibles sin permisos en DBFS (en producción usar Unity Catalog)

---

### 💾 Tablas en Databricks

**Tipos de tablas:**

1. **Managed Tables** (gestionadas)
   - Databricks controla datos + metadatos
   - Al borrar tabla, se borran los datos
   - Ubicación: `/user/hive/warehouse/`

2. **External Tables** (externas)
   - Datos en ubicación externa (S3, Azure Blob)
   - Al borrar tabla, datos persisten
   - Útil para datos compartidos

**Formatos soportados:**

* Delta Lake (recomendado) - transacciones ACID, time travel
* Parquet - columnar, comprimido
* CSV - simple, poco eficiente
* JSON - semi-estructurado

## 🐙 6. Integración con GitHub

---

### 📋 Requisitos

1. **Tener una cuenta de GitHub** (gratis): https://github.com/signup
2. **Tener un repositorio creado** en GitHub
3. **Generar un Personal Access Token (PAT)** de GitHub

---

### 🔑 Paso 1: Crear Personal Access Token (PAT)

**En GitHub:**

1. Ve a: **Settings → Developer settings → Personal Access Tokens → Tokens (classic)**
2. Click **"Generate new token (classic)"**
3. **Note:** "Databricks Access"
4. **Expiration:** 90 days (o más)
5. **Scopes** (permisos):
   * ✅ `repo` (acceso completo a repositorios)
   * ✅ `workflow` (si usas GitHub Actions)
6. Click **"Generate Token"**
7. **⚠️ IMPORTANTE:** Copia el token y guárdalo (solo se muestra una vez)

**Formato del token:** `ghp_xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx`

---

### 🔗 Paso 2: Conectar GitHub con Databricks

**Método 1: Databricks Repos (Recomendado) o Carpetas GIT**

1. En Databricks, panel izquierdo → **Repos**
2. Click **"Add Repo"**
3. **Git repository URL:** `https://github.com/tu-usuario/tu-repo.git`
4. **Git provider:** GitHub (Generalmente se completa solo, cuando se pega la URL)
5. **Git credential:** Paste tu Personal Access Token
6. Click **"Create"**

**Método 2: Clonar repositorio existente de GitHub**

1. Panel **Repos** → **"Add Repo"** o tambien se puede encontrar como **Carpeta GIT**
2. Pega URL del repositorio público de GitHub

## 🔄 7. Git Operations - Flujo de trabajo

### 🔄 Flujo de trabajo para las carpetas GIT

```text
1. PULL    → Traer cambios del repo remoto
2. EDIT    → Modificar notebooks
3. COMMIT  → Guardar cambios localmente
4. PUSH    → Enviar cambios al repo remoto
```

### 📏 Operaciones básicas

#### 1️⃣ Pull (Actualizar desde el repositorio en GitHub)

**Cuándo:** Antes de empezar a trabajar

**Cómo:**

1. En Databricks Repos, click en el nombre del repo
2. Click **"Pull"** (botón con flecha hacia abajo)
3. Confirma que estás en la rama correcta

#### 2️⃣ Commit (Guardar cambios)

**Cuándo:** Después de hacer cambios significativos

**Cómo:**

1. Click en el nombre del repo
2. Click **"Git..."** → **"Commit"**
3. Escribe un mensaje descriptivo:
   * ✅ Bien: "Agregar análisis de ventas por región"
   * ❌ Mal: "update", "fix", "cambios"
4. Click **"Commit & Push"**

#### 3️⃣ Push (Enviar a GitHub)

**Cuándo:** Después de commit, para sincronizar

**Cómo:**

* Si usaste "Commit & Push", ya está hecho
* Si no, click **"Git..."** → **"Push"**

#### 4️⃣ Branching (Ramas)

**Para qué:** Trabajar en features sin afectar `main`

**Crear rama:**

1. **"Git..."** → **"Create Branch"**
2. Nombre: `feature/analisis-ventas`
3. Work on that branch
4. Merge to main cuando esté listo

### ⚠️ Conflictos de Merge

**Pasan cuando:** Dos personas editan la misma línea

**Resolución:**

1. Databricks te avisa del conflicto
2. Abre el archivo con conflicto
3. Busca marcadores:

   ```text
   <<<<<<< HEAD
   Tu versión
   =======
   Versión remota
   >>>>>>> origin/main
   ```

4. Edita manualmente para resolver
5. Commit de la resolución

### 💡 Mejores Prácticas en Git

✅ Commit frecuentemente (cada feature pequeño).  
✅ Mensajes de commit descriptivos.  
✅ Pull antes de empezar a trabajar.  
✅ Usa branches para features grandes.  
✅ Haz code review antes de merge a main.  
❌ NO hagas commit de datos sensibles (contraseñas, API keys).  
❌ NO hagas commit de archivos grandes (>100 MB).

````python
# 🧪 9. Script de verificación completa del entorno

import sys
import os
from datetime import datetime

print("🔍 VERIFICACIÓN COMPLETA DEL ENTORNO")
print("=" * 70)

# Test 1: Python
print("\n✅ TEST 1: Python")
print(f"   Versión: {sys.version.split()[0]}")
print(f"   Executable: {sys.executable}")
assert sys.version_info >= (3, 9), "Python debe ser 3.9+"
print("   ✓ Python OK")
````

````python
# Test 2: Bibliotecas Core
print("\n✅ TEST 2: Bibliotecas Esenciales")
try:
    import pandas as pd
    import numpy as np
    from pyspark.sql import SparkSession
    print(f"   Pandas: {pd.__version__}")
    print(f"   NumPy: {np.__version__}")
    print("   PySpark: Disponible")
    print("   ✓ Bibliotecas OK")
except ImportError as e:
    print(f"   ✗ Error: {e}")
````

````python
# Test 3: Spark Session
print("\n✅ TEST 3: Spark Session")
try:
    df_test = spark.range(10).selectExpr("id", "id * 2 as doble")
    count = df_test.count()
    print(f"   Spark activo: {spark.version}")
    print(f"   Test DataFrame: {count} filas")
    print("   ✓ Spark OK")
except Exception as e:
    print(f"   ✗ Error: {e}")
````

````python
# Test 4: SQL
print("\n✅ TEST 4: SQL")
try:
    result = spark.sql("SELECT 1+1 as resultado").collect()[0][0]
    assert result == 2
    print("   SQL ejecutado exitosamente")
    print("   ✓ SQL OK")
except Exception as e:
    print(f"   ✗ Error: {e}")
````
