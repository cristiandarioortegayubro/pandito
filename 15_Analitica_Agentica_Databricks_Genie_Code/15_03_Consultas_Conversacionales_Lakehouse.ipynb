### 🔍 Ejemplos Prácticos

#### Ejemplo 1: Ranking de Sucursales

```
Pregunta: "¿Cuáles son las 5 sucursales con mayores ventas en 2024?"

Genie Code genera:
SELECT 
    sucursal_nombre,
    SUM(ventas) as ventas_totales
FROM ventas_mensuales_mendoza_h3
WHERE YEAR(fecha) = 2024
GROUP BY sucursal_nombre
ORDER BY ventas_totales DESC
LIMIT 5

Resultado:
1. Sucursal Centro    $850K
2. Sucursal Norte     $720K
3. Sucursal Plaza     $680K
4. Sucursal Oeste     $640K
5. Sucursal Este      $590K
```

---

#### Ejemplo 2: Comparación Temporal

```
Pregunta: "Compara ventas de Q1 2024 vs Q1 2023"

Genie Code genera:
WITH ventas_q1 AS (
  SELECT 
    YEAR(fecha) as año,
    QUARTER(fecha) as trimestre,
    SUM(ventas) as ventas_totales
  FROM ventas_mensuales_mendoza_h3
  WHERE QUARTER(fecha) = 1
    AND YEAR(fecha) IN (2023, 2024)
  GROUP BY YEAR(fecha), QUARTER(fecha)
)
SELECT * FROM ventas_q1
ORDER BY año

Resultado:
año | trimestre | ventas_totales
2023 | 1         | $1.2M
2024 | 1         | $1.5M  (+25%)
```

---

#### Ejemplo 3: Análisis de Anomalías

```
Pregunta: "¿Qué días tuvieron ventas atípicamente bajas?"

Genie Code genera:
WITH stats AS (
  SELECT 
    AVG(ventas) as promedio,
    STDDEV(ventas) as desv_std
  FROM ventas_mensuales_mendoza_h3
)
SELECT 
    fecha,
    ventas,
    (ventas - s.promedio) / s.desv_std as z_score
FROM ventas_mensuales_mendoza_h3, stats s
WHERE (ventas - s.promedio) / s.desv_std < -2
ORDER BY z_score

Resultado:
fecha      | ventas | z_score
2024-01-15 | $8K    | -2.8
2024-03-22 | $9K    | -2.5
```
