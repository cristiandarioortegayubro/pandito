## 📚 Analítica Agéntica: El Futuro del Análisis de Datos

### 🤖 ¿Qué es Analítica Agéntica?

**Analítica Agéntica:** Uso de agentes de IA para automatizar análisis de datos.

**Componentes:**
```
Pregunta en Lenguaje Natural
   ↓
Agente IA (Genie Code / Assistant)
   ↓
Generación de Código (PySpark/SQL)
   ↓
Ejecución Automática
   ↓
Resultado + Visualización
```

---

### 🧞 Databricks Genie Code

**Genie Code:** Asistente conversacional para análisis de datos.

**Capacidades:**
* ✅ **Lenguaje Natural:** "Muéstrame ventas por mes"
* ✅ **Generación SQL/PySpark:** Código automático
* ✅ **Visualizaciones:** Gráficos automáticos
* ✅ **Iterativo:** Refinar preguntas y respuestas

**Ejemplo:**
```
Usuario: "¿Cuál fue la sucursal con mayores ventas en 2024?"

Genie Code genera:
SELECT 
    sucursal_nombre,
    SUM(ventas) as ventas_totales
FROM ventas_mensuales_mendoza_h3
WHERE YEAR(fecha) = 2024
GROUP BY sucursal_nombre
ORDER BY ventas_totales DESC
LIMIT 1

Resultado:
sucursal_nombre | ventas_totales
Sucursal Centro | 2,450,000
```

---

### 💬 Databricks Assistant

**Assistant:** Ayudante IA integrado en notebooks.

**Funciones:**

**1️⃣ Generar código**
```
Prompt: "Crea un DataFrame con ventas filtradas por zona Centro"

Assistant genera:
df_centro = df.filter(df['zona'] == 'Centro')
```

**2️⃣ Explicar código**
```
Código: df.groupBy('zona').agg(F.sum('ventas'))

Assistant explica:
"Este código agrupa el DataFrame por 'zona' y suma las ventas 
de cada grupo usando la función de agregación sum()."
```

**3️⃣ Debugging**
```
Error: AttributeError: 'DataFrame' object has no attribute 'summari'

Assistant sugiere:
"Typo detectado: 'summari' → 'summary'. 
Prueba: df.summary()"
```

**4️⃣ Optimización**
```
Código lento: df.groupBy('zona').count().collect()

Assistant optimiza:
df.groupBy('zona').count().show()  # Más eficiente
```

---

### 🎯 Diferencia: Genie Code vs Assistant

| Característica | Genie Code | Assistant |
|----------------|------------|----------|
| **Interfaz** | Chat separado | Dentro del notebook |
| **Input** | Preguntas de negocio | Instrucciones de código |
| **Output** | SQL + resultados | Código Python/SQL |
| **Uso** | Análisis exploratorio | Desarrollo de código |
| **Target** | Analistas de negocio | Data engineers/scientists |

### 📝 Cómo Escribir Buenas Preguntas

**✅ BUENAS preguntas:**

```
1. "Muéstrame las ventas totales por zona en 2024"
   → Específica, con contexto temporal

2. "Compara las ventas de sucursal_id SUC001 vs SUC002"
   → Clara, con identificadores exactos

3. "Dame un gráfico de barras de ventas por mes"
   → Específica sobre visualización

4. "Calcula el ticket promedio por zona"
   → Clara sobre métrica deseada
```

**❌ MALAS preguntas:**

```
1. "Ventas"
   → Demasiado vaga

2. "Haz algo con los datos"
   → Sin objetivo claro

3. "Dame todo"
   → Inespecífica

4. "¿Cuál es mejor?"
   → Sin criterio de comparación
```
