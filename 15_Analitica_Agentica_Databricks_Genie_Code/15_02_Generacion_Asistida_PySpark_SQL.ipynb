
---

#### Ejemplo 2: Join complejo

**Prompt:**
```
"Join ventas con sucursales y muestra solo sucursales con > 1M en ventas"
```

**Assistant genera:**
```sql
SELECT 
    s.sucursal_nombre,
    s.zona,
    SUM(v.ventas) as ventas_totales
FROM ventas v
JOIN sucursales s ON v.sucursal_id = s.sucursal_id
GROUP BY s.sucursal_nombre, s.zona
HAVING SUM(v.ventas) > 1000000
ORDER BY ventas_totales DESC
```

---

#### Ejemplo 3: Window Functions SQL

**Prompt:**
```
"Calcula ventas acumuladas por sucursal ordenadas por fecha"
```

**Assistant genera:**
```sql
SELECT 
    sucursal_id,
    fecha,
    ventas,
    SUM(ventas) OVER (
        PARTITION BY sucursal_id 
        ORDER BY fecha
    ) as ventas_acumuladas
FROM ventas
ORDER BY sucursal_id, fecha
```

### ⚡ Optimización

#### Ejemplo: Código lento

**Código original:**
```python
# Lento: múltiples collect()
resultado1 = df.filter(...).collect()
resultado2 = df.filter(...).collect()
resultado3 = df.filter(...).collect()
```

**Prompt a Assistant:**
```
"Optimiza este código"
```

**Assistant sugiere:**
```python
# Rápido: cache una vez, filtrar después
df_cached = df.cache()

resultado1 = df_cached.filter(...).collect()
resultado2 = df_cached.filter(...).collect()
resultado3 = df_cached.filter(...).collect()

df_cached.unpersist()  # Liberar memoria
```

**Prompt:**

```
def calcular_kpis(df, zona):
```

**Assistant genera:**
```python
def calcular_kpis(df, zona):
    """
    Calcula KPIs de ventas por sucursal para una zona específica.
    
    Parámetros:
    -----------
    df : pyspark.sql.DataFrame
        DataFrame con columnas 'zona', 'sucursal_id', 'ventas'
    zona : str
        Nombre de la zona a filtrar (ej: 'Centro', 'Norte')
    
    Retorna:
    --------
    pyspark.sql.DataFrame
        DataFrame con columnas:
        - sucursal_id: ID de sucursal
        - total: Suma de ventas
        - promedio: Promedio de ventas
    
    Ejemplo:
    --------
    >>> kpis_centro = calcular_kpis(df, 'Centro')
    >>> kpis_centro.show()
    """
    return df.filter(df['zona'] == zona) \
             .groupBy('sucursal_id') \
             .agg(
                 F.sum('ventas').alias('total'),
                 F.avg('ventas').alias('promedio')
             )
```
