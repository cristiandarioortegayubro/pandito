## 📚 Los 3 Gráficos Fundamentales con Plotly Express

### 📈 1. Gráficos de Líneas (Line Charts)

**Propósito:** Mostrar evolución temporal o tendencias.

**Cuándo usar:**
* Series de tiempo (ventas mensuales, precio de acciones)
* Tendencias y patrones temporales
* Comparar múltiples series en el tiempo

**Sintaxis:**
```python
import plotly.express as px
fig = px.line(df, x='fecha', y='ventas', color='sucursal')
fig.show()
```

**Ventajas interactivas:**
* 🔍 Zoom en períodos específicos
* 📌 Hover para ver valores exactos
* 🎯 Comparar múltiples series fácilmente

---

### 📊 2. Gráficos de Barras (Bar Charts)

**Propósito:** Comparar valores entre categorías.

**Cuándo usar:**
* Comparación entre regiones, productos, vendedores
* Rankings (top 10, bottom 5)
* Composición por categorías

**Sintaxis:**
```python
fig = px.bar(df, x='sucursal', y='ventas', color='zona')
fig.show()
```

**Tipos:**
* **Vertical:** Categorías en eje X (default)
* **Horizontal:** `orientation='h'` (mejor para muchas categorías)
* **Apiladas:** `barmode='stack'` (mostrar composición)
* **Agrupadas:** `barmode='group'` (comparar subcategorías)

---

### 🔵 3. Scatter Plots (Gráficos de Dispersión)

**Propósito:** Mostrar relación entre dos variables continuas.

**Cuándo usar:**
* Detectar correlaciones
* Identificar outliers
* Análisis exploratorio de datos
* Visualización geoespacial (lat, lon)

**Sintaxis:**
```python
fig = px.scatter(df, x='lat', y='lon', 
                 size='ventas',      # Tamaño de puntos
                 color='zona',       # Color por categoría
                 hover_name='sucursal')  # Tooltip
fig.show()
```

**Parámetros clave:**
* `size`: Tamaño de puntos (proporcional a una variable)
* `color`: Color por categoría o valor continuo
* `symbol`: Forma de puntos por categoría
* `hover_data`: Información adicional en tooltip

---

### 🎨 Personalización Básica

**Títulos y etiquetas:**
```python
fig.update_layout(
    title='Ventas por Sucursal',
    xaxis_title='Mes',
    yaxis_title='Ventas ($)',
    template='plotly_white'  # Tema
)
```

**Temas disponibles:**
* `'plotly'` (default)
* `'plotly_white'` (fondo blanco)
* `'plotly_dark'` (modo oscuro)
* `'simple_white'` (minimalista)

## 📈 Los 3 gráficos fundamentales con datos reales

### 🎯 Aplicando px.line, px.bar y px.scatter

Con los datos de **Los Andes Market** vamos a crear los 3 gráficos fundamentales:

1. **Líneas (px.line):** Evolución mensual de ventas por zona
2. **Barras (px.bar):** Comparación de ventas por sucursal y zona (agrupadas y apiladas)
3. **Scatter (px.scatter):** Relación entre lat/lon y ventas (geoespacial)

---

### 📊 Datos preparados para visualización

```python
# Para líneas: ventas por fecha y zona
ventas_zona = df_ventas.groupby(['fecha', 'zona'])['ventas'].sum().reset_index()

# Para barras: ventas totales por sucursal, coloreadas por zona
ventas_sucursal = df_ventas.groupby(['sucursal_nombre', 'zona'])['ventas'].sum().reset_index()

# Para scatter: una fila por sucursal con lat, lon y ventas totales
sucursales_geo = df_ventas.groupby(['sucursal_nombre', 'lat', 'lon', 'zona'])['ventas'].sum().reset_index()
```

---

### 💡 Personalización aplicada
* `template='plotly_white'` → fondo limpio
* `color_continuous_scale='Blues'` → escala profesional
* `barmode='group'` vs `'stack'` → comparar o componer
* `size='ventas'` en scatter → tamaño proporcional a ventas
* `hover_name='sucursal_nombre'` → tooltip informativo

````python
import pandas as pd
import plotly.express as px

print("📈 LOS 3 GRÁFICOS FUNDAMENTALES CON DATOS REALES")
print("="*70)

if USAR_DATOS_REALES and df_ventas is not None:
    print("\n1️⃣  LÍNEAS: Evolución de ventas por zona")
    print("-"*70)

    ventas_zona = df_ventas.groupby(['fecha', 'zona'])['ventas'].sum().reset_index()
    fig_line = px.line(
        ventas_zona, x='fecha', y='ventas', color='zona',
        title='📈 Evolución de Ventas por Zona Comercial',
        labels={'ventas': 'Ventas ($)', 'fecha': 'Fecha', 'zona': 'Zona'},
        template='plotly_white'
    )
    fig_line.show()

    print("\n" + "="*70)
    print("\n2️⃣  BARRAS AGRUPADAS: Ventas por sucursal y zona")
    print("-"*70)

    ventas_sucursal = df_ventas.groupby(['sucursal_nombre', 'zona'])['ventas'].sum().reset_index()
    fig_bar_group = px.bar(
        ventas_sucursal, x='sucursal_nombre', y='ventas', color='zona',
        barmode='group',
        title='🏪 Ventas por Sucursal y Zona (Agrupadas)',
        labels={'ventas': 'Ventas ($)', 'sucursal_nombre': 'Sucursal'},
        template='plotly_white'
    )
    fig_bar_group.update_layout(xaxis_tickangle=-45)
    fig_bar_group.show()

    print("\n" + "="*70)
    print("\n3️⃣  BARRAS APILADAS: Composición por zona")
    print("-"*70)

    fig_bar_stack = px.bar(
        ventas_sucursal, x='sucursal_nombre', y='ventas', color='zona',
        barmode='stack',
        title='🏪 Ventas por Sucursal (Apiladas por Zona)',
        labels={'ventas': 'Ventas ($)', 'sucursal_nombre': 'Sucursal'},
        template='plotly_white'
    )
    fig_bar_stack.update_layout(xaxis_tickangle=-45)
    fig_bar_stack.show()

    print("\n" + "="*70)
    print("\n4️⃣  SCATTER GEOSPACIAL: Ventas por ubicación")
    print("-"*70)

    sucursales_geo = df_ventas.groupby(['sucursal_nombre', 'lat', 'lon', 'zona'])['ventas'].sum().reset_index()
    fig_scatter = px.scatter(
        sucursales_geo, x='lon', y='lat',
        size='ventas', color='zona',
        hover_name='sucursal_nombre',
        title='🗺️ Distribución Geográfica de Sucursales',
        labels={'lat': 'Latitud', 'lon': 'Longitud', 'zona': 'Zona', 'ventas': 'Ventas ($)'},
        template='plotly_white', size_max=30
    )
    fig_scatter.show()

    print("\n" + "="*70)
    print("\n5️⃣  SCATTER TEMPORAL: Ventas por año y mes")
    print("-"*70)

    ventas_mes = df_ventas.groupby(['año', 'mes', 'zona'])['ventas'].sum().reset_index()
    fig_scatter_time = px.scatter(
        ventas_mes, x='mes', y='ventas', color='zona',
        size='ventas',
        title='📅 Ventas por Mes y Zona (todos los años)',
        labels={'mes': 'Mes', 'ventas': 'Ventas ($)', 'zona': 'Zona'},
        template='plotly_white', size_max=20,
        hover_data=['año']
    )
    fig_scatter_time.show()
    print("\n   💡 El scatter temporal revela estacionalidad y patrones mensuales")
else:
    print("⚠️  No hay datos reales disponibles")

print("\n" + "="*70)
````

### 🎯 Reglas de Oro

👉 **Regla #1: Elegir el gráfico según la pregunta**
```python
# ¿Cómo evoluciona en el tiempo? → Líneas
fig = px.line(df, x='fecha', y='ventas', color='sucursal')

# ¿Cuál es mayor/menor? → Barras
fig = px.bar(df, x='sucursal', y='ventas', color='zona')

# ¿Existe correlación? → Scatter
fig = px.scatter(df, x='lat', y='lon', size='ventas', color='zona')
```

👉 **Regla #2: Siempre especificar title y labels**
```python
# MALO: gráfico sin contexto
fig = px.bar(df, x='sucursal', y='ventas')

# BUENO: título y etiquetas descriptivas
fig = px.bar(df, x='sucursal', y='ventas',
             title='Ventas por Sucursal — Los Andes Market',
             labels={'ventas': 'Ventas ($)', 'sucursal': 'Sucursal'})
fig.show()
```

👉 **Regla #3: barmode según objetivo**
```python
# Comparar subcategorías lado a lado
fig = px.bar(df, x='zona', y='ventas', color='sucursal', barmode='group')

# Mostrar composición total
fig = px.bar(df, x='zona', y='ventas', color='sucursal', barmode='stack')
```

---

### 📊 Guía de Decisión

| Situación | Método |
|-----------|--------|
| Evolución temporal de una serie | `px.line(df, x='fecha', y='val')` |
| Comparar múltiples series en el tiempo | `px.line(df, x='fecha', y='val', color='serie')` |
| Comparar valores entre categorías | `px.bar(df, x='cat', y='val')` |
| Composición por subcategorías | `px.bar(df, x='cat', y='val', color='sub', barmode='stack')` |
| Muchas categorías legibles | `px.bar(df, x='val', y='cat', orientation='h')` |
| Relación entre dos variables | `px.scatter(df, x='var1', y='var2')` |
| Tamaño proporcional a una variable | `px.scatter(df, x='x', y='y', size='val')` |
| Datos geoespaciales (lat/lon) | `px.scatter(df, x='lon', y='lat', size='val')` |
| Personalizar tema del gráfico | `fig.update_layout(template='plotly_white')` |
