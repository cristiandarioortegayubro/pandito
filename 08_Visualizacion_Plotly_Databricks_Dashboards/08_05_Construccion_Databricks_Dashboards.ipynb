````python
import pandas as pd
import numpy as np

print("💾 CARGANDO DATOS REALES DESDE UNITY CATALOG")
print("="*70)

CATALOG = "pandito_ds"
SCHEMA = "default"

try:
    # Cargar tabla de ventas de Los Andes Market
    df_ventas = spark.table(f"{CATALOG}.{SCHEMA}.ventas_mensuales_mendoza_h3").toPandas()
    df_ventas['fecha'] = pd.to_datetime(df_ventas['fecha'])
    df_ventas['año'] = df_ventas['fecha'].dt.year
    df_ventas['mes'] = df_ventas['fecha'].dt.month
    df_ventas['año_mes'] = df_ventas['fecha'].dt.to_period('M').astype(str)
    
    # Agregar métricas clave para dashboard
    df_ventas['trimestre'] = df_ventas['fecha'].dt.quarter
    
    print(f"\n✅ Datos reales cargados exitosamente")
    print(f"   📊 Registros: {len(df_ventas):,}")
    print(f"   📅 Período: {df_ventas['fecha'].min().strftime('%Y-%m-%d')} a {df_ventas['fecha'].max().strftime('%Y-%m-%d')}")
    print(f"   🏪 Sucursales: {df_ventas['sucursal_id'].nunique()}")
    print(f"   📍 Ubicación: Mendoza, Argentina")
    
    print(f"\n📈 Métricas disponibles para dashboard:")
    print(f"   • Ventas totales: ${df_ventas['ventas'].sum():,.2f}")
    print(f"   • Ticket promedio: ${df_ventas['ventas'].mean():,.2f}")
    print(f"   • Dimensiones: año, mes, trimestre, zona, sucursal")
    
    print(f"\n🎯 Este notebook creará un dashboard con datos REALES")
    print(f"   Listo para publicar en Databricks Dashboards")
    
    USAR_DATOS_REALES = True
    
except Exception as e:
    print(f"\n⚠️  No se pudo cargar la tabla de Unity Catalog")
    print(f"   Error: {e}")
    print(f"\n📝 Solución:")
    print(f"   1. Ejecuta primero: 00_05_Preparacion_Datos_Empresariales.ipynb")
    print(f"   2. Verifica que la tabla exista: {CATALOG}.{SCHEMA}.ventas_mensuales_mendoza_h3")
    print(f"\n   Continuando con datos sintéticos...")
    
    df_ventas = None
    USAR_DATOS_REALES = False

print("\n" + "="*70)
````

## 📚 Databricks Dashboards: Visualizaciones en Producción

### 🌐 ¿Qué son Databricks Dashboards?

**Databricks Dashboards** es la plataforma nativa de Databricks para publicar visualizaciones interactivas basadas en notebooks.

**Flujo:**
```
Notebook + Visualizaciones
          ↓
    Publicar Dashboard
          ↓
URL pública compartible
```

---

### 🔄 Flujo de Trabajo

#### 1️⃣ **Desarrollo en Notebook**
* Escribir código de análisis
* Crear visualizaciones (Plotly, matplotlib, tablas)
* Probar y validar

#### 2️⃣ **Publicación**
* Click en "Publicar" → "Dashboard"
* Seleccionar celdas a incluir
* Configurar layout

#### 3️⃣ **Compartir**
* Generar URL pública
* Configurar permisos
* Compartir con stakeholders

#### 4️⃣ **Automatización**
* Programar actualizaciones (diarias, semanales)
* Notificaciones automáticas

---

### 🎨 Elementos de un Dashboard

**Componentes:**

1. **Visualizaciones:**
   * Gráficos Plotly/matplotlib
   * Tablas
   * Métricas (KPI cards)

2. **Parámetros (Widgets):**
   * Filtros dinámicos
   * Dropdowns
   * Date pickers

3. **Texto y Markdown:**
   * Títulos
   * Descripciones
   * Insights

## 🌐 Dashboard ejecutivo con datos reales de Los Andes Market

### 🎯 Componentes del dashboard

Un dashboard ejecutivo completo necesita:

1. **KPI cards** → Métricas clave (ventas totales, promedio, growth)
2. **Gráfico de líneas** → Evolución temporal
3. **Gráfico de barras** → Ranking de sucursales
4. **Tabla resumen** → Detalle por zona

---

### 🛠️ Widgets para filtros dinámicos

```python
# Filtro por zona

dbutils.widgets.dropdown('año', 'Todos', ['Todos'] + sorted(df_ventas['año'].unique().tolist()))

```

### 💡 Flujo de publicación

1. Ejecutar celdas con visualizaciones
2. Click en **Publicar** → **Dashboard**
3. Arrastrar celdas al layout
4. Compartir URL con stakeholders
5. Programar actualizaciones automáticas

````python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

print("🌐 DASHBOARD EJECUTIVO CON DATOS REALES DE LOS ANDES MARKET")
print("="*70)

if USAR_DATOS_REALES and df_ventas is not None:
    # --- KPI CARDS ---
    print("\n1️⃣  KPI CARDS: Métricas clave")
    print("-"*70)

    total_ventas = df_ventas['ventas'].sum()
    promedio_mensual = df_ventas.groupby('fecha')['ventas'].sum().mean()
    num_sucursales = df_ventas['sucursal_id'].nunique()
    num_zonas = df_ventas['zona'].nunique()

    print(f"   💰 Ventas Totales:   ${total_ventas:>14,.0f}")
    print(f"   📊 Promedio Mensual: ${promedio_mensual:>14,.0f}")
    print(f"   🏪 Sucursales:       {num_sucursales:>14}")
    print(f"   📍 Zonas:            {num_zonas:>14}")

    print("\n" + "="*70)
    print("\n2️⃣  GRÁFICO DE LÍNEAS: Evolución temporal")
    print("-"*70)

    ventas_mes = df_ventas.groupby('fecha')['ventas'].sum().reset_index()
    fig_line = px.line(ventas_mes, x='fecha', y='ventas',
                       title='📈 Evolución de Ventas Totales',
                       template='plotly_white', labels={'ventas': 'Ventas ($)'})
    fig_line.update_traces(line_color='#2563eb', line_width=2)
    fig_line.show()

    print("\n" + "="*70)
    print("\n3️⃣  GRÁFICO DE BARRAS: Ranking de sucursales")
    print("-"*70)

    ventas_suc = df_ventas.groupby(['sucursal_nombre', 'zona'])['ventas'].sum().reset_index()
    ventas_suc = ventas_suc.sort_values('ventas', ascending=False)
    fig_bar = px.bar(ventas_suc, x='sucursal_nombre', y='ventas', color='zona',
                     title='🏪 Ranking de Sucursales por Ventas',
                     template='plotly_white', labels={'ventas': 'Ventas ($)'})
    fig_bar.update_layout(xaxis_tickangle=-45)
    fig_bar.show()

    print("\n" + "="*70)
    print("\n4️⃣  TABLA RESUMEN: Ventas por zona y año")
    print("-"*70)

    resumen = df_ventas.pivot_table(
        values='ventas', index='zona',
        columns='año', aggfunc='sum', fill_value=0
    )
    resumen['TOTAL'] = resumen.sum(axis=1)
    resumen = resumen.sort_values('TOTAL', ascending=False)
    display(resumen.round(0))

    print("\n" + "="*70)
    print("\n5️⃣  DASHBOARD COMBINADO (subplots)")
    print("-"*70)

    fig_combo = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Evolución Mensual', 'Ventas por Zona',
                        'Top 5 Sucursales', 'Distribución por Zona'),
        specs=[[{'type': 'scatter'}, {'type': 'bar'}],
               [{'type': 'bar'}, {'type': 'pie'}]]
    )

    # Líneas: evolución
    fig_combo.add_trace(go.Scatter(x=ventas_mes['fecha'], y=ventas_mes['ventas'],
                                    name='Ventas', line=dict(color='#2563eb', width=2)),
                        row=1, col=1)

    # Barras: por zona
    ventas_zona = df_ventas.groupby('zona')['ventas'].sum().reset_index()
    fig_combo.add_trace(go.Bar(x=ventas_zona['zona'], y=ventas_zona['ventas'],
                               name='Por Zona', marker_color='#60a5fa'),
                        row=1, col=2)

    # Top 5 sucursales
    top5 = ventas_suc.head(5)
    fig_combo.add_trace(go.Bar(x=top5['sucursal_nombre'], y=top5['ventas'],
                               name='Top 5', marker_color='#10b981'),
                        row=2, col=1)

    # Pie: distribución
    fig_combo.add_trace(go.Pie(labels=ventas_zona['zona'], values=ventas_zona['ventas'],
                                name='Distribución'),
                        row=2, col=2)

    fig_combo.update_layout(
        title='📊 Dashboard Ejecutivo - Los Andes Market',
        template='plotly_white', height=700, showlegend=False
    )
    fig_combo.show()
    print("\n   ✅ Dashboard combinado listo para publicar")
else:
    print("⚠️  No hay datos reales disponibles")

print("\n" + "="*70)
````
