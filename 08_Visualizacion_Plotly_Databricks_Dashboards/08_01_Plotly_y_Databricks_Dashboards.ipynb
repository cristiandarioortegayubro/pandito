## 📚 Plotly: Visualización Interactiva

### 📊 ¿Qué es Plotly?

**Plotly** es una biblioteca de visualización **interactiva** y **ejecutiva** para Python.

**Creada por:** Plotly Technologies (2012)  
**Licencia:** Open Source (MIT) + Cloud Platform

---

### 🆚 Plotly vs Matplotlib

| Característica | Matplotlib | Plotly |
|----------------|------------|--------|
| **Interactividad** | ❌ Estática | ✅ Interactiva (zoom, hover, pan) |
| **Estética** | 🟡 Básica | ✅ Moderna, profesional |
| **Curva aprendizaje** | 🔴 Compleja | ✅ Fácil (Plotly Express) |
| **Web-ready** | ❌ Requiere conversión | ✅ HTML nativo |
| **Dashboards** | ❌ No nativo | ✅ Dash framework |
| **Uso típico** | Papers académicos | Presentaciones ejecutivas |

---

### 🎯 Cuándo usar cada una

#### **Matplotlib** 👍
* Papers académicos
* Gráficos estáticos para reportes PDF
* Control total a bajo nivel
* Integración con publicaciones científicas

#### **Plotly** 👍
* Dashboards ejecutivos
* Presentaciones interactivas
* Publicación web (HTML)
* Exploración de datos
* Reportes para stakeholders

---

### 🔧 Dos Formas de Usar Plotly

#### 1️⃣ **Plotly Express** (Alto nivel - Recomendado)

```python
import plotly.express as px
fig = px.line(df, x='fecha', y='ventas', color='sucursal')
fig.show()
```

✅ **Ventajas:** Código conciso, rápido, ideal para exploración

---

#### 2️⃣ **Plotly Graph Objects** (Bajo nivel - Control total)

```python
import plotly.graph_objects as go
fig = go.Figure()
fig.add_trace(go.Scatter(x=df['fecha'], y=df['ventas'], mode='lines'))
fig.update_layout(title='Ventas')
fig.show()
```

✅ **Ventajas:** Control total, personalización avanzada

### 🛠️ Plotly Express vs Graph Objects con datos reales

```python
# Express: una línea para un gráfico completo
fig = px.line(df_ventas_agg, x='fecha', y='ventas', color='sucursal_nombre')

# Graph Objects: control total para dashboards

    datos = df_ventas[df_ventas['sucursal_nombre'] == sucursal]

```

---

### 💡 Preguntas que los gráficos responden
* ¿Hay estacionalidad en las ventas? (líneas)
* ¿Qué sucursal vende más? (barras)
* ¿Las sucursales crecen al mismo ritmo? (líneas múltiples)

````python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

print("📊 PRIMEROS GRÁFICOS CON DATOS REALES DE LOS ANDES MARKET")
print("="*70)

if USAR_DATOS_REALES and df_ventas is not None:
    print("\n1️⃣  GRÁFICO DE LÍNEAS: Evolución temporal de ventas")
    print("-"*70)

    # Agregar ventas totales por mes
    ventas_mensuales = df_ventas.groupby(['fecha', 'año_mes'])['ventas'].sum().reset_index()
    ventas_mensuales = ventas_mensuales.sort_values('fecha')

    fig_line = px.line(
        ventas_mensuales,
        x='fecha',
        y='ventas',
        title='📈 Evolución de Ventas Totales - Los Andes Market',
        labels={'ventas': 'Ventas ($)', 'fecha': 'Fecha'},
        template='plotly_white'
    )
    fig_line.update_traces(line_color='#2563eb', line_width=2)
    fig_line.show()

    print("\n" + "="*70)
    print("\n2️⃣  GRÁFICO DE BARRAS: Ventas totales por sucursal")
    print("-"*70)

    ventas_sucursal = df_ventas.groupby('sucursal_nombre')['ventas'].sum().reset_index()
    ventas_sucursal = ventas_sucursal.sort_values('ventas', ascending=False)

    fig_bar = px.bar(
        ventas_sucursal,
        x='sucursal_nombre',
        y='ventas',
        title='🏪 Ventas Totales por Sucursal',
        labels={'ventas': 'Ventas ($)', 'sucursal_nombre': 'Sucursal'},
        color='ventas',
        color_continuous_scale='Blues',
        template='plotly_white'
    )
    fig_bar.update_layout(xaxis_tickangle=-45)
    fig_bar.show()

    print("\n" + "="*70)
    print("\n3️⃣  GRÁFICO DE LÍNEAS MÚLTIPLES: Ventas por sucursal (top 5)")
    print("-"*70)

    top5 = ventas_sucursal.head(5)['sucursal_nombre'].tolist()
    df_top5 = df_ventas[df_ventas['sucursal_nombre'].isin(top5)]
    ventas_top5 = df_top5.groupby(['fecha', 'sucursal_nombre'])['ventas'].sum().reset_index()

    fig_multi = px.line(
        ventas_top5,
        x='fecha',
        y='ventas',
        color='sucursal_nombre',
        title='📈 Top 5 Sucursales - Evolución de Ventas',
        labels={'ventas': 'Ventas ($)', 'fecha': 'Fecha', 'sucursal_nombre': 'Sucursal'},
        template='plotly_white'
    )
    fig_multi.show()

    print("\n" + "="*70)
    print("\n4️⃣  GRAPH OBJECTS: Control total con go.Figure()")
    print("-"*70)

    fig_go = go.Figure()
    for sucursal in top5[:3]:
        datos = ventas_top5[ventas_top5['sucursal_nombre'] == sucursal]
        fig_go.add_trace(go.Scatter(
            x=datos['fecha'],
            y=datos['ventas'],
            mode='lines+markers',
            name=sucursal,
            line=dict(width=2)
        ))
    fig_go.update_layout(
        title='Top 3 Sucursales (Graph Objects)',
        xaxis_title='Fecha',
        yaxis_title='Ventas ($)',
        template='plotly_white',
        hovermode='x unified'
    )
    fig_go.show()
    print("\n   💡 Graph Objects da control total: hover unificado, markers, etc.")
else:
    print("⚠️  No hay datos reales disponibles")

print("\n" + "="*70)
````

1. **Plotly Express (alto nivel):**
   - `px.line(df, x='fecha', y='ventas', color='sucursal')`

   - Matplotlib: estático, académico, control a bajo nivel

5. **Databricks Dashboards:**
   - Visualizaciones de notebooks → dashboards compartibles

   - Stakeholders pueden interactuar sin acceder al notebook

### 🎯 Reglas de Oro

👉 **Regla #1: Empezar con Plotly Express, refinear con Graph Objects**
```python
# MALO: escribir 20 líneas de Graph Objects para un gráfico simple
fig = go.Figure()
fig.add_trace(go.Scatter(x=df['fecha'], y=df['ventas']))
fig.update_layout(title='Ventas')
fig.show()

# BUENO: una línea de Plotly Express
fig = px.line(df, x='fecha', y='ventas', title='Ventas')
fig.show()
```

👉 **Regla #2: Siempre especificar title y labels**
```python
# MALO: gráfico sin contexto
fig = px.bar(df, x='sucursal', y='ventas')

# BUENO: título y etiquetas descriptivas
fig = px.bar(df, x='sucursal', y='ventas',
             title='Ventas por Sucursal — Los Andes Market',
             labels={'ventas': 'Ventas ($)', 'sucursal': 'Sucursal'})
fig.show()
```

👉 **Regla #3: Exportar como HTML para compartir fuera de Databricks**
```python
# Los stakeholders no siempre tienen acceso al notebook
fig.write_html('dashboard_ventas.html')
# Se abre en cualquier navegador, conservando interactividad
```

---

### 📊 Guía de Decisión

| Situación | Método |
|-----------|--------|
| Exploración rápida de datos | `px.line()`, `px.bar()`, `px.scatter()` |
| Dashboard ejecutivo complejo | `go.Figure()` con múltiples traces |
| Comparación entre categorías | `px.bar(df, x='cat', y='val', color='grupo')` |
| Evolución temporal | `px.line(df, x='fecha', y='val', color='serie')` |
| Relación entre variables | `px.scatter(df, x='var1', y='var2', size='val')` |
| Reporte académico / PDF | Matplotlib (estático) |
| Presentación a stakeholders | Plotly (interactivo) |
| Compartir fuera de Databricks | `fig.write_html('archivo.html')` |
| Publicar en Databricks Dashboard | Plotly en notebook → Dashboard nativo |
