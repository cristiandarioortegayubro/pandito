```python
import plotly.graph_objects as go

# Gráfico Waterfall para Estado de Resultados (P&L)
fig = go.Figure(go.Waterfall(
    name = "P&L", orientation = "v",
    measure = ["relative", "relative", "total", "relative", "total"],
    x = ["Ventas Brutas", "Costos Directos", "Margen Bruto", "Gastos Operativos", "EBITDA"],
    textposition = "outside",
    text = ["+$1,000,000", "-$600,000", "$400,000", "-$150,000", "$250,000"],
    y = [1000000, -600000, 0, -150000, 0],
    connector = {"line":{"color":"rgb(63, 63, 63)"}},
))

fig.update_layout(
    title = "Estado de Resultados P&L - Ejemplo Ejecutivo",
    showlegend = False,
    template = "plotly_white"
)

# En Jupyter / Colab / Databricks: fig.show()
print("Código de gráfico Waterfall de P&L generado correctamente.")

```
