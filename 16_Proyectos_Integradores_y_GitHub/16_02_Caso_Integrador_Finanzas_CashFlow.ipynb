## 📚 Análisis Financiero con Datos

### 💰 Cash Flow: El Rey

**Cash Flow** = Flujo de efectivo (entradas - salidas)

**Fórmula básica:**
```
Cash Flow = Ingresos - Gastos + Financiamiento - Inversiones
```

**Componentes:**
* **Operativo:** Negocio día a día
* **Inversión:** Compra de activos
* **Financiamiento:** Deuda/capital

---

### 📊 Proyección de Cash Flow

**Método: Promedio Móvil**

```python
# Promedio de últimos 3 meses
df['proyeccion'] = df['ingresos'].rolling(window=3).mean()
```

**Método: Tendencia Lineal**

```python
from sklearn.linear_model import LinearRegression

# Modelo simple
X = np.arange(len(df)).reshape(-1, 1)
y = df['ingresos'].values

model = LinearRegression().fit(X, y)
df['proyeccion'] = model.predict(X)
```

### 📈 Ratios Financieros Clave

**1️⃣ Liquidez Corriente**
```
Activo Corriente / Pasivo Corriente

> 1.0 = Buena liquidez
< 1.0 = Problemas de liquidez
```

**2️⃣ Prueba Ácida**
```
(Activo Corriente - Inventario) / Pasivo Corriente

> 1.0 = Excelente
```

**3️⃣ Días de Cash**
```
(Cash / Gastos Diarios Promedio)

Días que la empresa puede operar sin ingresos
```

**4️⃣ Burn Rate**
```
Gastos Mensuales - Ingresos Mensuales

Si negativo = "quemando" efectivo
```

---

### 🚨 Alertas de Liquidez

**Condiciones de alerta:**

```python
# Alerta CRÍTICA: Cash < 30 días de gastos
if dias_de_cash < 30:
    print("🔴 ALERTA CRÍTICA: Liquidez crítica")

# Alerta MEDIA: Cash < 60 días
elif dias_de_cash < 60:
    print("🟠 ALERTA MEDIA: Monitorear liquidez")

# OK: Cash > 60 días
else:
    print("🟢 LIQUIDEZ SALUDABLE")
```

### 📈 Proyecciones Dinámicas

**Escenarios:**

**Base Case:**
```python
ingresos_proyectados = ingresos_actuales * 1.0  # Sin cambio
```

**Best Case:**
```python
ingresos_proyectados = ingresos_actuales * 1.2  # +20%
```

**Worst Case:**
```python
ingresos_proyectados = ingresos_actuales * 0.8  # -20%
```

**Sensibilidad:**
```python
for crecimiento in [-0.2, -0.1, 0, 0.1, 0.2]:
    proyeccion = ingresos * (1 + crecimiento)
    print(f"Crecimiento {crecimiento:+.0%}: ${proyeccion:,.0f}")
```
